# Deep Learning Interview Questions

**Notebook Purpose:** Comprehensive interview preparation for Deep Learning topics.

**Organization:** Questions are grouped by category with difficulty tags.
- **[Easy]** -- Commonly asked, expected knowledge for any ML role
- **[Medium]** -- Requires deeper understanding, common in mid/senior interviews
- **[Hard]** -- Advanced topics, expected for senior/research roles

**Categories:**
1. Fundamentals & Architecture
2. Activation Functions
3. Training & Optimization
4. Regularization
5. Convolutional Neural Networks (CNNs)
6. Loss Functions
7. Advanced Topics

---

# 1. FUNDAMENTALS & ARCHITECTURE

---

### Q1. What is the role of weights and bias in a neural network? [Easy]

**30-second answer:** Weights determine the strength of connections between neurons (slope of the decision boundary), while biases allow the decision boundary to shift, giving the model flexibility to fit data that does not pass through the origin.

**Detailed explanation:**
- **Weights** control how much influence a given input has on the neuron's output. During training, weights are adjusted via backpropagation to minimize the loss.
- **Bias** acts as an offset, similar to the intercept in linear regression. It allows neurons to activate even when all inputs are zero.
- For a single neuron: `output = activation(w1*x1 + w2*x2 + ... + bias)`
- A single neuron with no non-linearity can be considered as a linear regression function.

---

### Q2. What is the meaning of weight initialization in neural networks? [Easy]

**30-second answer:** Weight initialization sets the starting values of network parameters before training begins. Good initialization (e.g., Xavier, He) prevents vanishing/exploding gradients and enables faster convergence, while bad initialization can prevent a network from learning entirely.

**Detailed explanation:**
- A bad weight initialization prevents a network from learning.
- A good weight initialization helps in giving quicker convergence and better overall error.
- Biases can be initialized to zero.
- The standard rule is to set weights close to zero without being too small.
- Common strategies: **Xavier/Glorot** (for sigmoid/tanh), **He initialization** (for ReLU).

---

### Q3. Why is zero initialization not a good weight initialization process? [Easy]

**30-second answer:** If all weights are zero, every neuron produces the same output and receives the same gradient during backpropagation, so the network cannot break symmetry and cannot learn different features.

**Detailed explanation:**
- All neurons at each layer produce the same output and the same gradients during backpropagation.
- The network cannot learn because there is no source of asymmetry between neurons -- they all update identically.
- We need randomness in weight initialization to break this symmetry.
- **Exception:** For a single-layer model (linear/logistic regression), the cost function is convex, so weights will always converge to the optimal point regardless of initial value (though convergence may be slower).

---

### Q4. What is a Computational Graph? [Easy]

**30-second answer:** A computational graph is a directed graph where nodes represent mathematical operations and edges represent tensors (data). Frameworks like TensorFlow use this "DataFlow Graph" to organize and optimize computations.

**Detailed explanation:**
- Everything in TensorFlow is based on creating a computational graph.
- Nodes represent operations (add, multiply, activation functions).
- Edges represent tensors flowing between operations.
- Enables automatic differentiation for backpropagation.
- Allows optimization like operator fusion and parallelization.

---

### Q5. What do you mean by Tensor in TensorFlow? [Easy]

**30-second answer:** A tensor is a multi-dimensional array that serves as the fundamental data structure in TensorFlow. Scalars are 0-D tensors, vectors are 1-D, matrices are 2-D, and higher-dimensional arrays are N-D tensors.

---

### Q6. What is an Auto-encoder? [Medium]

**30-second answer:** An autoencoder is a neural network trained to reconstruct its own input through a bottleneck layer, learning a compressed (latent) representation of the data in an unsupervised manner.

**Detailed explanation:**
- Has three parts: encoder, bottleneck (latent space), decoder.
- Input neurons equal output neurons -- the target output is the same as the input.
- Uses dimensionality reduction to learn efficient representations.
- Works by **compressing input to a latent space representation, then reconstructing the output from this representation**.
- Applications: denoising, anomaly detection, generative models (VAE).

---

### Q7. What is a dead unit in a neural network? [Medium]

**30-second answer:** A dead unit is a neuron that always outputs zero (or a constant) for all inputs and never updates during training because its gradient is perpetually zero, commonly seen with ReLU activations.

---

### Q8. Is a neural network a universal function approximator? [Medium]

**30-second answer:** Yes. The Universal Approximation Theorem states that a feedforward neural network with a single hidden layer containing a finite number of neurons can approximate any continuous function on compact subsets of R^n, given appropriate activation functions.

**Note:** The following are all universal approximators:
- Kernel SVM
- Neural Networks
- Boosted Decision Trees

---

### Q9. Why is it necessary to introduce non-linearities in a neural network? [Easy]

**30-second answer:** Without non-linear activation functions, a neural network of any depth would simply compute a linear transformation of its input, making it no more powerful than a single-layer linear model.

**Detailed explanation:**
- A composition of linear functions is still a linear function.
- A linear model has a much smaller number of effective parameters and is limited in the complexity it can model.
- Non-linearities allow the network to learn complex decision boundaries and approximate arbitrary functions.

---

### Q10. What are the disadvantages of deep learning? [Easy]

**30-second answer:** Deep learning requires large datasets, is computationally expensive (training can take days), lacks interpretability, and is prone to overfitting on small datasets.

**Key disadvantages:**
- Long training times -- can take days depending on model complexity.
- Not effective for small datasets -- needs large amounts of data to generalize well.
- Requires significant computational resources (GPUs/TPUs).
- Acts as a "black box" -- difficult to interpret.
- Hyperparameter tuning can be tedious.

---

### Q11. Suppose training error is high and validation error is almost equal to it. What does it mean? [Medium]

**30-second answer:** This indicates **underfitting** -- the model is too simple to capture the underlying patterns in the data.

**What to do:**
- Add more parameters / increase model complexity (more layers, more neurons).
- Reduce regularization strength.
- Train for more epochs.
- Use a more expressive architecture.

---

### Q12. In training a neural network, the loss does not decrease in the first few epochs. What could be the reason? [Medium]

**30-second answer:** Common causes are: learning rate is too low, regularization parameter is too high, or the optimization is stuck at a poor initialization / local minimum.

**Additional causes:**
- Wrong loss function for the task.
- Data preprocessing issues (e.g., not normalized).
- Bug in the data pipeline (labels mismatched).

# 2. ACTIVATION FUNCTIONS

---

### Q13. What is the use of the activation function? [Easy]

**30-second answer:** Activation functions introduce non-linearity into neural networks, enabling them to learn complex, non-linear mappings from inputs to outputs. Without them, the network would only compute linear transformations regardless of depth.

**Detailed explanation:**
- The activation function decides whether a neuron should be activated by calculating the weighted sum and adding bias.
- Without activation functions, a neural network would only learn linear combinations of its input data.
- Networks that use the rectifier function (ReLU) for hidden layers are called "rectified networks."
- ReLU helps solve vanishing gradient problems compared to sigmoid/tanh.

---

### Q14. What are the pros and cons of ReLU? [Easy]

**30-second answer:** ReLU is computationally efficient and does not saturate in the positive region, leading to faster convergence. However, it suffers from the "dying ReLU" problem where neurons can permanently output zero.

**Pros:**
- Does not saturate in the positive region (no vanishing gradient for positive inputs).
- Computationally very efficient -- just a threshold at zero.
- Models with ReLU neurons converge much faster than with sigmoid/tanh.

**Cons:**
- **Dying ReLU problem:** If a neuron's activation becomes zero, its gradient is zero in backpropagation. A large gradient can cause weights to update such that the neuron never activates again -- the gradient flowing through it will be forever zero.
- Not zero-centered -- outputs are always non-negative.

---

### Q15. What is the dying ReLU problem and how to fix it? [Medium]

**30-second answer:** The dying ReLU problem occurs when neurons get stuck outputting zero for all inputs because large negative weight updates push the pre-activation permanently below zero. Fixes include using Leaky ReLU, ELU, or careful learning rate tuning and He initialization.

**Detailed explanation:**
- Once a ReLU neuron outputs zero, its gradient is zero, so it receives no updates -- it is "dead."
- A large gradient flowing through a ReLU neuron can cause weights to update such that the neuron never activates on any data point again.
- **Fixes:**
  - **Leaky ReLU:** f(x) = x if x > 0, else alpha * x (typically alpha = 0.01). Allows a small gradient for negative inputs.
  - **Parametric ReLU (PReLU):** The leak coefficient alpha is learned during training.
  - **ELU (Exponential Linear Unit):** Smooth transition for negative values, with mean activations closer to zero.
  - **GELU / SiLU / Swish:** Modern smooth alternatives used in transformers.
  - **Careful initialization:** Use He initialization with ReLU.
  - **Lower learning rate:** Prevents large gradient updates that kill neurons.

---

### Q16. What is the use of the Leaky ReLU function? [Easy]

**30-second answer:** Leaky ReLU allows small negative values when the input is less than zero (f(x) = alpha * x for x < 0, typically alpha = 0.01), preventing the dying ReLU problem by ensuring neurons always have a non-zero gradient.

---

### Q17. What is the softmax function? [Easy]

**30-second answer:** Softmax converts a vector of raw scores (logits) into a probability distribution where all values are between 0 and 1 and sum to 1, making it the standard output activation for multi-class classification.

**Detailed explanation:**
- Calculates the probability distribution of an event over 'n' different events.
- Output probabilities range from 0 to 1, and the sum of all probabilities equals 1.
- When used for multi-class classification, the target class will have the highest probability.
- Formula: softmax(z_i) = exp(z_i) / sum(exp(z_j)) for all j.

---

### Q18. What happens if we use ReLU activation and then sigmoid as the final layer? [Medium]

**30-second answer:** Since ReLU always outputs non-negative values, all inputs to the sigmoid layer will be >= 0, meaning sigmoid outputs will always be >= 0.5. The network will be biased toward predicting one class and may constantly predict that class for all inputs.

---

### Q19. Describe two ways of dealing with the vanishing gradient problem. [Medium]

**30-second answer:** (1) Use ReLU activation instead of sigmoid/tanh -- ReLU does not saturate for positive values. (2) Use proper weight initialization like Xavier (for sigmoid/tanh) or He (for ReLU) to keep gradients in a healthy range.

**Additional approaches:**
- Batch normalization to stabilize activations.
- Residual connections (skip connections) to provide gradient shortcuts.
- LSTM/GRU cells instead of vanilla RNN for sequential data.
- Gradient clipping to prevent exploding gradients (the related problem).

---

### Q20. Why was sigmoid replaced by ReLU as the preferred activation? [Easy]

**30-second answer:** Sigmoid saturates for large positive/negative inputs, causing vanishing gradients that make deep networks hard to train. ReLU does not saturate in the positive region and is computationally cheaper, enabling faster and more effective training.

**Detailed explanation:**
- Sigmoid was the most commonly used activation function until the saturation issue was identified.
- When gradients are too large in positive or negative direction, the resulting gradients from sigmoid get "squashed" -- this is called **saturation of the neuron**.
- ReLU (max(0, x)) provides constant gradient of 1 for positive inputs, enabling deeper networks to train effectively.

# 3. TRAINING & OPTIMIZATION

---

### Q21. Explain Stochastic, Batch, and Mini-batch Gradient Descent. [Easy]

**30-second answer:** Batch GD computes gradients over the entire dataset per update (stable but slow), Stochastic GD uses a single sample per update (fast but noisy), and Mini-batch GD uses small batches (best of both worlds -- efficient and reasonably stable).

**Detailed explanation:**

| Variant | Samples per Update | Pros | Cons |
|---|---|---|---|
| **Batch GD** | Entire dataset | Stable convergence, exact gradient | Slow for large datasets, memory-intensive |
| **Stochastic GD** | 1 sample | Fast updates, can escape local minima | Very noisy, high variance |
| **Mini-batch GD** | Small batch (32-512) | Computationally efficient, GPU-friendly | Requires batch size tuning |

---

### Q22. What are the main benefits of Mini-batch Gradient Descent? [Easy]

**30-second answer:** Mini-batch GD is computationally efficient (GPU parallelism), improves generalization by finding flat minima through gradient noise, and approximates the full gradient well enough to avoid local minima.

---

### Q23. Is it necessary to shuffle training data with Batch Gradient Descent? [Easy]

**30-second answer:** No, because the gradient is calculated using the entire training set at each epoch, so the order of samples does not affect the gradient computation.

---

### Q24. Why is shuffling important for Mini-batch Gradient Descent? [Medium]

**30-second answer:** Shuffling ensures that each mini-batch is a random, representative sample of the data, preventing the model from learning spurious patterns based on data ordering and improving convergence.

**Detailed explanation:**
- By shuffling rows and training on subsets, the mini-batch composition changes every iteration.
- It is possible that no two iterations across the entire training will use the exact same batch.
- Without shuffling, if data is sorted (e.g., by class), consecutive mini-batches would be homogeneous, causing biased gradient estimates.

---

### Q25. What is RMSprop? [Medium]

**30-second answer:** RMSprop (Root Mean Square Propagation) is an adaptive learning rate optimizer that divides the gradient by a running average of recent gradient magnitudes, preventing oscillations and enabling faster convergence on non-stationary objectives.

**Core formula:**

$$
E[g^2]_t = \gamma E[g^2]_{t-1} + (1 - \gamma) g_t^2
$$
$$
\theta_{t+1} = \theta_t - \frac{\eta}{\sqrt{E[g^2]_t + \epsilon}} \cdot g_t
$$

Where gamma (decay rate) is typically 0.9.

**Key insight:** It dampens the learning rate for parameters with frequent large gradients and keeps it higher for sparse updates.

**Why RMSprop was needed:**
- Standard SGD uses the same learning rate for all parameters -- can zigzag on steep slopes.
- Adagrad's learning rate keeps shrinking and eventually becomes too small.
- RMSprop fixes Adagrad's decay issue by using an exponential moving average.

**Best for:** RNNs, time-series data, deep networks with noisy gradients.

| Optimizer | Adaptive LR | Momentum | Best For |
|---|---|---|---|
| **SGD** | No | Optional | Large datasets, convex loss |
| **Adagrad** | Yes | No | Sparse features (NLP) |
| **RMSprop** | Yes (EMA) | No | RNNs, non-stationary objectives |
| **Adam** | Yes | Yes | General purpose |

---

### Q26. Explain the idea behind the Adam optimizer. [Medium]

**30-second answer:** Adam (Adaptive Moment Estimation) combines two key ideas: per-parameter adaptive learning rates (like RMSprop) and momentum (moving average of gradients), giving it fast convergence while avoiding saddle points.

**Detailed explanation:**
- Maintains two moving averages: first moment (mean of gradients) and second moment (mean of squared gradients).
- Includes bias correction for the initial steps when estimates are biased toward zero.
- Adam = RMSprop + Momentum + Bias Correction.
- Default hyperparameters (beta1=0.9, beta2=0.999, lr=0.001) work well for most problems.

---

### Q27. What is momentum in gradient descent? [Medium]

**30-second answer:** Momentum accelerates gradient descent by accumulating a moving average of past gradients and using that to update weights, helping the optimizer move faster through consistent gradient directions and dampening oscillations.

**Key concepts:**
- We define a velocity term, which is a moving average of gradients, then use it to update weights.
- If the momentum at step t-1 is incorrect with respect to t, the modified gradient will compensate for this error in the same update step.
- **Nesterov Momentum (look-ahead):** First looks at where the current momentum is pointing, then computes gradients from that point -- a "look-ahead" gradient step that provides better convergence.

---

### Q28. What are saddle points and why do they matter? [Medium]

**30-second answer:** A saddle point is where the gradient is zero but it is neither a local minimum nor maximum -- it curves up in some dimensions and down in others. In high-dimensional spaces, saddle points are far more common than local minima and can slow down optimization.

**Detailed explanation:**
- In low dimensions, local minima are common.
- In high dimensions, the critical points are predominantly saddle points, not local minima.
- **Pros:** Some saddle points have low loss and are close to many local minima -- they can be "good enough." Getting stuck in a true local minimum is actually rare in high dimensions.
- **Cons:** Gradient descent can slow down near saddle points because the gradient is near zero.
- **Solutions:** Momentum-based optimizers (SGD+momentum, Adam) help escape saddle points. Learning rate jittering (changing LR for a few epochs) can also help.

---

### Q29. Explain mixed precision training. [Hard]

**30-second answer:** Mixed precision training uses both FP16 (half precision) for forward/backward passes and FP32 (full precision) for weight updates, reducing memory usage by ~50% and speeding up training on modern GPUs while maintaining model accuracy.

**Detailed explanation:**
- **Forward pass and gradient computation** use FP16 -- faster on tensor cores, less memory.
- **Master copy of weights** stays in FP32 for accurate accumulation of small gradient updates.
- **Loss scaling:** Multiply loss by a large factor before backprop to prevent small gradients from underflowing to zero in FP16, then unscale before weight update.
- **Benefits:** ~2x speedup on GPUs with tensor cores (V100, A100, H100), ~50% memory reduction.
- **Implementation:** PyTorch `torch.cuda.amp`, TensorFlow `tf.keras.mixed_precision`.

---

### Q30. What is the difference between model parallelism and data parallelism? [Hard]

**30-second answer:** Data parallelism replicates the entire model on multiple GPUs and splits the data batch across them, while model parallelism splits the model itself across GPUs when it is too large to fit on a single device.

**Detailed explanation:**

| Aspect | Data Parallelism | Model Parallelism |
|---|---|---|
| **What is split** | Data (mini-batches) | Model (layers/parameters) |
| **Model copies** | Full copy on each GPU | Partial model on each GPU |
| **Communication** | Gradient synchronization (AllReduce) | Activations passed between GPUs |
| **Use case** | Model fits on one GPU, need faster training | Model too large for one GPU (e.g., LLMs) |
| **Example** | PyTorch DDP | Megatron-LM, pipeline parallelism |

- **Pipeline parallelism** is a form of model parallelism where different layers run on different GPUs, with micro-batches flowing through the pipeline.
- **Tensor parallelism** splits individual layers (e.g., attention heads) across GPUs.
- Modern LLM training combines all three: data + tensor + pipeline parallelism (3D parallelism).

# 4. REGULARIZATION

---

### Q31. What is Weight Decay / L2 Regularization? [Easy]

**30-second answer:** Weight decay adds a penalty proportional to the sum of squared weights to the loss function, discouraging large weights and encouraging simpler models that generalize better. It is essentially equivalent to L2 regularization.

**How it works:**

`L_total = L_original + lambda * sum(w_i^2)`

- The optimizer minimizes both the prediction error AND the size of the weights.
- Larger weights incur a larger penalty, pushing the optimizer toward smaller, more distributed weights.

**Why smaller weights help:**
- Large weights cause drastic output changes with small input changes -- a sign of fitting noise.
- Smaller weights lead to smoother functions, indicating better generalization (Occam's Razor).
- Weight decay encourages the model to use a wider range of features moderately rather than relying heavily on a few.

**Key parameter lambda:**
- lambda = 0: No regularization.
- Small lambda: Weak regularization.
- Large lambda: Strong regularization (risk of underfitting).
- Must be tuned via grid search or random search on a validation set.

**L2 vs Weight Decay nuance:** For standard SGD, they are equivalent. For Adam, they differ -- **AdamW** implements true decoupled weight decay and is preferred.

---

### Q32. What is Dropout and how does it work? [Easy]

**30-second answer:** Dropout randomly sets a fraction of neuron activations to zero during each training step, forcing the network to learn redundant representations and preventing co-adaptation of neurons. It acts like an ensemble of sub-networks.

**Detailed explanation:**
- During training: each neuron is "dropped" (set to zero) with probability p (typically 0.2-0.5).
- During inference: all neurons are active, but outputs are scaled by (1-p) to compensate (or use inverted dropout during training).
- Can be seen as an extreme form of bagging where each model is trained on a single case.
- Can be applied to hidden layers and even the input (visible) layer.
- Dropout gives a way to approximately combine many different architectures.

---

### Q33. What is Early Stopping? [Easy]

**30-second answer:** Early stopping monitors validation loss during training and stops when it starts increasing (or stops improving for a set number of epochs called "patience"), preventing the model from overfitting to the training data.

**How it works:**
- Evaluate the network on a validation dataset after every epoch.
- Stop training when the generalization error (validation loss) starts to increase.
- Restore the model weights from the epoch with the best validation performance.

---

### Q34. How does Batch Normalization work during training vs inference? [Medium]

**30-second answer:** During training, BatchNorm normalizes each mini-batch using the batch's own mean and variance, then applies learnable scale (gamma) and shift (beta) parameters. During inference, it uses running averages of mean and variance accumulated during training instead of batch statistics.

**Detailed explanation:**

**During Training:**
1. Compute mean and variance of the current mini-batch.
2. Normalize: `x_hat = (x - batch_mean) / sqrt(batch_var + epsilon)`
3. Scale and shift: `y = gamma * x_hat + beta` (gamma and beta are learnable).
4. Update running mean and running variance using exponential moving average.

**During Inference:**
1. Use the accumulated running mean and running variance (not the current batch's statistics).
2. Apply the same learned gamma and beta.
3. This makes inference deterministic and independent of batch size.

**Why it helps:**
- Reduces internal covariate shift (activations distribution changes less between layers).
- Allows higher learning rates and faster convergence.
- Acts as a mild regularizer (batch statistics introduce noise).
- Reduces sensitivity to weight initialization.

# 5. CONVOLUTIONAL NEURAL NETWORKS (CNNs)

---

### Q35. What are the three key architectural features of CNNs? [Easy]

**30-second answer:** CNNs use (1) local connectivity -- neurons connect only to spatially nearby inputs, (2) shared weights -- the same filter is applied across the entire input (convolution), and (3) pooling -- downsampling that provides translation invariance and reduces dimensionality.

**Detailed explanation:**

**Local Connectivity:**
- Neurons in one layer connect only to a local region of the input volume.
- The spatial extent of this connectivity is called the **receptive field** (equivalently the filter size).
- Assumption: spatially close pixels are more related than distant ones.
- This drastically reduces the number of parameters compared to fully connected layers.

**Shared Weights (Weight Sharing):**
- The same filter (weight vector) is used across the entire input -- this is what makes CNNs "convolutional."
- The forward pass becomes equivalent to convolving a filter over the image to produce a feature map.
- Training becomes the task of learning useful filters (what features to look for).
- Weight sharing occurs in CNNs and RNNs but NOT in fully connected networks.

**Pooling and Non-linearity (ReLU):**
- Pooling layers (typically max pooling) consider a block of input data and pass on the maximum value.
- Reduces spatial dimensions without additional learnable parameters.
- Provides **translation invariance** -- small shifts in the input do not change the output.
- ReLU introduces non-linearity: ReLU(x) = max(0, x).

### Q36. What is the convolution output size formula? [Easy]

**30-second answer:** Output size = ((I - F + 2P) / S) + 1, where I = input size, F = filter size, P = padding, S = stride.

**Example:** Input 28x28, Filter 7x7, Padding 0, Stride 1:
- C = ((28 - 7 + 0) / 1) + 1 = **22x22**

### Q37. What is the receptive field? [Medium]

**30-second answer:** The receptive field is the region of the input that influences a particular neuron's output. In CNNs, deeper layers have larger receptive fields because they aggregate information from progressively larger input regions through successive convolutions and pooling.

### Q38. Can backpropagation be applied to pooling layers? [Medium]

**30-second answer:** Yes. For max pooling, the gradient is routed only to the neuron that had the maximum value (the "winner"), and all other neurons in the pooling window receive zero gradient. For average pooling, the gradient is distributed equally.

### Q39. How will a network that uses PCA for dimensionality reduction behave compared to hidden layers? [Medium]

**30-second answer:** A network using PCA will not always give the same output as one using hidden layers, because PCA operates on correlated features (maximizing variance), whereas hidden layers optimize for predictive capacity of features (task-specific representations).

### Q41. Compare MAE, MSE, and Huber Loss. [Medium]

**30-second answer:** MSE penalizes large errors heavily and has smooth gradients (easy to optimize) but is sensitive to outliers. MAE is robust to outliers but has a constant gradient (harder to optimize near the minimum). Huber loss combines the best of both -- MSE-like behavior near zero and MAE-like behavior for large errors.

**Detailed comparison:**

| Property | MAE (L1 Loss) | MSE (L2 Loss) | Huber Loss |
|---|---|---|---|
| **Formula** | abs(y - y_hat) | (y - y_hat)^2 | MSE if abs(diff) < delta, else MAE |
| **Outlier sensitivity** | Robust | Very sensitive | Robust (configurable via delta) |
| **Gradient near minimum** | Constant (jumps to 0) | Gradually decreases | Gradually decreases |
| **Optimization** | Harder | Easier | Easy |
| **Differentiability** | Not at 0 | Everywhere | Everywhere |

**Key insights:**
- **MAE advantage:** Handles outlier samples better than MSE.
- **MAE disadvantage:** Gradient is constant until the minimum, where it becomes 0 immediately -- makes it hard to anticipate how close the minimum is.
- **MSE advantage:** Slope gradually decreases near the minimum, making optimization easier.
- **Huber loss:** When `|y - y_hat| < delta`, behaves like MSE. When `|y - y_hat| >= delta`, behaves like MAE. Best of both worlds -- less sensitive to outliers while being properly differentiable.

### Q42. Word2Vec vs GloVe -- What is the key difference? [Medium]

**30-second answer:** Word2Vec only uses local context windows to learn word embeddings, while GloVe (Global Vectors) additionally leverages global word co-occurrence statistics across the entire corpus, capturing both local and global semantic relationships.

**Detailed explanation:**
- **Word2Vec** (Skip-gram / CBOW): Predicts context words from a target word (or vice versa) using a sliding window. Only captures local context.
- **GloVe:** Constructs a global co-occurrence matrix and factorizes it to learn embeddings. Incorporates both local and global statistics.
- GloVe was designed specifically to address the limitation that Word2Vec loses valuable global co-occurrence information.

### Q43. Explain the Attention Mechanism. [Hard]

**30-second answer:** Attention allows a model to dynamically focus on the most relevant parts of the input when producing each element of the output, by computing a weighted sum of values where the weights (attention scores) are derived from the compatibility between a query and a set of keys.

**Detailed explanation:**

**Core components (Query, Key, Value):**
- **Query (Q):** What we are looking for (from the decoder or current position).
- **Key (K):** What each input element offers (used to compute relevance).
- **Value (V):** The actual information to be aggregated.

**Scaled Dot-Product Attention:**
$$
\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right) V
$$

- The dot product QK^T measures similarity between queries and keys.
- Division by sqrt(d_k) prevents the dot products from becoming too large (which would push softmax into saturation).
- Softmax converts scores to probabilities (attention weights).
- Weighted sum of values gives the output.

**Multi-Head Attention:**
- Run attention in parallel with different learned projections (heads).
- Each head can attend to different aspects of the input (syntax, semantics, position, etc.).
- Concatenate and project the outputs.

**Types of Attention:**
- **Self-attention:** Q, K, V all come from the same sequence (used in Transformers).
- **Cross-attention:** Q comes from one sequence, K and V from another (e.g., encoder-decoder attention).

### Q44. What is Transfer Learning and when to use it? [Medium]

**30-second answer:** Transfer learning uses a model pre-trained on a large dataset (e.g., ImageNet, large text corpora) as a starting point for a new task, dramatically reducing training time and data requirements. Use it when you have limited labeled data but the pre-trained model's domain is related to your task.

**Detailed explanation:**

**Two main strategies:**
1. **Feature extraction:** Freeze pre-trained layers, only train new head/classifier layers on top.
2. **Fine-tuning:** Unfreeze some or all pre-trained layers and train end-to-end with a small learning rate.

**When to use:**
- Limited labeled data for your task (most common scenario).
- Your task domain overlaps with the pre-training domain.
- You want faster convergence and better performance than training from scratch.

**When NOT to use:**
- Your domain is very different from the pre-training domain (e.g., medical images from ImageNet).
- You have abundant labeled data and compute to train from scratch.

**Examples:**
- **Vision:** ResNet/EfficientNet pre-trained on ImageNet, fine-tuned for medical imaging.
- **NLP:** BERT/GPT pre-trained on large text, fine-tuned for sentiment analysis, NER, etc.

### Q45. What are Residual Connections and why do they help? [Medium]

**30-second answer:** Residual (skip) connections add the input of a layer directly to its output (y = F(x) + x), creating shortcut paths for gradients during backpropagation. They solve the degradation problem in deep networks by allowing gradients to flow directly through the identity path, enabling training of networks with hundreds of layers.

**Detailed explanation:**

**The degradation problem:**
- As networks get deeper, training accuracy *decreases* (not just overfitting -- actual training performance degrades).
- This is NOT the vanishing gradient problem per se, but rather that deeper networks find it hard to learn identity mappings.

**How residual connections help:**
- If the optimal transformation is close to identity, the network only needs to learn a small residual F(x) near zero, which is much easier.
- Gradients flow through the skip connection unattenuated: `dL/dx = dL/dy * (dF/dx + 1)`. The "+1" from the identity ensures gradients always have a path.
- Enables training of very deep networks (ResNet-152, ResNet-1000).

**Used in:** ResNet, Transformers (every attention and FFN sub-layer has a residual connection), DenseNet (dense connections), U-Net.

### Q46. How do you choose between CNN, RNN, and Transformer? [Medium]

**30-second answer:** Use CNNs for spatial/grid data (images), RNNs/LSTMs for short sequential data with strong temporal dependencies, and Transformers for long sequences and tasks requiring global context (NLP, long-range dependencies). Transformers have largely replaced RNNs for most sequence tasks due to parallelization and superior performance.

**Decision guide:**

| Architecture | Best For | Strengths | Weaknesses |
|---|---|---|---|
| **CNN** | Images, spatial data, short text classification | Local pattern detection, parameter sharing, translation invariance | Limited global context, not naturally sequential |
| **RNN/LSTM/GRU** | Short sequences, time-series, streaming data | Natural sequential processing, variable length input | Slow (sequential), vanishing gradients, poor long-range memory |
| **Transformer** | NLP, long sequences, multimodal tasks, vision (ViT) | Parallelizable, captures long-range dependencies, scalable | Quadratic memory in sequence length, needs more data |

**Modern trends:**
- Transformers are increasingly used for vision (ViT, Swin Transformer) and even time-series.
- 1D CNNs can be effective for some sequence tasks (TCN -- Temporal Convolutional Networks).
- State-space models (Mamba, S4) are emerging as efficient alternatives for very long sequences.

### Q47. What is Knowledge Distillation? [Hard]

**30-second answer:** Knowledge distillation trains a smaller "student" model to mimic the outputs (soft probability distributions) of a larger "teacher" model, transferring the teacher's learned knowledge into a compact model that is faster and cheaper to deploy while retaining most of the teacher's performance.

**Detailed explanation:**

**How it works:**
1. Train a large, accurate teacher model.
2. Generate soft targets: run training data through the teacher and capture its output probabilities (using a temperature parameter T > 1 to soften the distribution).
3. Train the student on a combined loss: hard label loss (cross-entropy with true labels) + soft label loss (KL divergence between student and teacher soft outputs).

**Why soft labels help:**
- Soft probabilities contain "dark knowledge" -- relationships between classes that hard labels do not convey.
- Example: A "cat" image might get teacher probabilities [cat: 0.7, dog: 0.2, car: 0.01] -- telling the student that cats look more like dogs than cars.

**Temperature parameter:**
- Higher T produces softer probability distributions, revealing more inter-class relationships.
- T = 1 gives the original probabilities.

**Applications:**
- Model compression for edge/mobile deployment.
- DistilBERT (40% smaller, 60% faster than BERT, retains 97% performance).
- Training specialized small models from large foundation models.

### Q48. Explain the Lottery Ticket Hypothesis. [Hard]

**30-second answer:** The Lottery Ticket Hypothesis (Frankle & Carlin, 2019) states that dense neural networks contain sparse sub-networks ("winning tickets") that, when trained in isolation from the same initialization, can match the full network's accuracy. This suggests we can achieve the same performance with far fewer parameters if we find the right sub-network.

**Detailed explanation:**

**Key claims:**
1. A randomly initialized dense network contains a sub-network (the "winning ticket") that can be trained to full accuracy.
2. This sub-network achieves this accuracy when trained from its **original initialization** (not from a random re-initialization).
3. The winning ticket is typically 10-20% of the original network size.

**Finding the winning ticket (Iterative Magnitude Pruning):**
1. Train the full network to convergence.
2. Prune the smallest-magnitude weights (e.g., remove 20%).
3. Reset remaining weights to their original initialization.
4. Repeat until desired sparsity is achieved.

**Implications:**
- Suggests that the over-parameterization of large networks is needed to "search" for good sub-networks during training.
- Opens the door to more efficient training and inference if we can identify winning tickets early.
- Related to pruning, sparse training, and neural architecture search.

**Limitations:**
- Finding winning tickets requires training the full network first (expensive).
- The hypothesis is harder to confirm in very large models (though extensions like "late rewinding" help).
