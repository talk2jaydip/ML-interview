### Building an Instruction Dataset

Curating an instruction dataset involves careful steps to ensure that it meets the requirements of how the model will be used. Below is a structured breakdown:

#### Step 1: Curating Representative Data
- The dataset should reflect the tasks and queries the model is expected to handle.
- Collect a wide range of examples, ensuring variety in topics, contexts, and interaction styles.

---

### Key Dimensions of High-Quality Data

#### Accuracy
- **Definition**: Refers to how factually correct and relevant the data samples are.
- **Importance**: Ensures that:
  - Model responses are reliable and factually correct.
  - Responses are relevant to their associated instructions.
- **Best Practices**:
  1. Verify factual correctness against trusted sources.
  2. Remove irrelevant or inaccurate samples.

---

#### Diversity
- **Definition**: Encompasses the range of topics, writing styles, and text lengths within the dataset.
- **Importance**: Enables:
  - The model to handle a wide array of queries effectively.
  - Robustness across different contexts and styles.
- **Best Practices**:
  1. Cover topics ranging from technical to conversational.
  2. Include samples of varying complexities -- both simple and intricate.

---

#### Complexity
- **Definition**: Indicates the level of reasoning and challenge in the samples.
- **Importance**: Pushes the model to:
  - Tackle real-world multi-step reasoning tasks.
  - Excel in challenging scenarios.
- **Best Practices**:
  1. Incorporate multi-step tasks and logical problems.
  2. Avoid trivial or overly simplistic examples.

# **Instruction Dataset Formats & Chat Templates**

## **1. Instruction Dataset Formats**

Instruction datasets store structured instruction-response pairs. The most common formats are:

### **Alpaca**

-   **Structure:**
    
    ```json
    {"instruction": "...", "input": "...", "output": "..."}
    
    ```
    
-   **Example:**
    
    ```json
    {
      "instruction": "What is the capital of France?",
      "output": "The capital of France is Paris."
    }
    
    ```
    

### **ShareGPT**

-   **Structure:**
    
    ```json
    {"conversations": [{"from": "...", "value": "..."}, ...]}
    
    ```
    
-   **Example:**
    
    ```json
    {
      "conversations": [
        {"from": "user", "value": "What is the capital of France?"},
        {"from": "assistant", "value": "The capital of France is Paris."}
      ]
    }
    
    ```
### **OpenAI**

-   **Structure:**
    
    ```json
    {"conversations": [{"role": "...", "content": "..."}, ...]}
    
    ```
    
-   **Example:**
    
    ```json
    {
      "conversations": [
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "What is the capital of France?"},
        {"role": "assistant", "content": "The capital of France is Paris."}
      ]
    }
    
    ```
    

### **OASST**

-   **Structure:**
    
    ```json
    {"INSTRUCTION": "...", "RESPONSE": "..."}
    
    ```
    
-   **Example:**
    
    ```json
    {
      "INSTRUCTION": "What is the capital of France?",
      "RESPONSE": "The capital of France is Paris."
    }
    
    ```
    

### **Raw Text**

-   **Structure:**
    
    ```json
    {"text": "..."}
    
    ```
    
-   **Example:**
    
    ```json
    {
      "text": "What is the capital of France? The capital of France is Paris."
    }
    
    ```
    

----------

## **2. Chat Templates**

After parsing dataset formats, **chat templates** structure model input/output for consistency.

### **ChatML (OpenAI)**

-   **Example:**
    
    ```
    <|im_start|>system
    You are a helpful assistant.<|im_end|>
    <|im_start|>user
    What is the capital of France?<|im_end|>
    <|im_start|>assistant
    The capital of France is Paris.<|im_end|>
    
    ```
    

### **Llama 3**

-   **Example:**
    
    ```
    <|begin_of_text|><|start_header_id|>user<|end_header_id|>
    What is the capital of France?<|eot_id|>
    <|start_header_id|>assistant<|end_header_id|>
    The capital of France is Paris.<|eot_id|>
    
    ```
    

### **Phi-3**

-   **Example:**
    
    ```
    <|user|>
    What is the capital of France?<|end|>
    <|assistant|>
    The capital of France is Paris.<|end|>
    
    ```
    

### **Gemma**

-   **Example:**
    
    ```
    <bos><start_of_turn>user
    What is the capital of France?<end_of_turn>
    <start_of_turn>model
    The capital of France is Paris.<end_of_turn>
    
    ```
    

### **Alpaca (Chat Template)**

-   **Example:**
    
    ```
    ### Instruction: What is the capital of France?
    ### Response: The capital of France is Paris.<EOS>
    
    ```

## **Key Considerations**

-   **Single-turn formats** (e.g., **Alpaca, OASST**) are best for **isolated queries**.
-   **Multi-turn formats** (e.g., **ShareGPT, OpenAI**) allow for **conversational AI**.
-   **Precise formatting is critical** to avoid tokenization issues.

## Dataset Format Comparison Table

| **Feature** | **Alpaca** | **ShareGPT** | **OpenAI** | **OASST** | **Raw Text** |
|---|---|---|---|---|---|
| **Structure** | instruction/input/output fields | conversations array with from/value | messages array with role/content | INSTRUCTION/RESPONSE fields | Plain text string |
| **Multi-turn Support** | No (single-turn only) | Yes (native multi-turn) | Yes (native multi-turn) | No (single-turn only) | No |
| **System Prompt** | No (via instruction field workaround) | Yes (from: "system") | Yes (role: "system") | No | N/A |
| **Tool/Function Use** | No | No (not natively) | Yes (role: "tool", function_call) | No | No |
| **Complexity** | Low -- simple 3-field format | Medium -- conversation threads | High -- supports roles, tools, metadata | Low -- simple 2-field format | Minimal -- unstructured |
| **Best For** | Simple instruction-following tasks | Multi-turn chatbot training | Production chat systems, tool-augmented LLMs | Quick prototyping, simple Q&A | Continual pre-training, domain adaptation |
| **Popular Tools** | Stanford Alpaca, Axolotl | ShareGPT datasets, LMSys | OpenAI fine-tuning API, vLLM | Open Assistant | Hugging Face datasets |
| **Typical Dataset Size** | 10K-100K | 50K-500K | 10K-1M | 50K-200K | Millions of tokens |

### When to Choose Which Format
- **Alpaca**: Best for beginners; simple single-turn instruction tasks with minimal setup.
- **ShareGPT**: Ideal for training conversational agents from real user interactions.
- **OpenAI**: Choose when you need tool use, function calling, or system prompts in production.
- **OASST**: Good for community-sourced human feedback datasets.
- **Raw Text**: Use for continual pre-training or domain adaptation (not instruction-tuning).

## Data Quality Checklist for Fine-Tuning

Before starting any fine-tuning job, run through this checklist to ensure your dataset is production-ready.

### 1. Diversity
- [ ] **Topic coverage**: Does the dataset cover all expected use cases and edge cases?
- [ ] **Response length distribution**: Are there short, medium, and long responses? (Plot a histogram of token counts.)
- [ ] **Instruction variety**: Are instructions phrased in multiple ways, or are they templated and repetitive?
- [ ] **Language/style mix**: If multilingual or multi-style, is each category adequately represented?

### 2. Decontamination
- [ ] **Benchmark leakage**: Remove any examples that overlap with evaluation benchmarks (e.g., MMLU, HellaSwag, GSM8K).
- [ ] **Train/test overlap**: Ensure no test-set examples appear in training data.
- [ ] **Near-duplicate removal**: Use MinHash or embedding-based dedup to remove near-duplicates (aim for <1% duplication rate).

### 3. Format Consistency
- [ ] **Uniform structure**: Every example follows the same schema (e.g., all have instruction + output, no missing fields).
- [ ] **Chat template alignment**: Verify the dataset format matches the model's expected chat template exactly.
- [ ] **Special tokens**: Confirm no stray or misplaced special tokens (EOS, BOS, pad tokens).
- [ ] **Encoding**: All text is valid UTF-8 with no corrupt characters.

### 4. Length Distribution
- [ ] **Max length check**: Truncate or filter examples exceeding `max_seq_length` -- silently truncated examples cause training artifacts.
- [ ] **Min length check**: Remove trivially short examples (<10 tokens) that add noise without signal.
- [ ] **Packing efficiency**: If using packing, verify the length distribution allows good packing ratios (>80% utilization).

### 5. Content Quality
- [ ] **Factual accuracy**: Spot-check 50-100 random examples for correctness.
- [ ] **Instruction-output alignment**: Verify outputs actually answer the instructions (not generic filler).
- [ ] **No PII or sensitive data**: Scan for emails, phone numbers, API keys, personal names.
- [ ] **Label balance**: For classification tasks, check class distribution and apply resampling if needed.

> **Rule of Thumb**: 1,000 high-quality, diverse examples often outperform 100,000 noisy ones (LIMA paper). Invest in quality first.



## Memory Optimization Parameters

- **Gradient Checkpointing**  
  - Saves memory by not storing all activations during the forward pass.  
  - Missing activations are recomputed during the backward pass, trading extra computation for reduced memory usage.

- **Gradient Accumulation**  
  - Simulates a larger batch size by accumulating gradients over multiple smaller batches.  
  - Reduces memory requirements at the cost of slower training updates.

- **Quantization**  
  - Uses reduced precision (e.g., bf16) to save memory without significant performance loss.

- **Example Settings (Political Promises Detector):**  
  ```python
  gradient_accumulation_steps = 4
  bf16 = True
  gradient_checkpointing = True
  ```

---

## Regularization Parameters

- **Label Smoothing**  
  - Softens hard target labels (e.g., from 0/1) to prevent overconfidence.  
  - Improves model calibration and is useful with noisy or subjective labels.
> Label Smoothing improves model robustness, especially under noisy labels, by distributing some probability mass to incorrect classes.
This results in better-calibrated and less overfit models—especially helpful in classification tasks.
- **Noise Embeddings**  
  - Adds noise to input embeddings to reduce overfitting on stylistic features (formatting, wording, length).  
  - Employed only during training; can lead to longer, more informative outputs.  
  - Implemented using techniques such as NEFTune in Hugging Face.

- **Example Settings (Political Promises Detector):**  
  ```python
  label_smoothing_factor = 0.1
  neftune_noise_alpha = 5
  ```

---

## Batch Size and Learning Rate Relationship

- **Batch Size Considerations:**  
  - Larger batch sizes speed up training but require more memory.  
  - Smaller batches can introduce higher variance, possibly aiding convergence to flatter minima.  
  - Best practice: choose batch sizes as powers of two to optimize GPU I/O.

- **Learning Rate Trade-offs:**  
  - High learning rates may lead to faster convergence but risk overshooting minima.  
  - Lower learning rates may ensure careful convergence but require more steps.
  
- **Example Settings (Political Promises Detector):**  
  ```python
  per_device_train_batch_size = 8
  per_device_eval_batch_size = 8
  ```

- **Tip:** Reducing maximum sequence length can allow for larger batch sizes.

  - Updates only a small subset of parameters while maintaining performance.  
  - **Key Hyperparameters:**

## Working with Reduced Precision

- **Bitsandbytes Library:**  
  - Supports working in reduced precision modes (e.g., FP4).  
  - **Key Settings:**  
    - `use_4bit = True`  
    - `bnb_4bit_compute_dtype = 'float16'`  
    - `bnb_4bit_quant_type = 'nf4'`  
    - `use_nested_quant = False`  
    - `llm_int8_threshold`: We need to specify a threshold of values beyond which fp16 will be
used. This is because int8 quantization works well only for values lesser than 5–6.
  - `llm_int8_skip_modules`: This is used to specify the exceptions for which we do not want int8
  quantization.
  - `llm_int8_enable_fp32_cpu_offload`: If we want parts of the model to be run in int8 on GPU and the rest in
  FP32 on CPU, this parameter facilitates it. This is used in cases where
  the model is too large to fit on our GPU.
  - `bnb_4bit_compute_dtype`: This sets the computational type, regardless of the input type.
  - `bnb_4bit_quant_type`: The options here are FP4 or NF4. This is used to set the quantization
  type in the 4-bit layers.


### **FP4 (Fixed-Point 4)**
- **Summary**: Fixed-point representation using `4 bits for compact data` with limited precision and range, commonly used in embedded systems or resource-constrained environments.
- **Takeaway**: Ideal for systems with `limited hardware resources where floating-point operations are too costly`.

### **NF4 (Normalized Floating-Point 4)**
- **Summary**: A 4-bit floating-point format `offering a broader dynamic range by utilizing an exponent`, suitable for applications requiring `a mix of precision and range` like graphics or AI.
- **Takeaway**: Useful in scenarios where a larger range of values is needed, with a tradeoff in precision due to the limited bit count.

### **BF16 (BFloat16)**

- **Summary (One-liner for Interview)**: A 16-bit floating-point format with reduced precision compared to FP32 but with similar range, designed for machine learning and AI applications.
  
### **Explanation**:
- **Precision**: BF16 uses 1 sign bit, 8 exponent bits, and 7 fraction bits (instead of the typical 23 in FP32), which provides less precision but maintains a similar dynamic range.
- **Range**: The range is similar to FP32 due to the 8-bit exponent, making BF16 capable of handling large values and a wide dynamic range, which is crucial for deep learning models.
- **Use Case**: BF16 is widely used in AI and machine learning tasks, especially for training neural networks, as it offers a good trade-off between performance (lower memory usage and faster processing) and accuracy.

# 🎯 Label Smoothing: Improving Calibration & Reducing Overfitting  

## 🔍 What is Label Smoothing?  
Label smoothing is a **regularization technique** that helps:  
- **Reduce overfitting**  
- **Improve model calibration** (ensuring confidence scores reflect true accuracy)  

### 📌 Why Does Calibration Matter?  
A model is **well-calibrated** if its confidence scores match the actual probability of correctness.  

🔹 **Example:** Sentence classification (abusive vs. non-abusive)  
- If a well-calibrated model predicts **0.9 probability**, ~90% of those cases should be correct.  
- If it predicts **0.6 probability**, ~60% of cases should be correct.  

✅ **Good Calibration** → Output probabilities are a realistic measure of confidence.  
❌ **Poor Calibration** → Model is **overconfident**, leading to incorrect high-confidence predictions.  

> ⚠️ **Larger Models & Overconfidence**  
> Studies (Li et al.) show that **larger models** (e.g., GPT-like models) tend to be **less calibrated** compared to smaller models like BERT.  
> **Implication:** If uncertainty estimates matter, consider using **smaller models** or applying calibration techniques like label smoothing!  

---

## 🛠️ How Does Label Smoothing Work?  
### 🚀 Standard Training vs. Label Smoothing  
- **Without label smoothing:** The model learns **hard target labels** (`0` or `1`).  
- **With label smoothing:** The model **softens the labels** by subtracting a small value (`α`) from `1` and redistributing it across all labels.

### 📌 Example (Binary Classification)  
#### ❌ Standard Labels:  


- Label 0 will be transformed to label_smoothing_
- Label 1 will be transformed to 1 - label_smooth
> label_smoothing_factor/num_labels

#### Multi-Task vs. Single-Task Fine-Tuning

- Multi-task fine-tuning (e.g., instruction-tuning) helps models generalize better.
- Single-task fine-tuning specializes the model for a particular use case.

#### Example Process:

- Start with a base model (e.g., T5).
- Perform instruction-tuning (e.g., with FLAN) -- Produces FLAN-T5.
- Fine-tune FLAN-T5 on your specific dataset for optimal results.

##### Why Use Instruction-Tuning?

> Acts as an intermediate step to improve performance.
> Proven to be more effective than direct single-task fine-tuning.

### **Explanation & Key Takeaways**  

#### **Understanding Instruction-Tuning Datasets**  
Unlike traditional datasets (which contain just **input-output pairs**), instruction-tuning datasets also include **task descriptions in natural language**.  

##### **Structure of an Instruction-Tuning Dataset:**  
1. **Instruction** – Defines the task, specifies output format, and may include examples or constraints.  
2. **Optional Input** – The actual text for the model to process (e.g., a sentence or paragraph).  
3. **Output** – The correct response based on the instruction.  

---

### **Key Points to Note**  

✅ **Fine-Tuning Types:**  
- **Multi-task** (used for instruction-tuning) vs. **Single-task** (for task-specific adaptation).  
- Instruction-tuning acts as an **intermediate step** before single-task fine-tuning.  

✅ **Why Instruction-Tuning?**  
- LLMs are trained on **next-token prediction/denoising**, which doesn’t match real-world tasks.  
- Instruction-tuning **bridges this gap**, making models more **controllable** and **structured**.  

✅ **Why Customize Instruction-Tuning?**  
- Public datasets are **too general** and **lack domain-specific details**.  
- Domain expertise allows **tailored datasets** for better model performance.  

✅ **Methods to Create Instruction-Tuning Datasets:**  
1. **Use public instruction-tuning datasets.**  
2. **Convert traditional fine-tuning datasets** into instruction-based datasets.  
3. **Manually craft seed examples** and expand using an LLM.

# SFT (Supervised Fine-Tuning) Pipeline Walkthrough

The SFT pipeline is the most common approach to adapting a pre-trained LLM for a specific task or behavior. Below is an end-to-end walkthrough of each stage.

---

## Stage 1: Define the Objective
- **What task?** Classification, summarization, code generation, chatbot, etc.
- **What behavior?** Tone, format, safety constraints, response length.
- **Success metric?** Human eval, BLEU/ROUGE, accuracy, win-rate vs. baseline.

---

## Stage 2: Data Collection & Preparation
1. **Collect seed data**: Gather 500-5,000 high-quality examples (human-written or curated).
2. **Augment with LLMs**: Use Evol-Instruct or Self-Instruct to expand to 10K-100K examples.

# Hyperparameter Tuning Guide for LLM Fine-Tuning

## Learning Rate

| **Method** | **Typical Range** | **Notes** |
|---|---|---|
| Full Fine-Tuning | 1e-5 to 5e-5 | Lower LR since all parameters update; higher LR causes catastrophic forgetting |
| LoRA | 1e-4 to 3e-4 | Can use higher LR because only a small subset of parameters is updated |
| QLoRA | 1e-4 to 3e-4 | Same as LoRA; quantization does not change the LR requirement significantly |
| Continual Pre-Training | 1e-5 to 1e-4 | Start low; use re-warming strategy if training on substantially new data |

**Key insight**: If the loss is not decreasing, your LR is likely too low. If the loss is unstable or diverging, your LR is too high.

---

## Batch Size

- **Effective batch size** = `per_device_batch_size` x `num_GPUs` x `gradient_accumulation_steps`
- **Recommended effective batch size**: 16-64 for most fine-tuning tasks.
- **Small datasets (<5K)**: Use smaller batch sizes (4-8 effective) to get more gradient updates per epoch.
- **Large datasets (>50K)**: Larger batch sizes (32-64) are more efficient and stable.
- **Rule**: When you double the batch size, increase the learning rate by ~1.4x (square root scaling).

---

## Epochs

| **Dataset Size** | **Recommended Epochs** | **Rationale** |
|---|---|---|
| <1K examples | 5-10 | Small data needs more passes, but watch for overfitting |
| 1K-10K examples | 3-5 | Sweet spot for most instruction-tuning tasks |
| 10K-100K examples | 1-3 | Large datasets often converge in 1-2 epochs |
| >100K examples | 1 | One epoch is usually sufficient; more causes overfitting |

**Warning**: More epochs is NOT always better. Monitor eval loss and stop early if it increases.

---

## Warmup Ratio

- **Typical value**: 0.03 to 0.10 (3% to 10% of total training steps).
- **Why it matters**: Warmup gradually increases the learning rate from 0 to the target LR, preventing early training instability.
- **For LoRA**: 0.03-0.05 is usually sufficient.
- **For full fine-tuning**: Use 0.05-0.10 for more stability.

---

## Learning Rate Scheduler: Cosine vs. Linear

| **Scheduler** | **When to Use** | **Behavior** |
|---|---|---|
| **Cosine** | Default choice for most fine-tuning | Smoothly decays LR following a cosine curve; ends near 0. Good for convergence. |
| **Linear** | Short training runs (<1 epoch) | Linearly decays LR to 0. Simpler but less optimal for longer runs. |
| **Cosine with restarts** | Continual pre-training over multiple data batches | Periodically resets LR to allow adaptation to new data distributions. |
| **Constant with warmup** | Very short runs or hyperparameter search | Keeps LR constant after warmup; useful for quick experiments. |

**Default recommendation**: Use **cosine scheduler** with warmup unless you have a specific reason not to.

## Decision Flowchart

```
START: What are your constraints?
  |
  |-- Q1: How much GPU memory do you have?
  |     |
  |     |-- < 16 GB (single consumer GPU)
  |     |     --> Use QLoRA (4-bit quantization + LoRA)
  |     |     Best for: 7B-13B models on a single GPU
  |     |
  |     |-- 16-48 GB (single A100/A6000)
  |     |     --> Use LoRA (16-bit)
  |     |     Best for: 7B-13B models with better quality than QLoRA
  |     |
  |     |-- 48+ GB or multi-GPU
  |           --> Consider Full Fine-Tuning
  |           Go to Q2
  |
  |-- Q2: How much training data do you have?
  |     |
  |     |-- < 100 examples
  |     |     --> Use Prompt Tuning or Few-shot In-Context Learning
  |     |     Fine-tuning may overfit with this little data
  |     |
  |     |-- 100 - 10K examples
  |     |     --> Use LoRA or QLoRA
  |     |     Sweet spot for parameter-efficient methods
  |     |
  |     |-- > 10K examples + sufficient compute
  |           --> Full Fine-Tuning is viable
  |           More data justifies updating all parameters
  |
  |-- Q3: How different is your task from the base model's training?
        |
        |-- Similar (general chat, Q&A, summarization)
        |     --> LoRA is sufficient
        |     Small adaptations work well with few parameters
        |
        |-- Very different (new domain, new language, new format)
              --> Full Fine-Tuning or Continual Pre-Training + LoRA
              Major distribution shifts need more parameter updates
```

## Key Takeaways for Interviews

- **LoRA** is the default choice for most fine-tuning scenarios -- it offers the best tradeoff of quality, speed, and memory.
- **QLoRA** is for when you are memory-constrained (e.g., fine-tuning a 13B model on a single 24GB GPU).

### **Why Update the Parameters of an LLM?**  

Updating an LLM's parameters is essential for improving its performance in specialized use cases. The primary reasons for updating LLM parameters include:  

---

### **1. Domain Adaptation (DAPT)**  
- Many domains (e.g., law, finance, medicine) have **unique linguistic characteristics** that general-purpose LLMs may not handle well.  
- Continual pre-training on **domain-specific data** helps the LLM **understand jargon, structure, and conventions** better.  
- Example: Training a general LLM on **legal documents** to improve its ability to summarize court rulings.  

---

### **2. Task Adaptation (TAPT)**  
- Even within a domain, LLMs may struggle with **specific tasks** (e.g., sentiment analysis in financial news).  
- **Task-adaptive pre-training** fine-tunes the model on **data related to specific tasks**.  
- Example: Pre-training an LLM on **financial market reports** before fine-tuning it for **stock price prediction**.  

---

### **3. Knowledge Updating (Lifelong Learning)**  
- Pre-trained LLMs have **fixed knowledge** that can become **obsolete over time**.  
- Continual pre-training on **new data** helps the LLM stay **up-to-date**.  
- Example: Updating an LLM trained before **COVID-19** with **pandemic-related medical literature**.  

---

### **4. Controllability & Steerability (Alignment Training)**  
- Fine-tuning can help **control the model's behavior**, ensuring it follows specific guidelines.  
- Useful for **reinforcing safety mechanisms** (e.g., rejecting unethical requests).  
- Techniques like **RLHF (Reinforcement Learning from Human Feedback)** can **steer** the LLM's responses.  

---
### Task-Specific vs. Domain-Specific Models 

1. **Task-Specific Models**:
   - Focus: Optimized for a single task (e.g., translation, sentiment analysis).
   - Advantages: Efficient with smaller model sizes (<8B parameters); requires 100-100,000 data samples.
   - Use Case: Ideal for specialized, resource-constrained applications.

2. **Domain-Specific Models**:
   - Focus: Fine-tuned for a specific field (e.g., medicine, law, e-commerce).
   - Advantages: Enhances field-specific vocabulary and expertise.
   - Data Needs: Varies by domain complexity and representation in pre-training data -- specialized fields may need more.

3. **Key Factors for Domain-Specific Fine-Tuning**:
   - Domain size (knowledge breadth and vocabulary).
   - Representation in pre-training data.
   - Challenges due to closed-source pre-training datasets.

## **Techniques for Updating LLM Parameters**  

### **1. Continual Pre-Training**  
- Continues pre-training a general LLM on **new, domain-specific, or updated data**.  
- Helps with **domain adaptation** and **knowledge updates**.  

**Challenges:**  
- **Catastrophic forgetting**: The model might **lose its previous knowledge** while learning new data.  
- **Solution:** Replay, distillation, and regularization techniques (explained later).  

---

### **2. Parameter-Efficient Fine-Tuning (PEFT)**  
- Updates only **a small subset of parameters**, reducing memory and compute costs.  
- Techniques include **LoRA (Low-Rank Adaptation)** and **Adapters**.  
- Ideal for **task adaptation** without requiring full model fine-tuning.  

---

### **3. Model Merging (Model Fusion)**  
- Combines multiple LLMs to **merge their capabilities**.  
- Can be used to **integrate domain knowledge** into a general-purpose model.  
- Still a **new and experimental** area ("dark arts" of NLP).  

---

Here's a comparison of **Task-Adaptive Pre-Training (TAPT)** and **Domain-Adaptive Pre-Training (DAPT)** in a table format:

| **Aspect**                        | **Task-Adaptive Pre-Training (TAPT)**                          | **Domain-Adaptive Pre-Training (DAPT)**                            |
|-----------------------------------|---------------------------------------------------------------|---------------------------------------------------------------------|
| **Purpose**                       | Adapt the model to specific downstream tasks.                 | Adapt the model to a specialized domain or vocabulary.              |
| **Data Type**                     | Smaller, task-specific unsupervised dataset.                  | Larger domain-specific unsupervised dataset.                        |
| **Focus**                          | Tailored to specific tasks (e.g., classification, summarization). | Tailored to the linguistic structure and jargon of a domain.        |
| **Order of Execution**            | Performed after DAPT to fine-tune on task-specific data.       | Performed first to adapt to domain-related content.                 |
| **Data Selection Method**         | Data selected using clustering methods based on task relevance. | Data selected using clustering methods based on domain relevance.   |
| **Size of Dataset**               | Typically smaller than DAPT datasets.                         | Typically larger and domain-specific.                               |
| **Goal**                          | Enhance the model's ability on specific tasks.                 | Expose the model to domain-specific text to capture specialized knowledge. |
| **Model Adaptation**              | Adjusts the model to perform well on the target task.          | Adjusts the model to understand domain-specific structure and terminology. |

This table highlights the key differences between TAPT and DAPT.

## **Continual Pre-Training in Detail**  

### **Why Not Train an LLM from Scratch?**  
- Training an LLM **from scratch** on domain-specific data is **expensive and inefficient**.  
- **Solution:** Take an **already pre-trained** LLM and fine-tune it on domain-specific data.  

### **Common Use Cases of Continual Pre-Training**  
- **Domain-Adaptive Pre-Training (DAPT)**: Exposing an LLM to an **entire domain** (e.g., legal or financial text).  
- **Task-Adaptive Pre-Training (TAPT)**: Pre-training on **data relevant to specific tasks** within a domain.  
- **Lifelong Learning**: Keeping the model up-to-date with **new knowledge**.  

---

### **The Catastrophic Forgetting Problem**  
- When an LLM is trained on **new data**, it may forget **previously learned** knowledge.  
- Continual pre-training can degrade performance on **general tasks**.  

### **Solutions for Catastrophic Forgetting**  
- **Replay (Memory)**: Mixes original pre-training data with new data to retain past knowledge.  
- **Distillation**: Takes an older checkpoint of the model and during training compares the KL-divergence between the older and the current representations and penalizes it.  
- **Regularization**: Penalizes large parameter changes to preserve prior learning.  
- **Parameter Expansion**: Adds new parameters (width/depth) to retain old knowledge while learning new information.

---

## **Practical Implementation of Continual Pre-Training**  
### **Steps to Continually Pre-Train an LLM on Domain-Specific Data**  
1. **Select a pre-trained LLM** (e.g., LLaMA-2, Mistral, GPT-J).  
2. **Curate domain-specific data** (e.g., legal texts, medical papers).  
3. **Tokenize and format the data** (match the LLM's pre-training format).  
4. **Train the model using self-supervised learning** (next-word prediction, masked token prediction).  
5. **Evaluate performance** using:  
   - Benchmark tests (e.g., perplexity, accuracy on domain-specific tasks).  
   - General tasks (to detect catastrophic forgetting).  
6. **Apply replay, distillation, or parameter expansion** if forgetting occurs.

## **Future Directions in Continual Learning**  
- **Mixing domain and general data** to avoid forgetting.  
- **Automating replay and distillation techniques**.  
- **Combining continual learning with parameter-efficient fine-tuning (PEFT)**.

### **Replay (Memory) Strategy**  
- Mix pre-training data with continual training data to reduce data drift.  
- Start with ~25% new data and gradually increase up to 80%.  
- If the original dataset is monolithic, identify domains to ensure balanced sampling.  

### **Learning Rate Strategies**  
- **Lowering LR over time**: Reduces forgetting but may slow down learning.  
- **Re-warming strategy (Gupta et al.)**:  
  1. Increase LR to a maximum at the start.  
  2. Decay it using a cosine schedule.  
  3. Keep LR constant at a minimum to maintain balance between old and new knowledge.
   

### **Parameter Expansion**  
- **Basic approach**: Add new layers to the model and only train those new parameters during continual pre-training.
- **Domain-specific modules**: Insert and train specialized parameter modules (adapters) within existing layers to tailor the model to specific domains.
- **Further exploration**: Adapter-based approaches will be discussed in "Parameter-Efficient Fine-Tuning."


### **Task-Adaptive Pre-Training (TAPT)**
- **Supplement to Domain-Adaptive Pre-Training (DAPT)**: TAPT helps to continually pre-train an LLM on a smaller, task-specific unsupervised dataset.
- **Prevention of catastrophic forgetting**: Perform DAPT first to adapt the model to the domain, followed by TAPT, and then fine-tune the model on the downstream tasks using supervised data.
- **Data selection for TAPT**: Use embedding-based methods to select unsupervised data, focusing on data that is clustered similarly to gold-truth sentences.

### **Key Tips**
- **Domain-specific jargon**: Domain-specific text often includes specialized terms. A useful strategy is to add extra tokens to represent this jargon, improving model understanding and handling of domain-specific language.


## ** Breakdown: Reinforcement Learning & RLHF**

-   **Traditional Reinforcement Learning (RL)**
    
    -   **Agent-Environment Interaction:**
        -   The agent interacts with its environment by taking actions.
    -   **Trial and Error:**
        -   The agent learns by experimenting—trying different actions.
    -   **Reward Function:**
        -   After actions, a reward is provided if the actions lead towards the goal.
        -   The objective is to maximize these rewards.
    -   **Challenge:**
        -   In real-world scenarios, it can be difficult to define what "success" looks like and how to craft an appropriate reward function.
-   **Reinforcement Learning from Human Feedback (RLHF)**
    
    -   **Human-in-the-Loop:**
        -   Instead of relying solely on a pre-defined reward function, a human provides iterative feedback.
    -   **Integrating Preferences:**
        -   A reward model is trained using human feedback to better align the agent's actions with human preferences.
-   **Types of Human Feedback**
    
    -   **Binary Feedback:**
        -   The feedback is a simple yes/no or accept/reject decision.
    -   **Binary Comparisons:**
        -   Two outputs are compared, and the preferred one is chosen.
    -   **Ranking:**
        -   A set of outputs is ranked according to preference.
    -   **Corrective Feedback:**
        -   The human provides a corrected or ideal output, often described in natural language.
** Breakdown: RLHF Setup by OpenAI**

-   **Phase 1: Supervised Fine-Tuning**
    
    -   **Prompt Dataset:**
        -   Assemble a diverse set of potential user requests.
    -   **Human Annotation:**
        -   Annotators provide desired responses for each prompt.
    -   **Fine-Tuning Dataset:**
        -   Combine prompts and human responses to form the training dataset.
    -   **Model Update:**
        -   Fine-tune the pre-trained model using this dataset.
    -   **Resource-Intensive:**
        -   Gathering quality annotations requires significant effort and resources.
-   **Phase 2: Reward Modeling**
    
    -   **Query Generation:**
        -   Use a variety of prompts to generate multiple responses from the model.
    -   **Feedback Collection:**
        -   Annotators review the responses and offer feedback via:
            -   Rank-ordering responses or
            -   Selecting the best generation.
    -   **Reward Model Training:**
        -   Train a model to predict which output a human would prefer based on the provided feedback.
-   **Phase 3: Proximal Policy Optimization (PPO)**
    
    -   **Model Optimization:**
        -   Utilize the trained reward model to further refine the fine-tuned model.
    -   **PPO Algorithm:**
        -   Apply Proximal Policy Optimization to adjust the model’s actions in alignment with human preferences.
    -   **Outcome:**
        -   The optimized model delivers outputs that better reflect the human-in-the-loop guidance.
        
## **Interactive Breakdown: In-Depth PPO Training & RLHF Process**

-   **1. PPO Training Process**
    
    -   **Response Generation:**
        -   The language model generates a response (or continuation) given a prompt.
    -   **Reward Calculation:**
        -   The reward model evaluates the query-response pair and `outputs a scalar reward representing` how well the response fits the desired behavior.
    -   **Log-Probability & KL-Divergence Calculation:**
        -   **Log-Probabilities:**
            -   Calculate the log-probabilities of each token in the query-response sequence using two models:
                -   `The fine-tuned model (SFT model) being adjusted`.
                -   `A reference model (the original pre-trained model before fine-tuning)`.
        -   **KL-Divergence:**
            -   Compute the Kullback-Leibler (KL) divergence between these log-probabilities.
            -   This divergence acts as a regularization signal, ensuring that the new outputs do not deviate too far from the reference model.
            
    -   **Model Optimization Using PPO:**
        -   Combine the reward signal and the KL-divergence regularization.
        -   Train the model with the Proximal Policy Optimization (PPO) algorithm, updating its behavior based on these signals.
-   **2. Training the Reward Model with Hugging Face TRL Library**
    
    -   **Objective:**
        -   Train a reward model that can compare two outputs generated by a language model and predict which one aligns better with human preferences.
    -   **Dataset: Anthropic’s hh-rlhf**
        -   **Dataset Composition:**
            -   Approximately 161,000 pairs of examples.
            -   Each pair includes:
                -   A **chosen** response (preferred by human annotators).
                -   A **rejected** response.
        -   **Context of Examples:**
            -   Each pair corresponds to different responses by the LLM to the same prompt.
            -   Human annotators select the response that best aligns with the intended values and principles.
    -   **Example from the Dataset:**
        -   **Chosen Response Example:**
            -   A conversation where the assistant advises caution or curiosity in response to a sensitive query.
        -   **Rejected Response Example:**
            -   A conversation where the assistant’s response is less aligned with safe or ethical practices.
        -   _(The provided example involved a query about hacking emails, with responses differing in approach.)_

The model being updated with PPO is the fine-tuned model (often called the SFT model). Here’s how it works:

-   **Fine-Tuned (SFT) Model:**
    
    -   This model starts as a base LLM that has been adapted through supervised fine-tuning with human-provided responses.
    -   It’s the one generating responses during the RLHF process.
-   **Reference Model:**
    
    -   This is usually the original pre-trained base model before any fine-tuning.
    -   It remains unchanged and is used to compute the KL-divergence to regularize the fine-tuned model.
-   **PPO Application:**
    
    -   The PPO algorithm uses the reward signal (from the reward model) and the KL-divergence (comparing the SFT model to the reference model) to iteratively update the SFT model.
    -   The objective is to improve the SFT model's alignment with human preferences while keeping it close to the robust language capabilities of the base LLM.



## ** Sampling Methods for Addressing Hallucination**

-   **Decoding Methods & Hallucination Risk**
    
    -   **Greedy Decoding:**
        -   Always picks the highest-probability token.
        -   Tends to produce less hallucination because it minimizes randomness.
    -   **Top-p (Nucleus) Sampling:**
        -   Samples from a pool of tokens whose cumulative probability meets a threshold pp.
        -   `Introduces randomness, which can sometimes result in choosing an incorrect token—thus increasing hallucination risk`.
-   **Factual-Nucleus Sampling**
    - Factual‐nucleus sampling is a decoding strategy used during inference—not during the training phase.
    -   **Core Idea:**
        -   Dynamically reduce randomness as the generated sequence gets longer.
        -   Fewer valid candidate tokens remain in later parts of a sentence; hence, less randomness is desired.
    -   **Formula:**  $$ pt=max⁡{ω, p×λt−1}p_t = \max\{\omega,\, p \times \lambda^{t-1}\} $$
        -   tt: The generation step (i.e., token index within the sentence).
-   **Tunable Parameters**
    
    -   **Decay Rate (λ\lambda):**
        -   Multiplies the initial pp value at each step.
        -   Controls how quickly randomness decays as more tokens are generated.
    -   **Reset (p):**
        -   Option to reset pp at the start of each new sentence.
        -   Prevents pp from decaying too far within long sequences, which could overly mimic greedy decoding.
    -   **Lower-bound (ω\omega):**
        -   Sets a minimum threshold for ptp_t.
        -   Ensures that some level of diversity is maintained, preventing the algorithm from becoming completely deterministic.
-   **Trade-Offs**
    
    -   **Reducing Hallucination:**
        -   A lower pp value (via decay) makes the output more factual and reduces hallucinations.
    -   **Loss of Diversity:**
        -   Too low a pp can reduce the diversity of token generation, potentially making responses repetitive or less creative.

This method allows fine-tuning of the randomness in token sampling dynamically, aiming to strike a balance between reducing hallucination risk and maintaining output diversity.



### 🧠 **What is Nucleus Sampling (a.k.a. Top-p Sampling)?**

**Nucleus sampling** is a **probabilistic decoding strategy** used to select the next token in text generation. Instead of always picking the highest-probability token (like greedy decoding), it introduces **controlled randomness** by sampling from a **subset of the most likely tokens**.

---

### 🧩 **How It Works:**

1. The model outputs a **probability distribution** over all possible next tokens.
2. Sort the tokens by their probability (from highest to lowest).
3. Select the **smallest set of tokens whose cumulative probability ≥ *p***.
   - This subset is called the **nucleus**.
   - The threshold *p* is usually between **0.8 to 0.95**.
4. **Sample** from this nucleus (instead of all tokens).

---

### 📌 **Example:**

Say the model predicts the next word with this distribution:

| Token  | Probability |
|--------|-------------|
| "the"  | 0.35        |
| "a"    | 0.25        |
| "cat"  | 0.20        |
| "dog"  | 0.10        |
| "ran"  | 0.05        |
| Others | 0.05        |

If `p = 0.90`, we take the smallest set of tokens whose cumulative probability is at least 90%:

→ **{ "the", "a", "cat" }** → 0.35 + 0.25 + 0.20 = 0.80 → add "dog" → now 0.90 ✅  
Then randomly sample from this set.

---

### ⚖️ **Why Use Nucleus Sampling?**

| Method             | Behavior                              | Use Case                              |
|--------------------|----------------------------------------|----------------------------------------|
| **Greedy**         | Always picks most likely token         | Deterministic but dull                 |
| **Random**         | Samples from all tokens                | Too chaotic                            |
| **Top-k Sampling** | Samples from top *k* most probable     | Fixed-size randomness                 |
| **Nucleus (Top-p)**| Samples from dynamic token set based on probability mass | Balanced creativity & coherence ✅ |

---

### ✅ **Pros:**
- Dynamic: Adapts the size of token candidates based on content.
- More fluent and creative outputs than greedy decoding.
- Less chaotic than pure random or top-*k* sampling.

### ❌ **Cons:**
- Still has randomness—may occasionally produce irrelevant text.
- Requires tuning of *p* for best performance.

---

### 🧵 Summary (One-liner for Interview):

> **Nucleus sampling selects the next token by sampling from the smallest group of likely candidates whose cumulative probability exceeds a threshold *p*, balancing randomness with coherence.**

:

> Nucleus sampling filters tokens by cumulative probability mass, offering a smart balance between diversity and factuality. It’s a flexible alternative to rigid greedy or top-*k* decoding methods.

### **Decoding by Contrasting Layers (DoLa)**

DoLa is a decoding strategy for improving the **factual accuracy** of outputs from **large language models (LLMs)** by leveraging the internal structure of Transformer architectures.

> 🧠 It builds on the hypothesis that **upper layers** encode **factual knowledge**, while **lower layers** handle **syntactic patterns**.

#### **Key Components:**

- **Logit Differencing:**  
  For each token, DoLa computes the **difference between logits** from a **higher ("expert") layer** and a **lower ("amateur") layer**, amplifying factual signals for generation.

- **Layer Selection:**  
  Typically, the **final layer** acts as the **expert**, and a **mid-layer** is chosen as the amateur. The right layer pair selection is crucial for optimal contrast and output quality.



### 🧠 How DoLa Modifies This Process

DoLa introduces a twist to enhance factual accuracy:

1.  **Dual Logit Extraction**:
    
    -   **Expert Logits**: Obtained from the final (higher) layer, rich in factual content.
        
    -   **Amateur Logits**: Derived from an intermediate (lower) layer, containing more syntactic information.
        
2.  **Logit Differencing**:
    
    -   Compute the difference: `Expert Logits - Amateur Logits`.
        
    -   This highlights the factual components by amplifying signals from the higher layer and suppressing less relevant syntactic cues from the lower layer.
        
3.  **Adjusted Token Selection**:
    
    -   The resulting logits are used to generate a new probability distribution.
        
    -   Token selection proceeds as before (greedy or sampling), but now guided by this adjusted distribution that emphasizes factual accuracy.
        
#### **Benefits of DoLa:**

- ✅ **Reduced Hallucinations:**  
  Encourages the model to prefer factually grounded outputs by emphasizing representations from higher layers.

- ⚡ **Training-Free Deployment:**  
  Applied **only at inference time**, requiring no re-training, making it computationally efficient and easy to adopt.

---

### ✅ **Summary **  
> DoLa improves factual accuracy in LLMs by contrasting logits from high and mid transformer layers during inference, with no need for retraining.

> DoLa leverages the layer-wise knowledge hierarchy in Transformers to boost factual grounding. It's a lightweight, inference-only technique that effectively reduces hallucinations.



### **Beam Search Explanation**

**Beam Search** is an optimization technique for generating sequences (like text) in language models (LLMs). It's used to improve the quality of the generated text by keeping track of the top possible sequences rather than just choosing the single most probable token at each step, like in **greedy decoding**.

---

### 🧠 **How Beam Search Works:**

Imagine you're generating a sequence of tokens (e.g., a sentence). At each step, you can choose the next token from a set of possible tokens. In greedy decoding, you’d simply pick the most likely token each time. But with beam search, instead of just picking one token, you explore multiple possible sequences and select the **best path**.

---

#### **Step-by-Step Process of Beam Search**:

1. **Start with an initial token (e.g., <start>)**.
2. **Expand all possible next tokens**: The model computes the probability distribution over the entire vocabulary for the next token.
3. **Keep top *k* candidates**: Instead of keeping just one candidate, you store the *k* most probable sequences. This *k* is called the **beam width**.
4. **Repeat for each step**:
   - For each of the *k* sequences, expand it by considering all possible next tokens.
   - Add the new tokens to the sequences and score the sequences based on the cumulative probability (product of the probabilities at each step).
   - Keep the top *k* sequences after this step.
5. **End when a stopping condition is met**: This can be when a specific token is generated (like <end>), or after a certain number of steps.

---

### 📊 **Example with Beam Width = 3:**

Let’s say you’re generating a 3-token sequence and you have a beam width of 3. The model is generating the next word after the input "The cat".

#### **Step 1: Initial Token**
Start with the token “The”.

- Next token probabilities could be:
  - **"cat"**: 0.6
  - **"dog"**: 0.3
  - **"sat"**: 0.1

#### **Step 2: Expand Tokens**
Now, expand each of the top 3 candidates.

For "The cat" (with probability 0.6):
- Next tokens: 
  - **"sat"**: 0.5
  - **"jumped"**: 0.4
  - **"ran"**: 0.1

For "The dog" (with probability 0.3):
- Next tokens:
  - **"barked"**: 0.4
  - **"ran"**: 0.3
  - **"sat"**: 0.2

For "The sat" (with probability 0.1):
- Next tokens:
  - **"on"**: 0.7
  - **"mat"**: 0.2
  - **"roof"**: 0.1

#### **Step 3: Keep Top *k* Sequences (Beam Width = 3)**

After expanding each sequence, you keep the top 3 highest scoring sequences. For example:

1. **"The cat sat"** (0.6 * 0.5 = 0.3)
2. **"The dog barked"** (0.3 * 0.4 = 0.12)
3. **"The sat on"** (0.1 * 0.7 = 0.07)

#### **Step 4: Continue Expansion**

This process continues until you reach a stopping condition, like the sequence `<end>` or a maximum length.

---

### 🎯 **Key Parameters**:

- **Beam Width (k)**: The number of possible sequences to keep at each step.
  - A larger beam width improves output quality but is computationally more expensive.
  - A smaller beam width might lead to less diversity or poor-quality output.
- **Probability Scoring**: Sequences are scored based on the cumulative product of the probabilities at each step.

---

### 🔍 **Advantages of Beam Search**:

1. **Better Sequences**: By exploring multiple hypotheses at each step, beam search tends to produce more fluent and coherent sequences than greedy decoding.
2. **Deterministic**: Unlike methods like sampling, beam search produces the same output for the same input every time.

---

### ❌ **Disadvantages of Beam Search**:

1. **Computationally Expensive**: As the beam width increases, the number of sequences grows exponentially, leading to higher computational cost.
2. **Risk of Repetition**: It might still suffer from repeated or generic sequences if the beam width is too small or doesn't consider diversity.

---

### 🧵 **Summary (One-liner for Interview)**:

> **Beam search is an optimized decoding technique that keeps track of the top *k* most probable sequences at each step to improve text generation quality, balancing exploration with computational efficiency.**

---

### 💡 **2-liner Takeaway**:

Beam search is more efficient than greedy decoding because it explores multiple potential token sequences, not just the most likely one. However, it’s computationally expensive, and the beam width needs careful tuning for the best results.

# Common Fine-Tuning Failures & Debugging Guide

## 1. Loss Not Decreasing
| **Possible Cause** | **Diagnosis** | **Fix** |
|---|---|---|
| Learning rate too low | Loss flat from the start | Increase LR by 3-10x; try 2e-5 for full FT or 2e-4 for LoRA |
| Learning rate too high | Loss spikes or oscillates | Reduce LR; add warmup steps (5-10% of total) |
| Data formatting error | Loss stays at random baseline (~ln(vocab_size)) | Check that labels are correctly aligned with inputs; verify chat template |
| Frozen parameters | Loss does not move at all | Verify which parameters require grad; check PEFT target modules |

## 2. Loss Decreases But Output Quality is Poor
| **Possible Cause** | **Diagnosis** | **Fix** |
|---|---|---|
| Overfitting | Train loss drops, eval loss increases | Reduce epochs (1-3), add dropout, use label smoothing, increase data |
| Data contamination | Model memorizes responses verbatim | Deduplicate dataset; remove benchmark/test set leaks |
| Poor data quality | Outputs mimic bad examples | Audit random samples; filter low-quality entries |
| Wrong chat template | Outputs include template tokens or gibberish | Match template exactly to the base model's expected format |

## 3. Catastrophic Forgetting
| **Symptom** | **Fix** |
|---|---|
| Good on fine-tuned task, terrible on general tasks | Mix 5-20% general-purpose data into training set |
| Model loses coherence or language ability | Use LoRA instead of full fine-tuning; lower LR |
| Domain knowledge disappears | Apply replay strategy; use DAPT before SFT |

## 4. Out-of-Memory (OOM) Errors
**Quick fixes (ordered by impact):**
1. Enable gradient checkpointing (`gradient_checkpointing=True`)
2. Reduce batch size and increase gradient accumulation steps
3. Reduce max sequence length
4. Switch to QLoRA (4-bit quantization + LoRA)
5. Use DeepSpeed ZeRO Stage 2 or 3
6. Enable CPU offloading for optimizer states

## 5. Training is Extremely Slow
- **Check GPU utilization**: Use `nvidia-smi` -- if GPU util < 80%, the bottleneck is data loading
- **Enable `group_by_length=True`** to reduce padding waste
- **Use `packing=True`** (TRL) to pack multiple short examples per sequence
- **Increase batch size** if GPU memory allows
- **Use bf16** instead of fp32 for 2x speedup on Ampere+ GPUs

## 6. LoRA-Specific Issues
| **Issue** | **Fix** |
|---|---|
| LoRA rank too low, poor performance | Increase `r` from 8 to 32 or 64 |
| LoRA only targets attention, missing MLP | Add `gate_proj`, `up_proj`, `down_proj` to target modules |
| Merging LoRA degrades quality | Check for dtype mismatches during merge; merge in fp16/bf16 |

---

> **Interview Tip**: When asked about debugging fine-tuning, structure your answer as: (1) check the loss curve, (2) check data quality, (3) check hyperparameters, (4) check compute constraints.

# Top 10 Fine-Tuning Interview Q&A (30-Second Answers)

---

**Q1: What is the difference between fine-tuning and prompt engineering?**

> **A:** Prompt engineering adapts the model's behavior at inference time by crafting better inputs -- no weights change. Fine-tuning actually updates model parameters on task-specific data, producing a permanently modified model. Fine-tuning gives deeper behavioral changes and better consistency, but requires data and compute. Use prompt engineering first; fine-tune when it is not enough.

---

**Q2: When would you choose LoRA over full fine-tuning?**

> **A:** LoRA is preferred when you have limited GPU memory, moderate data (1K-50K examples), or need to maintain multiple task-specific adapters. It trains only 0.1-1% of parameters, reducing memory by 3-10x. Full fine-tuning is better when you have abundant data (>50K), sufficient compute, and the task requires significant behavioral shifts from the base model.

---

**Q3: What is catastrophic forgetting and how do you prevent it?**

> **A:** Catastrophic forgetting is when a model loses previously learned knowledge while learning new tasks. Prevention strategies: (1) use LoRA instead of full fine-tuning, (2) mix 5-20% general-purpose data into training, (3) use lower learning rates, (4) apply replay/distillation techniques, (5) perform continual pre-training (DAPT) before SFT.

---

**Q4: How do you decide on the learning rate for fine-tuning?**

> **A:** For full fine-tuning, use 1e-5 to 5e-5 (lower because all parameters update). For LoRA/QLoRA, use 1e-4 to 3e-4 (higher because fewer parameters update). Always use warmup (3-10% of steps) and a cosine scheduler. If loss does not decrease, increase LR. If loss is unstable, decrease LR.

---

**Q5: What is the difference between DAPT and TAPT?**

> **A:** DAPT (Domain-Adaptive Pre-Training) exposes the model to a large corpus of domain-specific text (e.g., all medical literature) using self-supervised learning. TAPT (Task-Adaptive Pre-Training) further narrows this to data similar to the specific downstream task. The recommended pipeline is: Base Model --> DAPT --> TAPT --> SFT on labeled data. DAPT is done first because it is broader.

---

**Q6: How much data do you need for fine-tuning?**

> **A:** It depends on the method and task. The LIMA paper showed 1,000 high-quality examples can be sufficient for instruction-tuning. General guidelines: Prompt tuning needs <100 examples, LoRA works well with 1K-50K, full fine-tuning benefits from 10K+. Quality matters more than quantity -- 1K curated examples often beat 100K noisy ones.

---

**Q7: What is QLoRA and when should you use it?**

> **A:** QLoRA combines 4-bit quantization of the base model with LoRA adapters trained in higher precision. It reduces memory by ~4x compared to standard LoRA, enabling fine-tuning of 7-13B models on a single 24GB consumer GPU. Use QLoRA when you are memory-constrained. The quality trade-off is minimal (typically <1% degradation on benchmarks).

---

**Q8: How do you evaluate a fine-tuned model?**

> **A:** Use a multi-pronged approach: (1) automated metrics (perplexity, BLEU/ROUGE for generation, accuracy for classification), (2) human evaluation on 50-100 examples rated for relevance, correctness, and fluency, (3) A/B comparison against the base model, (4) regression testing on general benchmarks to check for catastrophic forgetting. No single metric is sufficient.

---

**Q9: What are chat templates and why do they matter?**

> **A:** Chat templates define the exact token format the model expects for system/user/assistant messages (e.g., ChatML uses `<|im_start|>` tokens, Llama 3 uses `<|start_header_id|>`). Using the wrong template is one of the most common fine-tuning mistakes -- it causes the model to output gibberish or template tokens. Always match your data format to the base model's expected template exactly.

---

**Q10: Walk me through the SFT pipeline end-to-end.**

> **A:** (1) Define the objective and success metrics. (2) Collect and curate 1K-50K high-quality instruction-response pairs. (3) Format data to match the model's chat template. (4) Choose a base model based on task complexity and compute budget. (5) Configure hyperparameters (LR, batch size, epochs, scheduler). (6) Train using SFTTrainer with LoRA, monitoring loss curves. (7) Evaluate with automated metrics + human eval. (8) Merge LoRA weights, quantize for serving, deploy. (9) Collect user feedback and iterate.
