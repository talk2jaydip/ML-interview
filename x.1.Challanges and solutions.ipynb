<tr style="background:#f8fbff;">
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>1. Embedding Space Misalignment</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Fine-tuned embeddings don’t reflect task-specific similarity.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Leads to irrelevant document retrieval → affects the generator’s output.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">✅ Use <strong>task-specific hard negatives</strong> (closely related but wrong context).&lt;br&gt;✅ Fine-tune with <strong>triplet or cosine loss</strong> on labeled or semi-labeled data.</td>
</tr>
<tr>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>2. Domain Drift in Pretrained Models</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Generic embeddings can't capture domain-specific terms/jargon.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Pretrained models may retrieve unrelated content in specialized fields.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">✅ Fine-tune using <strong>domain-specific pairs</strong>.&lt;br&gt;✅ Include <strong>jargon-heavy examples</strong> in your dataset.</td>
</tr>
<tr style="background:#f8fbff;">
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>3. Limited Labeled Data</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Annotating similarity or relevance is costly/time-consuming.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Hard to create positive/negative training pairs at scale.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">✅ Use <strong>heuristics</strong>: cosine threshold, BM25 ranking as pseudo-labels.&lt;br&gt;✅ Try <strong>semi-supervised bootstrapping</strong>.</td>
</tr>
<tr>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>4. Overfitting on Small Datasets</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Fine-tuned model memorizes training examples but generalizes poorly.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Retrieval accuracy drops on unseen queries.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">✅ Use <strong>early stopping</strong>, <strong>dropout</strong>, and <strong>regular validation</strong>.&lt;br&gt;✅ Inject <strong>diversity in examples</strong> (hard negatives, near-miss pairs).</td>
</tr>
<tr style="background:#f8fbff;">
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>5. Suboptimal Retrieval Latency</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">FAISS retrieval becomes slow with large-scale embeddings.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Hurts real-time RAG pipeline latency.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">✅ Use <strong>HNSW index</strong> instead of flat index.&lt;br&gt;✅ Use <strong>quantized or compressed indexes (PQ)</strong>.&lt;br&gt;✅ Enable <strong>FAISS GPU acceleration</strong>.</td>
</tr>
<tr>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>6. Mismatched Query/Document Format</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Training pairs don’t reflect production usage.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Model learns a different similarity pattern than what's used at inference.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">✅ Ensure <strong>query/document formatting in training</strong> matches retrieval use case (e.g., questions vs. passages).</td>
</tr>
<tr style="background:#f8fbff;">
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>7. Evaluation Blind Spots</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Relying only on metrics like Recall@k or MRR.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Can miss edge cases or mislead tuning efforts.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">✅ Combine <strong>quantitative (Recall@K, MRR)</strong> and <strong>qualitative evaluations</strong>.&lt;br&gt;✅ Regularly inspect retrieval outputs manually.</td>
</tr>
</tbody>
</table>

<table style="width:100%;border-collapse:collapse;background:#ffffff;font-size:0.96rem;margin:16px 0;border:1px solid #dde6ec;border-radius:16px;overflow:hidden;">
<thead>
<tr>
<th style="background:#f3f6f8;color:#24415c;padding:10px 12px;text-align:left;border:1px solid #dde6ec;font-weight:700;"><strong>Challenge</strong></th>
<th style="background:#f3f6f8;color:#24415c;padding:10px 12px;text-align:left;border:1px solid #dde6ec;font-weight:700;"><strong>What Can Go Wrong</strong></th>
<th style="background:#f3f6f8;color:#24415c;padding:10px 12px;text-align:left;border:1px solid #dde6ec;font-weight:700;"><strong>Solutions &amp; Best Practices</strong></th>
</tr>
</thead>
<tbody>
<tr style="background:#f8fbff;">
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>1. Noisy or Unclean Text Data</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Misspellings, inconsistent formatting, duplicates → noisy embeddings.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">✅ Apply text normalization (remove HTML, special characters).&lt;br&gt;✅ Deduplicate entries.</td>
</tr>
<tr>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>2. Lack of Diversity in Examples</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Biased training leads to poor generalization.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">✅ Include a mix of easy, medium, and <strong>hard negative pairs</strong>.&lt;br&gt;✅ Cover multiple query styles.</td>
</tr>
<tr style="background:#f8fbff;">
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>3. Ambiguous or Irrelevant Pairs</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Misleading labels confuse the model during training.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">✅ Manually review a small % of dataset.&lt;br&gt;✅ Use heuristics + confidence thresholds.</td>
</tr>
<tr>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>4. Dataset Domain Mismatch</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Using general datasets for highly specialized use cases.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">✅ Build datasets from <strong>internal KBs</strong>, documentation, or forums from your domain.</td>
</tr>
<tr style="background:#f8fbff;">
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>5. Poor Tokenization Strategy</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Long inputs get truncated → loss of meaning.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">✅ Keep <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">max_length</code> consistent (128–256 tokens works best).&lt;br&gt;✅ Test truncation impact.</td>
</tr>
<tr>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>6. Unbalanced Positive/Negative Ratio</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Dominance of one type skews model’s similarity judgment.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">✅ Maintain roughly 1:1 ratio of positive:negative pairs.&lt;br&gt;✅ Sample smart negatives.</td>
</tr>
<tr style="background:#f8fbff;">
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>7. Static Dataset Curation</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">No adaptation to evolving queries in production.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">✅ Collect real production queries and periodically <strong>refresh training data</strong>.</td>
</tr>
</tbody>
</table>
