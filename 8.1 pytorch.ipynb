**A.  Tensors: The Foundation**

As described in the provided text, tensors are the fundamental data structure in PyTorch.  They are multi-dimensional arrays, similar to NumPy's `ndarray`, but with crucial additions for deep learning:

*   **GPU Support:**  PyTorch tensors can reside on GPUs, enabling significant acceleration of computations. This is *essential* for training large models like LLMs.
*   **Automatic Differentiation:** PyTorch tracks operations performed on tensors, allowing for automatic gradient computation (more on this later).

### PyTorch `squeeze()` and `unsqueeze()` Operations Explained

These two operations are used to manipulate tensor dimensions, specifically adding or removing dimensions of size 1.

---

### **1. `torch.squeeze()`**
#### **Functionality**: Removes dimensions of size **1** from a tensor.
#### **Syntax**:
```python
torch.squeeze(input, dim=None)
```
- `input`: The input tensor.
- `dim` (optional): The specific dimension to squeeze. If not provided, all dimensions of size `1` are removed.

#### **Example 1: Removing all dimensions of size 1**
```python
import torch

x = torch.rand(1, 3, 1, 5)  # Shape: [1, 3, 1, 5]
y = torch.squeeze(x)        # Shape: [3, 5]

print(y.shape)  # Output: torch.Size([3, 5])
```

#### **Example 2: Removing a specific dimension**
```python
x = torch.rand(1, 3, 1, 5)   # Shape: [1, 3, 1, 5]
y = torch.squeeze(x, dim=0)  # Removes the first dimension
print(y.shape)  # Output: torch.Size([3, 1, 5])
```
> If the specified `dim` is **not** of size `1`, the tensor remains unchanged.

---

### **2. `torch.unsqueeze()`**
#### **Functionality**: Adds a new dimension of size **1** at the specified index.
#### **Syntax**:
```python
torch.unsqueeze(input, dim)
```
- `input`: The input tensor.
- `dim`: The dimension where a new axis (size 1) should be inserted.

#### **Example 1: Adding a new dimension**
```python
x = torch.rand(3, 5)       # Shape: [3, 5]
y = torch.unsqueeze(x, 0)  # Adds a new dimension at index 0

print(y.shape)  # Output: torch.Size([1, 3, 5])
```

#### **Example 2: Adding a new dimension at the end**
```python
y = torch.unsqueeze(x, -1)  # Adds a new dimension at the last index
print(y.shape)  # Output: torch.Size([3, 5, 1])
```

---

### **Use Cases**
- `squeeze()`: Useful for removing unnecessary singleton dimensions.
- `unsqueeze()`: Useful when reshaping tensors for operations like broadcasting or passing inputs to models expecting specific shapes.

### Quick Review

````python
import torch

# Creating a tensor of shape (64, 3, 32, 32) - let's say it represents a batch of 64 images with 3 channels (RGB) and size 32x32.
batch = torch.randn(64, 3, 32, 32)

# Flattening the tensor to shape (64, -1), where -1 means infer the size based on other dimensions
flattensor = batch.view(64, -1)

# Printing the shape of the flattened tensor
print(f"Flattened tensor shape: {flattensor.shape}")  # Expected: torch.Size([64, 3072])
````

````python
import torch

# Create a tensor of shape (16, 3, 64, 64) representing a batch of 16 RGB images of size 64x64
batch = torch.randn(16, 3, 64, 64)

# Flatten the tensor to shape (16, -1)
flattened_batch = batch.view(16, -1)

# Print the shape of the flattened tensor
print(f"Flattened batch shape: {flattened_batch.shape}")  # Expected: torch.Size([16, 12288])
````

````python
import torch

# Create a tensor of size (2, 3, 4)
tensor = torch.randn(2, 3, 4)

# Use .view() to flatten this tensor into a 2D tensor
flattened_tensor = tensor.view(2, -1)

print(f"Flattened tensor: {flattened_tensor.shape}")  # Expected: torch.Size([2, 12])
````

````python
import torch

# Create a tensor of shape (64, 2, 2)
tensor = torch.randn(64, 2, 2)

# Reshaping the tensor into (64, -1) (flattening the last 2 dimensions)
reshaped_tensor = tensor.view(64, -1)

# Print the shape of the reshaped tensor
print(f"Reshaped tensor: {reshaped_tensor.shape}")  # Expected: torch.Size([64, 4])
````

````python
import torch

# Create a 3D tensor of shape (128, 3, 256)
tensor_3d = torch.randn(128, 3, 256)

# Flatten the tensor into 2D with 128 rows and the product of 3 * 256 columns
flattened_tensor = tensor_3d.view(128, -1)

print(f"Flattened tensor: {flattened_tensor.shape}")  # Expected: torch.Size([128, 768])
````

````python
import torch

# Create a tensor of shape (4, 3, 6)
tensor = torch.randn(4, 3, 6)

# Use .view() with `-1` to flatten it along a specific dimension
flattened_tensor = tensor.view(4, -1)  # Automatically infer the correct size

print(f"Flattened tensor shape: {flattened_tensor.shape}")  # Expected: torch.Size([4, 18])
````

````python
import torch

# Create a tensor of shape (2, 3, 4)
tensor = torch.randn(2, 3, 4)

# Transpose the tensor, changing its layout
transposed_tensor = tensor.transpose(1, 2)  # Swap the 2nd and 3rd dimensions

# Make the tensor contiguous (needed after transposition for reshaping)
contiguous_tensor = transposed_tensor.contiguous()

# Now reshape it
reshaped_tensor = contiguous_tensor.view(2, -1)

print(f"Reshaped tensor after transposition: {reshaped_tensor.shape}")  # Expected: torch.Size([2, 12])
````

````python
import torch
import torch.nn as nn

# Create a sample input tensor of shape (1, 256, 8, 8) simulating a batch of images
input_tensor = torch.randn(1, 256, 8, 8)

# Define a simple linear layer after flattening the input
class FlattenAndLinear(nn.Module):
    def __init__(self):
        super(FlattenAndLinear, self).__init__()
        self.flatten = nn.Flatten()  # Flatten the input tensor
        self.fc = nn.Linear(256*8*8, 10)  # Fully connected layer (10 output classes)

    def forward(self, x):
        x = self.flatten(x)  # Flatten the input
        x = self.fc(x)  # Pass through a fully connected layer
        return x

# Instantiate the model
model = FlattenAndLinear()

# Pass the input tensor through the model
output = model(input_tensor)

print(f"Output shape: {output.shape}")  # Expected: torch.Size([1, 10])
````

|  | `squeeze()`   | Remove dimensions of size 1. | - Removing singleton dimensions from model outputs (e.g., removing a batch dimension of 1 after processing a single example).  - Getting rid of unnecessary dimensions after certain operations. | - Useful for cleaning up tensor shapes. ## A.3 Seeing models as computation graphs

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/appendix-a_compressed/7.webp" width="600px">

> PyTorch builds such a computation graph in the background, and we can use this to calculate gradients of a loss function with respect to the model parameters (here w1 and b) to train the model.

> If we carry out computations in PyTorch, it will build a computational graph internally by default if one of its terminal nodes has the requires_grad attribute set to True.

````python
import torch.nn.functional as F

y = torch.tensor([1.0])  # true label
x1 = torch.tensor([1.1]) # input feature
w1 = torch.tensor([2.2]) # weight parameter
b = torch.tensor([0.0])  # bias unit

z = x1 * w1 + b          # net input
a = torch.sigmoid(z)     # activation & output

loss = F.binary_cross_entropy(a, y)
print(loss)
````

## A.4 Automatic differentiation made easy

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/appendix-a_compressed/8.webp" width="600px">

##### Figure A.8 The most common way of computing the loss gradients in a computation graph involves applying the chain rule from right to left, also called reverse-model automatic differentiation or backpropagation. We start from the output layer (or the loss itself) and work backward through the network to the input layer. We do this to compute the gradient of the loss with respect to each parameter (weights and biases) in the network, which informs how we update these parameters during training.

````python
import torch.nn.functional as F
from torch.autograd import grad

y = torch.tensor([1.0])
x1 = torch.tensor([1.1])
w1 = torch.tensor([2.2], requires_grad=True)
b = torch.tensor([0.0], requires_grad=True)

z = x1 * w1 + b 
a = torch.sigmoid(z)

loss = F.binary_cross_entropy(a, y)

grad_L_w1 = grad(loss, w1, retain_graph=True)
grad_L_b = grad(loss, b, retain_graph=True)

print(grad_L_w1)
print(grad_L_b)
````

> we can call .backward on the loss, and PyTorch will compute the gradients of all the leaf nodes in the graph, which will be stored via the tensors’ .grad attributes:

````python
loss.backward()

print(w1.grad)
print(b.grad)
````

## A.5 Implementing multilayer neural networks

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/appendix-a_compressed/9.webp" width="500px">

````python
class NeuralNetwork(torch.nn.Module):
    def __init__(self, num_inputs, num_outputs):
        super().__init__()

        self.layers = torch.nn.Sequential(
                
            # 1st hidden layer
            torch.nn.Linear(num_inputs, 30),
            torch.nn.ReLU(),

            # 2nd hidden layer
            torch.nn.Linear(30, 20),
            torch.nn.ReLU(),

            # output layer
            torch.nn.Linear(20, num_outputs),
        )

    def forward(self, x):
        logits = self.layers(x)
        return logits
````

````python
model = NeuralNetwork(50, 3)
````

> for example, if we use it for `prediction after training—constructing this computational graph for backpropagation can be wasteful as it performs unnecessary computations and consumes additional memory`. So, when we use a model for inference (for instance, making predictions) rather than training, `the best practice is to use the torch.no_grad() context manager`. `This tells PyTorch that it doesn’t need to keep track of the gradients, which can result in significant savings in memory and computation`:

````python
with torch.no_grad():
    out = torch.softmax(model(X), dim=1)
print(out)
````

## A.6 Setting up efficient data loaders

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/appendix-a_compressed/10.webp" width="600px">

````python
X_train = torch.tensor([
    [-1.2, 3.1],
    [-0.9, 2.9],
    [-0.5, 2.6],
    [2.3, -1.1],
    [2.7, -1.5]
])

y_train = torch.tensor([0, 0, 0, 1, 1])
````

````python
X_test = torch.tensor([
    [-0.8, 2.8],
    [2.6, -1.6],
])

y_test = torch.tensor([0, 1])
````

````python
from torch.utils.data import Dataset


class ToyDataset(Dataset):
    def __init__(self, X, y):
        self.features = X
        self.labels = y

    def __getitem__(self, index):
        one_x = self.features[index]
        one_y = self.labels[index]        
        return one_x, one_y

    def __len__(self):
        return self.labels.shape[0]

train_ds = ToyDataset(X_train, y_train)
test_ds = ToyDataset(X_test, y_test)
````

````python
from torch.utils.data import DataLoader

torch.manual_seed(123)

train_loader = DataLoader(
    dataset=train_ds,
    batch_size=2,
    shuffle=True,
    num_workers=0
)
````

````python
test_ds = ToyDataset(X_test, y_test)

test_loader = DataLoader(
    dataset=test_ds,
    batch_size=2,
    shuffle=False,
    num_workers=0
)
````

## A.7 A typical training loop

````python
import torch.nn.functional as F


torch.manual_seed(123)
model = NeuralNetwork(num_inputs=2, num_outputs=2)
optimizer = torch.optim.SGD(model.parameters(), lr=0.5)

num_epochs = 3

for epoch in range(num_epochs):
    
    model.train()
    for batch_idx, (features, labels) in enumerate(train_loader):

        logits = model(features)
        
        loss = F.cross_entropy(logits, labels) # Loss function
        
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
    
        ### LOGGING
        print(f"Epoch: {epoch+1:03d}/{num_epochs:03d}"
              f" | Batch {batch_idx:03d}/{len(train_loader):03d}"
              f" | Train/Val Loss: {loss:.2f}")

    model.eval()
    # Optional model evaluation
````

> To prevent undesired gradient accumulation, it is important to include an optimizer.zero_grad() call in each update round to reset the gradients to 0. Otherwise, the gradients will accumulate, which may be undesired.

````python
model.eval()

with torch.no_grad():
    outputs = model(X_train)

print(outputs)
````

````python
torch.set_printoptions(sci_mode=False)
probas = torch.softmax(outputs, dim=1)
print(probas)

predictions = torch.argmax(probas, dim=1)
print(predictions)
````

````python
def compute_accuracy(model, dataloader):

    model = model.eval()
    correct = 0.0
    total_examples = 0
    
    for idx, (features, labels) in enumerate(dataloader):
        
        with torch.no_grad():
            logits = model(features)
        
        predictions = torch.argmax(logits, dim=1)
        compare = labels == predictions
        correct += torch.sum(compare)
        total_examples += len(compare)

    return (correct / total_examples).item()
````

````python
compute_accuracy(model, train_loader)
````

````python
compute_accuracy(model, test_loader)
````
