## 1. Multimodal PDF Processor

Class to extract text, images, and tables from PDFs


````python
class MultimodalPDFProcessor:
    """Process PDFs to extract text, images, and tables"""
    
    def __init__(self, pdf_path: str):
        self.pdf_path = pdf_path
        self.doc = fitz.open(pdf_path)
        
    def extract_text_chunks(self, chunk_size: int = 1000, chunk_overlap: int = 200) -> List[Document]:
        """Extract and chunk text from PDF"""
        logger.info("Extracting text from PDF...")
        
        # Extract full text
        full_text = ""
        for page_num in range(len(self.doc)):
            page = self.doc[page_num]
            full_text += f"\n\n--- Page {page_num + 1} ---\n\n"
            full_text += page.get_text()
        
        # Split into chunks
        text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=chunk_size,
            chunk_overlap=chunk_overlap,
            length_function=len,
        )
        
        chunks = text_splitter.create_documents([full_text])
        
        # Add metadata
        for i, chunk in enumerate(chunks):
            chunk.metadata['chunk_id'] = i
            chunk.metadata['type'] = 'text'
            chunk.metadata['source'] = self.pdf_path
        
        logger.info(f"Extracted {len(chunks)} text chunks")
        return chunks
    
    def extract_images(self) -> List[Dict[str, Any]]:
        """Extract images from PDF with metadata"""
        logger.info("Extracting images from PDF...")
        
        images = []
        for page_num in range(len(self.doc)):
            page = self.doc[page_num]
            image_list = page.get_images()
            
            for img_index, img in enumerate(image_list):
                try:
                    xref = img[0]
                    base_image = self.doc.extract_image(xref)
                    image_bytes = base_image["image"]
                    
                    images.append({
                        'page_num': page_num + 1,
                        'image_index': img_index,
                        'image_bytes': image_bytes,
                        'ext': base_image['ext'],
                        'width': base_image['width'],
                        'height': base_image['height']
                    })
                except Exception as e:
                    logger.warning(f"Error extracting image {img_index} from page {page_num + 1}: {e}")
        
        logger.info(f"Extracted {len(images)} images")
        return images
    
    def extract_tables_structured(self) -> List[Dict[str, Any]]:
        """Extract tables using unstructured library"""
        logger.info("Extracting tables from PDF...")
        
        try:
            # Use unstructured to extract elements
            elements = partition_pdf(
                filename=self.pdf_path,
                strategy="hi_res",
                infer_table_structure=True,
            )
            
            tables = []
            for element in elements:
                if isinstance(element, Table):
                    # Convert table to markdown
                    table_md = element.metadata.text_as_html if hasattr(element.metadata, 'text_as_html') else str(element)
                    
                    tables.append({
                        'type': 'table',
                        'content': table_md,
                        'metadata': {
                            'page_number': getattr(element.metadata, 'page_number', None),
                            'source': self.pdf_path
                        }
                    })
            
            logger.info(f"Extracted {len(tables)} tables")
            return tables
            
        except Exception as e:
            logger.error(f"Error extracting tables: {e}")
            return []
    
    def generate_image_captions(self, images: List[Dict], llm) -> List[Document]:
        """Generate captions for images using LLM with vision capabilities"""
        logger.info("Generating image captions...")
        
        # Note: This is a placeholder. In production, use a vision model like GPT-4V or CLIP
        # For this demo, we'll create metadata-based captions
        
        image_docs = []
        for i, img_data in enumerate(images):
            caption = f"Image from page {img_data['page_num']}, dimensions: {img_data['width']}x{img_data['height']}"
            
            doc = Document(
                page_content=caption,
                metadata={
                    'chunk_id': f"img_{i}",
                    'type': 'image',
                    'page_num': img_data['page_num'],
                    'source': self.pdf_path,
                    'image_index': img_data['image_index']
                }
            )
            image_docs.append(doc)
        
        logger.info(f"Generated {len(image_docs)} image captions")
        return image_docs
    
    def process(self, generate_captions: bool = False, llm=None) -> List[Document]:
        """Process PDF and return all document chunks"""
        all_docs = []
        
        # Extract text
        text_chunks = self.extract_text_chunks()
        all_docs.extend(text_chunks)
        
        # Extract images
        images = self.extract_images()
        if generate_captions and images:
            image_docs = self.generate_image_captions(images, llm)
            all_docs.extend(image_docs)
        
        # Extract tables
        tables = self.extract_tables_structured()
        for i, table in enumerate(tables):
            doc = Document(
                page_content=table['content'],
                metadata={
                    'chunk_id': f"table_{i}",
                    'type': 'table',
                    'source': self.pdf_path,
                    **table['metadata']
                }
            )
            all_docs.append(doc)
        
        logger.info(f"Total documents extracted: {len(all_docs)}")
        return all_docs
    
    def close(self):
        """Close the PDF document"""
        self.doc.close()
````

````python
class MultimodalRAG:
    """Complete Multimodal RAG System"""
    
    def __init__(self, embedding_model_name: str = "sentence-transformers/all-MiniLM-L6-v2"):
        """Initialize the RAG system"""
        self.embedding_model_name = embedding_model_name
        
        # Initialize embeddings
        logger.info(f"Loading embedding model: {embedding_model_name}")
        self.embeddings = HuggingFaceEmbeddings(
            model_name=embedding_model_name,
            model_kwargs={'device': 'cpu'},
            encode_kwargs={'normalize_embeddings': True}
        )
        
        self.vectorstore = None
        self.retriever = None
        
    def create_vectorstore(self, documents: List[Document], use_faiss: bool = True):
        """Create vector store from documents"""
        logger.info(f"Creating vector store from {len(documents)} documents...")
        
        if use_faiss:
            self.vectorstore = FAISS.from_documents(documents, self.embeddings)
        else:
            self.vectorstore = Chroma.from_documents(
                documents=documents,
                embedding=self.embeddings,
                persist_directory="./chroma_db"
            )
        
        logger.info("Vector store created successfully")
        
    def setup_retriever(self, search_type: str = "similarity", k: int = 4):
        """Setup retrieval mechanism"""
        if self.vectorstore is None:
            raise ValueError("Vector store not created. Call create_vectorstore first.")
        
        if search_type == "similarity":
            self.retriever = self.vectorstore.as_retriever(
                search_type="similarity",
                search_kwargs={"k": k}
            )
        elif search_type == "mmr":
            self.retriever = self.vectorstore.as_retriever(
                search_type="mmr",
                search_kwargs={"k": k, "fetch_k": k*2}
            )
        else:
            raise ValueError(f"Unknown search type: {search_type}")
        
        logger.info(f"Retriever setup with {search_type} search, k={k}")
        
    def save_vectorstore(self, path: str):
        """Save vector store to disk"""
        if self.vectorstore is None:
            raise ValueError("Vector store not created")
        
        if isinstance(self.vectorstore, FAISS):
            self.vectorstore.save_local(path)
            logger.info(f"Vector store saved to {path}")
    
    def load_vectorstore(self, path: str):
        """Load vector store from disk"""
        self.vectorstore = FAISS.load_local(path, self.embeddings, allow_dangerous_deserialization=True)
        logger.info(f"Vector store loaded from {path}")
````

## 3. LLM Setup and QA Chain

Configure the language model for answer generation


````python
def setup_llm(model_name: str = "gpt-3.5-turbo", temperature: float = 0.0):
    """Setup the LLM for question answering"""
    if model_name.startswith("gpt"):
        llm = ChatOpenAI(
            model_name=model_name,
            temperature=temperature,
        )
    else:
        # For local models, use HuggingFace
        from langchain_community.llms import HuggingFacePipeline
        from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
        
        tokenizer = AutoTokenizer.from_pretrained(model_name)
        model = AutoModelForCausalLM.from_pretrained(model_name)
        
        hf_pipeline = pipeline(
            "text-generation",
            model=model,
            tokenizer=tokenizer,
            max_new_tokens=512,
            temperature=temperature,
        )
        llm = HuggingFacePipeline(pipeline=hf_pipeline)
    
    return llm

def create_qa_chain(llm, retriever, chain_type: str = "stuff"):
    """Create a QA chain with retriever"""
    
    # Custom prompt template
    prompt_template = """Use the following pieces of context to answer the question at the end.
If you don't know the answer, just say that you don't know, don't try to make up an answer.

Context may include:
- Text passages from the document
- Information from tables
- Descriptions of images

Context: {context}

Question: {question}

Provide a detailed answer based only on the context provided:"""
    
    prompt = ChatPromptTemplate.from_template(prompt_template)
    
    qa_chain = RetrievalQA.from_chain_type(
        llm=llm,
        chain_type=chain_type,
        retriever=retriever,
        return_source_documents=True,
        chain_type_kwargs={"prompt": prompt}
    )
    
    return qa_chain
````

## 5. Complete End-to-End Workflow

Run the complete multimodal RAG pipeline


````python
# ============================================
# COMPLETE MULTIMODAL RAG WORKFLOW
# ============================================

def run_multimodal_rag(pdf_path: str, questions: List[str]):
    """
    Complete end-to-end multimodal RAG workflow
    
    Args:
        pdf_path: Path to the PDF file
        questions: List of questions to ask
    """
    
    print("=" * 60)
    print("MULTIMODAL RAG - END TO END WORKFLOW")
    print("=" * 60)
    
    # Step 1: Process PDF
    print("\n[Step 1] Processing PDF: Extracting text, images, and tables...")
    processor = MultimodalPDFProcessor(pdf_path)
    documents = processor.process(generate_captions=True)
    processor.close()
    
    print(f"✓ Extracted {len(documents)} document chunks")
    print(f"  - Text chunks: {sum(1 for d in documents if d.metadata.get('type') == 'text')}")
    print(f"  - Image chunks: {sum(1 for d in documents if d.metadata.get('type') == 'image')}")
    print(f"  - Table chunks: {sum(1 for d in documents if d.metadata.get('type') == 'table')}")
    
    # Step 2: Create Vector Store
    print("\n[Step 2] Creating vector store and embeddings...")
    rag_system = MultimodalRAG()
    rag_system.create_vectorstore(documents, use_faiss=True)
    print("✓ Vector store created")
    
    # Step 3: Setup Retriever
    print("\n[Step 3] Setting up retriever...")
    rag_system.setup_retriever(search_type="similarity", k=4)
    print("✓ Retriever configured")
    
    # Step 4: Setup LLM
    print("\n[Step 4] Initializing LLM...")
    try:
        llm = setup_llm(model_name="gpt-3.5-turbo", temperature=0.0)
        print("✓ LLM initialized (GPT-3.5-turbo)")
    except Exception as e:
        print(f"⚠ Could not initialize OpenAI LLM: {e}")
        print("Using HuggingFace local model instead...")
        llm = setup_llm(model_name="microsoft/DialoGPT-medium", temperature=0.0)
        print("✓ LLM initialized (DialoGPT)")
    
    # Step 5: Create QA Chain
    print("\n[Step 5] Creating QA chain...")
    qa_chain = create_qa_chain(llm, rag_system.retriever)
    print("✓ QA chain ready")
    
    # Step 6: Answer Questions
    print("\n[Step 6] Answering questions...")
    print("=" * 60)
    
    results = []
    for i, question in enumerate(questions, 1):
        print(f"\n❓ Question {i}: {question}")
        print("-" * 60)
        
        try:
            result = qa_chain.invoke({"query": question})
            
            answer = result['result']
            sources = result['source_documents']
            
            print(f"💡 Answer: {answer}")
            print(f"\n📚 Sources ({len(sources)} retrieved):")
            
            for j, doc in enumerate(sources, 1):
                doc_type = doc.metadata.get('type', 'unknown')
                page_num = doc.metadata.get('page_num', 'N/A')
                content_preview = doc.page_content[:150] + "..." if len(doc.page_content) > 150 else doc.page_content
                
                print(f"  {j}. Type: {doc_type} | Page: {page_num}")
                print(f"     Content: {content_preview}")
            
            results.append({
                'question': question,
                'answer': answer,
                'sources': sources
            })
            
        except Exception as e:
            print(f"❌ Error: {e}")
            results.append({
                'question': question,
                'answer': f"Error: {e}",
                'sources': []
            })
    
    # Step 7: Save Vector Store
    print("\n[Step 7] Saving vector store...")
    try:
        rag_system.save_vectorstore("./multimodal_rag_db")
        print("✓ Vector store saved to ./multimodal_rag_db")
    except Exception as e:
        print(f"⚠ Could not save vector store: {e}")
    
    print("\n" + "=" * 60)
    print("✅ WORKFLOW COMPLETE")
    print("=" * 60)
    
    return results, rag_system

# Example usage
# questions = [
#     "What is the main topic of this document?",
#     "What tables are present in the document?",
#     "Describe the images in the document",
#     "Summarize the key findings"
# ]
# 
# pdf_path = "your_document.pdf"  # Replace with your PDF path
# results, rag_system = run_multimodal_rag(pdf_path, questions)
````

````python
# Advanced retrieval with metadata filtering
class AdvancedRetriever:
    """Enhanced retriever with metadata filtering and reranking"""
    
    def __init__(self, vectorstore, embedding_model):
        self.vectorstore = vectorstore
        self.embedding_model = embedding_model
    
    def retrieve_with_filters(self, query: str, k: int = 4, doc_type: str = None):
        """Retrieve documents with optional metadata filtering"""
        # Get raw documents
        docs = self.vectorstore.similarity_search(query, k=k*2)
        
        # Filter by document type if specified
        if doc_type:
            docs = [doc for doc in docs if doc.metadata.get('type') == doc_type]
        
        # Return top k
        return docs[:k]
    
    def rerank_documents(self, query: str, documents: List[Document], top_k: int = 3):
        """Rerank documents using cross-encoder (if available)"""
        # Simple reranking based on embedding similarity
        # In production, use a cross-encoder model
        
        query_embedding = self.embedding_model.embed_query(query)
        
        scores = []
        for doc in documents:
            doc_embedding = self.embedding_model.embed_query(doc.page_content)
            # Cosine similarity
            import numpy as np
            score = np.dot(query_embedding, doc_embedding) / (
                np.linalg.norm(query_embedding) * np.linalg.norm(doc_embedding)
            )
            scores.append((score, doc))
        
        # Sort by score descending
        scores.sort(key=lambda x: x[0], reverse=True)
        
        return [doc for _, doc in scores[:top_k]]

# Example usage
# advanced_retriever = AdvancedRetriever(rag_system.vectorstore, rag_system.embeddings)
# filtered_docs = advanced_retriever.retrieve_with_filters(
#     "What tables show statistics?", 
#     k=5, 
#     doc_type="table"
# )
# reranked_docs = advanced_retriever.rerank_documents("statistics", filtered_docs)
````
