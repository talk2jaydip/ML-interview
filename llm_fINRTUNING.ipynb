## Memory Optimization Parameters

- **Gradient Checkpointing**  
  - Saves memory by not storing all activations during the forward pass.  
  - Missing activations are recomputed during the backward pass, trading extra computation for reduced memory usage.

- **Gradient Accumulation**  
  - Simulates a larger batch size by accumulating gradients over multiple smaller batches.  
  - Reduces memory requirements at the cost of slower training updates.

- **Quantization**  
  - Uses reduced precision (e.g., bf16) to save memory without significant performance loss.

- **Example Settings (Political Promises Detector):**  
  ```python
  gradient_accumulation_steps = 4
  bf16 = True
  gradient_checkpointing = True
  ```

## Regularization Parameters

- **Label Smoothing**  
  - Softens hard target labels (e.g., from 0/1) to prevent overconfidence.  
  - Improves model calibration and is useful with noisy or subjective labels.

- **Noise Embeddings**  
  - Adds noise to input embeddings to reduce overfitting on stylistic features (formatting, wording, length).  
  - Employed only during training; can lead to longer, more informative outputs.  
  - Implemented using techniques such as NEFTune in Hugging Face.

## Batch Size and Learning Rate Relationship

- **Batch Size Considerations:**  
  - Larger batch sizes speed up training but require more memory.  
  - Smaller batches can introduce higher variance, possibly aiding convergence to flatter minima.  
  - Best practice: choose batch sizes as powers of two to optimize GPU I/O.

- **Learning Rate Trade-offs:**  
  - High learning rates may lead to faster convergence but risk overshooting minima.  
  - Lower learning rates may ensure careful convergence but require more steps.
  
- **Example Settings (Political Promises Detector):**

# 🎯 Label Smoothing: Improving Calibration & Reducing Overfitting  

## 🔍 What is Label Smoothing?  
Label smoothing is a **regularization technique** that helps:  
- **Reduce overfitting**  
- **Improve model calibration** (ensuring confidence scores reflect true accuracy)  

### 📌 Why Does Calibration Matter?  
A model is **well-calibrated** if its confidence scores match the actual probability of correctness.  

🔹 **Example:** Sentence classification (abusive vs. non-abusive)  
- If a well-calibrated model predicts **0.9 probability**, ~90% of those cases should be correct.  
- If it predicts **0.6 probability**, ~60% of cases should be correct.  

✅ **Good Calibration** → Output probabilities are a realistic measure of confidence.  
❌ **Poor Calibration** → Model is **overconfident**, leading to incorrect high-confidence predictions.  

> ⚠️ **Larger Models & Overconfidence**  
> Studies (Li et al.) show that **larger models** (e.g., GPT-like models) tend to be **less calibrated** compared to smaller models like BERT.  
> **Implication:** If uncertainty estimates matter, consider using **smaller models** or applying calibration techniques like label smoothing!  

---

## 🛠️ How Does Label Smoothing Work?  
### 🚀 Standard Training vs. Label Smoothing  
- **Without label smoothing:** The model learns **hard target labels** (`0` or `1`).  
- **With label smoothing:** The model **softens the labels** by subtracting a small value (`α`) from `1` and redistributing it across all labels.

# 🚀 Noise Embeddings in Fine-Tuning

Fine-tuning datasets are typically small (<50,000 examples), making models prone to overfitting to stylistic characteristics such as formatting, wording, and text length. A solution to this issue is **adding noise to input embeddings**.

## 🔍 Why Use Noise Embeddings?
Jain et al. found that adding noise embeddings:
- **Reduces overfitting** to formatting and wording.
- **Encourages longer, more informative text generation.**
- **Increases token diversity** without simply adding repetition.

## 🛠️ NEFTune: Noisy Embedding Instruction Fine-Tuning
Hugging Face supports **NEFTune**, a noise addition technique:
1. A **noise vector** is added to each embedding vector.
2. Elements in the noise vector are **randomly sampled** from \([-1,1]\).
3. The noise vector is **scaled** using a scaling factor.
4. The final noisy vector is added **only during training**, not inference.

> ⚠️ **Warning:**  
> The full impact of noise embeddings is not yet well understood. Gains in fine-tuning performance may come at the cost of other model capabilities. **Always test for regressions!**

#### **Understanding Instruction-Tuning Datasets**  
Unlike traditional datasets (which contain just **input-output pairs**), instruction-tuning datasets also include **task descriptions in natural language**.  

##### **Structure of an Instruction-Tuning Dataset:**  
1. **Instruction** – Defines the task, specifies output format, and may include examples or constraints.  
2. **Optional Input** – The actual text for the model to process (e.g., a sentence or paragraph).  
3. **Output** – The correct response based on the instruction.

✅ **Fine-Tuning Types:**  
- **Multi-task** (used for instruction-tuning) vs. **Single-task** (for task-specific adaptation).

✅ **Why Instruction-Tuning?**  
- LLMs are trained on **next-token prediction/denoising**, which doesn’t match real-world tasks.  
- Instruction-tuning **bridges this gap**, making models more **controllable** and **structured**.  

✅ **Why Customize Instruction-Tuning?**

✅ **Methods to Create Instruction-Tuning Datasets:**

### **2. Input-First vs. Output-First Data Generation**  
✅ **Input-First Approach**:  
1. Given an **instruction**, the LLM generates an **input instance**.  
2. The LLM then generates the **corresponding output**.  
🚨 *Risk: Certain labels may be overrepresented, leading to data imbalance.*  

✅ **Output-First Approach**:  
1. Given an **instruction**, the LLM generates an **output label** first.  
2. Then, the LLM generates an **input text** that satisfies this label.  
💡 *Combining both approaches helps balance label distribution.*

#### **2️⃣ Response Generation**  
- Responses to the evolved instructions can be **LLM-generated or human-verified**.  

#### **3️⃣ Candidate Filtering**  
- Low-quality outputs are removed using **heuristics or LLMs**.

### **Why Update the Parameters of an LLM?**  

Updating an LLM's parameters is essential for improving its performance in specialized use cases. The primary reasons for updating LLM parameters include:  

---

### **1️⃣ Domain Adaptation (DAPT)**  
- Many domains (e.g., law, finance, medicine) have **unique linguistic characteristics** that general-purpose LLMs may not handle well.  
- Continual pre-training on **domain-specific data** helps the LLM **understand jargon, structure, and conventions** better.  
- Example: Training a general LLM on **legal documents** to improve its ability to summarize court rulings.  

---

### **2️⃣ Task Adaptation (TAPT)**  
- Even within a domain, LLMs may struggle with **specific tasks** (e.g., sentiment analysis in financial news).  
- **Task-adaptive pre-training** fine-tunes the model on **data related to specific tasks**.  
- Example: Pre-training an LLM on **financial market reports** before fine-tuning it for **stock price prediction**.  

---



### **3️⃣ Knowledge Updating (Lifelong Learning)**  
- Pre-trained LLMs have **fixed knowledge** that can become **obsolete over time**.  
- Continual pre-training on **new data** helps the LLM stay **up-to-date**.  
- Example: Updating an LLM trained before **COVID-19** with **pandemic-related medical literature**.  

---

### **4️⃣ Controllability & Steerability (Alignment Training)**  
- Fine-tuning can help **control the model’s behavior**, ensuring it follows specific guidelines.  
- Useful for **reinforcing safety mechanisms** (e.g., rejecting unethical requests).  
- Techniques like **RLHF (Reinforcement Learning from Human Feedback)** can **steer** the LLM’s responses.  

---
### Task-Specific vs. Domain-Specific Models 

1. **Task-Specific Models**:
   - Focus: Optimized for a single task (e.g., translation, sentiment analysis).
   - Advantages: Efficient with smaller model sizes (<8B parameters); requires 100–100,000 data samples.
   - Use Case: Ideal for specialized, resource-constrained applications.

2. **Domain-Specific Models**:
   - Focus: Fine-tuned for a specific field (e.g., medicine, law, e-commerce).
   - Advantages: Enhances field-specific vocabulary and expertise.
   - Data Needs: Varies by domain complexity and representation in pre-training data—specialized fields may need more.

3. **Key Factors for Domain-Specific Fine-Tuning**:
   - Domain size (knowledge breadth and vocabulary).
   - Representation in pre-training data.
   - Challenges due to closed-source pre-training datasets.

## **Techniques for Updating LLM Parameters**  

### **1️⃣ Continual Pre-Training**  
✅ Continues pre-training a general LLM on **new, domain-specific, or updated data**.  
✅ Helps with **domain adaptation** and **knowledge updates**.  

🚨 **Challenges:**  
- **Catastrophic forgetting**: The model might **lose its previous knowledge** while learning new data.  
- **Solution:** Replay, distillation, and regularization techniques (explained later).  

---

### **2️⃣ Parameter-Efficient Fine-Tuning (PEFT)**  
✅ Updates only **a small subset of parameters**, reducing memory and compute costs.  
✅ Techniques include **LoRA (Low-Rank Adaptation)** and **Adapters**.  
✅ Ideal for **task adaptation** without requiring full model fine-tuning.  

---

### **3️⃣ Model Merging (Model Fusion)**  
✅ Combines multiple LLMs to **merge their capabilities**.  
✅ Can be used to **integrate domain knowledge** into a general-purpose model.  
✅ Still a **new and experimental** area ("dark arts" of NLP).  

---

Here’s a comparison of **Task-Adaptive Pre-Training (TAPT)** and **Domain-Adaptive Pre-Training (DAPT)** in a table format:

| **Aspect**                        | **Task-Adaptive Pre-Training (TAPT)**                          | **Domain-Adaptive Pre-Training (DAPT)**                            |
|-----------------------------------|---------------------------------------------------------------|---------------------------------------------------------------------|
| **Purpose**                       | Adapt the model to specific downstream tasks.                 | Adapt the model to a specialized domain or vocabulary.              |
| **Data Type**                     | Smaller, task-specific unsupervised dataset.                  | Larger domain-specific unsupervised dataset.                        |
| **Focus**                          | Tailored to specific tasks (e.g., classification, summarization). | Tailored to the linguistic structure and jargon of a domain.        |
| **Order of Execution**            | Performed after DAPT to fine-tune on task-specific data.       | Performed first to adapt to domain-related content.                 |
| **Data Selection Method**         | Data selected using clustering methods based on task relevance. | Data selected using clustering methods based on domain relevance.   |
| **Size of Dataset**               | Typically smaller than DAPT datasets.                         | Typically larger and domain-specific.                               |
| **Goal**                          | Enhance the model’s ability on specific tasks.                 | Expose the model to domain-specific text to capture specialized knowledge. |
| **Model Adaptation**              | Adjusts the model to perform well on the target task.          | Adjusts the model to understand domain-specific structure and terminology. |

This table highlights the key differences between TAPT and DAPT.

## **Continual Pre-Training in Detail**  

### **🔹 Why Not Train an LLM from Scratch?**  
- Training an LLM **from scratch** on domain-specific data is **expensive and inefficient**.  
- **Solution:** Take an **already pre-trained** LLM and fine-tune it on domain-specific data.  

### **🔹 Common Use Cases of Continual Pre-Training**  
📌 **Domain-Adaptive Pre-Training (DAPT)**: Exposing an LLM to an **entire domain** (e.g., legal or financial text).  
📌 **Task-Adaptive Pre-Training (TAPT)**: Pre-training on **data relevant to specific tasks** within a domain.  
📌 **Lifelong Learning**: Keeping the model up-to-date with **new knowledge**.  

---

### **⚠️ The Catastrophic Forgetting Problem**  
🚨 When an LLM is trained on **new data**, it may forget **previously learned** knowledge.  
🚨 Continual pre-training can degrade performance on **general tasks**.  

### **🛠️ Solutions for Catastrophic Forgetting**  
✅ **Replay (Memory)**: Mixes original pre-training data with new data to retain past knowledge.  
✅ **Distillation**: Takes an older checkpoint of the model and during training compares the
KL-divergence between the older and the current representations and
penalizes i.  
✅ **Regularization**: Penalizes large parameter changes to preserve prior learning.  
✅ **Parameter Expansion**: Adds new parameters (width/depth) to retain old knowledge while learning new information.

## **Practical Implementation of Continual Pre-Training**  
### **🔹 Steps to Continually Pre-Train an LLM on Domain-Specific Data**  
1️⃣ **Select a pre-trained LLM** (e.g., LLaMA-2, Mistral, GPT-J).  
2️⃣ **Curate domain-specific data** (e.g., legal texts, medical papers).  
3️⃣ **Tokenize and format the data** (match the LLM’s pre-training format).  
4️⃣ **Train the model using self-supervised learning** (next-word prediction, masked token prediction).  
5️⃣ **Evaluate performance** using:  
   - Benchmark tests (e.g., perplexity, accuracy on domain-specific tasks).  
   - General tasks (to detect catastrophic forgetting).  
6️⃣ **Apply replay, distillation, or parameter expansion** if forgetting occurs.

### **Replay (Memory) Strategy**  
- Mix pre-training data with continual training data to reduce data drift.  
- Start with ~25% new data and gradually increase up to 80%.  
- If the original dataset is monolithic, identify domains to ensure balanced sampling.  

### **Learning Rate Strategies**  
- **Lowering LR over time**: Reduces forgetting but may slow down learning.  
- **Re-warming strategy (Gupta et al.)**:  
  1. Increase LR to a maximum at the start.  
  2. Decay it using a cosine schedule.  
  3. Keep LR constant at a minimum to maintain balance between old and new knowledge.
   

### **Parameter Expansion**  
- **Basic approach**: Add new layers to the model and only train those new parameters during continual pre-training.
- **Domain-specific modules**: Insert and train specialized parameter modules (adapters) within existing layers to tailor the model to specific domains.
- **Further exploration**: Adapter-based approaches will be discussed in "Parameter-Efficient Fine-Tuning."


### **Task-Adaptive Pre-Training (TAPT)**
- **Supplement to Domain-Adaptive Pre-Training (DAPT)**: TAPT helps to continually pre-train an LLM on a smaller, task-specific unsupervised dataset.
- **Prevention of catastrophic forgetting**: Perform DAPT first to adapt the model to the domain, followed by TAPT, and then fine-tune the model on the downstream tasks using supervised data.
- **Data selection for TAPT**: Use embedding-based methods to select unsupervised data, focusing on data that is clustered similarly to gold-truth sentences.

### **Key Tips**
- **Domain-specific jargon**: Domain-specific text often includes specialized terms. A useful strategy is to add extra tokens to represent this jargon, improving model understanding and handling of domain-specific language.


**Interactive Breakdown: Reinforcement Learning & RLHF**

-   **Traditional Reinforcement Learning (RL)**
    
    -   **Agent-Environment Interaction:**
        -   The agent interacts with its environment by taking actions.
    -   **Trial and Error:**
        -   The agent learns by experimenting—trying different actions.
    -   **Reward Function:**
        -   After actions, a reward is provided if the actions lead towards the goal.
        -   The objective is to maximize these rewards.
    -   **Challenge:**
        -   In real-world scenarios, it can be difficult to define what "success" looks like and how to craft an appropriate reward function.
-   **Reinforcement Learning from Human Feedback (RLHF)**
    
    -   **Human-in-the-Loop:**
        -   Instead of relying solely on a pre-defined reward function, a human provides iterative feedback.
    -   **Integrating Preferences:**
        -   A reward model is trained using human feedback to better align the agent's actions with human preferences.
-   **Types of Human Feedback**
    
    -   **Binary Feedback:**
        -   The feedback is a simple yes/no or accept/reject decision.
    -   **Binary Comparisons:**
        -   Two outputs are compared, and the preferred one is chosen.
    -   **Ranking:**
        -   A set of outputs is ranked according to preference.
    -   **Corrective Feedback:**
        -   The human provides a corrected or ideal output, often described in natural language.
**Interactive Breakdown: RLHF Setup by OpenAI**

-   **Phase 1: Supervised Fine-Tuning**
    
    -   **Prompt Dataset:**
        -   Assemble a diverse set of potential user requests.
    -   **Human Annotation:**
        -   Annotators provide desired responses for each prompt.
    -   **Fine-Tuning Dataset:**
        -   Combine prompts and human responses to form the training dataset.
    -   **Model Update:**
        -   Fine-tune the pre-trained model using this dataset.
    -   **Resource-Intensive:**
        -   Gathering quality annotations requires significant effort and resources.
-   **Phase 2: Reward Modeling**
    
    -   **Query Generation:**
        -   Use a variety of prompts to generate multiple responses from the model.
    -   **Feedback Collection:**
        -   Annotators review the responses and offer feedback via:
            -   Rank-ordering responses or
            -   Selecting the best generation.
    -   **Reward Model Training:**
        -   Train a model to predict which output a human would prefer based on the provided feedback.
-   **Phase 3: Proximal Policy Optimization (PPO)**
    
    -   **Model Optimization:**
        -   Utilize the trained reward model to further refine the fine-tuned model.
    -   **PPO Algorithm:**
        -   Apply Proximal Policy Optimization to adjust the model’s actions in alignment with human preferences.
    -   **Outcome:**
        -   The optimized model delivers outputs that better reflect the human-in-the-loop guidance.
**Interactive Breakdown: In-Depth PPO Training & RLHF Process**

-   **1. PPO Training Process**
    
    -   **Response Generation:**
        -   The language model generates a response (or continuation) given a prompt.
    -   **Reward Calculation:**
        -   The reward model evaluates the query-response pair and `outputs a scalar reward representing` how well the response fits the desired behavior.
    -   **Log-Probability & KL-Divergence Calculation:**
        -   **Log-Probabilities:**
            -   Calculate the log-probabilities of each token in the query-response sequence using two models:
                -   `The fine-tuned model (SFT model) being adjusted`.
                -   `A reference model (the original pre-trained model before fine-tuning)`.
        -   **KL-Divergence:**
            -   Compute the Kullback-Leibler (KL) divergence between these log-probabilities.
            -   This divergence acts as a regularization signal, ensuring that the new outputs do not deviate too far from the reference model.

The model being updated with PPO is the fine-tuned model (often called the SFT model). Here’s how it works:

-   **Fine-Tuned (SFT) Model:**
    
    -   This model starts as a base LLM that has been adapted through supervised fine-tuning with human-provided responses.
    -   It’s the one generating responses during the RLHF process.
-   **Reference Model:**
    
    -   This is usually the original pre-trained base model before any fine-tuning.
    -   It remains unchanged and is used to compute the KL-divergence to regularize the fine-tuned model.
-   **PPO Application:**
    
    -   The PPO algorithm uses the reward signal (from the reward model) and the KL-divergence (comparing the SFT model to the reference model) to iteratively update the SFT model.
    -   The objective is to improve the SFT model's alignment with human preferences while keeping it close to the robust language capabilities of the base LLM.

** Sampling Methods for Addressing Hallucination**

-   **Decoding Methods & Hallucination Risk**
    
    -   **Greedy Decoding:**
        -   Always picks the highest-probability token.
        -   Tends to produce less hallucination because it minimizes randomness.
    -   **Top-p (Nucleus) Sampling:**
        -   Samples from a pool of tokens whose cumulative probability meets a threshold pp.
        -   `Introduces randomness, which can sometimes result in choosing an incorrect token—thus increasing hallucination risk`.
-   **Factual-Nucleus Sampling**
    - Factual‐nucleus sampling is a decoding strategy used during inference—not during the training phase.
    -   **Core Idea:**
        -   Dynamically reduce randomness as the generated sequence gets longer.
        -   Fewer valid candidate tokens remain in later parts of a sentence; hence, less randomness is desired.
    -   **Formula:**  $$ pt=max⁡{ω, p×λt−1}p_t = \max\{\omega,\, p \times \lambda^{t-1}\} $$
        -   tt: The generation step (i.e., token index within the sentence).
-   **Tunable Parameters**
    
    -   **Decay Rate (λ\lambda):**
        -   Multiplies the initial pp value at each step.
        -   Controls how quickly randomness decays as more tokens are generated.
    -   **Reset (p):**
        -   Option to reset pp at the start of each new sentence.
        -   Prevents pp from decaying too far within long sequences, which could overly mimic greedy decoding.
    -   **Lower-bound (ω\omega):**
        -   Sets a minimum threshold for ptp_t.
        -   Ensures that some level of diversity is maintained, preventing the algorithm from becoming completely deterministic.
-   **Trade-Offs**
    
    -   **Reducing Hallucination:**
        -   A lower pp value (via decay) makes the output more factual and reduces hallucinations.
    -   **Loss of Diversity:**
        -   Too low a pp can reduce the diversity of token generation, potentially making responses repetitive or less creative.

This method allows fine-tuning of the randomness in token sampling dynamically, aiming to strike a balance between reducing hallucination risk and maintaining output diversity.

Decoding by Contrasting Layers (DoLa) is a decoding strategy designed to enhance the factual accuracy of outputs from large language models (LLMs). `It operates on the premise that factual knowledge is predominantly encoded in the upper layers of Transformer architectures`, `while lower layers capture more syntactic information`. 
By contrasting the outputs (logits) from higher and lower layers, DoLa aims to amplify the factual content during text generation. 
**Key Components of DoLa:**

-   **Logit Differencing:** For each token prediction, DoLa computes the difference between `the logits from a higher (expert) layer and a lower (amateur) layer`. This contrast emphasizes the factual information encoded in the higher layers, guiding the model towards more accurate token selection.
    
-   **Layer Selection:** The choice of which layers to contrast is crucial. Typically, `the final layer serves as the expert layer, while an intermediate layer acts as the amateur layer.` This selection leverages the hierarchical encoding of information within the Transformer model. citeturn0search1
    

**Benefits of DoLa:**

-   **Reduced Hallucinations:** By focusing on the factual knowledge in the upper layers, DoLa helps mitigate the generation of content that deviates from known facts, addressing a common issue in LLM outputs. citeturn0search1
    
-   **Training-Free Implementation:** DoLa is applied during the `inference phase` and does not require additional training or fine-tuning of the model, making it a practical approach for improving factuality without extensive computational resources.
