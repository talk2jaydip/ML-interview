<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">For a specific RFM group, we want to recommend product to customers based on their purchase patterns. To achieve this, we run <strong>Market basket analysis</strong> to find out what products are frequently purchased together.
<code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">, Market Basket Analysis identifies purchasing patterns: purchased item X =&gt; also purchase item Y.</code></p>
</blockquote>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">The support measures how frequently X and Y are purchased together.</li>
<li style="margin:6px 0;">The confidence is the frequency of purchasing X and Y together
dividing the frequency of purchasing X.</li>
<li style="margin:6px 0;">Confidence equals to 1 means that if customer purchased X, they definitely purchased Y.  And the  lift the the support of purchasing X and Y together over support of X
and support of Y.</li>
</ul>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">Market basket analysis is a method to gain insights into granular behavior of customers. This is helpful in devising strategies which uncovers deeper understanding of purchase decisions taken by the customers.</code></p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">

<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">After you identify the purchasing pattern, say purchased X and also purchased Y, potential customers of product Y can be identified as customers who purchased product X but haven't purchased product Y.</p>

<h2 style="background:linear-gradient(90deg,#22313d 0%,#273743 100%);border-left:8px solid #72aa9b;color:#e7eff3;padding:11px 16px;border-radius:14px;border:1px solid #3a4b58;box-shadow:0 6px 18px rgba(0,0,0,0.18);margin:22px 0 14px 0;"><strong>Customer Segmentation Models</strong></h2>
<h2 style="background:linear-gradient(90deg,#22313d 0%,#273743 100%);border-left:8px solid #72aa9b;color:#e7eff3;padding:11px 16px;border-radius:14px;border:1px solid #3a4b58;box-shadow:0 6px 18px rgba(0,0,0,0.18);margin:22px 0 14px 0;"><strong>1. Demographic Segmentation Model</strong></h2>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">Demographic segmentation means dividing people into groups based on certain demographic factors, including age, income, marital status, and occupation.</p>
<h2 style="background:linear-gradient(90deg,#22313d 0%,#273743 100%);border-left:8px solid #72aa9b;color:#e7eff3;padding:11px 16px;border-radius:14px;border:1px solid #3a4b58;box-shadow:0 6px 18px rgba(0,0,0,0.18);margin:22px 0 14px 0;"><strong>2. Geographic Segmentation Model</strong></h2>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">With geographic segmentation, you categorize your audience based on where they work, live, and shop.</p>
<h2 style="background:linear-gradient(90deg,#22313d 0%,#273743 100%);border-left:8px solid #72aa9b;color:#e7eff3;padding:11px 16px;border-radius:14px;border:1px solid #3a4b58;box-shadow:0 6px 18px rgba(0,0,0,0.18);margin:22px 0 14px 0;"><strong>3. Psychographic Segmentation</strong></h2>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">We each have unique personalities, but we share traits or characteristics. Psychographic segmentation means forming groups based on common traits such as hobbies, lifestyle choices, personality traits, cultural beliefs, and values.</p>
<h2 style="background:linear-gradient(90deg,#22313d 0%,#273743 100%);border-left:8px solid #72aa9b;color:#e7eff3;padding:11px 16px;border-radius:14px;border:1px solid #3a4b58;box-shadow:0 6px 18px rgba(0,0,0,0.18);margin:22px 0 14px 0;"><strong>4. Technographic Segmentation</strong></h2>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">Technographic segmentation means categorizing people depending on the devices, hardware, and software they use. Why does this data matter? Well, according to  <a style="color:#8fd1ff;" href="https://www.outerboxdesign.com/web-design-articles/mobile-ecommerce-statistics">statistics</a>:</p>
<h2 style="background:linear-gradient(90deg,#22313d 0%,#273743 100%);border-left:8px solid #72aa9b;color:#e7eff3;padding:11px 16px;border-radius:14px;border:1px solid #3a4b58;box-shadow:0 6px 18px rgba(0,0,0,0.18);margin:22px 0 14px 0;"><strong>5. Behavioral Segmentation</strong></h2>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">Want to know how your audience interacts with your business? Try behavioral segmentation.</p>
<h2 style="background:linear-gradient(90deg,#22313d 0%,#273743 100%);border-left:8px solid #72aa9b;color:#e7eff3;padding:11px 16px;border-radius:14px;border:1px solid #3a4b58;box-shadow:0 6px 18px rgba(0,0,0,0.18);margin:22px 0 14px 0;"><strong>6. Needs-Based Segmentation</strong></h2>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">Successful marketing often comes down to showing prospects how your goods or services meet their needs. That’s where needs-based segmentation comes in.</p>
<h2 style="background:linear-gradient(90deg,#22313d 0%,#273743 100%);border-left:8px solid #72aa9b;color:#e7eff3;padding:11px 16px;border-radius:14px;border:1px solid #3a4b58;box-shadow:0 6px 18px rgba(0,0,0,0.18);margin:22px 0 14px 0;">Techniques of Segmentation</h2>
<h3 style="display:inline-block;color:#d8e4ea;background:#22303a;border-left:5px solid #7fa89e;padding:8px 14px;border-radius:12px;border:1px solid #3a4a52;margin:18px 0 10px 0;">Objective Segmentation</h3>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">Segmentation to identify the type of customers <strong>who would respond to a particular offer</strong></li>
<li style="margin:6px 0;">Segmentation to identify <strong>high spenders among customers</strong> who will use the e-commerce channel for festive shopping</li>
<li style="margin:6px 0;">Segmentation to identify customers <strong>who will default on their credit</strong> obligation for a loan or credit card</li>
</ul>
<blockquote style="margin:14px 0;padding:12px 16px;background:#24313d;border-left:5px solid #90b7a2;border-radius:14px;border:1px solid #3c4d5a;color:#d9e4ec;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">The most common techniques used for building an <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">objective segmentation are CHAID and CRT</code>.</p>
</blockquote>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">Each of these techniques attempt to maximize the difference among segments with regards to the stated objective (sometimes referred to as the target for segmentation).</p>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">CHAID uses a chi square statistic for the same while CRT uses Gini impurity</code>.</p>
<h3 style="display:inline-block;color:#d8e4ea;background:#22303a;border-left:5px solid #7fa89e;padding:8px 14px;border-radius:12px;border:1px solid #3a4a52;margin:18px 0 10px 0;">Non-Objective Segmentation</h3>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">Segmentation of the customer base to understand the specific profiles which exist within the customer base so that <strong>multiple marketing actions</strong> can be personalized for each segment</li>
<li style="margin:6px 0;">Segmentation of geographies on the basis of <strong>affluence and lifestyle of people living in each geograph</strong>y so that sales and distribution strategies can be formulated accordingly</li>
<li style="margin:6px 0;">Segmentation of <strong>web site visitors on the basis of browsing behavior</strong> to understand the level of engagement and affinity towards the brand.</li>
</ul>
<h2 style="background:linear-gradient(90deg,#22313d 0%,#273743 100%);border-left:8px solid #72aa9b;color:#e7eff3;padding:11px 16px;border-radius:14px;border:1px solid #3a4b58;box-shadow:0 6px 18px rgba(0,0,0,0.18);margin:22px 0 14px 0;">## What is Cohort and Cohort Analysis?</h2>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">A cohort is a group of people sharing something in common, such as the sign-up date to an app, the month of the first purchase, geographical location, acquisition channel (organic users, coming from performance marketing, etc.) and so on.</code></p>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">Cohort analysis is a descriptive analytics technique in cohort analysis. Customers are divided into mutually exclusive cohorts, which are then tracked over time.</p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">Customer Lifetime Value</li>
</ul>
</li>
</ul>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">Customer lifetime value is how much money a customer will bring your brand throughout their entire time as a paying customer.</p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">Time cohorts: customers who signed up for a product or service during a particular time frame.<code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">Groups are divided by first activity.</code></li>
<li style="margin:6px 0;">Behavior cohorts: customers who purchased a product or subscribed to service in the past.</li>
<li style="margin:6px 0;">Size cohorts: refer to the various sizes of customers who purchase the company’s products or services.</li>
</ul>
