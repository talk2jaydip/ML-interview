## 1.1 Evolution of RAG

RAG (Retrieval-Augmented Generation) was introduced by Lewis et al. (2020) at Meta AI to ground LLM outputs in external knowledge. Since then, RAG has evolved through three distinct generations:

```
EVOLUTION OF RAG ARCHITECTURES
================================

Generation 1: Naive RAG (2020-2022)
------------------------------------
  Query --> Retrieve (Top-K) --> Concatenate --> Generate
  
  Problems: No quality check, irrelevant chunks, lost context,
            "Lost in the Middle" problem

Generation 2: Advanced RAG (2023-2024)
---------------------------------------
  Query --> [Pre-Retrieval Optimization] --> Retrieve 
        --> [Post-Retrieval Processing] --> Generate
  
  Additions: Query rewriting, HyDE, re-ranking, 
             sentence-window retrieval, parent-child chunks

Generation 3: Modular / Agentic RAG (2024-2026)
-------------------------------------------------
  Query --> [Router/Agent] --> {Decision Engine}
        --> Retrieve / Web Search / No Retrieval / Multi-hop
        --> [Quality Check] --> Generate / Re-retrieve
        --> [Self-Reflection] --> Final Answer
  
  Additions: Self-correction, adaptive retrieval, graph-based,
             multi-modal, agent-orchestrated pipelines
```

### Key Paradigm Shifts

| Era | Approach | Retrieval | Generation | Quality Control |
|-----|----------|-----------|------------|-----------------|
| **Naive RAG** | Fixed pipeline | Single-pass top-K | Direct concatenation | None |
| **Advanced RAG** | Optimized pipeline | Query expansion + re-rank | Windowed/hierarchical context | Post-retrieval filtering |
| **Modular RAG** | Composable modules | Adaptive, multi-source | Module-specific generation | Self-reflection loops |
| **Agentic RAG** | Agent-orchestrated | On-demand, tool-based | Agent-controlled generation | Multi-step verification |

> **Interview Tip:** When discussing RAG evolution, emphasize that each generation addresses specific failure modes of the previous one. Naive RAG fails on quality; Advanced RAG fails on adaptability; Modular/Agentic RAG introduces decision-making intelligence.

**Reference:** Lewis et al., "Retrieval-Augmented Generation for Knowledge-Intensive NLP Tasks," NeurIPS 2020.

<a id="1.2"></a>
## 1.2 GraphRAG (Microsoft)

**Paper:** "From Local to Global: A Graph RAG Approach to Query-Focused Summarization" (Microsoft Research, 2024)
**Repository:** https://github.com/microsoft/graphrag

### What is GraphRAG?

GraphRAG is Microsoft Research's structured, hierarchical approach to RAG that constructs **knowledge graphs** from source documents and uses **community detection** to enable both local (entity-specific) and global (dataset-wide) question answering.

### Why GraphRAG?

Traditional (naive) RAG fails catastrophically on **global questions** -- questions that require understanding the entire dataset rather than specific text chunks. For example:
- "What are the main themes discussed across all documents?"
- "How do different departments' strategies relate to each other?"

Microsoft's evaluation showed naive RAG gives **misleading answers** to such global questions because it only retrieves semantically similar chunks, missing the big picture.

### How It Works

```
GraphRAG Pipeline
==================

INDEXING PHASE:
                                                          
  Raw Text --> [LLM Entity Extraction] --> Entities & Relationships
                                                |
                                                v
                                     [Knowledge Graph Construction]
                                                |
                                                v
                                     [Leiden Community Detection]
                                                |
                                                v
                              [Hierarchical Community Summaries]
                                    /           |           \
                                Level 0     Level 1      Level 2
                              (granular)  (mid-level)   (themes)

QUERY PHASE:

  +------------------+     +------------------+     +------------------+
  |   Local Search   |     |  Global Search   |     |  DRIFT Search    |
  |  Entity-focused  |     |  Community-based |     |  Local + Global  |
  |  "Who is X?"     |     |  "Main themes?"  |     |  Combined        |
  +------------------+     +------------------+     +------------------+
         |                        |                        |
         v                        v                        v
  [Entity subgraph    [Map-Reduce over      [Dynamic retrieval
   + local context]    community summaries]   with exploration]
```

### Three Query Modes

| Mode | Best For | How It Works | Cost |
|------|----------|-------------|------|
| **Local Search** | Specific entity questions | Retrieves entity subgraph + neighboring context | Low |
| **Global Search** | Broad thematic questions | Map-reduce over community summaries at chosen level | High |
| **DRIFT Search** | Mixed questions | Starts local, dynamically broadens scope | Medium |

### Key Results
- **70-80% win rate** over naive RAG on comprehensiveness and diversity metrics
- Intermediate-level summaries use only **20-70% of tokens** compared to full source text
- Excels on questions requiring synthesis across many documents

> **Interview Tip:** GraphRAG is a favorite interview topic. Be prepared to explain (1) why naive RAG fails on global questions, (2) the Leiden community detection algorithm, and (3) the trade-off between indexing cost (many LLM calls for entity extraction) and query quality. A common follow-up: "When would you NOT use GraphRAG?" Answer: When your questions are always entity-specific and your corpus is small -- the indexing overhead isn't justified.

**Reference:** Edge et al., "From Local to Global: A Graph RAG Approach to Query-Focused Summarization," arXiv:2404.16130, 2024.

<a id="1.3"></a>
## 1.3 Corrective RAG (CRAG)

**Paper:** Yan et al., "Corrective Retrieval Augmented Generation," arXiv:2401.15884, January 2024

### Core Idea

CRAG adds a **self-correction mechanism** to the retrieval step. Instead of blindly trusting retrieved documents, it evaluates their quality and takes corrective action when retrieval confidence is low.

### How CRAG Works

```
CRAG Architecture
==================

  User Query
      |
      v
  [Standard Retriever] --> Retrieved Documents
      |
      v
  [Lightweight Retrieval Evaluator]
      |
      +--- Confidence: HIGH -----> Use retrieved docs as-is
      |                                    |
      +--- Confidence: MEDIUM ---> [Knowledge Refinement]
      |                            Strip irrelevant parts,
      |                            keep key information
      |                                    |
      +--- Confidence: LOW ------> [Web Search Trigger]
                                   Query reformulation
                                   Web search execution
                                   Results integration
                                           |
                                           v
                                   [Knowledge Refinement]
                                   Decompose into strips
                                   Filter irrelevant content
                                   Recompose refined knowledge
                                           |
                                           v
                                      [Generator]
                                           |
                                           v
                                      Final Answer
```

### Three Confidence Actions

| Confidence Level | Action | Description |
|-----------------|--------|-------------|
| **Correct** (High) | Use retrieved docs | Documents are relevant, proceed normally |
| **Ambiguous** (Medium) | Knowledge refinement | Extract useful parts, discard noise |
| **Incorrect** (Low) | Web search fallback | Trigger external web search, reformulate query |

### Knowledge Refinement Algorithm
1. **Decompose** retrieved documents into fine-grained "knowledge strips"
2. **Score** each strip for relevance to the query using the evaluator
3. **Filter** out irrelevant strips
4. **Recompose** the remaining strips into refined context

### Key Results
- Significant improvements across **4 datasets** (short-form and long-form QA)
- Works as a **plug-and-play** enhancement -- can be added to any existing RAG pipeline
- The retrieval evaluator is lightweight (T5-based), adding minimal latency

> **Interview Tip:** CRAG is an excellent example of "fail-safe" design in ML systems. In interviews, frame it as: "What happens when your retriever returns garbage?" CRAG's three-tier response (use/refine/replace) is a mature engineering pattern. Compare it to circuit breakers in microservices architecture.

**Reference:** Yan et al., "Corrective Retrieval Augmented Generation," arXiv:2401.15884, 2024.

<a id="1.4"></a>
## 1.4 Self-RAG

**Paper:** Asai et al., "Self-RAG: Learning to Retrieve, Generate, and Critique through Self-Reflection," ICLR 2024 (arXiv:2310.11511, October 2023)

### Core Idea

Self-RAG trains a single LM to **adaptively decide** when to retrieve, **evaluate** retrieved passages, and **critique** its own generation -- all through special **reflection tokens** embedded during training.

### How Self-RAG Works

```
Self-RAG Flow with Reflection Tokens
======================================

  User Query: "What causes aurora borealis?"
      |
      v
  [LM generates initial tokens...]
      |
      v
  [RETRIEVE token decision]
      |
      +--- [Retrieve] = YES --> Trigger retrieval
      |         |
      |         v
      |    Retrieved passages {d1, d2, d3}
      |         |
      |         v
      |    [ISREL] token: Is passage relevant?
      |         |
      |         +--- RELEVANT --> Generate with passage
      |         |        |
      |         |        v
      |         |   [ISSUP] token: Is generation supported?
      |         |        |
      |         |        +--- FULLY SUPPORTED --> Continue
      |         |        +--- PARTIALLY SUPPORTED --> Flag
      |         |        +--- NOT SUPPORTED --> Discard
      |         |
      |         +--- IRRELEVANT --> Skip passage
      |
      +--- [Retrieve] = NO --> Generate without retrieval
                |
                v
           [ISUSE] token: Is output useful?
                |
                +--- Score 1-5
```

### Four Reflection Token Types

| Token | Purpose | Values | When Evaluated |
|-------|---------|--------|----------------|
| **[Retrieve]** | Should I retrieve? | Yes / No | Before generation of each segment |
| **[ISREL]** | Is passage relevant? | Relevant / Irrelevant | After retrieval, per passage |
| **[ISSUP]** | Is generation supported by evidence? | Fully / Partially / No Support | After generation with passage |
| **[ISUSE]** | Is the overall output useful? | Score 1-5 | After complete generation |

### Key Innovations
1. **On-demand retrieval**: Unlike standard RAG that always retrieves, Self-RAG retrieves only when needed
2. **Learned self-critique**: The reflection tokens are trained into the model, not prompted
3. **Inference-time control**: You can adjust the weight of each reflection token at inference to prioritize factuality vs. creativity
4. **No separate retriever training**: The same LM handles both generation and critique

### Key Results
- Self-RAG 7B and 13B **outperform ChatGPT** and retrieval-augmented Llama2-chat
- Significant improvements on open-domain QA, reasoning, and fact verification
- Superior **citation accuracy** on long-form generation tasks
- The model learns to retrieve less frequently on tasks where parametric knowledge suffices

> **Interview Tip:** Self-RAG vs CRAG is a common comparison question. Key difference: CRAG adds a separate evaluator module on top of any RAG pipeline (plug-and-play). Self-RAG trains the LM itself to decide and evaluate (requires fine-tuning). CRAG is easier to deploy; Self-RAG is more deeply integrated.

**Reference:** Asai et al., "Self-RAG: Learning to Retrieve, Generate, and Critique through Self-Reflection," ICLR 2024.

## 1.5 Adaptive RAG

**Paper:** Jeong et al., "Adaptive-RAG: Learning to Adapt Retrieval-Augmented Large Language Models through Question Complexity," NAACL 2024 (arXiv:2403.14403)

### Core Idea

Adaptive RAG uses a **lightweight query complexity classifier** to dynamically route each query to the most appropriate retrieval strategy, avoiding unnecessary computation for simple queries and ensuring sufficient depth for complex ones.

### How It Works

```
Adaptive RAG Routing
=====================

  User Query
      |
      v
  [Query Complexity Classifier]  <-- Small LM (e.g., T5-small)
      |
      +--- SIMPLE -----------> [No Retrieval]
      |                        LLM answers from parametric knowledge
      |                        Example: "What is the capital of France?"
      |
      +--- MODERATE ---------> [Single-Step Retrieval]
      |                        Standard RAG: retrieve once, generate
      |                        Example: "What was GDP growth in 2024?"
      |
      +--- COMPLEX ----------> [Iterative Multi-Step Retrieval]
                               Multi-hop RAG: retrieve, reason, retrieve again
                               Example: "How did the policy changes in 2023
                                         affect 2024 economic indicators
                                         compared to predictions?"
```

### Complexity Classification Strategy

The classifier is trained using **automatically collected silver labels**:
1. Run each query through all three strategies
2. The simplest strategy that produces a correct answer defines the query's complexity label
3. Train a small classifier on these labels

| Query Complexity | Strategy | Example | Latency | Cost |
|-----------------|----------|---------|---------|------|
| **Simple** | No retrieval | Factual recall, definitions | Very Low | Minimal |
| **Moderate** | Single-step RAG | Fact lookup, recent events | Medium | Low |
| **Complex** | Iterative RAG | Multi-hop reasoning, comparison | High | Higher |

### Why This Matters
- **Efficiency**: Simple queries skip retrieval entirely (saves latency + cost)
- **Quality**: Complex queries get multi-hop retrieval they need
- **Adaptability**: The router learns from actual system performance, not heuristics

> **Interview Tip:** Adaptive RAG demonstrates a key production principle: **not all queries are equal**. In a system design interview, proposing an adaptive routing layer shows maturity. Mention that 60-70% of real-world queries are often simple enough for direct LLM answers, so routing saves significant infrastructure cost.

**Reference:** Jeong et al., "Adaptive-RAG: Learning to Adapt Retrieval-Augmented Large Language Models through Question Complexity," NAACL 2024.

## 1.6 Modular RAG

**Paper:** Gao et al., "Modular RAG: Transforming RAG Systems into LEGO-like Reconfigurable Frameworks," arXiv:2407.21059, July 2024

### Core Idea

Modular RAG decomposes RAG systems into **independent, reusable modules** (like LEGO blocks) that can be composed into different configurations. This moves beyond the rigid "retrieve-then-generate" pipeline toward flexible, reconfigurable architectures.

### The Module Taxonomy

```
Modular RAG: Composable Building Blocks
=========================================

RETRIEVAL MODULES          PROCESSING MODULES        GENERATION MODULES
+----------------+        +------------------+       +------------------+
| Dense Retriever|        | Query Rewriter   |       | Standard LLM Gen |
| Sparse (BM25)  |        | Re-Ranker        |       | Attributed Gen   |
| Hybrid Search  |        | Knowledge Refiner|       | Summarizer       |
| Web Search     |        | Compressor       |       | Multi-step Gen   |
| Graph Retriever|        | Deduplicator     |       | Verifier         |
| SQL Retriever  |        | Merger/Fusion    |       | Critic           |
+----------------+        +------------------+       +------------------+

ROUTING MODULES            MEMORY MODULES            ORCHESTRATION
+----------------+        +------------------+       +------------------+
| Query Router   |        | Conversation Mem |       | Sequential       |
| Complexity Det.|        | Long-term Store  |       | Conditional      |
| Intent Classif.|        | Episodic Memory  |       | Branching        |
| Source Selector |        | Entity Memory    |       | Looping          |
+----------------+        +------------------+       +------------------+
```

### Four Flow Patterns

| Pattern | Description | Use Case | Example |
|---------|-------------|----------|---------|
| **Linear** | A -> B -> C -> D | Simple QA | Retrieve -> Rerank -> Generate |
| **Conditional** | A -> (if X then B else C) -> D | Adaptive systems | Route based on query type |
| **Branching** | A -> {B, C, D} -> Merge -> E | Multi-source | Query multiple DBs in parallel |
| **Looping** | A -> B -> C -> (if not good, goto A) | Self-corrective | Generate -> Verify -> Re-retrieve |

### Why Modular RAG Matters for Production

1. **Testability**: Each module can be tested and evaluated independently
2. **Swappability**: Replace a re-ranker without touching the retriever
3. **A/B Testing**: Test different module combinations easily
4. **Scalability**: Scale bottleneck modules independently
5. **Debugging**: Isolate failures to specific modules

### Example Configurations

```
Config 1: Simple QA Pipeline
  Query -> [Dense Retriever] -> [Reranker] -> [Generator]

Config 2: Self-Corrective Pipeline  
  Query -> [Dense Retriever] -> [Evaluator] --(low confidence)--> [Web Search]
                                             --(high confidence)--> [Generator]

Config 3: Multi-Source Knowledge Pipeline
  Query -> [Router] --> [SQL DB] --------+
                    --> [Vector DB] ------+--> [Merger] -> [Reranker] -> [Generator]
                    --> [Knowledge Graph]-+

Config 4: Agentic Loop Pipeline
  Query -> [Planner] -> [Retriever] -> [Generator] -> [Verifier]
                ^                                         |
                +---- (if verification fails) <-----------+
```

> **Interview Tip:** Modular RAG shows systems-thinking ability. In design interviews, sketch out your RAG system as modules with clear interfaces. This demonstrates that you think about maintainability, not just accuracy. Mention that real production systems evolve: you start with Config 1 and add modules as failure modes are discovered.

**Reference:** Gao et al., "Modular RAG: Transforming RAG Systems into LEGO-like Reconfigurable Frameworks," arXiv:2407.21059, 2024.

## 1.7 Agentic RAG

**Key Source:** LangChain Blog, "Agentic RAG with LangGraph," 2024; Andrew Ng's Agentic Design Patterns, 2024

### Core Idea

Agentic RAG replaces the fixed RAG pipeline with an **AI agent** that autonomously decides **when**, **what**, and **how** to retrieve. The agent treats retrieval as one of many tools it can invoke, enabling dynamic multi-step reasoning.

### Three Cognitive Architectures for RAG

```
Architecture 1: CHAIN (Traditional RAG)
=========================================
  Query --> Retrieve --> Generate --> Answer
  (No feedback, no decision-making)

Architecture 2: ROUTER (Adaptive RAG)
=======================================
  Query --> [Router Decision]
              |
              +--> Vector DB --> Generate --> Answer
              +--> Web Search --> Generate --> Answer
              +--> SQL Query --> Generate --> Answer
  (Single decision point, no loops)

Architecture 3: STATE MACHINE / AGENT (Agentic RAG)
=====================================================
  Query --> [Agent with Tools]
              |
              v
        +--[Think/Plan]<--+
        |                  |
        v                  |
  [Choose Action]          |
    |   |   |              |
    v   v   v              |
  [Retrieve] [Search] [Compute]
    |   |   |              |
    v   v   v              |
  [Evaluate Results]-------+
        |
        v (when satisfied)
      Answer
  (Multiple loops, self-reflection, tool orchestration)
```

### How Agentic RAG Differs from Standard RAG

| Feature | Standard RAG | Agentic RAG |
|---------|-------------|-------------|
| Retrieval | Always retrieves | Retrieves only when needed |
| Strategy | Fixed pipeline | Dynamic, query-dependent |
| Tools | Single retriever | Multiple tools (search, SQL, API, code) |
| Iteration | Single pass | Multi-step with loops |
| Self-correction | None | Evaluates and re-retrieves |
| Query handling | Treats all queries equally | Adapts strategy per query |
| Planning | None | Decomposes complex queries into sub-tasks |

### Agentic RAG with LangGraph Implementation Pattern

The LangGraph framework implements Agentic RAG as a **state machine** with:
- **Nodes**: Retrieval, grading, generation, query rewriting, web search
- **Edges**: Conditional transitions based on quality assessments
- **State**: Accumulated documents, query history, generation attempts

Key decision points in the agent loop:
1. **Grade retrieved documents**: Are they relevant to the query?
2. **Check for hallucination**: Is the generation grounded in the documents?
3. **Assess answer quality**: Does the answer actually address the question?
4. **Decide next action**: Re-retrieve, rewrite query, search web, or return answer

> **Interview Tip:** Agentic RAG is the state-of-the-art. When asked "How would you build a RAG system?", don't just describe retrieve-then-generate. Describe an agent loop with quality gates. Mention that Andrew Ng showed GPT-3.5 with an agentic loop (95.1% on HumanEval) outperforms GPT-4 zero-shot (67%) -- the architecture matters more than the model.

**References:**
- LangChain Blog, "Agentic RAG with LangGraph," 2024
- Andrew Ng, "Agentic Design Patterns," The Batch / DeepLearning.AI, 2024

<a id="1.8"></a>
## 1.8 Multi-Modal RAG

### Core Idea

Multi-modal RAG extends retrieval-augmented generation beyond text to handle **images, tables, charts, audio, and video**. This is critical for enterprise documents (PDFs with charts), medical imaging, and multimedia knowledge bases.

### Architecture Approaches

```
APPROACH 1: Multi-Modal Embeddings
====================================
  Documents (text + images + tables)
      |
      v
  [Multi-Modal Embedding Model]  (e.g., CLIP, GPT-4V embeddings)
      |
      v
  [Unified Vector Store]
      |
      v
  Query --> [Same Embedding Model] --> Similarity Search --> Multi-modal Context --> LLM

APPROACH 2: Separate Pipelines with Fusion
=============================================
  Documents
      |
      +---> Text --> [Text Embeddings] --> [Text Vector DB]
      |
      +---> Images --> [Vision Model Caption] --> [Text Embeddings] --> [Image Vector DB]
      |                     or
      |              [CLIP Embeddings] --> [Image Vector DB]
      |
      +---> Tables --> [Table Serialization] --> [Text Embeddings] --> [Table Vector DB]
      |
      v
  Query --> [Multi-Index Search] --> [Result Fusion] --> [Multi-Modal LLM] --> Answer

APPROACH 3: Vision-Language Model Direct Processing
=====================================================
  Documents (as images/screenshots)
      |
      v
  [Document Screenshots / PDF Pages as Images]
      |
      v
  Query + Retrieved Page Images --> [VLM like GPT-4V, Claude] --> Answer
```

### Multi-Modal RAG Comparison

| Approach | Pros | Cons | Best For |
|----------|------|------|----------|
| **Multi-modal embeddings** | Unified search space | Limited model availability | Text-heavy docs with occasional images |
| **Separate pipelines** | Specialized per modality | Complex fusion logic | Large-scale multi-modal corpora |
| **Vision-LLM direct** | Preserves layout/formatting | Expensive, high latency | Complex documents (PDFs, slides) |
| **Table extraction** | Structured data queries | Extraction errors | Financial reports, data tables |

### Key Techniques for Each Modality

**Images:** Use CLIP or SigLIP embeddings, or generate captions with a VLM and embed the caption text.

**Tables:** Serialize to markdown/HTML, or use specialized table understanding models (TableFormer, TAPAS). Store both raw table and natural language summary.

**Audio:** Transcribe with Whisper, then apply standard text RAG on transcripts. Store timestamps for source attribution.

**Video:** Extract keyframes + transcript. Embed keyframe descriptions and transcript chunks. Retrieve by text or visual similarity.

> **Interview Tip:** When discussing multi-modal RAG, mention the "ColPali" approach (2024) -- using vision-language models to directly embed document page images without OCR/text extraction. This is a paradigm shift because it preserves layout, formatting, and visual context that text extraction loses. It is especially valuable for complex PDFs with charts and diagrams.

**References:**
- Faysse et al., "ColPali: Efficient Document Retrieval with Vision Language Models," arXiv:2407.01449, 2024
- LangChain, "Multi-Modal RAG," documentation and cookbooks, 2024

<a id="1.9"></a>
## 1.9 Long-Context RAG vs Traditional Chunked RAG

### The Debate

With models like Gemini 1.5 Pro (1M tokens), Claude 3.5 (200K tokens), and GPT-4 Turbo (128K tokens) offering massive context windows, a critical question emerges: **Do we still need RAG, or can we just stuff everything into the context window?**

### Head-to-Head Comparison

| Dimension | Chunked RAG | Long-Context (Stuff-it-all) | Long-Context + RAG (Hybrid) |
|-----------|-------------|---------------------------|--------------------------|
| **Accuracy** | Good (if chunks are relevant) | Good (all info available) | Best (relevant + full context) |
| **Cost** | Low (small context) | Very High (huge input tokens) | Medium |
| **Latency** | Low | High (processing 100K+ tokens) | Medium |
| **Scalability** | Scales to billions of docs | Limited by context window | Best of both worlds |
| **"Needle in Haystack"** | Depends on retrieval quality | Degrades in the "middle" | Strong with retrieval |
| **Freshness** | Real-time (re-index) | Must re-send everything | Real-time |
| **Multi-doc reasoning** | Weak (limited chunks) | Strong (sees everything) | Strong |

### The "Lost in the Middle" Problem

Research by Liu et al. (2023) showed that LLMs struggle to use information placed in the **middle** of long contexts. Performance is best when relevant information is at the **beginning** or **end** of the context.

```
Performance vs. Position of Relevant Information
==================================================

  High |*                                           *
       | *                                         *
       |  *                                       *
  Perf |   *                                     *
       |    *          *  *  *  *  *            *
       |     *       *                *       *
  Low  |      * * *                     * * *
       +-------------------------------------------
        Start      Middle of Context        End
```

### When to Use What

| Scenario | Recommended Approach | Reason |
|----------|---------------------|--------|
| Small corpus (<50 pages) | Long context | All fits, no retrieval needed |
| Large corpus (>1000 docs) | Chunked RAG | Can't fit in any context window |
| Medium corpus + complex queries | Long-context + RAG | Retrieve relevant docs, add surrounding context |
| Cost-sensitive production | Chunked RAG | 10-100x cheaper than stuffing |
| Multi-hop reasoning | Agentic RAG | Needs iterative retrieval |
| Real-time knowledge updates | Chunked RAG | Just re-index changed docs |

### The Emerging Consensus (2025-2026)

The best production systems use a **hybrid approach**:
1. **Retrieve** the most relevant chunks using RAG
2. **Expand** each chunk to include surrounding context (parent document, neighboring sections)
3. **Feed** the expanded, relevant context to a long-context model
4. This gives you the **precision** of RAG with the **reasoning ability** of long-context models

> **Interview Tip:** This is one of the most frequently debated topics. The interviewer wants to see nuance, not dogma. Don't say "RAG is dead" or "Long context doesn't work." Say: "It depends on corpus size, cost constraints, and query complexity. In production, we often combine both -- RAG for precision retrieval, long-context models for reasoning over the retrieved set."

**Reference:** Liu et al., "Lost in the Middle: How Language Models Use Long Contexts," TACL 2024.

## 1.10 ColBERT / Late Interaction Models

**Paper:** Khattab & Zaharia, "ColBERT: Efficient and Effective Passage Search via Contextualized Late Interaction over BERT," SIGIR 2020; Santhanam et al., "ColBERTv2: Effective and Efficient Retrieval via Lightweight Late Interaction," NAACL 2022

### The Retrieval Model Spectrum

```
RETRIEVAL MODEL COMPARISON
============================

Sparse Retrieval (BM25)          Dense Retrieval (DPR/E5)       Late Interaction (ColBERT)
========================         ==========================     ==========================

Query: "climate change"          Query: "climate change"        Query: "climate change"
   |                                |                              |  |
   v                                v                              v  v
[Term Matching]                  [Single Vector]               [Token Vectors]
"climate" -> doc scores          q = [0.2, 0.5, ...]          q1=[..], q2=[..]
"change"  -> doc scores                |                          |   |
   |                                   v                          v   v
   v                             [Single Vector]               [Token Vectors per Doc]
[BM25 Score]                     d = [0.3, 0.1, ...]          d1=[..], d2=[..], d3=[..]
                                       |                          |
                                       v                          v
                                 [Dot Product]                [MaxSim Operation]
                                 score = q . d                For each qi, find max
                                                              similarity across all dj
                                                              Sum these max scores

Speed:    FAST                   FAST (ANN search)            MEDIUM
Quality:  MODERATE               GOOD                         BEST
Storage:  LOW                    LOW (1 vec/doc)              HIGH (1 vec/token)
```

### How ColBERT's Late Interaction Works

1. **Encode query** tokens independently: each query token gets its own vector
2. **Encode document** tokens independently (can be pre-computed and cached)
3. **Score** using MaxSim: for each query token, find its maximum similarity with any document token, then sum

```
MaxSim Scoring Example:
========================
Query tokens:  q1="climate"  q2="change"  q3="effects"

Document tokens: d1="global"  d2="warming"  d3="climate"  d4="impact"

MaxSim(q1, D) = max(sim(q1,d1), sim(q1,d2), sim(q1,d3), sim(q1,d4)) = sim(q1,d3) = 0.95
MaxSim(q2, D) = max(sim(q2,d1), sim(q2,d2), sim(q2,d3), sim(q2,d4)) = sim(q2,d4) = 0.72
MaxSim(q3, D) = max(sim(q3,d1), sim(q3,d2), sim(q3,d3), sim(q3,d4)) = sim(q3,d4) = 0.81

Final Score = 0.95 + 0.72 + 0.81 = 2.48
```

### ColBERTv2 Improvements
- **Residual compression**: Reduces storage by 6-10x using centroid-based compression
- **Denoised supervision**: Trains with harder negatives from a cross-encoder teacher
- **State-of-the-art** on in-domain and out-of-domain benchmarks

### Why ColBERT Matters for RAG

| Feature | Bi-Encoder (DPR) | Cross-Encoder | ColBERT |
|---------|-----------------|---------------|---------|
| **Pre-computation** | Full (1 vec/doc) | None | Partial (token vecs/doc) |
| **Query-time cost** | Very low | Very high | Low-medium |
| **Quality** | Good | Best | Near-best |
| **Use case** | First-stage retrieval | Re-ranking only | Retrieval + re-ranking |
| **Token-level matching** | No | Yes | Yes |

> **Interview Tip:** ColBERT is the sweet spot between speed and quality. In interviews, explain the three-way trade-off: BM25 (fast, keyword-only), bi-encoders (fast, semantic), cross-encoders (slow, highest quality), ColBERT (fast enough, near-cross-encoder quality). For production RAG, a common pattern is: BM25 + bi-encoder retrieval -> ColBERT or cross-encoder re-ranking.

**References:**
- Khattab & Zaharia, "ColBERT," SIGIR 2020
- Santhanam et al., "ColBERTv2," NAACL 2022

## 1.11 Hypothetical Document Embeddings (HyDE)

**Paper:** Gao et al., "Precise Zero-Shot Dense Retrieval without Relevance Labels," ACL 2023 (arXiv:2212.10496, December 2022)

### Core Idea

HyDE addresses the **query-document mismatch** problem: queries are short and abstract, while relevant documents are long and detailed. Instead of directly embedding the query, HyDE asks the LLM to **generate a hypothetical answer document**, then uses that document's embedding for retrieval.

### How HyDE Works

```
Standard Dense Retrieval:
==========================
  Query: "What causes northern lights?"
      |
      v
  [Embed Query] --> q_vec = [0.2, 0.5, ...]
      |
      v
  [Search against doc embeddings]  <-- May miss relevant docs because
      |                                query embedding != doc embedding
      v
  Results (potentially poor recall)


HyDE Retrieval:
================
  Query: "What causes northern lights?"
      |
      v
  [LLM generates hypothetical document]
      |
      v
  "The aurora borealis is caused by charged particles
   from the sun interacting with Earth's magnetic field.
   These particles excite atmospheric gases, causing them
   to emit light at various wavelengths..."
      |
      v
  [Embed hypothetical document] --> h_vec = [0.3, 0.6, ...]
      |                                     ^
      v                                     |
  [Search against doc embeddings]    Now in the SAME embedding
      |                              space as real documents!
      v
  Results (much better recall)
```

### Why It Works

The key insight is that **documents are more similar to other documents than queries are to documents**. The hypothetical document:
- Contains factual-sounding content in document style (even if inaccurate)
- Has the right vocabulary, length, and structure
- The encoder's "dense bottleneck filters out incorrect details" through similarity matching

### HyDE Trade-offs

| Aspect | Benefit | Limitation |
|--------|---------|------------|
| **Recall** | Significantly improved | Slight overhead from LLM call |
| **Zero-shot** | No training data needed | Requires instruction-following LLM |
| **Multilingual** | Works across languages | LLM must support target language |
| **Latency** | N/A | Adds LLM generation latency (~1-3s) |
| **Accuracy** | Better retrieval | Hypothetical doc may bias toward wrong subtopic |
| **Cost** | N/A | Extra LLM API call per query |

### When to Use HyDE

- **Great for**: Complex queries, zero-shot scenarios, cross-lingual retrieval, academic search
- **Skip if**: Queries are simple keyword lookups, latency is critical (<100ms), or you have a well-fine-tuned retriever

> **Interview Tip:** HyDE is a clever "no-training-needed" technique. In interviews, explain the core insight in one sentence: "We bridge the query-document semantic gap by generating a fake document that looks like what a real answer would look like, then search for documents similar to that fake one." Also mention that HyDE can be combined with query expansion: generate multiple hypothetical documents and use their embeddings together.

**Reference:** Gao et al., "Precise Zero-Shot Dense Retrieval without Relevance Labels," ACL 2023.

## 1.12 RAPTOR - Tree-Organized Retrieval

**Paper:** Sarthi et al., "RAPTOR: Recursive Abstractive Processing for Tree-Organized Retrieval," ICLR 2024 (arXiv:2401.18059, January 2024)

### Core Idea

RAPTOR addresses a fundamental limitation of standard RAG: it only retrieves **short, contiguous text chunks**, losing the ability to reason about broader themes and high-level concepts. RAPTOR builds a **hierarchical tree of summaries** from the bottom up, enabling retrieval at different levels of abstraction.

### How RAPTOR Works

```
RAPTOR Tree Construction
==========================

Step 1: Start with leaf chunks
  [Chunk 1] [Chunk 2] [Chunk 3] [Chunk 4] [Chunk 5] [Chunk 6] ... [Chunk N]

Step 2: Embed and cluster similar chunks (using GMM soft clustering)
  {Chunk 1, 2, 3}    {Chunk 4, 5}    {Chunk 6, ..., N}

Step 3: Summarize each cluster with LLM
  [Summary A]          [Summary B]     [Summary C]
     |                    |                |
  [C1] [C2] [C3]      [C4] [C5]       [C6]...[CN]

Step 4: Recursively cluster and summarize summaries
           [Meta-Summary X]              [Meta-Summary Y]
              /        \                    /         \
        [Sum A]     [Sum B]           [Sum C]      [Sum D]
        / | \        / \               / \          / | \
      C1 C2 C3    C4  C5            C6  C7       C8 C9 C10

Step 5: Final tree with multiple retrieval levels
                    [Root Summary: Full Document Theme]
                    /                                  \
         [Meta-Summary X]                    [Meta-Summary Y]
         /             \                     /              \
    [Summary A]    [Summary B]         [Summary C]     [Summary D]
    /   |   \       /     \            /      \         /   |   \
  C1   C2   C3    C4     C5          C6      C7      C8   C9   C10
```

### Two Retrieval Strategies

| Strategy | How It Works | Best For |
|----------|-------------|----------|
| **Tree Traversal** | Start at root, descend to relevant branches | Known hierarchy, broad-to-narrow queries |
| **Collapsed Tree** | Flatten all nodes (all levels) into one index, retrieve by similarity | General-purpose, mixed abstraction queries |

### Key Innovation: Multi-Level Abstraction

```
Query: "What is the overall argument of this book?"
  --> Retrieves from Level 2-3 (high-level summaries)

Query: "What specific data was presented in Chapter 3?"
  --> Retrieves from Level 0 (leaf chunks)

Query: "How does the argument in Chapter 3 relate to the conclusion?"
  --> Retrieves from Level 0 (Chapter 3 details) + Level 2 (conclusion summary)
```

### Key Results
- Combined with GPT-4, RAPTOR improved QuALITY benchmark by **20 percentage points** absolute accuracy
- Particularly strong on questions requiring **multi-step reasoning** across sections
- Outperforms standard chunk-based retrieval on questions about themes, arguments, and cross-section relationships

### Trade-offs
- **Indexing cost**: Requires multiple LLM summarization calls during indexing
- **Storage**: Stores original chunks + all summary levels
- **Freshness**: Re-indexing requires re-building the tree (or incremental updates)

> **Interview Tip:** RAPTOR is great for demonstrating understanding of abstraction hierarchies. In interviews, compare it to how humans read: we don't just remember sentences -- we build mental models at different abstraction levels. When asked about handling long documents, RAPTOR is a strong answer. Mention that it's complementary to GraphRAG: RAPTOR captures hierarchical document structure, GraphRAG captures entity relationships.

**Reference:** Sarthi et al., "RAPTOR: Recursive Abstractive Processing for Tree-Organized Retrieval," ICLR 2024.

## 2.1 What Are AI Agents?

### Definition

An **AI agent** is an autonomous system powered by a large language model that can:
1. **Perceive** its environment (receive inputs, read context)
2. **Reason** about goals and plan actions
3. **Act** using tools and APIs
4. **Learn** from feedback and adjust its approach

Unlike simple LLM calls (stateless, single-turn), agents maintain **state**, make **decisions**, and operate in **loops** until a task is complete.

### Core Components of an AI Agent

```
AI AGENT ARCHITECTURE
======================

                    +----------------------------------+
                    |           AI AGENT               |
                    |                                  |
                    |  +----------+    +-----------+   |
  User Goal ------->  | PLANNING |    |  MEMORY   |   |
                    |  |          |    |           |   |
                    |  | - Task   |    | - Short   |   |
                    |  |   decomp |    |   term    |   |
                    |  | - CoT    |    |   (context|   |
                    |  | - ReAct  |    |    window)|   |
                    |  | - Tree   |    | - Long    |   |
                    |  |   of     |    |   term    |   |
                    |  |   Thought|    |   (vector |   |
                    |  +----+-----+    |    DB)    |   |
                    |       |          | - Episodic|   |
                    |       v          |   (past   |   |
                    |  +----------+    |   actions)|   |
                    |  |  LLM    |    +-----------+   |
                    |  | (Brain) |          ^         |
                    |  +----+-----+         |         |
                    |       |               |         |
                    |       v               |         |
                    |  +----------+   +----------+    |
                    |  |  TOOLS  |   | PROFILE/ |    |
                    |  |         |   | PERSONA  |    |
                    |  | - Search|   |          |    |
                    |  | - Code  |   | - Role   |    |
                    |  | - APIs  |   | - Rules  |    |
                    |  | - DB    |   | - Goals  |    |
                    |  +----+-----+   +----------+    |
                    |       |                         |
                    +-------+-------------------------+
                            |
                            v
                     Action / Output
```

### Andrew Ng's Four Agentic Design Patterns (2024)

| Pattern | Description | Impact | Example |
|---------|-------------|--------|---------|
| **Reflection** | LLM examines its own output and improves it | Self-critique loop | Code review agent checks its own code |
| **Tool Use** | LLM calls external tools (search, code exec, APIs) | Extends capabilities beyond text | Agent searches web for current prices |
| **Planning** | LLM creates multi-step plans before acting | Structured problem-solving | Research agent outlines, then writes |
| **Multi-Agent** | Multiple specialized agents collaborate | Division of labor | Coder agent + reviewer agent + tester agent |

### The Power of Agentic Patterns (Andrew Ng's Key Insight)

**HumanEval Coding Benchmark Results:**
- GPT-3.5 (zero-shot): **48.1%** accuracy
- GPT-4 (zero-shot): **67.0%** accuracy
- GPT-3.5 with agentic loop: **95.1%** accuracy

This means a **weaker model with better architecture beats a stronger model without it**. The architecture matters as much as (or more than) the model.

> **Interview Tip:** Start any agent discussion with this framework: Planning + Memory + Tools + Action. Interviewers want to see that you understand agents as systems, not just "LLMs that call functions." The Andrew Ng benchmark is a powerful data point to cite.

**Reference:** 
- Lilian Weng, "LLM Powered Autonomous Agents," lilianweng.github.io, June 2023

## 2.2 Agent Architectures: ReAct, Plan-and-Execute, LATS, Reflexion

### 2.2.1 ReAct (Reasoning + Acting)

**Paper:** Yao et al., "ReAct: Synergizing Reasoning and Acting in Language Models," ICLR 2023

ReAct interleaves **reasoning traces** (thinking) with **actions** (tool calls) in a single prompt trajectory.

```
ReAct Loop
============
  
  Query: "What is the GDP of the country where the Eiffel Tower is located?"
  
  Thought 1: I need to find which country has the Eiffel Tower.
  Action 1:  Search("Eiffel Tower location")
  Observation 1: The Eiffel Tower is in Paris, France.
  
  Thought 2: The Eiffel Tower is in France. Now I need France's GDP.
  Action 2:  Search("France GDP 2024")
  Observation 2: France's GDP in 2024 was approximately $3.1 trillion.
  
  Thought 3: I now have the answer.
  Action 3:  Finish("The GDP of France, where the Eiffel Tower is located, 
                      is approximately $3.1 trillion as of 2024.")
```

### 2.2.2 Plan-and-Execute

**Concept:** Separate the planning phase from the execution phase. An LLM creates a full plan upfront, then a separate executor follows the plan step by step.

```
Plan-and-Execute Architecture
===============================

  User Query
      |
      v
  [PLANNER LLM]
      |
      v
  Plan:
    Step 1: Search for Eiffel Tower location
    Step 2: Extract country name
    Step 3: Search for that country's GDP
    Step 4: Format and return answer
      |
      v
  [EXECUTOR] -- executes each step sequentially
      |              |
      |         Uses tools as needed
      |              |
      v              v
  [RE-PLANNER] -- adjusts plan if a step fails
      |
      v
  Final Answer
```

### 2.2.3 LATS (Language Agent Tree Search)

**Paper:** Zhou et al., "Language Agent Tree Search Unifies Reasoning Acting and Planning in Language Models," 2023

LATS combines **Monte Carlo Tree Search (MCTS)** with LLM agents, enabling the agent to explore multiple action paths and backtrack.

```
LATS: Tree Search for Agents
==============================

                        [Root: User Query]
                       /        |         \
                   Action A   Action B   Action C
                   /    \        |        /    \
                 A1     A2      B1      C1     C2
                 /       \      |       |
               A1a      A1b   B1a    C1a (BEST)
                              |
                         [Backtrack!]
                         [Try C1a instead]

  Key Operations:
  1. SELECT:   Choose most promising node (UCB score)
  2. EXPAND:   Generate possible next actions
  3. EVALUATE: Score leaf nodes with LLM feedback
  4. BACKPROP: Update scores up the tree
  5. REFLECT:  Learn from failed paths
```

### 2.2.4 Reflexion

**Paper:** Shinn et al., "Reflexion: Language Agents with Verbal Reinforcement Learning," NeurIPS 2023

Reflexion adds **verbal self-reflection** after task failures, storing reflections in memory for future attempts.

```
Reflexion Loop
================

  Attempt 1:
    [Agent tries task] --> [Evaluator: FAIL]
         |
         v
    [Self-Reflection]: "I failed because I didn't check the date range.
                         Next time I should verify temporal constraints first."
         |
         v
    [Store reflection in memory]
  
  Attempt 2:
    [Agent retries with reflection in context] --> [Evaluator: PASS]
```

### Architecture Comparison

| Architecture | Planning | Execution | Error Recovery | Best For | Limitation |
|-------------|----------|-----------|----------------|----------|------------|
| **ReAct** | Interleaved | Step-by-step | Limited (linear) | Simple multi-step tasks | No backtracking |
| **Plan-and-Execute** | Upfront plan | Follow plan | Re-plan on failure | Well-structured tasks | Plan may be wrong |
| **LATS** | Tree search | Explore branches | Full backtracking | Complex decisions | High compute cost |
| **Reflexion** | Trial-and-error | Multiple attempts | Verbal reflection | Learning from mistakes | Needs evaluator |

> **Interview Tip:** Know when to recommend each architecture. ReAct is your default -- simple, effective, widely used. Plan-and-Execute when tasks have clear structure. LATS when you need to explore alternatives (e.g., code generation). Reflexion when the agent needs to improve over multiple attempts. A great answer weaves these together: "I'd start with ReAct for prototyping, add Reflexion for learning from failures, and consider LATS for the most complex sub-tasks."

**References:**
- Yao et al., "ReAct," ICLR 2023
- Zhou et al., "LATS," 2023
- Shinn et al., "Reflexion," NeurIPS 2023

### LangGraph: Graph-Based Agent Orchestration

```
LangGraph Architecture
========================

  State = {messages: [], documents: [], query: "..."}
  
  Nodes (Functions):
    [retrieve] --> [grade_documents] --> [generate] --> [check_hallucination]
  
  Edges (Transitions):
    grade_documents --relevant--> generate
    grade_documents --not_relevant--> rewrite_query
    check_hallucination --grounded--> check_answer_quality  
    check_hallucination --not_grounded--> generate (retry)
    check_answer_quality --useful--> END
    check_answer_quality --not_useful--> rewrite_query

  Key Features:
  - Stateful execution with checkpointing
  - Human-in-the-loop breakpoints
  - Streaming support
  - Persistence (save/resume workflows)
```

## 2.4 Orchestration Patterns

### Four Major Multi-Agent Orchestration Patterns

```
PATTERN 1: SEQUENTIAL (Pipeline)
===================================

  [Agent A] --> [Agent B] --> [Agent C] --> [Agent D]
  Researcher    Writer       Editor       Publisher
  
  Pros: Simple, predictable, easy to debug
  Cons: No parallelism, slow for independent tasks
  Use:  Content creation, data processing pipelines


PATTERN 2: PARALLEL (Fan-out / Fan-in)
========================================

                     +--> [Agent B: Analyze data] --+
                     |                               |
  [Agent A: Plan] ---+--> [Agent C: Search web]  ---+--> [Agent E: Synthesize]
                     |                               |
                     +--> [Agent D: Query DB]    ---+
  
  Pros: Fast (parallel execution), good for independent sub-tasks
  Cons: Complex merging logic, harder error handling
  Use:  Research tasks, multi-source data gathering


PATTERN 3: HIERARCHICAL (Supervisor)
======================================

              [Supervisor Agent]
              /       |        \
             v        v         v
        [Worker A] [Worker B] [Worker C]
        (Research)  (Code)    (Analysis)
             \        |        /
              v       v       v
              [Supervisor Agent]
              (Reviews, delegates more, or finishes)
  
  Pros: Clear authority, specialized workers, scalable
  Cons: Supervisor is bottleneck, single point of failure
  Use:  Complex enterprise tasks, project management


PATTERN 4: DEBATE / CONSENSUS
================================

  [Agent A: Advocate Position 1]
         |
         v
  [Agent B: Advocate Position 2] -- critiques A's position
         |
         v
  [Agent A: Responds to B's critique] -- refines position
         |
         v
  [Judge Agent: Evaluates both positions]
         |
         v
  [Final Decision]
  
  Pros: Higher quality through adversarial critique
  Cons: Expensive (many LLM calls), may not converge
  Use:  Critical decisions, safety review, fact-checking
```

### Orchestration Pattern Comparison

| Pattern | Latency | Cost | Complexity | Quality | Reliability |
|---------|---------|------|------------|---------|-------------|
| **Sequential** | High (serial) | Low | Low | Moderate | High |
| **Parallel** | Low (concurrent) | Medium | Medium | Moderate | Medium |
| **Hierarchical** | Medium | Medium-High | High | High | Medium |
| **Debate** | Very High | High | Medium | Highest | Low-Medium |

### Emerging Pattern: Dynamic Orchestration

In 2025-2026, the trend is toward **dynamic orchestration** where the orchestration pattern itself is chosen by an agent:

```
Dynamic Orchestration
======================

  [Meta-Agent / Orchestrator]
      |
      v
  "This task needs parallel research 
   followed by sequential writing"
      |
      v
  [Dynamically constructs workflow]:
    Step 1: Fan-out to 3 research agents (parallel)
    Step 2: Merge findings (sequential)
    Step 3: Writer agent (sequential)
    Step 4: Reviewer agent (sequential)
```

> **Interview Tip:** When designing multi-agent systems in interviews, always state your orchestration pattern explicitly. "I'll use a hierarchical pattern with a supervisor agent because the task requires dynamic delegation" shows architectural thinking. Bonus points for mentioning failure handling: "If a worker agent fails, the supervisor can retry with a different approach or escalate to a human."

## 2.5 Memory Systems

### Three Types of Agent Memory

```
AGENT MEMORY ARCHITECTURE
===========================

+------------------------------------------------------------------+
|                        AI AGENT                                   |
|                                                                   |
|  SHORT-TERM MEMORY              LONG-TERM MEMORY                 |
|  (Context Window)               (Persistent Storage)              |
|  +--------------------+        +---------------------------+     |
|  | Current conversation|        | SEMANTIC MEMORY            |     |
|  | Recent tool outputs |        | (Facts & Knowledge)        |     |
|  | Working scratchpad  |        | Vector DB of past info     |     |
|  |                    |        |                           |     |
|  | Capacity: 4K-200K  |        | EPISODIC MEMORY            |     |
|  | tokens             |        | (Past Experiences)         |     |
|  | Volatile: Lost     |        | Successful action traces   |     |
|  | between sessions   |        | Failed attempts + lessons  |     |
|  +--------------------+        |                           |     |
|                                | PROCEDURAL MEMORY          |     |
|                                | (How-To Knowledge)         |     |
|                                | Tool usage patterns        |     |
|                                | Learned workflows          |     |
|                                +---------------------------+     |
+------------------------------------------------------------------+
```

### Memory System Deep Dive

| Memory Type | Analogy | Storage | Access Pattern | Duration | Implementation |
|-------------|---------|---------|---------------|----------|----------------|
| **Short-term** | Working memory | Context window | Direct in prompt | Session only | Chat history buffer |
| **Semantic** | Factual knowledge | Vector DB | Similarity search | Permanent | Pinecone, Chroma, pgvector |
| **Episodic** | Personal experiences | Structured DB | Time + relevance | Permanent | Reflexion traces, action logs |
| **Procedural** | Muscle memory | Code/rules | Direct lookup | Permanent | Tool configs, learned prompts |

### Memory Management Strategies

**1. Sliding Window**: Keep last N messages in context
```
Messages: [m1, m2, m3, m4, m5, m6, m7, m8]
Window=5: [m4, m5, m6, m7, m8]  (oldest dropped)
```

**2. Summary Memory**: LLM summarizes old conversations, keeps summary + recent messages
```
[Summary of m1-m5] + [m6, m7, m8]  (compressed but preserves key info)
```

**3. Entity Memory**: Extract and track entities mentioned across conversations
```
Entities: {
  "Project Alpha": "ML pipeline, deadline March, team of 5",
  "John": "Lead engineer, prefers Python, working on retrieval module"
}
```

**4. Retrieval-Based Memory**: Store all messages in vector DB, retrieve relevant ones
```
Query: "What did we decide about the database?"
Retrieved: [m3: "We chose PostgreSQL with pgvector", m7: "Added read replicas"]
Context: [Retrieved memories] + [Recent messages]
```

### Generative Agents Memory (Stanford/Google, 2023)

The influential "Generative Agents" paper introduced a sophisticated memory architecture:
- **Memory Stream**: Timestamped log of all observations and actions
- **Retrieval**: Scores memories by **recency** (time decay), **importance** (LLM-rated significance), and **relevance** (embedding similarity to current context)
- **Reflection**: Periodically synthesizes memories into higher-level observations ("John has been stressed lately because of the project deadline")

> **Interview Tip:** Memory is often the weakest part of agent systems and a great area to show depth. Explain the trade-off: short-term memory is fast but limited, long-term memory scales but adds retrieval latency and relevance challenges. Mention that memory management is essentially a RAG problem within the agent itself -- you're retrieving relevant past experiences.

**References:**
- Park et al., "Generative Agents: Interactive Simulacra of Human Behavior," 2023
- Lilian Weng, "LLM Powered Autonomous Agents," 2023

## 2.6 Tool Use Patterns and Function Calling

### How Function Calling Works

```
FUNCTION CALLING FLOW
======================

1. Define tools (JSON schema):
   {
     "name": "search_database",
     "description": "Search product database by query",
     "parameters": {
       "query": {"type": "string"},
       "limit": {"type": "integer", "default": 10}
     }
   }

2. LLM receives user message + tool definitions
3. LLM decides to call a tool (returns structured JSON):
   {"tool": "search_database", "args": {"query": "wireless headphones", "limit": 5}}
4. Application executes the tool
5. Tool result fed back to LLM
6. LLM generates final response using tool output
```

### Tool Use Design Patterns

| Pattern | Description | Example | When to Use |
|---------|-------------|---------|-------------|
| **Single tool** | LLM calls one tool | Calculator for math | Simple augmentation |
| **Sequential tools** | Chain of tool calls | Search -> Summarize -> Email | Multi-step workflows |
| **Parallel tools** | Multiple simultaneous calls | Search 3 DBs at once | Independent data gathering |
| **Tool selection** | LLM chooses from many tools | Pick from 20+ APIs | General-purpose assistants |
| **Tool creation** | LLM writes code as a tool | Generate Python function | Novel computations |
| **Human-in-the-loop** | Pause for human approval | "Shall I send this email?" | High-stakes actions |

### Model Context Protocol (MCP) -- The Emerging Standard (2024-2025)

MCP, introduced by Anthropic, is becoming the **universal standard** for connecting LLMs to tools:

```
MCP Architecture
==================

  [LLM / Agent]
      |
      v
  [MCP Client]  <-- Standardized protocol
      |
      +---> [MCP Server: File System]     (read/write files)
      +---> [MCP Server: Database]        (query SQL/NoSQL)
      +---> [MCP Server: Web Search]      (search APIs)
      +---> [MCP Server: GitHub]          (repos, PRs, issues)
      +---> [MCP Server: Slack]           (messages, channels)
      
  Key Benefits:
  - One protocol, many tools (like USB for AI)
  - Server-side tool definitions (no client changes)
  - Secure: server controls access
  - Discoverable: client can list available tools
```

### Best Practices for Tool Design

1. **Clear descriptions**: The LLM selects tools based on description quality
2. **Typed parameters**: Use JSON Schema with types, enums, and defaults
3. **Error messages**: Return informative errors the LLM can reason about
4. **Idempotent tools**: Prefer tools that can be safely retried
5. **Least privilege**: Give tools minimal permissions needed
6. **Rate limiting**: Prevent runaway agent loops from making excessive API calls

> **Interview Tip:** Tool use is the bridge between reasoning and action. In interviews, discuss the "tool selection problem" -- with 50+ tools, how does the LLM choose the right one? Mention strategies: (1) good descriptions, (2) few-shot examples in the prompt, (3) hierarchical tool selection (first pick a category, then a specific tool), (4) fine-tuning for tool selection.

**Reference:** Anthropic, "Model Context Protocol," 2024; OpenAI, "Function Calling," API documentation, 2023-2024.

## 2.7 Agent Evaluation and Benchmarks

### Why Agent Evaluation Is Hard

Unlike traditional ML (clear metrics on static test sets), agents are:
- **Non-deterministic**: Same input can produce different action sequences
- **Multi-step**: Errors compound across steps
- **Tool-dependent**: External tools may fail or return stale data
- **Subjective**: Quality of a research report is hard to score automatically

### Key Agent Benchmarks

| Benchmark | What It Tests | Tasks | Key Metric |
|-----------|--------------|-------|------------|
| **SWE-bench** | Code generation + debugging | Fix real GitHub issues | % resolved |
| **WebArena** | Web navigation agents | Complete tasks on real websites | Task success rate |
| **GAIA** | General AI assistants | Questions needing tools + reasoning | Accuracy |
| **AgentBench** | Multi-domain agent tasks | OS, DB, web, games, coding | Success rate |
| **HumanEval** | Code generation | Write Python functions | pass@k |
| **ToolBench** | Tool selection and use | 16K+ real APIs | Win rate |
| **MINT** | Multi-turn tool interaction | Math, science with tools | Task success |

### Evaluation Dimensions for Agents

```
AGENT EVALUATION FRAMEWORK
============================

  1. TASK SUCCESS          Did the agent complete the task?
     |                     Binary or graded success metric
     |
  2. EFFICIENCY            How many steps/tokens/API calls?
     |                     Fewer is better (cost + latency)
     |
  3. TOOL ACCURACY         Did the agent call the right tools
     |                     with the right parameters?
     |
  4. REASONING QUALITY     Was the thought process logical?
     |                     Trace evaluation (human or LLM judge)
     |
  5. SAFETY                Did the agent avoid harmful actions?
     |                     Guardrail compliance, no data leaks
     |
  6. ROBUSTNESS            Does it handle edge cases, tool failures,
     |                     ambiguous queries gracefully?
     |
  7. USER EXPERIENCE       Response time, clarity, helpfulness
                           Human satisfaction scores
```

### Practical Agent Evaluation Strategy

| Level | What to Evaluate | Method | Frequency |
|-------|-----------------|--------|-----------|
| **Unit tests** | Individual tools, prompts | Automated assertions | Every commit |
| **Integration tests** | Multi-step flows | Golden datasets with expected outputs | Daily |
| **LLM-as-judge** | Overall quality | GPT-4/Claude scores agent outputs | Daily |
| **Human evaluation** | Subjective quality | Annotator ratings | Weekly |
| **A/B testing** | User preference | Live traffic split | Continuous |
| **Red teaming** | Safety, adversarial | Dedicated adversarial prompts | Before release |

> **Interview Tip:** Agent evaluation is an unsolved problem and interviewers know it. Show maturity by acknowledging the challenges: "There's no single metric for agents. I'd use a multi-level approach: unit tests for tools, golden-set integration tests for flows, LLM-as-judge for quality, and human eval for subjective assessment. I'd also track cost-per-task and latency as first-class metrics."

## 3.1 Chunking Optimization

### Why Chunking Matters

Chunking is the **most impactful yet underrated** component of a RAG system. Poor chunking leads to: retrieved chunks missing key context, chunks containing irrelevant information mixed with relevant, and broken tables/code blocks.

### Chunking Strategies Comparison

| Strategy | How It Works | Chunk Size | Pros | Cons | Best For |
|----------|-------------|------------|------|------|----------|
| **Fixed-size** | Split every N characters/tokens | 256-1024 tokens | Simple, predictable | Breaks mid-sentence | Quick prototypes |
| **Recursive** | Split by hierarchy: paragraph -> sentence -> word | Varies | Respects text structure | May produce uneven chunks | General text documents |
| **Semantic** | Split when embedding similarity drops | Varies | Semantically coherent chunks | Computationally expensive | High-quality requirements |
| **Sentence-window** | Retrieve sentence, expand to surrounding window | Sentence + context | Precise retrieval + context | Complex implementation | QA systems |
| **Parent-child** | Small chunks for retrieval, return parent doc | Small/Large | Best of both worlds | Storage overhead | Enterprise search |
| **Agentic** | LLM decides chunk boundaries | Varies | Optimal boundaries | Very expensive | Critical applications |
| **Document-structure** | Split by headings, sections, chapters | Section-level | Preserves document logic | Needs structured docs | Technical docs, manuals |

### Parent-Child Chunking (Most Recommended for Production)

```
PARENT-CHILD CHUNKING
=======================

  Original Document
  ==========================================
  | Chapter 1: Introduction                |
  |   Section 1.1: Background              |
  |     Paragraph 1 (sentence a, b, c)     |  <-- Child chunks (small)
  |     Paragraph 2 (sentence d, e, f)     |      used for RETRIEVAL
  |   Section 1.2: Methods                 |
  |     Paragraph 3 (sentence g, h, i)     |
  ==========================================

  Index Structure:
    Child chunk "sentence a, b, c" --> points to --> Parent: "Section 1.1"
    Child chunk "sentence d, e, f" --> points to --> Parent: "Section 1.1"
    
  At retrieval time:
    Query matches child chunk (precise)
    System returns parent chunk (full context)
```

### Optimal Chunk Size Guidelines

| Use Case | Recommended Size | Overlap | Rationale |
|----------|-----------------|---------|-----------|
| **Factual QA** | 256-512 tokens | 50 tokens | Precise, focused chunks |
| **Summarization** | 1024-2048 tokens | 100 tokens | More context for synthesis |
| **Code** | Function/class level | 0 | Natural boundaries |
| **Legal/Medical** | Section level | Full section | Must preserve complete clauses |
| **Conversational** | 512-1024 tokens | 100 tokens | Balance precision + context |

> **Interview Tip:** When asked about chunking, start with: "The right chunking strategy depends on the use case. For factual QA, I'd use parent-child chunking with small children (256 tokens) for precise retrieval and larger parents (1024 tokens) for context. I'd always include overlap to avoid losing information at boundaries."

### How to Choose an Embedding Model

```
Decision Tree for Embedding Model Selection
=============================================

  What is your deployment constraint?
      |
      +--- API-only (no GPU) ----> OpenAI text-embedding-3-small/large
      |                             or Voyage AI (best quality)
      |
      +--- Self-hosted (GPU) ----> NV-Embed-v2 (best quality)
      |                             or GTE-Qwen2 (longest context)
      |                             or e5-mistral-7b (good balance)
      |
      +--- Edge/Mobile ---------> bge-small, all-MiniLM-L6-v2
      
  Secondary considerations:
      - Need multilingual? --> Cohere embed-v4, e5-multilingual
      - Need code? --> Voyage Code, text-embedding-3-large
      - Need long docs? --> GTE-Qwen2 (128K), Voyage (32K)
      - Budget-constrained? --> text-embedding-3-small ($0.02/1M)
```

### Fine-Tuning Embeddings for Your Domain

**When to fine-tune:**
- Domain-specific jargon (legal, medical, financial)
- Significant quality gap with off-the-shelf models
- Enough training data (>1000 query-document pairs)

**Fine-tuning approaches:**
1. **Contrastive learning**: Train with (query, positive_doc, negative_doc) triplets
2. **Matryoshka Representation Learning (MRL)**: Train model to produce useful embeddings at multiple dimensions (128, 256, 512, 1024)
3. **LoRA fine-tuning**: Efficient adapter-based fine-tuning (e.g., jina-embeddings-v3 supports task-specific LoRA heads)
4. **Synthetic data generation**: Use LLM to generate query-doc pairs from your corpus

> **Interview Tip:** Don't just say "I'd use OpenAI embeddings." Show you know the landscape: "I'd start with text-embedding-3-small for prototyping due to cost efficiency. For production, I'd evaluate against our domain-specific test set. If off-the-shelf models underperform, I'd fine-tune using contrastive learning with synthetic query-doc pairs generated from our corpus. I'd also consider Matryoshka training for flexible dimension reduction."

## 3.4 Hybrid Search (Dense + Sparse Retrieval)

### Why Hybrid Search?

Neither dense (embedding) nor sparse (keyword) retrieval is universally better:

| Query Type | Dense (Embeddings) | Sparse (BM25) | Winner |
|-----------|-------------------|---------------|--------|
| "What causes headaches?" | Finds semantic matches about migraine, pain | Matches "headaches" literally | Dense |
| "error code ERR_SSL_PROTOCOL" | May miss exact code | Matches exact error string | Sparse |
| "Python list comprehension syntax" | Finds conceptually similar | Matches exact terms | Tie |
| "latest iPhone price" | Finds phone-related content | Matches "iPhone" + "price" | Sparse |
| "How to handle employee burnout" | Finds related HR content | Needs exact terms | Dense |

### Hybrid Search Architecture

```
HYBRID SEARCH PIPELINE
========================

  User Query: "ERR_SSL_PROTOCOL error in nginx"
      |
      +------------------+-------------------+
      |                  |                   |
      v                  v                   v
  [BM25/SPLADE]    [Dense Embeddings]    [Optional: ColBERT]
  Sparse search     Semantic search       Late interaction
      |                  |                   |
      v                  v                   v
  Results_sparse     Results_dense        Results_colbert
  (10 docs)          (10 docs)            (10 docs)
      |                  |                   |
      +------------------+-------------------+
                         |
                         v
                  [Score Fusion]
                  (RRF or weighted linear)
                         |
                         v
                  Merged ranked list
                         |
                         v
                  [Optional: Re-Ranker]
                  Cross-encoder or Cohere Rerank
                         |
                         v
                  Final Top-K results
```

### Score Fusion Methods

**1. Reciprocal Rank Fusion (RRF)** -- Most common, parameter-free:
```
RRF_score(doc) = sum( 1 / (k + rank_in_system_i) ) for each retrieval system i

Where k = 60 (constant to prevent domination by top-ranked docs)

Example:
  Doc X: rank 2 in BM25, rank 5 in dense
  RRF = 1/(60+2) + 1/(60+5) = 0.0161 + 0.0154 = 0.0315

  Doc Y: rank 1 in BM25, rank 20 in dense  
  RRF = 1/(60+1) + 1/(60+20) = 0.0164 + 0.0125 = 0.0289

  Doc X wins (consistently ranked in both systems)
```

**2. Weighted Linear Combination**:
```
score(doc) = alpha * normalize(sparse_score) + (1-alpha) * normalize(dense_score)

Typical alpha = 0.3-0.5 (tune on validation set)
```

### When to Use Hybrid Search

| Scenario | Use Hybrid? | Why |
|----------|------------|-----|
| Technical documentation | **Yes** | Mix of exact terms (error codes) and concepts |
| Legal documents | **Yes** | Exact statute numbers + semantic meaning |
| Customer support | **Yes** | Product names (sparse) + issue descriptions (dense) |
| Creative writing search | **No** | Pure semantic, keywords less important |
| Log analysis | **No** | Pure keyword/regex, embeddings add noise |

> **Interview Tip:** Hybrid search is a production must-have. Say: "In production, I always start with hybrid search using RRF fusion because it's simple, parameter-free, and consistently outperforms either method alone. The beauty of RRF is that it doesn't need score normalization -- it only uses ranks." Mention SPLADE as an advanced sparse model that learns sparse representations (better than raw BM25).

**Reference:** Cormack et al., "Reciprocal Rank Fusion outperforms Condorcet and individual Rank Learning Methods," SIGIR 2009.

<a id="3.5"></a>
## 3.5 Re-Ranking Strategies

### Why Re-Rank?

First-stage retrieval (BM25, bi-encoder) optimizes for **recall** (find all relevant docs) but sacrifices **precision**. Re-ranking applies a more expensive model to re-order the top candidates for precision.

```
TWO-STAGE RETRIEVAL PIPELINE
==============================

  Stage 1: RETRIEVAL (fast, high recall)          Stage 2: RE-RANKING (slow, high precision)
  ====================================            =======================================
  
  Query --> [Bi-Encoder / BM25]                   Top-100 docs --> [Cross-Encoder / Reranker]
            Process millions of docs                               Process only 100 docs
            ~1-10ms                                                ~50-500ms
            Returns top-100                                        Returns re-ordered top-10
```

### Re-Ranking Model Comparison

| Reranker | Type | Quality | Latency | Cost | Best For |
|----------|------|---------|---------|------|----------|
| **Cohere Rerank v3** | API | Excellent | ~100ms | $2/1K queries | Production, easy integration |
| **cross-encoder/ms-marco** | Cross-encoder | Very Good | ~200ms | Self-hosted | Open-source production |
| **ColBERTv2** | Late interaction | Very Good | ~50ms | Self-hosted | Speed + quality balance |
| **bge-reranker-v2-m3** | Cross-encoder | Excellent | ~150ms | Self-hosted | Multilingual |
| **RankGPT** | LLM-based | Best | ~2-5s | LLM API cost | When quality > latency |
| **FlashRank** | Lightweight | Good | ~10ms | Self-hosted | Ultra-low latency |

### RankGPT: LLM-as-Reranker

A novel approach where an LLM directly re-ranks documents using pairwise or listwise comparison:

```
RankGPT Prompt (Simplified):
==============================

"Given the query: 'How to optimize database queries'
Please rank the following passages from most to least relevant:

[1] Database indexing strategies for faster query execution...
[2] Introduction to SQL databases and their history...
[3] Query optimization techniques including EXPLAIN ANALYZE...
[4] NoSQL vs SQL: A comprehensive comparison...

Output the ranking as a permutation: [3, 1, 4, 2]"
```

### Practical Re-Ranking Tips

1. **Retrieve more, re-rank to fewer**: Retrieve top-100, re-rank to top-5
2. **Set a relevance threshold**: After re-ranking, discard documents below a score threshold
3. **Consider latency budget**: Cross-encoders add 100-500ms; decide if your SLA allows it
4. **Batch processing**: Re-rank in batches for throughput
5. **Cache re-ranked results**: Cache query -> re-ranked results for frequent queries

> **Interview Tip:** Re-ranking is a high-ROI, low-effort improvement. In interviews, say: "The single biggest improvement I'd make to any RAG system is adding a re-ranker. In my experience, Cohere Rerank or a cross-encoder model can improve answer quality by 10-20% with minimal code changes -- just add it between retrieval and generation."

<a id="3.6"></a>
## 3.6 Caching Strategies for RAG

### Why Cache?

RAG systems are expensive (embedding calls + vector DB queries + LLM generation). Caching can reduce latency by 10-100x and cost by 50-90% for repeated or similar queries.

### Caching Layers

```
RAG CACHING ARCHITECTURE
==========================

  User Query
      |
      v
  [Layer 1: EXACT CACHE]
  Key: hash(query)
  Hit? --> Return cached answer immediately (<1ms)
      |
      v (miss)
  [Layer 2: SEMANTIC CACHE]
  Key: embedding(query), threshold: 0.95
  Hit? --> Return cached answer for similar query (~5ms)
      |
      v (miss)
  [Layer 3: RETRIEVAL CACHE]
  Key: hash(query), Value: retrieved chunks
  Hit? --> Skip retrieval, go straight to generation (~50ms saved)
      |
      v (miss)
  [Layer 4: EMBEDDING CACHE]
  Key: hash(text), Value: embedding vector
  Hit? --> Skip embedding API call (~20ms saved)
      |
      v (miss)
  [Full RAG Pipeline]
  Embed -> Retrieve -> Rerank -> Generate
```

### Cache Strategy Comparison

| Cache Type | Hit Rate | Latency Saved | Implementation | Invalidation |
|-----------|----------|---------------|----------------|-------------|
| **Exact query cache** | Low (5-15%) | ~500ms-2s | Redis/Memcached hash lookup | TTL-based |
| **Semantic cache** | Medium (20-40%) | ~500ms-2s | Vector similarity on queries | Threshold tuning |
| **Retrieval cache** | Medium (15-30%) | ~100-300ms | Query hash -> doc IDs | On re-indexing |
| **Embedding cache** | High (60-80%) | ~20-50ms per call | Text hash -> vector | On model change |
| **LLM response cache** | Low-Medium | ~1-5s | Prompt hash -> response | TTL-based |

### Semantic Cache Deep Dive

Semantic caching uses embedding similarity to serve cached results for **semantically similar** (not just identical) queries:

```
Example:
  Cached: "What is the return policy?" -> "Our return policy is..."
  
  New query: "How do I return an item?"
  Similarity: 0.96 (above 0.95 threshold)
  --> Serve cached answer! No RAG pipeline needed.
  
  New query: "What is the refund timeline?"
  Similarity: 0.88 (below 0.95 threshold)
  --> Cache miss, run full pipeline.
```

**Tools:** GPTCache, LangChain SemanticCache, Redis with vector search

> **Interview Tip:** Caching is a production differentiator. In interviews, mention the layered approach: "I'd implement embedding caching first (highest hit rate, easiest), then add semantic caching for similar queries. For a customer support bot, 30-40% of queries are near-duplicates -- semantic caching handles these with sub-10ms latency instead of 2-3 seconds."

## 3.7 Latency Optimization

### RAG Latency Breakdown

```
TYPICAL RAG LATENCY BREAKDOWN
================================

  Component              Latency        % of Total
  -------------------------------------------
  Query preprocessing     10-50ms         2%
  Embedding generation    20-100ms        5%
  Vector search           5-50ms          3%
  Re-ranking             50-500ms        20%
  LLM generation        500-3000ms       70%
  -------------------------------------------
  TOTAL                 585-3700ms      100%
```

### Optimization Strategies (Ordered by Impact)

| Strategy | Latency Reduction | Complexity | Cost Impact |
|----------|------------------|------------|-------------|
| **Streaming responses** | Perceived: 80% | Low | None |
| **Semantic caching** | Up to 95% (on hit) | Medium | Reduces cost |
| **Smaller LLM for simple queries** | 50-70% | Medium | Reduces cost |
| **Parallel retrieval** | 30-50% (on multi-source) | Medium | None |
| **Quantized embedding models** | 20-40% | Medium | None |
| **Pre-computed answers** | 95% (for known queries) | High | Increases storage |
| **Reduce context length** | 20-40% | Low | None |
| **GPU-accelerated vector search** | 50-80% (on search) | High | Increases infra cost |
| **Edge deployment** | Network latency eliminated | Very High | Increases complexity |

### Key Optimizations in Detail

**1. Streaming**: Don't wait for full generation. Stream tokens as they're generated. Users perceive the first token latency (TTFT), not total latency.

**2. Adaptive model routing**: Use a fast, small model (GPT-4o-mini, Claude Haiku) for simple queries and a large model (GPT-4o, Claude Opus) for complex ones. This cuts average latency by 50%.

**3. Reduce retrieved context**: Instead of stuffing 10 chunks into the prompt, use aggressive re-ranking to select only the top 3-5 most relevant. Shorter context = faster generation.

**4. Async and parallel**: Run embedding generation, BM25 search, and metadata filtering in parallel instead of sequentially.

> **Interview Tip:** Always profile before optimizing. The #1 mistake is optimizing vector search latency (3% of total) when LLM generation (70%) is the real bottleneck. Show you know where the time goes: "I'd first add streaming to improve perceived latency, then implement adaptive model routing to reduce generation time for simple queries."

## 3.9 Guardrails and Safety in RAG Systems

### Safety Concerns in RAG

| Risk | Description | Impact | Mitigation |
|------|-------------|--------|------------|
| **Prompt injection** | User crafts input to override system instructions | Data leak, harmful output | Input sanitization, guardrail models |
| **Data poisoning** | Malicious content in knowledge base | Wrong/harmful answers | Content filtering at indexing |
| **Hallucination** | LLM generates info not in retrieved context | Misinformation | Faithfulness checking, citations |
| **PII leakage** | RAG returns documents containing PII | Privacy violation | PII detection + masking |
| **Topic drift** | System answers questions outside its scope | Liability, misinformation | Topic classifiers, scope guardrails |
| **Bias amplification** | Biased training data/documents | Discriminatory outputs | Bias detection, diverse sources |

### Guardrail Architecture

```
GUARDRAIL LAYERS
==================

  User Input
      |
      v
  [INPUT GUARDRAILS]
  - Prompt injection detection (classifier or LLM)
  - Topic/scope validation ("Is this within our domain?")
  - PII detection in query
  - Rate limiting
      |
      v
  [RAG Pipeline] --> Retrieved docs --> [RETRIEVAL GUARDRAILS]
                                        - PII masking in retrieved content
                                        - Relevance threshold filtering
                                        - Source authority validation
      |
      v
  [OUTPUT GUARDRAILS]
  - Hallucination detection (cross-reference with retrieved docs)
  - Toxicity/harm classifier
  - PII detection in output
  - Citation verification
  - Confidence scoring
      |
      v
  User-facing Response (with citations)
```

### Guardrail Tools

| Tool | Focus | How It Works |
|------|-------|-------------|
| **Guardrails AI** | Structured output validation | Validators for format, PII, toxicity |
| **NeMo Guardrails (NVIDIA)** | Conversational safety | Programmable dialog rails |
| **LLM Guard** | Input/output safety | Prompt injection, PII, toxicity |
| **Rebuff** | Prompt injection | Multi-layer injection detection |

> **Interview Tip:** Safety shows production maturity. Always mention guardrails when designing a RAG system: "I'd add input guardrails for prompt injection detection, PII masking on retrieved documents, and output guardrails for hallucination detection. Every answer would include citations to the source documents so users can verify."

<a id="5.1"></a>
## 5.1 Top 20 Advanced RAG & Agentic Systems Interview Questions with Answers

### Question 1: What is RAG and why is it needed?
**Answer**: RAG (Retrieval-Augmented Generation) augments an LLM with external knowledge retrieved at query time. It's needed because LLMs have knowledge cutoffs, hallucinate, can't access private data, and can't be easily updated. RAG grounds the LLM's responses in actual documents, reducing hallucination and enabling domain-specific answers.

---

### Question 2: What are the main failure modes of naive RAG?
**Answer**: 
1. **Irrelevant retrieval**: Top-K chunks may not be relevant to the query
2. **Lost in the middle**: LLMs ignore information in the middle of long contexts
3. **Missing context**: Small chunks lose surrounding context needed for understanding
4. **No quality control**: System blindly trusts retrieved documents
5. **Single strategy**: Same pipeline for simple and complex queries
6. **No self-correction**: No mechanism to detect or fix bad retrievals/generations

---

### Question 3: Explain GraphRAG and when you'd use it.
**Answer**: GraphRAG (Microsoft) builds knowledge graphs from documents using LLM entity extraction, applies Leiden community detection for hierarchical clustering, and generates community summaries. It excels at **global questions** ("What are the main themes across all documents?") where naive RAG fails because it only retrieves individual chunks. I'd use it when the corpus is entity-rich and users ask both specific and broad questions. The trade-off is high indexing cost (many LLM calls).

---

### Question 4: Compare Self-RAG and CRAG.
**Answer**: Both add self-correction, but differently:
- **CRAG**: Adds a separate lightweight evaluator that grades retrieval quality (correct/ambiguous/incorrect) and triggers web search as fallback. Plug-and-play -- no model fine-tuning needed.
- **Self-RAG**: Trains special reflection tokens into the LM itself to decide when to retrieve, evaluate relevance, and check if generation is supported. Requires fine-tuning but is more deeply integrated.

CRAG is easier to deploy; Self-RAG gives better quality but requires training.

---

### Question 5: What is Agentic RAG and how does it differ from standard RAG?
**Answer**: Agentic RAG replaces the fixed pipeline with an AI agent that autonomously decides WHEN to retrieve (not always), WHAT to retrieve (choosing among multiple tools/sources), and HOW to process results (with self-correction loops). Standard RAG always retrieves, uses one source, and runs a single pass. Agentic RAG can decompose complex queries, try multiple strategies, and verify its own answers.

---

### Question 6: How would you implement hybrid search in a RAG system?
**Answer**: I'd combine BM25 (sparse/keyword) with dense embeddings using Reciprocal Rank Fusion (RRF). BM25 handles exact matches (error codes, product names), while dense retrieval handles semantic meaning. RRF merges results by rank position: `score = sum(1/(k + rank_i))` where k=60. This is parameter-free and consistently outperforms either method alone. I'd add a cross-encoder re-ranker on top for the final ordering.

---

### Question 7: Explain the "Lost in the Middle" problem and how to mitigate it.
**Answer**: Liu et al. (2023) showed LLMs perform best when relevant information is at the beginning or end of the context, but struggle with information in the middle. Mitigations:
1. Place most relevant chunks first (re-rank by relevance)
2. Use fewer, higher-quality chunks instead of stuffing many chunks
3. Use models trained to handle long contexts uniformly (e.g., Gemini, Claude)
4. Implement RAPTOR-style hierarchical summaries for better context organization

---

### Question 8: What is ColBERT and why is it significant for retrieval?
**Answer**: ColBERT uses "late interaction" -- it encodes each query token and document token into separate vectors, then scores relevance using MaxSim (for each query token, find its max similarity with any document token, sum the scores). This gives near-cross-encoder quality with bi-encoder-like speed, because document token vectors can be pre-computed and cached. It's the sweet spot between speed and quality.

---

### Question 9: Explain HyDE (Hypothetical Document Embeddings).
**Answer**: HyDE bridges the query-document embedding gap. Instead of embedding the short query directly, it asks the LLM to generate a hypothetical answer document, then embeds that document for retrieval. The insight is that a generated document is more similar to real documents in embedding space than a short query is. The encoder's dense bottleneck filters out factual errors. It works zero-shot and across languages.

---

### Question 10: What are the four agentic design patterns (Andrew Ng)?
**Answer**:
1. **Reflection**: LLM critiques its own output and iterates
2. **Tool Use**: LLM calls external tools (search, code, APIs)
3. **Planning**: LLM creates multi-step plans before execution
4. **Multi-Agent Collaboration**: Multiple specialized agents work together

Key insight: GPT-3.5 with an agentic loop (95.1% on HumanEval) beats GPT-4 zero-shot (67%). Architecture matters more than model.

---

### Question 11: Compare ReAct, Plan-and-Execute, LATS, and Reflexion.
**Answer**:
- **ReAct**: Interleaves thinking and acting in a linear trace. Default choice -- simple and effective.
- **Plan-and-Execute**: Creates a full plan upfront, then executes step by step. Re-plans on failure. Good for structured tasks.
- **LATS**: Uses Monte Carlo Tree Search to explore multiple action paths with backtracking. Best for complex decisions but computationally expensive.
- **Reflexion**: Adds verbal self-reflection after failures, stored in memory for future attempts. Learns from mistakes across attempts.

---

### Question 12: How do memory systems work in AI agents?
**Answer**: Three types:
- **Short-term**: Context window (current conversation, working memory). Lost between sessions.
- **Long-term semantic**: Vector DB storing facts and knowledge. Retrieved by similarity.
- **Episodic**: Past experiences (successful/failed action traces) used for reflection and improvement.

The key design challenge is deciding what to remember and what to forget, much like the RAG problem itself -- you're retrieving relevant memories from past experiences.

---

### Question 13: How would you evaluate a RAG system?
**Answer**: I'd use the RAGAS framework with these metrics:
- **Faithfulness**: Is the answer grounded in retrieved context? (Hallucination detection)
- **Context Recall**: Did we retrieve all relevant documents?
- **Context Precision**: Are relevant documents ranked higher?
- **Answer Correctness**: Is the answer factually correct?

I'd build a golden test set of 100-200 query-answer pairs and run automated evaluation on every pipeline change. I'd also track latency and cost per query as operational metrics.

---

### Question 14: What chunking strategy would you use for a production RAG system?
**Answer**: Parent-child chunking. Small child chunks (256-512 tokens) for precise retrieval, but return the larger parent chunk (1024-2048 tokens) for context. I'd use recursive text splitting with overlap (50-100 tokens) and preserve document structure (headings, sections). For code, I'd chunk at function/class boundaries. For tables, I'd serialize to markdown and store both raw table and natural language summary.

---

### Question 15: How would you handle multi-modal documents (PDFs with tables, images)?
**Answer**: Three approaches depending on the use case:
1. **Caption + embed**: Use a VLM to caption images, serialize tables to markdown, embed as text. Simple but loses visual information.
2. **Multi-modal embeddings**: Use CLIP/SigLIP for images alongside text embeddings. Separate indices with fusion at query time.
3. **ColPali / Vision-LLM**: Embed entire pages as images, retrieve visually. Preserves layout and formatting. Best for complex documents but more expensive.

I'd start with approach 1 and add approach 3 for documents where layout matters.

---

### Question 16: What vector database would you choose and why?
**Answer**: It depends on constraints:
- **Already using Postgres?** pgvector -- minimal new infrastructure
- **Need zero-ops SaaS?** Pinecone -- fully managed
- **Need hybrid search?** Weaviate or Qdrant -- built-in BM25 + dense
- **Need massive scale (100M+ vectors)?** Milvus with GPU acceleration
- **Prototyping?** Chroma (embedded, no server)
- **Research?** FAISS (in-memory, battle-tested)

I'd never choose based on benchmark numbers alone -- the operational considerations (managed vs. self-hosted, existing stack, team expertise) usually dominate.

---

### Question 17: How would you reduce hallucination in a RAG system?
**Answer**: Multi-layer approach:
1. **Improve retrieval**: Better chunks + re-ranking = more relevant context = fewer hallucinations
2. **Prompt engineering**: "Only answer based on the provided context. If unsure, say 'I don't know'."
3. **Citation requirement**: Force the model to cite specific passages for every claim
4. **Faithfulness checking**: Use RAGAS faithfulness metric or a separate LLM to verify grounding
5. **Self-RAG / CRAG**: Add self-correction mechanisms that detect unsupported generation
6. **Confidence scoring**: Output a confidence score and set thresholds for human escalation

---

### Question 18: What is the Model Context Protocol (MCP)?
**Answer**: MCP, introduced by Anthropic in 2024, is a universal standard protocol for connecting LLMs to tools -- like USB for AI. Instead of each tool requiring custom integration code, MCP provides a standardized client-server protocol. MCP servers expose tools (file system, database, APIs) and MCP clients (LLM applications) can discover and call them. Key benefits: standardized interface, server-side tool definitions, security, and discoverability.

---

### Question 19: How would you optimize a RAG system for cost?
**Answer**: Ordered by impact:
1. **Model routing** (50% savings): Send simple queries to GPT-4o-mini, complex to GPT-4o
2. **Semantic caching** (20-40%): Cache answers for similar queries
3. **Reduce context** (30-50% on LLM cost): Use aggressive re-ranking to send top-3 instead of top-10 chunks
4. **Embedding caching** (reduces embedding costs): Cache embeddings for repeated texts
5. **Open-source models** (60-90% savings): Self-host for high-volume use cases
I'd measure cost-per-query first, then target the biggest component (usually LLM generation at 70-80%).

---

### Question 20: What's the difference between multi-agent orchestration patterns?
**Answer**:
- **Sequential**: A->B->C pipeline. Simple, predictable. Good for content creation.
- **Parallel**: Fan-out to multiple agents, fan-in to merge. Fast for independent tasks.
- **Hierarchical**: Supervisor delegates to specialized workers. Scalable, clear authority.
- **Debate**: Agents argue positions, judge decides. Highest quality but expensive.

I'd choose sequential for simplicity, parallel for speed, hierarchical for complex enterprise tasks, and debate for safety-critical decisions where quality is paramount.

- Hybrid search (BM25 + dense) with top-50 candidates
- Cohere Rerank v3 to re-rank to top-5

**Generation:**
- Model routing: GPT-4o-mini for simple FAQs (70% of queries), GPT-4o for complex
- System prompt enforces citations, tone, and "I don't know" behavior
- Multi-turn: Sliding window of last 5 messages + RAG context

**Caching:**
- Semantic cache (GPTCache) with 0.95 similarity threshold
- Expected hit rate: 30-40% (many repeated questions)

## 5.3 System Design Interview: "Design a Multi-Agent Research Assistant"

### Requirements
- User provides a research topic
- System produces a comprehensive research report with citations
- Should search multiple sources (web, academic papers, databases)
- Must fact-check claims and identify conflicting information
- Output: Structured report with executive summary, findings, and references

### Architecture: Hierarchical Multi-Agent System

```
MULTI-AGENT RESEARCH ASSISTANT
================================

  User: "Research the impact of AI on healthcare in 2024"
      |
      v
  +================================+
  |     ORCHESTRATOR AGENT         |
  |  (Plans, delegates, reviews)   |
  +================================+
      |
      v
  [PLANNING PHASE]
  Orchestrator creates research plan:
    1. Search for recent AI healthcare applications
    2. Find clinical trial data and outcomes
    3. Identify regulatory developments
    4. Analyze economic impact
    5. Find expert opinions and critiques
      |
      v
  [EXECUTION PHASE - Parallel]
      |
      +---> [Research Agent 1: Web Search]
      |     Tools: Google Search, News API
      |     Focus: Recent developments, news
      |
      +---> [Research Agent 2: Academic]
      |     Tools: Semantic Scholar, PubMed, arXiv
      |     Focus: Papers, clinical trials
      |
      +---> [Research Agent 3: Data Analysis]
      |     Tools: Code execution, statistics
      |     Focus: Quantitative findings
      |
      +---> [Research Agent 4: Regulatory]
            Tools: Legal database, government sites
            Focus: FDA, EU AI Act, compliance
      |
      v
  [SYNTHESIS PHASE]
      |
      +---> [Fact Checker Agent]
      |     Cross-references claims across sources
      |     Flags contradictions
      |
      +---> [Writer Agent]
      |     Synthesizes findings into structured report
      |     Includes citations for every claim
      |
      +---> [Critic Agent]
            Reviews report for accuracy, completeness
            Sends back for revision if needed
      |
      v
  [FINAL REPORT]
  - Executive summary
  - Key findings with citations
  - Conflicting viewpoints noted
  - Confidence levels per section
  - Full reference list
```

### Key Design Decisions

| Decision | Choice | Rationale |
|----------|--------|-----------|
| **Orchestration** | Hierarchical (supervisor) | Need central coordination for complex task |
| **Research execution** | Parallel fan-out | Independent research areas, minimize latency |
| **Quality control** | Separate fact-checker + critic | Adversarial review improves accuracy |
| **Memory** | Shared scratchpad + per-agent working memory | Agents need to see each other's findings |
| **Model selection** | GPT-4o for orchestrator + critic, GPT-4o-mini for researchers | Critical thinking needs best model |
| **Framework** | LangGraph | Need fine-grained state management and loops |

### Handling Edge Cases
- **Conflicting sources**: Fact-checker highlights disagreements; writer presents both sides
- **Insufficient results**: Research agent retries with reformulated queries; orchestrator may add new research agents
- **Runaway loops**: Maximum 3 revision cycles before producing best-effort output
- **Rate limiting**: Stagger API calls, implement backoff

## 5.4 Common Pitfalls and How to Discuss Them

### RAG Pitfalls

| Pitfall | What Goes Wrong | How to Discuss |
|---------|----------------|---------------|
| **"My RAG just hallucinated"** | LLM generates text not in context | "I'd add faithfulness checking and enforce citations" |
| **"Retrieval returns irrelevant docs"** | Poor chunking or embedding model | "I'd evaluate chunking strategy, add re-ranking, and consider hybrid search" |
| **"It's too slow"** | Long context + large model | "I'd profile the pipeline, add streaming, implement caching, and use model routing" |
| **"It's too expensive"** | Over-engineering, no caching | "I'd measure cost/query, implement semantic caching, and route to smaller models" |
| **"It doesn't answer some questions"** | Knowledge gaps in corpus | "I'd implement CRAG-style web search fallback for low-confidence retrievals" |
| **"Users don't trust the answers"** | No citations, no explanations | "I'd add source attribution with links, confidence scores, and 'I don't know' responses" |

### Agent Pitfalls

| Pitfall | What Goes Wrong | How to Discuss |
|---------|----------------|---------------|
| **"Agent gets stuck in a loop"** | No termination condition | "I'd set max iterations, add timeout, and implement loop detection" |
| **"Agent uses the wrong tool"** | Poor tool descriptions | "I'd improve descriptions, add few-shot examples, and consider hierarchical tool selection" |
| **"Multi-agent system is chaotic"** | Unclear responsibilities | "I'd define clear roles, use structured communication, and add a supervisor agent" |
| **"Agent costs are exploding"** | Too many LLM calls | "I'd cache intermediate results, use cheaper models for simple steps, and add early termination" |
| **"Can't debug agent failures"** | No observability | "I'd implement detailed tracing (LangSmith), log every decision point, and add replay capability" |

## 5.5 Explain RAG to Different Audiences

### To a CEO (30 seconds)
"RAG lets our AI assistant give accurate, up-to-date answers by searching our company's documents in real-time. Instead of the AI making things up, it finds the right information first, then crafts the answer based on what it found. Think of it as giving the AI a search engine for our internal knowledge. The result: fewer wrong answers, lower legal risk, and customers get help instantly instead of waiting for a human."

### To a Product Manager (2 minutes)
"RAG is the architecture that connects an LLM to our knowledge base. When a user asks a question, the system first searches our documents to find relevant information, then generates an answer grounded in those documents -- with citations. The key metrics we track are: retrieval quality (are we finding the right docs?), faithfulness (is the answer actually based on those docs?), and answer correctness. We can improve each independently. The trade-offs are cost (each query costs $0.01-0.05), latency (2-5 seconds), and accuracy (currently ~92%, targeting 98%)."

### To an Engineer (5 minutes)
"RAG has three stages: indexing, retrieval, and generation. For indexing, we chunk documents using parent-child splitting, embed with text-embedding-3-small, and store in pgvector with BM25 via pg_search. For retrieval, we run hybrid search (BM25 + dense) with RRF fusion, then re-rank with a cross-encoder. For generation, we route between GPT-4o-mini and GPT-4o based on query complexity. We cache at three levels: embedding cache, retrieval cache, and semantic response cache. The whole pipeline is evaluated using RAGAS (faithfulness, context recall, answer correctness) on a golden test set of 200 Q&A pairs. We use LangSmith for tracing and LangGraph for the agentic flow with self-correction loops."
