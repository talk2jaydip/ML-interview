# Understanding Principal Component Analysis (PCA)



## 1. What is Dimensionality Reduction?

> Dimensionality reduction algorithms project high-dimensional data to a low-dimensional space, retaining as much salient information as possible while removing redundant information.

-   **Dimensionality:** Refers to the number of features or variables associated with each data sample (often the columns in a dataset).
-   **Goal:** Reduce the number of dimensions per sample *without losing significant information*.
-   **Why?**
    -   Removes noise and redundancy.
    *   Improves algorithm performance (efficiency and effectiveness).
    *   Reduces dataset size and hardware requirements.
    *   Helps combat the "Curse of Dimensionality".

---

## 2. The Curse Of Dimensionality

> The Curse Of Dimensionality Is A Phrase Commonly Used To Describe Issues That Arise When Working With Data That Has A High Number Of Dimensions In The Feature Space.

-   **Problem:** As the number of features (dimensions) increases, the data becomes increasingly sparse.
-   **Consequence:** It becomes harder to find statistically valid correlations or patterns. More features might not improve model performance due to sparsity.
-   **Solution:** Dimensionality reduction techniques aim to `keep the useful information provided by the extra features but minimize the negative effect of sparsity`.

---


-  Dimensionality Reduction Is Typically An Important Pre-Processing Step Used Before A Classification Stage
- A Process Of Feature Projection, Which Adjusts The Data From The Higher Dimensional Space Into A Space With Fewer Dimensions To Remove The Sparsity From The Data
- The Secondary Benefit That Can Be Obtained By Pre-Processing The Dataset With A Dimensionality Reduction Stage Is The Improved Computational Performance That Can Be Achieved.
- We are always trying to balance the information loss resulting from the projection into lower dimensional space with reducing the sparsity of the data
-  PCA, are essentially methods of compressing information into a smaller size for transfer, and, in many compression methods, some losses occur as a result of the compression process



## 2. Major Approaches to Dimensionality Reduction

There are two primary categories of dimensionality reduction techniques:

---

### 2.1 Linear Projection

These methods project high-dimensional data onto a lower-dimensional **linear** subspace.

- **Key Idea:** Assumes the structure of the data can be captured using **linear combinations** of the original features.

#### Techniques:

- **Principal Component Analysis (PCA):**
  - Identifies the directions (principal components) of maximum variance in data.
  - Unsupervised and commonly used for visualization and feature reduction.

- **Singular Value Decomposition (SVD):**
  - Matrix factorization technique: decomposes data into `UΣVᵀ`.
  - Related to PCA; reveals underlying structure in data.

- **Linear Discriminant Analysis (LDA):**
  - Can be used for both **classification** and **dimensionality reduction**.
  - Maximizes class separability by projecting onto axes that best separate the classes.
  - **Supervised**: requires class labels.

- **Non-negative Matrix Factorization (NNMF):**
  - Decomposes data into two **non-negative** matrices.
  - Often used in scenarios like image or document analysis (e.g., pixel intensities, word counts).

- **Random Projection:**
  - Projects data into a lower-dimensional space using a **random matrix**.
  - Computationally efficient, surprisingly effective.

---

### 2.2 Manifold Learning (Nonlinear Dimensionality Reduction)

These methods assume high-dimensional data lies on a **nonlinear, curved manifold** within the higher-dimensional space.

- **Key Idea:** Captures complex, nonlinear relationships that linear methods might miss.

#### Techniques:

- **Isomap:**
  - Preserves **geodesic distances** (along the manifold).
  - Unrolls data that lies on curved surfaces (e.g., Swiss Roll).

- **t-Distributed Stochastic Neighbor Embedding (t-SNE):**
  - Preserves **local structure** and similarities between neighboring points.
  - Excellent for **visualization** in 2D or 3D.

- **Independent Component Analysis (ICA):**
  - Decomposes multivariate signals into **statistically independent** components.
  - Unlike PCA, allows **non-Gaussian** distributions.
  - Useful in audio/signal separation (e.g., separating mixed voices).

- **Locally Linear Embedding (LLE):**
  - Preserves relationships within local neighborhoods.
  - Assumes data lies on locally linear patches of the manifold.

- **Multidimensional Scaling (MDS):**
  - Maps data to a lower-dimensional space while preserving **pairwise distances** as well as possible.

- **Dictionary Learning:**
  - Learns a set of **basis elements ("dictionary")**.
  - Represents data as **sparse linear combinations** of these elements.

- **Random Trees Embedding:**
  - Uses random decision trees to encode data into high-dimensional sparse binary vectors.
  - Often combined with other dimensionality reduction methods.

---

## 3. Key Considerations

- **Linear vs. Nonlinear:**
  - Use **linear methods** (e.g., PCA, LDA) if data structure is mostly linear or when speed matters.
  - Use **manifold learning** for complex, nonlinear relationships.

- **Supervised vs. Unsupervised:**
  - **LDA** is **supervised** (requires class labels).
  - Others (e.g., PCA, SVD, t-SNE, Isomap) are **unsupervised**.

- **Goals:**
  - **Visualization?** → Use t-SNE or MDS.
  - **Feature extraction or noise reduction?** → Use PCA, LDA, or SVD.
  - **Data interpretability?** → Consider NNMF or Dictionary Learning.

---

> 🔁 Remember: Dimensionality reduction is often a balance between **information preservation**, **computational efficiency**, and **interpretability**.


## PCA

> If the correlation is `very high` among a subset of the features, PCA will attempt to combine the highly correlated features and represent this data with a smaller number of `linearly uncorrelated features`.

The algorithm keeps performing this `correlation reduction`, finding the directions of maximum variance in the original high-dimensional data and projecting them onto a smaller dimensional space. `These newly derived components are known as principal components`.

 * It seems reasonable to select the axis that preserves the maximum amount of variance, as it will most likely lose less information than the other projections. Another way to justify this choice is that `it is the axis that minimizes the mean squared distance between the original dataset and its projection onto that axis.` This is the rather simple idea behind PCA.4

 * For each principal component, `PCA finds a zero-centered unit vector pointing in the direction of the PC`. Since two opposing unit vectors lie on the same axis, the direction of the unit vectors returned by PCA is not stable: if you perturb the training set slightly and run PCA again, the unit vectors may point in the opposite direction as the original vectors. However, they will generally still lie on the same axes. 

###  `PCA assumes that the dataset is centered around the origin.` 

**- Pca Reduces The Sparsity In The Dataset By Separating The Data Into A Series Of Components Where Each Component Represents A Source Of Information Within The Data.  
-  The First Component Produced In Pca, The Principal Component Comprises The Majority Of Information Or Variance Within The Data. 

> The Principal Component Can Often Be Thought Of As Contributing The Most Amount Of Interesting Information In Addition To The Mean. With Each Subsequent Component, Less Information, But More Subtlety, Is Contributed To The Compressed Data.

1. PCA is an unsupervised method.
2. It searches for the directions that data have the largest variance.
3. Maximum number of principal components <= number of features.
4. All principal components are orthogonal to each other.

> PCA is a deterministic algorithm which doesn’t have parameters to initialize and it doesn’t have local minima problem like most of the machine learning algorithms has.

### t-Distributed Stochastic Neighbor Embedding (t-SNE)

- **Overview:**  
  t-SNE is a nonlinear dimensionality reduction technique used for visualizing high-dimensional data. It maps each high-dimensional point into a two- or three-dimensional space, ensuring that similar points are modeled close to each other and dissimilar points are modeled farther apart.

---

### How t-SNE Works

- **Probability Distributions:**  
  t-SNE constructs **two probability distributions**:  
  1. One over pairs of points in the high-dimensional space.  
  2. One over pairs of points in the low-dimensional space.  
  
  The goal is that similar points have a high probability while dissimilar points have a lower probability in both spaces.

- **Conversion of Distances to Probabilities:**  
  > Convert the distances between datapoints in the high-dimensional space to conditional probabilities.

  - **Using a Gaussian Curve:**  
    > To define this probability, we use a Gaussian curve, and we see that the probability is high for nearby points, while it is very low for distant points.
    
    - A wide curve would suggest that many points are far away.
    - A narrow curve suggests that the points are tightly compacted.

- **Projection and Corresponding Probabilities:**  
  > Once we project the data into the low-dimensional space, we can also determine the corresponding probability between the corresponding low-dimensional data.

---

### Cost Function and Optimization

- **Kullback–Leibler Divergence:**  
  t-SNE minimizes the Kullback–Leibler divergence between the `two probability distributions` (high-dimensional and low-dimensional). This minimization helps to ensure that the `low-dimensional representation reflects the similarity relationships from the high-dimensional space.`

---

### Preprocessing and Computational Considerations

- **Preprocessing with PCA:**  
  > It is best to use another dimensionality reduction technique (such as PCA, as we do here) to reduce the number of dimensions before applying t-SNE.

  - **Benefits:**  
    - Reduces noise in the features fed into t-SNE.
    - Speeds up the computation of the algorithm.

- **Nonconvex Cost Function:**  
  > t-SNE has a nonconvex cost function, which means that different initializations of the algorithm will generate different results. There is no stable solution.






### Understanding the Crowding Problem

In high-dimensional spaces, data points can be equidistant from a reference point and still be well-separated due to the abundance of dimensions. However, when projecting this data into a lower-dimensional space, there's insufficient room to maintain these distances accurately. As a result, `points that were moderately distant in high-dimensional space become artificially crowded together in the low-dimensional representation`, making it difficult to interpret the true relationships between data points.

---

### t-SNE's Approach to Addressing the Crowding Problem

t-SNE mitigates the crowding problem through two key modifications:


1. **Use of Student's t-Distribution in Low-Dimensional Space**

   While the high-dimensional similarities between data points are modeled using a Gaussian distribution, t-SNE employs a Student's t-distribution with one degree of freedom (equivalent to a Cauchy distribution) in the low-dimensional space. This distribution has heavier tails than a Gaussian, allowing for a better representation of moderate distances and reducing the tendency of points to crowd together.

2. **Modified Kullback–Leibler (KL) Divergence Cost Function**

   t-SNE minimizes the KL divergence between the probability distributions of the high-dimensional and low-dimensional spaces. By doing so, it ensures that the local structure (i.e., the relative distances between neighboring points) is preserved as much as possible in the lower-dimensional embedding.
