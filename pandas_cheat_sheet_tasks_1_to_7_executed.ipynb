This notebook is a clean interview-prep cheat sheet using **one dataframe** and solving **Tasks 1 to 7** with:
- explanation
- code
- visible output
- interview meaning

Topics covered:
1. Department-wise salary summary  
2. Department and city wise average salary  
3. Add department average salary to each row  
4. Compare employee salary to department average  
5. Count high performers by department  
6. Percentage of high performers by department  
7. Top paid employee in each department


## 0. Setup

We start by importing pandas and numpy.
- `pandas` is for dataframe operations
- `numpy` helps with conditional columns like `np.where`


````python
import pandas as pd
import numpy as np
````

## 1. Create the dataframe

This is the single dataframe used for all questions.

Important columns:
- `dept`: department
- `city`: location
- `salary`: fixed pay
- `bonus`: extra pay
- `rating`: performance rating
- `skills`: list column
- `projects`: nested list of dictionaries

We also convert `joined` into datetime.


````python
df = pd.DataFrame({
    "emp_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Helen", "Ian", "Jane"],
    "dept": ["Data", "Data", "Web", "Web", "Data", "ML", "ML", "Web", "Data", "ML"],
    "city": ["Bangalore", "Bangalore", "Delhi", "Delhi", "Mumbai", "Bangalore", "Mumbai", "Delhi", "Mumbai", "Bangalore"],
    "salary": [120000, 95000, 110000, 105000, 130000, 150000, 140000, 99000, 125000, 155000],
    "bonus": [10000, 8000, 12000, 7000, 15000, 20000, 18000, 6000, 14000, 22000],
    "rating": [5, 3, 4, 2, 5, 4, 5, 3, 4, 5],
    "experience": [6, 3, 5, 2, 7, 8, 6, 3, 5, 9],
    "joined": ["2020-01-15", "2021-06-10", "2019-09-23", "2022-03-01", "2018-11-12",
               "2017-07-19", "2020-08-25", "2021-01-11", "2019-05-30", "2016-12-05"],
    "skills": [
        ["python", "sql", "pandas"],
        ["python", "excel"],
        ["javascript", "react"],
        ["javascript", "css"],
        ["python", "sql", "spark"],
        ["python", "ml", "deep-learning"],
        ["python", "ml", "sql"],
        ["html", "css", "javascript"],
        ["python", "sql", "tableau"],
        ["python", "ml", "nlp"]
    ],
    "projects": [
        [{"name": "fraud", "hours": 120}, {"name": "churn", "hours": 80}],
        [{"name": "sales_dashboard", "hours": 60}],
        [{"name": "website_revamp", "hours": 100}],
        [{"name": "landing_page", "hours": 40}, {"name": "seo_fix", "hours": 30}],
        [{"name": "recommendation", "hours": 150}],
        [{"name": "forecasting", "hours": 200}, {"name": "vision", "hours": 90}],
        [{"name": "classification", "hours": 170}],
        [{"name": "frontend_migration", "hours": 110}],
        [{"name": "reporting", "hours": 95}, {"name": "automation", "hours": 50}],
        [{"name": "llm_search", "hours": 220}, {"name": "chatbot", "hours": 140}]
    ]
})

df["joined"] = pd.to_datetime(df["joined"])
df
````

## Quick reminder: what `groupby` does

`groupby` means:
1. split rows into groups
2. apply a function to each group
3. combine the results

### Common patterns
- `agg()` -> one row per group
- `transform()` -> same number of rows as original dataframe
- `filter()` -> keep or remove whole groups
- `apply()` -> custom logic on each group


## 2. Task 1 — Department-wise salary summary

### Goal

For each `dept`, calculate:
- total salary
- average salary
- max salary
- min salary
- employee count

### Why this uses `agg`
We want **one summary row per department**, so `groupby(...).agg(...)` is the right tool.


````python
dept_salary_summary = df.groupby("dept").agg(
    total_salary=("salary", "sum"),
    avg_salary=("salary", "mean"),
    max_salary=("salary", "max"),
    min_salary=("salary", "min"),
    employee_count=("emp_id", "count")
).reset_index()

dept_salary_summary
````

### Explanation

For `Data` department, salaries are:
- 120000
- 95000
- 130000
- 125000

So:
- total = 470000
- average = 117500
- max = 130000
- min = 95000
- count = 4

This is a classic interview pattern: **group + multiple metrics**.


## 3. Task 2 — Department and city wise average salary

### Goal

Find average salary for each combination of:
- `dept`
- `city`

### Why this uses multi-column `groupby`
Now the group is not just department.  
The group is the combination of **department + city**.


````python
dept_city_avg_salary = df.groupby(["dept", "city"], as_index=False).agg(
    avg_salary=("salary", "mean")
)

dept_city_avg_salary
````

### Explanation

Example:
- `Data + Bangalore` has Alice and Bob
- average salary = `(120000 + 95000) / 2 = 107500`

This pattern is useful when interviewers ask:
- average by team and region
- sales by category and month
- score by class and section


## 4. Task 3 — Add department average salary to each row

### Goal

Create a new column:
- `dept_avg_salary`

### Why this uses `transform`
We want the **department average repeated back onto every original row**.

That means:
- same number of rows as original dataframe
- so use `transform`, not `agg`


````python
df["dept_avg_salary"] = df.groupby("dept")["salary"].transform("mean")

df[["emp_id", "name", "dept", "salary", "dept_avg_salary"]]
````

### Explanation

For the `Data` department, the average salary is `117500`.  
So every `Data` employee gets `117500` in the new column.

Memory trick:
- `agg()` shrinks
- `transform()` broadcasts back


## 5. Task 4 — Compare employee salary to department average

### Goal

Create a new column:
- `salary_vs_dept_avg`

Logic:
- `"above"` if employee salary > department average
- `"below_or_equal"` otherwise

### Why this uses `np.where`
`np.where(condition, value_if_true, value_if_false)` is a fast way to build conditional columns.


````python
df["salary_vs_dept_avg"] = np.where(
    df["salary"] > df["dept_avg_salary"],
    "above",
    "below_or_equal"
)

df[["emp_id", "name", "dept", "salary", "dept_avg_salary", "salary_vs_dept_avg"]]
````

### Explanation

For Bob in `Data`:
- salary = 95000
- department average = 117500
- result = `below_or_equal`

For Eva in `Data`:
- salary = 130000
- department average = 117500
- result = `above`


## 6. Task 5 — Count high performers by department

### Goal

For each department, count employees where:
- `rating >= 4`

### Why this uses conditional aggregation
We group by department, then inside each group we count how many rows satisfy a condition.


````python
high_performers_by_dept = df.groupby("dept").agg(
    high_performers=("rating", lambda x: (x >= 4).sum())
).reset_index()

high_performers_by_dept
````

### Explanation

Inside each department:
- `(x >= 4)` creates booleans
- `True` counts as `1`
- `False` counts as `0`
- `.sum()` gives total high performers

Example for `ML`, ratings are:
- 4, 5, 5

Condition `>= 4` becomes:
- True, True, True

Sum = `3`


## 7. Task 6 — Percentage of high performers by department

### Goal

For each department, compute:
- total employees
- high performers
- percentage of high performers

### Pattern
This is a two-step pattern:
1. aggregate counts
2. create a derived percentage column


````python
high_perf_pct = df.groupby("dept").agg(
    total_employees=("emp_id", "count"),
    high_performers=("rating", lambda x: (x >= 4).sum())
).reset_index()

high_perf_pct["pct_high_performers"] = (
    high_perf_pct["high_performers"] / high_perf_pct["total_employees"] * 100
)

high_perf_pct
````

### Explanation

For `Web`:
- total employees = 3
- high performers = 1
- percentage = `1 / 3 * 100 = 33.33`

This kind of question is common in interviews:
- % premium users by country
- % passed students by class
- % fraud cases by region


## 8. Task 7 — Top paid employee in each department

### Goal

Return the full row of the highest salary employee per department.

There are two good interview-friendly ways to solve this:
1. `groupby + nlargest`
2. `sort_values + drop_duplicates`


### Method 1 — `groupby + nlargest`

````python
top_paid_by_dept = df.groupby("dept", group_keys=False).apply(
    lambda g: g.nlargest(1, "salary")
)

top_paid_by_dept[["emp_id", "name", "dept", "salary"]]
````

### Explanation

Within each department group:
- sort by salary descending
- keep the largest 1 row

This returns the full employee row, not just the salary.


### Method 2 — `sort_values + drop_duplicates`

````python
top_paid_by_dept_2 = (
    df.sort_values(["dept", "salary"], ascending=[True, False])
      .drop_duplicates("dept")
)

top_paid_by_dept_2[["emp_id", "name", "dept", "salary"]]
````

### Explanation

This works because:
1. sort each department so the highest salary comes first
2. drop duplicates by department
3. first row kept per department is the top paid one

This method is also very common in interviews.


## Final cheat sheet summary

### Use these patterns clearly

#### 1. Summary per group
```python
df.groupby("dept").agg(total_salary=("salary", "sum"))
```

#### 2. Summary by multiple keys
```python
df.groupby(["dept", "city"], as_index=False).agg(avg_salary=("salary", "mean"))
```

#### 3. Group metric back to each row
```python
df["dept_avg_salary"] = df.groupby("dept")["salary"].transform("mean")
```

#### 4. Condition using a group metric
```python
df["flag"] = np.where(df["salary"] > df["dept_avg_salary"], "above", "below")
```

#### 5. Conditional count inside groupby
```python
df.groupby("dept").agg(high_performers=("rating", lambda x: (x >= 4).sum()))
```

#### 6. Top row per group
```python
df.groupby("dept", group_keys=False).apply(lambda g: g.nlargest(1, "salary"))
```

### Interview memory line
- `agg` = one row per group  
- `transform` = same row count  
- `apply` = custom group logic  
- `np.where` = if/else column creation
