# Missing Value Imputation in Time Series

## **Interpolation**

- Interpolation is a commonly used technique for time series missing value imputation. 
- It helps in estimating the missing data-point using the two surrounding known data points.

-   Time-Based Interpolation
-   Spline Interpolation
-   Linear Interpolation
>Imputations from these methods make more sense when t**he missing value window ( width of missing data) is small**. For instance, if several consecutive values are missing, it becomes harder for these methods to estimate them.

## **Denoising a Time Series**

-   **Rolling means**
	- The Rolling mean is simply the mean for a window of previous observations,
	- This can greatly help in minimizing the noise in time series data.

-   **Fourier Transform**
	- Fourier Transform can help remove the noise by converting the time series data into the frequency domain, and from there, we can filter out the noisy frequencies. 
	- Then, we can apply the inverse Fourier transform to obtain the filtered time series. 

#### **_Non-Stationarity_**

- Non-stationarity is when the statistical properties of a series, e.g the mean, variance, and covariance (or the process generating the series) changes over time.
- A proven method of stationarizing a non-stationary series is through the use of differencing.![enter image description here](https://miro.medium.com/max/750/0*JRvNGPeRl4RF9lRS.webp)


# **_ARIMA_**

ARIMA is one example of a traditional method of forecasting time series. ARIMA stands for **Auto-Regressive Integrated Moving Average** and is divided into 3 parts —

-   AR "p" 
	-  The **auto-regressive** part represents the number of time periods to apply lag our data for. 
	- A p term of 2 means we consider two time-steps before each observation as the explanatory variables for that observation for the autoregressive portion of the calculation. `T**he observation itself becomes the target variable.*`

-   I(d) — 
	- For the **Integrated** portion, d represents the number of differencing transformations applied to the series to transform a non-stationary time series to a stationary one.

-   MA(q)— A time series is thought to be a combination of 3 components: 
	- **Trend** refers to the gradual increase or decrease of the time series over time
	- **Seasonality** is a repeating short-term cycle in the series, and 
	- **Error** refers to noise not explained by seasonality or trend. 

	**Moving average** therefore refers to the number of lags of the error component to consider. .

AR & MA are quite similar, the difference is 
		- AR considers lagged time series values from previous time periods, while 
		- MA considers errors from previous periods.

**Lagged Correlation**

- With time series, we don’t only consider error metrics like MSE, MAE, etc. This is because such metrics don’t capture lag.
-  To get a better picture of how well our model is performing, we also look at **Lag Correlation (Cross-Correlation) between the true and predicted values**.

<ol style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Autocorrelation and Partial Autocorrelation:</strong> These functions help identify the order of autoregressive (AR) and moving average (MA) terms in an ARIMA model.</p>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Spectral Analysis:</strong> Tools like the periodogram and spectral density estimation help identify frequency components in stationary time series.</p>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>ARIMA Modeling:</strong> Autoregressive Integrated Moving Average (ARIMA) models are suita<strong>ble for modeling and forecasting stationary time series data.</strong></p>
</li>
</ol>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>For Nonstationary Time Series:</strong></p>
<ol style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Differencing:</strong> Transforming a nonstationary series into a stationary one by taking <em>differences between consecutive observations</em>. This is often done before applying <em>ARIMA modeling</em>.</p>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Decomposition:</strong> Decompose the time series into <strong>trend, seasonality, and residual components</strong> using methods like seasonal decomposition of time series (STL).</p>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Cointegration Analysis:</strong> In cases of multiple nonstationary time series, cointegration analysis helps identify long-term relationships among them.</p>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>State-Space Models:</strong> Techniques like the Kalman filter and structural time series modeling can capture complex dynamics in nonstationary data.</p>
</li>
</ol>
<h3 style="display:inline-block;color:#d8e4ea;background:#22303a;border-left:5px solid #7fa89e;padding:8px 14px;border-radius:12px;border:1px solid #3a4a52;margin:18px 0 10px 0;"><strong>ARIMA (AutoRegressive Integrated Moving Average):</strong></h3>
<ol style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>AutoRegressive (AR):</strong> The &quot;AR&quot; component of ARIMA refers to the autoregressive part of the model. <em>It captures the relationship between the current value of a time series and its past values</em>. Specifically, it models how the current value depends on previous values with a linear combination. <strong>The order of the autoregressive component, denoted as &quot;p,&quot; represents the number of past observations used for prediction.</strong></p>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Integrated (I):</strong> The &quot;I&quot; component signifies differencing, which is employed to make a nonstationary time series stationary. <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">Differencing involves subtracting the current value from a lagged value</code>, typically denoted as &quot;d&quot; times. The order of differencing, &quot;d,&quot; is determined by <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">the number of differences required to stabilize the mean and make the series stationary</code>.</p>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Moving Average (MA):</strong> The &quot;MA&quot; component deals with the moving average part of the model. <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">It models how the current value depends on past white noise or error terms</code>. The order of the moving average component, denoted as &quot;q,&quot; indicates <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">the number of past white noise terms considered in the model.</code></p>
</li>
</ol>
<h3 style="display:inline-block;color:#d8e4ea;background:#22303a;border-left:5px solid #7fa89e;padding:8px 14px;border-radius:12px;border:1px solid #3a4a52;margin:18px 0 10px 0;"><strong>ARIMA Plus (ARIMA+):</strong></h3>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">&quot;ARIMA Plus&quot; is not a standard term but can be used to describe an extended version of the ARIMA model that incorporates additional components or features to enhance its forecasting capabilities. Here are some possible extensions:</p>
<ol style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;"><strong>Seasonal Component (S):</strong> In many time series, there are recurring patterns or seasonal effects. ARIMA can be extended to SARIMA (Seasonal ARIMA) by including seasonal autoregressive and moving average terms. The seasonal component accounts for periodic patterns within the data.</li>
</ol>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;"><strong>Multivariate forecasting</strong></li>
</ul>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">Multivariate time series  consist of <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">more than one time series variable that is not only dependent on its past values but also has some dependency on the other variables</code>.</p>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">For example, a set of  macroeconomic indicators such as  <strong>gross domestic product</strong>  (<strong>GDP</strong>), inflation, and so on of a particular country can be considered as a multivariate time series. The aim of multivariate forecasting is to come up with a model that captures the interrelationship between the different variables along with its relationship with its past and forecast all the time series together in  the future.</p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;"><strong>Explanatory forecasting</strong></li>
</ul>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">In addition  to the past values of a time series, we might use some other information to predict the future values of a time series. For example, for predicting retail store sales, information regarding promotional offers (both historical and future ones) is usually helpful.</p>
<h2 style="background:linear-gradient(90deg,#22313d 0%,#273743 100%);border-left:8px solid #72aa9b;color:#e7eff3;padding:11px 16px;border-radius:14px;border:1px solid #3a4b58;box-shadow:0 6px 18px rgba(0,0,0,0.18);margin:22px 0 14px 0;">Time series preprocessing and feature engineering are two important steps in building predictive models for time series data.</h2>
<pre style="background:#1b2330;color:#e6edf3;padding:14px 16px;border-radius:14px;border:1px solid #334155;overflow-x:auto;line-height:1.7;margin:14px 0;"><code style="background:transparent;color:#e6edf3;border:0;padding:0;">Here is a breakdown and explanation of these steps as if I were giving an interview answer.
</code></pre>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">Time series preprocessing is the process of cleaning and transforming the raw time series data into a format that is suitable for modeling. Some common preprocessing steps are:</p>
<ol style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Handling missing values</strong>: Missing values can occur in time series data due to various reasons, such as sensor failures, data entry errors, or irregular sampling intervals. Missing values can affect the quality and accuracy of the models, so they need to be handled appropriately. Some common methods for handling missing values are:</p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;"><strong>Imputation</strong>: Replacing the missing values with some estimated values, such as the mean, median, or mode of the available data, or using more sophisticated methods, such as interpolation, regression, or machine learning.</li>
<li style="margin:6px 0;"><strong>Deletion</strong>: Removing the observations with missing values from the data, either by dropping the entire rows or columns, or by using a sliding window approach that only considers a fixed number of observations at a time.</li>
<li style="margin:6px 0;"><strong>Flagging</strong>: Marking the missing values with a special indicator, such as zero, negative one, or NaN, and treating them as a separate category in the models.</li>
</ul>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Resampling</strong>: <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">Resampling is the process of changing the frequency or granularity of the time series data</code>.
- For example, we can resample daily data into weekly, monthly, or yearly data, or vice versa.
&gt; Resampling can help reduce noise, smooth out fluctuations, and capture long-term trends or seasonal patterns in the data. Some common methods for resampling are:</p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;"><strong>Aggregation</strong>: Aggregating the data over a larger time interval by applying a summary statistic, such as <strong>sum, average, minimum, maximum, or count</strong>. For example, we can aggregate daily sales data into monthly sales data by taking the sum of sales for each month.</li>
<li style="margin:6px 0;"><strong>Disaggregation</strong>: Disaggregating the data over a smaller time interval by applying a splitting rule, such as equally dividing, proportionally allocating, or randomly assigning. For example, we can disaggregate monthly sales data into daily sales data by dividing the monthly sales by the number of days in each month.</li>
</ul>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Interpolation</strong>: Interpolating the data over a different time interval by using a mathematical function, such as <strong>linear, polynomial, spline, or exponential.</strong></p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">For example, we can interpolate hourly temperature data into minute temperature data by using a linear function that connects the adjacent hourly observatons.</li>
<li style="margin:6px 0;"></li>
</ul>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Normalization</strong>: Normalization is the process of scaling or transforming the time series data to have a common range or distribution. Normalization can help reduce the effect of outliers, improve the comparability of different time series, and enhance the performance of some models that assume certain properties of the data. Some common methods for normalization are:</p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;"><strong>Min-max scalin</strong>g: Scaling the data to have a fixed minimum and maximum value, usually between zero and one. This method preserves the shape and proportion of the data but is sensitive to outliers.</li>
<li style="margin:6px 0;">S<strong>tandardization</strong>: Scaling the data to have a zero mean and unit variance. This method makes the data follow a standard normal distribution but changes the shape and proportion of the data.</li>
<li style="margin:6px 0;"><strong>Log transformation</strong>: Transforming the data by taking the natural logarithm of each value. This method reduces the skewness and variance of the data but cannot handle zero or negative values.</li>
</ul>
</li>
</ol>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">Time series feature engineering is the process of creating new variables or features from the existing time series data that can capture useful information or patterns for modeling. Some common feature engineering techniques are:</p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Fourier Transforms:</strong></p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">Apply Fourier transforms to capture frequency domain information in the time series, which is especially useful for detecting periodic patterns.<code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">can filter out the noisy frequencies</code></li>
</ul>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Date-related features</strong>: <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">Extracting features from the date-time column that can reflect temporal aspects of the data, such as year, month, day, hour, minute, second, weekday, weekend, holiday, season, quarter, etc</code>. These features can help capture cyclical or seasonal patterns in the data that may affect the target variable.</p>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Time-based features</strong>: <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">Creating features based on the elapsed time or duration between different events or observations in the data</code>.</p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">For example, we can create features such as time since last purchase, time until next purchase, average time between purchases, etc. These features can help capture trends or changes in customer behavior over time.</li>
</ul>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Lag features:</strong> <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">Creating features based on previous values or observations of the target variable or other variables in the data</code>.</p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">For example, we can create features such as lagged sales, lagged temperature, lagged demand, etc. These features can help capture autocorrelation or dependency in the data that may affect future outcomes.</li>
</ul>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Rolling window features:</strong> Creating features based on statistics calculated over a fixed window of previous observations in the data. For example, we can create features such as rolling mean, rolling standard deviation, rolling minimum, rolling maximum, etc. These features can help capture <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">local patterns or variations in the data that may affect short-term outcomes</code>.</p>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Expanding window feature</strong>s: Creating features based on statistics calculated over an expanding window of all previous observations in the data. For example, we can create features such as <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">cumulative sum, cumulative count, cumulative average, cumulative maximum,</code> etc. These features can help capture <strong>global patterns or trends in the data that may affect long-term outcomes.</strong></p>
</li>
<li style="margin:6px 0;"></li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Domain-specific features</strong>: Creating features based on domain knowledge or external information that can provide additional context or insight for the data. For example, we can create features such as weather conditions, economic indicators, social media sentiment, etc. These features can help capture exogenous factors or influences that may affect the target variable.</p>
</li>
</ul>
<h3 style="display:inline-block;color:#d8e4ea;background:#22303a;border-left:5px solid #7fa89e;padding:8px 14px;border-radius:12px;border:1px solid #3a4a52;margin:18px 0 10px 0;"><strong>Fourier Transform:</strong></h3>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">The Fourier transform is a mathematical transformation <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">used to analyze the frequency components of a time series or signal</code>. It decomposes a signal into its constituent sinusoidal components, <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">revealing the frequencies and amplitudes present in the data.</code></p>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">Why we need it:</p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Frequency Analysis:</strong> Fourier transform helps in understanding the periodic patterns, cycles, and oscillations present in time series data. For example, it's used in signal processing to analyze sound waves and in analyzing the seasonality of economic data.</p>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Filtering and Denoising:</strong> By identifying the dominant frequencies, you can filter out noise or unwanted components from the signal, leaving behind the essential information.</p>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Feature Extraction:</strong> In some cases, Fourier transform can be used as a feature engineering technique to extract relevant frequency-domain features for machine learning models.</p>
</li>
</ul>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">Any time series can contain some or all of the  following components:</p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;"><strong>Trend</strong></li>
<li style="margin:6px 0;"><strong>Seasonal</strong></li>
<li style="margin:6px 0;"><strong>Cyclical</strong></li>
<li style="margin:6px 0;"><strong>Irregular</strong></li>
</ul>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">additive_  (<em>Y = Trend + Seasonal + Cyclical + Irregular</em>)</p>
<blockquote style="margin:14px 0;padding:12px 16px;background:#24313d;border-left:5px solid #90b7a2;border-radius:14px;border:1px solid #3c4d5a;color:#d9e4ec;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">Additive decomposition is appropriate when the magnitude of the <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">seasonal fluctuations is relatively constant across different levels of the time series</code>. In other words, if the amplitude of seasonality doesn't depend on the overall level of the data, an additive decomposition is suitable</p>
</blockquote>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">multiplicative_  (<em>Y = Trend * Seasonal * Cyclical *</em> <em>Irregular</em>).</p>
<blockquote style="margin:14px 0;padding:12px 16px;background:#24313d;border-left:5px solid #90b7a2;border-radius:14px;border:1px solid #3c4d5a;color:#d9e4ec;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">Multiplicative decomposition is suitable <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">when the magnitude of seasonal fluctuations varies with the overall level of the time series</code>. If the amplitude of seasonality increases or decreases as the level of the data changes, a multiplicative decomposition is often more appropriate.</p>
</blockquote>
<blockquote style="margin:14px 0;padding:12px 16px;background:#24313d;border-left:5px solid #90b7a2;border-radius:14px;border:1px solid #3c4d5a;color:#d9e4ec;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">The <strong>trend</strong> is a long-term change in the mean of a time series. It is the smooth and steady movement of a time series in a particular direction</p>
</blockquote>
<blockquote style="margin:14px 0;padding:12px 16px;background:#24313d;border-left:5px solid #90b7a2;border-radius:14px;border:1px solid #3c4d5a;color:#d9e4ec;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">When a time <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">series exhibits regular, repetitive, up-and-down fluctuations</code>, we call that <strong>seasonality</strong>. For instance, retail sales typically shoot up during the holidays,</p>
</blockquote>
<blockquote style="margin:14px 0;padding:12px 16px;background:#24313d;border-left:5px solid #90b7a2;border-radius:14px;border:1px solid #3c4d5a;color:#d9e4ec;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">the cyclical component also <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">exhibits a similar up-and-down pattern around the trend line</code>, but instead of repeating the pattern every period, the cyclical component is irregular.</p>
</blockquote>
<pre style="background:#1b2330;color:#e6edf3;padding:14px 16px;border-radius:14px;border:1px solid #334155;overflow-x:auto;line-height:1.7;margin:14px 0;"><code style="background:transparent;color:#e6edf3;border:0;padding:0;">	- A good example of this is economic recession, which happens over a 10-year cycle.
</code></pre>
<blockquote style="margin:14px 0;padding:12px 16px;background:#24313d;border-left:5px solid #90b7a2;border-radius:14px;border:1px solid #3c4d5a;color:#d9e4ec;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">The irregular component</p>
</blockquote>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">This component  is left after removing the <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">trends, seasonality, and cyclicity </code> from a time series.
Traditionally, this component is considered  <em>unpredictable</em>  and is also called the  <em>residual</em>  or  <em>error term</em>.</p>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">autocorrelation is the correlation between the values of a time series in successive periods</code></p>
<blockquote style="margin:14px 0;padding:12px 16px;background:#24313d;border-left:5px solid #90b7a2;border-radius:14px;border:1px solid #3c4d5a;color:#d9e4ec;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">partial autocorrelation removes any <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">indirect correlation that may be present before presenting the correlations</code></p>
</blockquote>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">If <em>t</em> is the current time step,
- let’s assume <em>t-1</em> is highly correlated to <em>t</em>.
- So, by extending this logic, <em>t-2</em> will be highly correlated with <em>t-1</em> and
<code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">because of this correlation, the autocorrelation between _t_ and _t-2_ would be high.</code></p>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">However, <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">partial autocorrelation corrects this and extracts the correlation, which can be purely attributed to _t-2_ and _t_</code>.</p>
<h3 style="display:inline-block;color:#d8e4ea;background:#22303a;border-left:5px solid #7fa89e;padding:8px 14px;border-radius:12px;border:1px solid #3a4a52;margin:18px 0 10px 0;"># Decomposing a time series</h3>
<ol style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Detrending</strong>: Here, we estimate the  <strong>trend component</strong>  (which is the smooth change in the time series) and  remove  it from the time  series, giving  us a  <strong>detrended</strong> <strong>time series</strong>.</p>
</li>
<li style="margin:6px 0;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Deseasonalizing</strong>: Here, we estimate the seasonality  component from  the <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">detrended time series.</code> After removing the seasonal component, what is left is the residual.</p>
</li>
</ol>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Log transforms</strong> are typically known to reduce the variance of the data and thereby remove heteroscedasticity in the data. Intuitively, we can think of a log transform as something that <em>pulls in</em> the extreme values on the right of the histogram, at the same time stretching back the very low values on the left of the histogram.</p>
<h3 style="display:inline-block;color:#d8e4ea;background:#22303a;border-left:5px solid #7fa89e;padding:8px 14px;border-radius:12px;border:1px solid #3a4a52;margin:18px 0 10px 0;"><strong>Deterministic Trend:</strong></h3>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">A deterministic trend in a time series refers to <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">a predictable and systematic long-term pattern or movement in the data that can be modeled and forecasted with a high degree of certainty</code>.</p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">This trend is driven by known, fixed factors or variables, and it follows a clear, consistent path over time. In other words, a deterministic trend has a well-defined and deterministic relationship with time.</li>
</ul>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Example of Deterministic Trend:</strong></p>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">Consider the monthly sales data of a popular smartphone brand over several years. If the company has been consistently launching a new model with improved features every year in September, the sales data is likely to exhibit a deterministic upward trend during the month of September each year. This predictable increase in sales can be attributed to the annual product release and can be modeled with a high degree of certainty.</p>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">In this case, the deterministic trend is driven by a known event (the product release), and it follows a clear pattern, making it possible to forecast the sales for the next September with a high level of confidence.</p>
<h3 style="display:inline-block;color:#d8e4ea;background:#22303a;border-left:5px solid #7fa89e;padding:8px 14px;border-radius:12px;border:1px solid #3a4a52;margin:18px 0 10px 0;"><strong>Stochastic Trend:</strong></h3>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">refers to a random and unpredictable long-term pattern or movement that cannot be precisely modeled or forecasted</code>. '</p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">Unlike a deterministic trend, a stochastic trend does not follow a fixed and known path over time. Instead, it exhibits fluctuations and variations that are not driven by identifiable, fixed factors.</li>
</ul>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Example of Stochastic Trend:</strong></p>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">Consider a daily <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">stock price of a company known for its volatility</code>. While there may be periods where the stock price generally increases due to positive news or strong financial performance, the long-term movement of the stock price is often characterized by random fluctuations, influenced by a multitude of unpredictable factors such as market sentiment, economic events, and investor behavior.</p>

<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>The sine and cosine transformation is a way of encoding the timestamp of a time series data into two new features that can capture the cyclical or seasonal patterns in the data.</strong></p>
<pre style="background:#1b2330;color:#e6edf3;padding:14px 16px;border-radius:14px;border:1px solid #334155;overflow-x:auto;line-height:1.7;margin:14px 0;"><code style="background:transparent;color:#e6edf3;border:0;padding:0;">- The idea is to map the timestamp to a point on a unit circle, where the x-coordinate is the cosine value and the y-coordinate is the sine value. This way, the timestamp can be represented by an angle that varies from 0 to 2π over a fixed period of time, such as a day, a week, or a year.
</code></pre>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">The effect of this transformation is that it can preserve the temporal information of the data while reducing the dimensionality and complexity of the data.</code></p>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;"> It can also help improve the performance of some models that assume linear relationships between the features and the target variable, such as linear regression or neural networks</code>.</p>
