````python

import matplotlib.pyplot as plt
import numpy as np

# original data set
X = [1, 2, 3]
y = [1, 2.5, 3.5]

# slope of best_fit_1 is 0.5
# slope of best_fit_2 is 1.0
# slope of best_fit_3 is 1.5

hyps = [0.5, 1.0, 1.5] 

# mutiply the original X values by the theta 
# to produce hypothesis values for each X
def multiply_matrix(mat, theta):
    mutated = []
    for i in range(len(mat)):
        mutated.append(mat[i] * theta)

    return mutated

# calculate cost by looping each sample
# subtract hyp(x) from y
# square the result
# sum them all together
def calc_cost(m, X, y):
    total = 0
    for i in range(m):
        squared_error = (y[i] - X[i]) ** 2
        total += squared_error
    
    return total * (1 / (2*m))

# calculate cost for each hypothesis
for i in range(len(hyps)):
    hyp_values = multiply_matrix(X, hyps[i])

    print("Cost for ", hyps[i], " is ", calc_cost(len(X), y, hyp_values))

````

- Notice that the `weight** value is too small**` then the **derivative is negative**, and the `weight value is **too large` **then the derivative is **positive**.

- Gradient Descent is a generic optimization algorithm capable of finding optimal solutions to a wide range of problems. The general idea of Gradient Descent is to tweak parameters iteratively in order to minimize a cost function.

- An important parameter in Gradient Descent is `the size of the steps`, determined by the `learning rate hyperparameter`. 
- If the learning rate is `too small`, then the algorithm will have to go through many iterations to converge, `which will take a long tim`

- On the other hand, if the learning rate i`s too high`, you might jump across the valley and end up on the other side,

-  If the random initialization starts the algorithm on the left, then it will converge to a local minimum, which is not as good as the global minimum. If it starts on the right, then it will take a very `long time to cross the plateau`. And if you stop too early, `you will never reach the global minimum.`


### 🔻 Local Minimum

A **point** where the loss is lower than nearby points, but **not necessarily the lowest overall**. Gradient descent might stop here if the gradient becomes (almost) zero.

----------

### 🌍 Global Minimum

The **absolute lowest** point of the loss function. This is the ideal point gradient descent is trying to reach.

----------

### ➖ Plateau

A **flat region** where gradients are **very small or zero**, causing gradient descent to **slow down** or get stuck temporarily.

##### Batch Gradient Descent
> This is why the algorithm is called Batch Gradient Descent: it uses the whole batch of training data at every step (actually, Full Gradient Descent would probably be a better name). `As a result it is terribly slow on very large training sets`. However, Gradient Descent scales well with the number of features; 

- To find a good learning rate, you can use grid search . However, you may want to limit the number of iterations so that grid search can eliminate models that take too long to converge.

> #### The main problem with Batch Gradient Descent is the fact that it uses the whole training set to compute the gradients at every step, which makes it very slow when the training set is large. 

#### Stochastic Gradient Descent
- At the opposite extreme, Stochastic Gradient Descent picks a` random instance in the training set at every step and computes the gradients based only on that single instance`. 

- On the other hand, due to its stochastic (i.e., random) nature, this algorithm is much `less regular than Batch Gradient Descent`: instead of gently decreasing until it reaches the minimum, `the cost function will bounce up and down`, decreasing only on average. Over time it will end up `very close to the minimum, but once it gets there it will continue to bounce around`, never settling down (see Figure 4-9). So once the algorithm stops, 
##### `the final parameter values are good, but not optimal.`




When the cost function is very irregular (as in Figure 4-6), this can actually help the algorithm jump out of local minima, `so Stochastic Gradient Descent has a better chance of finding the global minimum than Batch Gradient Descent does`.

Therefore,` randomness is good to escape from local optima`, but bad because it means that the algorithm can never settle at the minimum. 

**When using Stochastic Gradient Descent, the training instances must be independent and identically distributed (IID) to ensure that the parameters get pulled toward the global optimum, on average. A simple way to ensure this is to shuffle the instances during training (e.g., pick each instance randomly, or shuffle the training set at the beginning of each epoch). `If you do not shuffle the instances—for example, if the instances are sorted by label—then SGD will start by optimizing for one label, then the next, and so on, and it will not settle close to the global minimum.`**

##### Mini-batch Gradient Descent
//
- The last Gradient Descent algorithm we will look at is called Mini-batch Gradient Descent. It is simple to understand once you know Batch and Stochastic Gradient Descent: at each step, instead of computing the gradients based on the full training set (as in Batch GD) or based on just one instance (as in Stochastic GD), 
- `Mini-batch GD computes the gradients on small random sets of instances called mini-batches`. The main advantage of Mini-batch GD over Stochastic GD is that you can get a `performance boost` 



### 🔁 **Gradient Descent Types – Quick Comparison**

| **Type**               | **How It Works**                                | **Speed**    | **Memory** | **Noise**  | **Best For**                       |
|------------------------|--------------------------------------------------|--------------|------------|------------|------------------------------------|
| **SGD**                | 1 sample → 1 update                             | Fastest      | Low        | High       | Huge datasets, online learning     |
| **Batch GD**           | All samples → 1 update                          | Slow         | High       | Low        | Small datasets                     |
| **Mini-Batch GD**      | Batch of samples (e.g., 32, 64) → 1 update      | Balanced     | Medium     | Medium     | Most deep learning tasks           |

#### Assumptions and Their Importance Linear regression relies on several assumptions to ensure valid results:


#### LINE-HN


- Linearity: The relationship between yyy and each xix_ix_i is linear.
    - Multiple linear regression (or multivariable regression): One dependent variable predicted by multiple independent variables. [`L` – Linearity]
    - Multivariate linear regression: Multiple dependent variables are predicted simultaneously by the same set of independent variables.

- Independence: Errors are independent, meaning one observation’s error doesn’t affect another. [`I` – Independence]
- Normality: Errors follow a normal distribution, crucial for hypothesis testing. [`N` – Normal errors]

- Homoscedasticity: The variance of errors is constant across all levels of xix_ix_i [`E` – Equal variance (Homoscedasticity)]
.


- No Multicollinearity: Independent variables are not highly correlated, as this can destabilize coefficient estimates. [`N` – No multicollinearity]

> > **Heteroscedasticity** is exactly the opposite of homoscedasticity, which means that the error terms are not equally distributed. To correct this phenomenon, usually, a log function is used.

### Model Evaluation Metrics

*   **Mean Absolute Error (MAE):**
    *   Average of the absolute differences between actual and predicted values. `MAE = Σ|Yᵢ - Ŷᵢ| / n`.
    *   Units are the same as the target variable. Less sensitive to outliers than MSE/RMSE.

>** A regressor can be evaluated using many different metrics, such as the following:**
- Mean absolute error: This is the average of absolute errors of all the data points in the given dataset.

> - Median absolute error: `This is the median of all the errors in the given dataset.` The main advantage of this metric is that it's `robust to outliers`. A single bad point in the test dataset wouldn't skew the entire error metric, as opposed to a mean error metric.




*   **Mean Squared Error (MSE):**
    *   Average of the squared differences between actual and predicted values. `MSE = RSS / n`.
    *   Units are the square of the target variable's units. Harder to interpret directly. Sensitive to outliers.

*   **Root Mean Squared Error (RMSE):**
    *   Square root of MSE. `RMSE = sqrt(MSE)`.
    *   Units are the same as the target variable, making it more interpretable than MSE. Represents the typical prediction error. Sensitive to outliers.

> #### Root mean squared error (RMSE): It measures the standard deviation of the residuals. It should be as low as possible.




### 📊 **Explained Variance Score (a.k.a. Variance Score)**


#### How much of the variation in the actual data is **captured** by the model.

---

### ✅ **Score Range**  
- **1.0** → Perfect prediction  
- **0.0** → Model predicts only the **mean**  
- **< 0.0** → Worse than just predicting the mean

---

### 🔍 **Formula (Conceptual)**  
$$\text{Explained Variance} = 1 - \frac{\text{Var}(y - \hat{y})}{\text{Var}(y)}$$ 

- y : true values  
- $$ \hat{y}  : \text {predicted values} $$


---

### 💡 **Quick Intuition**  
If predictions vary **just like** actual values → score close to **1**.  
If predictions are flat or totally off → score drops.

---

### 📌 **Example**  
- Dataset: House prices  
- If your model predicts prices that rise and fall **similar to actual** prices → High explained variance.  
- If it predicts the **same price every time** → Low score.



- R2 score: This is pronounced as R-squared, and this score refers to the coefficient of determination. This tells us how well the unknown samples will be predicted by our model. The best possible score is 1.0, but the score can be negative as well.





- **R² Score** (Coefficient of Determination)  
  > Measures how well predictions match actual values.

  $$
  R^2 = 1 - \frac{\text{MSE}}{\text{Var}(y)}
  $$

  - Near **1** = Excellent predictions  
  - Near **0** = Predicts about as good as the **mean**  
  - **< 0** = Worse than predicting the mean

---

- **Explained Variance Score**  
  > Measures how much of the **variation** in the target is captured by the model.


Then:
$$
\text{Var}(e) = \frac{1}{n} \sum_{i=1}^{n}(e_i - \bar{e})^2
$$

- \( e_i \) {is each individual error  }
- \( \bar{e} \) \text{is the **mean of the errors** } 
  

  $$
  \text{Explained Variance} = 1 - \frac{\text{Var}(y - \hat{y})}{\text{Var}(y)}
  $$

  - Near **1** = Model captures most of the variance  
  - Does **not** penalize for bias (unlike R²)

---

### 📌 Good Practice
- ✅ **Low Mean Squared Error (MSE)**  
- ✅ **High R² Score**  
- ✅ **High Explained Variance Score**



🧠  Insight:
> If the residuals have low variance, your model predictions are close to actual values → high explained variance!

E.g
> If R2 is 80% means: 80% of variance Explained by Indeendent Variable

> When we compare the R2 Score with the Explained Variance Score, we are basically checking the Mean Error;
#### so if R2 = Explained Variance Score, that means: The Mean Error = Zero!

**Caution:** R² *always* increases or stays the same when you add more predictors, even if they are useless.

*   **Adjusted R-squared:**
    *   Modification of R² that adjusts for the number of predictors in the model.
    *   Penalizes adding predictors that don't improve the model significantly.
    *   More suitable for comparing models with different numbers of predictors.


---

**one main difference between R2 and the adjusted R2: **
> R2 assumes that every single variable explains the variation in the **dependent variable**. 

> The adjusted R2 tells you the `percentage of variation explained by` **only the independent** variables that actually affect the dependent variable.

1.** R2 increases with every predictor added to a model**. *AS R2 ALWAYS INCREASES AND NEVER DECREASES*, it can appear to be a better fit with the more terms you add to the model. This can be completely misleading.

2. Similarly, if your model has too many terms and too many high-order polynomials you can run into the problem of over-fitting the data. When you over-fit data, a misleadingly high R2 value can lead to misleading projections.


**Adjusted R2**
> Compared to R Squared which can only increase, Adjusted R Squared has the capability to decrease with the addition of less significant variables, thus resulting in a more reliable and accurate evaluation.- 

- Adjusted R Squared, however, makes use of the degree of freedom to compensate and penalize for the inclusion of a bad variable. Adjusted R Squared can be expressed as :

> The value of Adjusted R Squared decreases as k increases also while considering R Squared acting a penalization factor for a bad variable and rewarding factor for a good or significant variable. 

#### Adjusted R Squared is thus a better model evaluator and can correlate the variables more efficiently than R Squared.



3, The adjusted R2 will penalize you for adding independent variables (K in the equation) that do not fit the model.



Where,

- k is the number of independent variables
- n is the number of observations

## 🔄 **Multicollinearity in Linear Regression**

### ❓ What is it?
- **Multicollinearity** occurs when **independent variables** are **highly correlated** with each other.
- Violates a key assumption of linear regression: **independents should be independent**!

---

### ⚠️ Why is it a Problem?
1. 🎯 **Unreliable Coefficients**  
   - Model struggles to decide which variable is really affecting the outcome.
2. 💡 **Wrong Interpretations**  
   - You can’t trust individual feature importance or influence.
3. 🔄 **Feature Changes Chain-Reaction**  
   - Changing one variable indirectly changes another → messes up predictions.

---

### 🧪 **How to Detect It?**
- **VIF (Variance Inflation Factor)**  
  $$
  \text{VIF} = \frac{1}{1 - R^2}
$$
  - VIF > 5 or 10? 🚩 Problematic correlation  
  - Rule: **Higher VIF = Higher multicollinearity**

---

### 🧹 **How to Fix It?**
- ✅ **Remove** highly correlated features (with highest VIF)
- ✅ Use **dimensionality reduction** (e.g., PCA)
- ✅ Try **regularization techniques** like **Ridge regression** (shrinks coefficients)

---

#### 23. What is the generalized linear model?
- The generalized linear model is the derivative of the ordinary linear regression model. **GLM is more flexible in terms of residuals and can be used where linear regression does not seem appropriat**e. 
- GLM allows the distribution of residuals to be other than a normal distribution.
- It generalizes the linear regression by allowing the linear model to link to the target variable using the linking function. 
- Model estimation is done using the method of maximum likelihood estimation.


> A linking function in generalized linear models (GLMs) is `used to model the relationship between the linear predictor` (i.e., the sum of the product of the predictor variables and their respective coefficients) a`nd the response variable in a way that accounts for the distributional properties of the response variable`.

- In GLMs, **the response variable is assumed to follow a probability distribution from the exponential family,** such as the normal, binomial, Poisson, or gamma distribution. 
- - `The linking function is used to transform the linear predictor into the appropriate scale of the response variable`, depending on the distributional assumptions.

Here are some commonly used linking functions for different distributions in generalized linear models:

-   Binomial distribution: Logit, Probit, Complementary log-log (C-log-log), Log
-   Poisson distribution: Log, Identity, Sqrt
-   Normal distribution: Identity, Inverse, Log
-   Gamma distribution: Inverse, Log

#### VIF

**Q:** What is VIF (Variance Inflation Factor) and why is it important in linear regression?

**A:** VIF (Variance Inflation Factor) is a measure used to `detect multicollinearity` in linear regression. It quantifies `how much the variance of the estimated regression coefficients` is increased due to `multicollinearity among predictor variables`.

**Q:** How is VIF calculated and what are its typical ranges?

**A:** VIF is calculated for each predictor variable as 
$$ VIF = \frac{1}{1 - R^2} $$
, where \( R^2 \) is the coefficient of determination of a linear regression model predicting the current predictor from the remaining predictors. A VIF value of 1 indicates no multicollinearity. Typically, VIF values above 5 or 10 are considered indicative of multicollinearity.



**Q:** How can you use VIF to address multicollinearity?

**A:** If you find high VIF values, it's a sign of multicollinearity. You can take steps to address it:
1. **Remove one of the correlated variables:** If two or more variables are highly correlated, you can choose to keep the one that's more relevant to your analysis.
2. **Combine correlated variables:** If it makes sense, you can create a new variable that combines the information from correlated predictors.
3. **Regularization:** Techniques like Ridge or Lasso regression can help mitigate multicollinearity by adding a penalty to the coefficients.

**Q:** How do you interpret VIF values?

**A:** A VIF value of 1 indicates no multicollinearity (perfect independence). As VIF increases, it indicates increasing multicollinearity. Generally:
- VIF < 2: No or very low multicollinearity.
- VIF between 2 and 5: Moderate multicollinearity (may need attention).
- VIF > 5: High multicollinearity (requires action to mitigate).

**Q:** Can you provide an example of VIF interpretation?

**A:** Certainly. Let's say the VIF for a predictor variable "Income" is 6. 
> This means the variance of the estimated coefficient for "Income" is inflated by a factor of 6 due to multicollinearity. It suggests that "Income" has a high correlation with other predictor variables in the model.



````python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from statsmodels.stats.outliers_influence import variance_inflation_factor

# 🔧 Step 1: Simulate data
np.random.seed(42)
n = 100

# Independent variables
X1 = np.random.rand(n)
X2 = 2 * X1 + np.random.normal(0, 0.05, size=n)  # Highly correlated with X1
X3 = np.random.rand(n)

# Combine into a DataFrame
df = pd.DataFrame({
    'X1': X1,
    'X2': X2,
    'X3': X3
})

# 🔍 Step 2: Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)

# 🔁 Step 3: Compute VIF
vif_df = pd.DataFrame()
vif_df["Feature"] = df.columns
vif_df["VIF"] = [variance_inflation_factor(X_scaled, i) for i in range(X_scaled.shape[1])]

print(vif_df)
````



### 💥 Problem:
**Linear Regression** is sensitive to **outliers** — even a few can throw off the line completely.

---

### 🛠️ Solution: **RANSAC Regression**
RANSAC is a robust method that **ignores outliers** by focusing only on **inliers** (points that fit well).

---

### 🔁 **RANSAC Workflow – Step by Step**

1. **Random Sampling**  
   - Randomly select a **subset** of the data (just enough to fit the model).  
   - Check if it’s a valid set.

2. **Model Fitting**  
   - Fit a model (like linear regression) on this subset.

3. **Inlier Detection**  
   - Use this model to **predict on all data**.  
   - Calculate residuals (prediction error).  
   - Points with small residuals (below `residual_threshold`) are **inliers**.

4. **Model Selection**  
   - Keep the model if it has the **most inliers** so far.  
   - If tied, choose the one with the **better score** (e.g., R²).

5. **Repeat**  
   - Do this process for several iterations.  
   - Final model is the one with the most consistent inliers.

---

### ✅ Key Benefits:
- **Robust to outliers**  
- Finds a model that fits the **majority** of the data well  
- Often used in computer vision (e.g., line or plane detection)

---

### 📌 Example Use Case:
You're fitting a line to noisy sensor data, but there are random spikes (outliers).  
**RANSAC** will ignore the spikes and still find a good line.

---



*   **Bias:** Error from erroneous assumptions in the learning algorithm. High bias means the model is too simple and fails to capture the underlying relationship (underfitting). Standard Linear Regression can have high bias if the true relationship isn't linear.
*   **Variance:** Error from sensitivity to small fluctuations in the training set. High variance means the model is too complex and captures noise in the training data (overfitting). Standard Linear Regression typically has low variance.

<blockquote style="border-left: 5px solid #61dafb; padding-left: 10px; color: #ffffff; background-color: #333; padding: 10px;">
    <strong>Q4: Your R-squared is very high (e.g., 0.98). Is this always good? What could it mean?</strong>
</blockquote>

**A:** A very high R-squared isn't automatically good. It warrants investigation:
*   **Overfitting:** The model might fit the training data perfectly but fail to generalize to unseen data. Check performance on a separate test set and compare Adjusted R-squared to R-squared.
*   **Data Leakage:** A feature might inadvertently contain information about the target variable that wouldn't be available in a real-world prediction scenario. For example, including "Final Exam Score" when predicting "Midterm Exam Score" if the final happens later.
*   **Tautology / Feature Redundancy:** One feature might be directly calculated from the target or another feature (e.g., predicting weight in kg using weight in lbs as a feature).
*   **Low Variance in Target:** If the target variable itself has very little variance, even a poor model might achieve a high R-squared.
*   **Genuine Strong Relationship:** It *could* simply mean the predictors chosen have a very strong linear relationship with the target, but always be skeptical and validate rigorously.

<blockquote style="border-left: 5px solid #61dafb; padding-left: 10px; color: #ffffff; background-color: #333; padding: 10px;">
    <strong>Q6: How do you interpret the coefficients if you've log-transformed the dependent variable (Y) but kept the independent variables (X) in their original scale (log-lin model)?</strong>
</blockquote>

**A:** In a log-lin model (`log(Y) = β₀ + β₁X + ε`), the coefficient `β₁` is interpreted as follows: a one-unit increase in `X` is associated with approximately a `(100 * β₁)%` *percentage* change in `Y`.
*   **Derivation:** `d(log(Y))/dX = β₁`. Since `d(log(Y)) ≈ ΔY / Y`, we have `ΔY / Y ≈ β₁ * ΔX`. If `ΔX = 1`, then the relative change `ΔY / Y ≈ β₁`. Multiply by 100 for percentage change.
*   **More Precise:** The exact percentage change is `(exp(β₁) - 1) * 100%`. For small values of `β₁`, `exp(β₁) - 1 ≈ β₁`.

<blockquote style="border-left: 5px solid #61dafb; padding-left: 10px; color: #ffffff; background-color: #333; padding: 10px;">
    <strong>Q7: Why is feature scaling important for Gradient Descent but less so (or not at all) for the Normal Equation?</strong>
</blockquote>

  <strong>Q7: Why is feature scaling important for Gradient Descent ?</strong>
</blockquote>

**A:**
*   **Gradient Descent:** GD uses the gradient of the cost function to update parameters. If features have vastly different scales, `the cost function surface becomes elongated` (like a narrow valley). GD will oscillate back and forth and take a long time to converge, or might fail to converge `if the learning rate is too high`. 
*   Scaling features makes the cost function surface more `spherical`, allowing GD to converge much faster and more reliably. Regularization terms (L1/L2) also penalize coefficients based on magnitude, so scaling ensures features are penalized fairly relative to each other.


- ✅ **MAE** (Mean Absolute Error)  
- ✅ **MSE** (Mean Squared Error)  
- ✅ **RMSE** (Root Mean Squared Error)  
- ✅ **R² (R-squared)**  
- ✅ **Adjusted R²**  
- ✅ Other important ones: **MAPE**, **MSLE**, **RMSLE**, etc.



## 🔢 1. MAE – Mean Absolute Error

### 🧮 Formula:
$$\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} \left| y_i - \hat{y}_i \right|$$

### ✅ Intuition:
- Measures average **absolute** difference between actual and predicted.
- **Linear penalty** on errors.

### ✔️ Pros:
- Easy to interpret
- Robust to outliers compared to MSE

### ❌ Cons:
- Doesn’t penalize large errors strongly

---

## 🔢 2. MSE – Mean Squared Error

### 🧮 Formula:
$$\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

### ✅ Intuition:
- Squared penalty → amplifies larger errors
- Good for emphasizing large deviations

### ✔️ Pros:
- Differentiable (good for optimization)
- Sensitive to outliers (can be a pro)

### ❌ Cons:
- Unit is squared → hard to interpret
- Over-penalizes outliers

---

## 🔢 3. RMSE – Root Mean Squared Error

### 🧮 Formula:
$$\text{RMSE} = \sqrt{ \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 }$$

### ✅ Intuition:
- Same as MSE but in **same units** as target
- Penalizes large errors

### ✔️ Pros:
- Easier to interpret than MSE
- Still emphasizes large errors

### ❌ Cons:
- Still sensitive to outliers

---

## 🔢 4. R² – Coefficient of Determination

### 🧮 Formula:
$$
R^2 = 1 - \frac{\text{SS}_{\text{res}}}{\text{SS}_{\text{tot}}}
$$

Where:
- \( \text{SS}_{\text{res}} = \sum (y_i - \hat{y}_i)^2 \)
- \( \text{SS}_{\text{tot}} = \sum (y_i - \bar{y})^2 \)

### ✅ Intuition:
- Proportion of variance in the target explained by model.
- Ranges from **–∞ to 1**
  - R² = 1 → perfect
  - R² = 0 → no better than mean
  - R² < 0 → worse than mean

---

## 🔢 5. Adjusted R²

### 🧮 Formula:
$$
\text{Adjusted } R^2 = 1 - \left( \frac{(1 - R^2)(n - 1)}{n - p - 1} \right)
$$

Where:
- \( n \) = number of observations
- \( p \) = number of predictors

### ✅ Intuition:
- Penalizes R² for adding **irrelevant features**
- Prevents overfitting

---

## 🔢 6. MAPE – Mean Absolute Percentage Error

$$
\text{MAPE} = \frac{1}{n} \sum \left| \frac{y_i - \hat{y}_i}{y_i} \right| \times 100
$$

✅ Measures average **percentage** error  
❌ Undefined when \( y_i = 0 \)

---

## 🔢 7. MSLE / RMSLE – Mean Squared Logarithmic Error

$$
\text{MSLE} = \frac{1}{n} \sum \left( \log(1 + \hat{y}_i) - \log(1 + y_i) \right)^2
$$

✅ Use when:
- You care about **relative differences**
- Errors at small scales are more important  
❌ Can't be used with negative values

---

## 📊 Metric Comparison Table

| Metric   | Unit     | Penalizes Large Errors | Interpretable | Handles Zeros | Use Case                               |
|----------|----------|------------------------|---------------|---------------|----------------------------------------|
| MAE      | Same     | No                     | ✅ Easy        | ✅             | General-purpose, robust to outliers    |
| MSE      | Squared  | ✅ Yes                 | ❌ Hard        | ✅             | Optimization, focus on large errors    |
| RMSE     | Same     | ✅ Yes                 | ✅ Moderate    | ✅             | Emphasize large deviations             |
| R²       | None     | ❌ N/A                 | ✅ Moderate    | ✅             | Model explanatory power                |
| Adj. R²  | None     | ❌ N/A                 | ✅ Moderate    | ✅             | Feature selection, penalize complexity |
| MAPE     | %        | Moderate               | ✅ Easy        | ❌             | Business (sales/forecast), no 0s       |
| MSLE     | Log      | ✅ Small rel. errors   | ❌             | ❌             | Skewed data (e.g., growth rates)       |

---


---

## ✅ One-liner Summary:

> Use **MAE** for robustness, **MSE/RMSE** when large errors matter, and **R²/Adjusted R²** for explanatory power.

**Two-liner takeaway:**  
> Choose your error metric based on whether you care more about interpretability, large deviations, scale, or relative error.  
No one metric is best for all cases — align it with your domain goal and error tolerance.

---



````python
import numpy as np
import pandas as pd
from sklearn.metrics import (
    mean_absolute_error, mean_squared_error, r2_score, mean_absolute_percentage_error, mean_squared_log_error
)
import matplotlib.pyplot as plt

# Simulated actual vs predicted values
y_true = np.array([100, 200, 300, 400, 500])
y_pred = np.array([90, 210, 290, 430, 480])

# Compute metrics
mae = mean_absolute_error(y_true, y_pred)
mse = mean_squared_error(y_true, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_true, y_pred)
adj_r2 = 1 - (1 - r2) * (len(y_true) - 1) / (len(y_true) - 1 - 1)  # p = 1 predictor for simplicity
mape = mean_absolute_percentage_error(y_true, y_pred) * 100

# Only calculate MSLE if no values are negative
if np.all(y_true >= 0) and np.all(y_pred >= 0):
    msle = mean_squared_log_error(y_true, y_pred)
    rmsle = np.sqrt(msle)
else:
    msle = rmsle = np.nan

# Put in DataFrame
metrics_df = pd.DataFrame({
    'Metric': ['MAE', 'MSE', 'RMSE', 'R²', 'Adjusted R²', 'MAPE (%)', 'MSLE', 'RMSLE'],
    'Value': [mae, mse, rmse, r2, adj_r2, mape, msle, rmsle]
})

# Visualizing prediction error
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(y_true, label='Actual', marker='o')
ax.plot(y_pred, label='Predicted', marker='x')
ax.set_title('Actual vs Predicted')
ax.set_xlabel('Sample')
ax.set_ylabel('Value')
ax.legend()
ax.grid(True)

import ace_tools as tools; tools.display_dataframe_to_user(name="Regression Evaluation Metrics", dataframe=metrics_df)
plt.show()
````
