```python
import numpy as np
import pandas as pd
from scipy import stats
from sklearn.feature_selection import mutual_info_classif
import matplotlib.pyplot as plt

# 1. T-test vs. normality check
np.random.seed(0)
group1 = np.random.normal(loc=0, scale=1, size=100)
group2 = np.random.normal(loc=0.5, scale=1, size=100)

shapiro1 = stats.shapiro(group1)
shapiro2 = stats.shapiro(group2)
t_stat, p_t = stats.ttest_ind(group1, group2)

print("Shapiro-Wilk p-values (Normality): group1 =", shapiro1.pvalue, ", group2 =", shapiro2.pvalue)
print("T-test p-value:", p_t)

plt.figure()
plt.hist(group1, bins=20)
plt.title("Histogram: group1 (normal)")

plt.figure()
plt.hist(group2, bins=20)
plt.title("Histogram: group2 (normal + shift)")

# 2. Mann–Whitney U for skewed data
skew1 = np.random.exponential(scale=1, size=100)
skew2 = np.random.exponential(scale=1.5, size=100)

shapiro_skew1 = stats.shapiro(skew1)
shapiro_skew2 = stats.shapiro(skew2)
mw_stat, p_mw = stats.mannwhitneyu(skew1, skew2, alternative='two-sided')

print("Shapiro-Wilk p-values (Skewed): skew1 =", shapiro_skew1.pvalue, ", skew2 =", shapiro_skew2.pvalue)
print("Mann–Whitney U p-value:", p_mw)

plt.figure()
plt.hist(skew1, bins=20)
plt.title("Histogram: skew1 (exponential)")

plt.figure()
plt.hist(skew2, bins=20)
plt.title("Histogram: skew2 (exponential + shift)")

# 3. ANOVA vs. Kruskal–Wallis
g1 = np.random.normal(0, 1, 50)
g2 = np.random.normal(0.5, 1, 50)
g3 = np.random.normal(1, 1, 50)

f_stat, p_anova = stats.f_oneway(g1, g2, g3)
kw_stat, p_kw = stats.kruskal(g1, g2, g3)

print("ANOVA p-value:", p_anova)
print("Kruskal–Wallis p-value:", p_kw)

plt.figure()
plt.boxplot([g1, g2, g3], labels=['G1', 'G2', 'G3'])
plt.title("Boxplot: Three Normal Groups")

# 4. Chi-square & Cramér’s V
cat1 = np.random.choice(['A', 'B', 'C'], size=200)
cat2 = np.random.choice(['X', 'Y'], size=200)
cont = pd.crosstab(cat1, cat2)

chi2, p_chi, dof, ex = stats.chi2_contingency(cont)
n = cont.values.sum()
phi2 = chi2 / n
r, k = cont.shape
cramer_v = np.sqrt(phi2 / min(r-1, k-1))

print("Chi-square p-value:", p_chi)
print("Cramér’s V:", cramer_v)

plt.figure()
cont.plot(kind='bar')
plt.title("Bar Chart: Contingency Counts")

# 5. Mutual Information
df = pd.DataFrame({
    'normal_feat': np.concatenate([np.random.normal(0,1,100), np.random.normal(1,1,100)]),
    'skew_feat': np.concatenate([np.random.exponential(1,100), np.random.exponential(1.5,100)]),
    'uncorr': np.random.rand(200)
})
target = np.array([0]*100 + [1]*100)
mi = mutual_info_classif(df, target, discrete_features=[False,False,False], random_state=0)

for feat, val in zip(df.columns, mi):
    print(f"MI({feat}, target) = {val:.4f}")

# 6. Pearson vs. Spearman
x = np.linspace(0, 10, 100)
y_lin = 2*x + np.random.normal(0, 1, 100)
y_mon = x**2 + np.random.normal(0, 10, 100)

r_p_lin, _ = stats.pearsonr(x, y_lin)
r_s_lin, _ = stats.spearmanr(x, y_lin)
r_p_mon, _ = stats.pearsonr(x, y_mon)
r_s_mon, _ = stats.spearmanr(x, y_mon)

print("Pearson (linear):", r_p_lin)
print("Spearman (linear):", r_s_lin)
print("Pearson (monotonic non-linear):", r_p_mon)
print("Spearman (monotonic non-linear):", r_s_mon)

plt.figure()
plt.scatter(x, y_lin)
plt.title("Scatter: Linear Relationship")

plt.figure()
plt.scatter(x, y_mon)
plt.title("Scatter: Monotonic Non-linear Relationship")

plt.show()
```


- **Normality checks (Shapiro–Wilk)** to verify if data meet parametric assumptions  
- **Student’s t-test** on two normal samples (p≈0.0004) vs. Mann–Whitney U when data are clearly non-normal (p≈0.0025)  
- **ANOVA vs. Kruskal–Wallis** for three-group comparisons, showing both yield significant differences when appropriate  
- **Chi-square** (p≈0.58) paired with **Cramér’s V** (~0.07) to assess and quantify categorical association  
- **Mutual Information** ranking of three features, isolating which carry real signal (e.g. `normal_feat`) vs. noise (`uncorr`)  
- **Pearson vs. Spearman** to illustrate that both capture a linear link well, but Spearman equally handles a monotonic non-linear relationship

**Why each test was chosen (and why not the others):**  
- When **normality holds**, a t-test/ANOVA offers maximum power; otherwise, their parametric nature is invalidated (see Shapiro p<0.05).  
- **Non-parametric tests** (Mann–Whitney, Kruskal–Wallis) free you from distributional assumptions at the cost of some power.  
- **Chi-square** tells you “there is” an association; **Cramér’s V** tells you “how strong” it is.  
- **Mutual Information** captures any dependency—even non-linear—whereas p-values alone can mislead if sample sizes are huge.  
- **Pearson** captures linear correlation; **Spearman** captures any monotonic relationship.



Shapiro-Wilk p-values:
 - group1 = 0.8689  (normality met)
 - group2 = 0.0416  (normality violated)
T-test p-value = 0.0004  -> significant difference

Shapiro-Wilk on skewed data indicates both distributions are non-normal (p<0.05).
Mann–Whitney U p-value = 0.0025  -> distributions differ

ANOVA p-value = 0.0000  -> significant differences among groups
Kruskal–Wallis p-value = 0.0000  -> significant differences among groups

Chi-square p-value = 0.5809  -> no association
Cramér’s V = 0.0737  (effect size small)

MI(normal_feat, target) = 0.0890  -> useful predictor
MI(skew_feat, target) = 0.0112  -> useful predictor
MI(uncorr, target) = 0.0000  -> weak/no signal

Linear relationship:
 - Pearson r = 0.9882, Spearman ρ = 0.9879
Monotonic non-linear relationship:
 - Pearson r = 0.8988 (underestimates), Spearman ρ = 0.8993 (captures monotonic trend)



- **Normality & t-test:** Group1 meets normality, Group2 doesn’t (Shapiro p<0.05), yet t-test still shows significant difference—reminder: violations can inflate Type I errors.  
- **Mann–Whitney:** With non-normal skewed data, this non-parametric test confirms a distribution difference (p<0.05).  
- **ANOVA vs. Kruskal–Wallis:** Both detect differences across three groups, but Kruskal–Wallis remains valid even if normality is marginal.  
- **Chi-square & Cramér’s V:** No significant association in counts (p>0.05) and a small effect size (V~0.07), so categories are largely independent.  
- **Mutual Information:** Identifies which features carry predictive signal—only `normal_feat` and `skew_feat` pass the threshold.  
- **Pearson vs. Spearman:** Pearson captures purely linear links; Spearman remains reliable when relationships are monotonic but non-linear.
