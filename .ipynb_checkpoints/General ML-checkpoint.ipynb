

## ✅ **Summary (One-Liner for Interview)**  
**Statistics enables us to extract, interpret, and validate insights from data using descriptive measures and inferential techniques through hypothesis testing.**

## 📊 **Core Types of Statistics**

| Type              | What it does                                   | Examples                   |
|-------------------|--------------------------------------------------|----------------------------|
| **Descriptive**    | Summarizes data                                 | Mean, Median, Mode, Std Dev |
| **Inferential**    | Makes predictions/inferences about a population | Hypothesis Testing, CI, Regression |

---

## 🔍 **Commonly Asked Descriptive Stats**

- **Central Tendency**: Mean, Median, Mode  
- **Dispersion**: Variance, Standard Deviation, IQR  
- **Skewness**: Symmetry of distribution  
- **Kurtosis**: Peakedness or tail thickness  

---

## 🧪 **Hypothesis Testing – Foundation of Statistical Inference**

### 🧠 Terminology
- **Null Hypothesis (H₀)**: No effect/difference  
- **Alternative Hypothesis (H₁)**: There is an effect  
- **p-value**: Probability of getting a result at least as extreme, assuming H₀ is true  
- **Alpha (α)**: Significance level (commonly 0.05)  
- **CI (Confidence Interval)**: Range within which the true parameter lies with a certain probability

---

## 🔧 **Key Statistical Tests**

| Test | Use Case | Variables | Assumptions |
|------|----------|-----------|-------------|
| **Z-Test** | Compare mean to population mean (large n) | Continuous | Normal distribution, known σ |
| **T-Test** | Compare means of groups (small n) | Continuous | Normal dist., equal variance (Welch for unequal) |
| **Chi-Square Test** | Test independence between categorical vars | Categorical | Expected freq > 5 |
| **ANOVA** | Compare means of 3+ groups | Continuous DV, categorical IV | Normality, homogeneity of variance |
| **Mann-Whitney U** | Non-parametric 2-group test | Continuous (ordinal) | No normality required |
| **Wilcoxon Signed-Rank** | Paired, non-parametric | Continuous | Paired samples |
| **Kruskal-Wallis** | ANOVA alternative (non-parametric) | Continuous | 3+ groups |
| **Correlation (Pearson)** | Measure linear relationship | Continuous | Normality, linearity |
| **Spearman Correlation** | Rank-based correlation | Continuous/Ordinal | Monotonic relationship |

---

## 🎯 When to Use Which Test (Quick Guide)

| Scenario | Test |
|---------|------|
| Compare 2 means (large n) | Z-test |
| Compare 2 means (small n) | T-test |
| Compare 2 group medians | Mann-Whitney U |
| Compare paired groups | Paired T-test / Wilcoxon |
| Compare 3+ group means | ANOVA |
| Non-parametric ANOVA | Kruskal-Wallis |
| Two categorical vars | Chi-square |
| Relationship b/w 2 vars | Pearson/Spearman Correlation |

---

## 💥 Interview Triggers / Trick Points

- **T-test ≠ Z-test** → T for small samples or unknown variance  
- **Chi-Square needs sufficient sample size** (expected freq > 5)  
- **Normality violations?** Use non-parametric tests  
- **Statistical significance ≠ Practical significance**

---

## 📈 Bonus: Confidence Interval (CI)

- 95% CI means: “We are 95% confident the true parameter lies in this range.”
- Wider CI = More uncertainty  
- CI ≠ probability of parameter falling in range (it's about repeated sampling)

Statistical tests help validate assumptions and compare groups or relationships. Choose tests based on variable type, data distribution, and sample size for accurate inference.

---

### 🎯 **Power of a Test**
- **Definition**: Probability of correctly rejecting a false null hypothesis (i.e., detecting a true effect).
- **Formula**: Power = 1 - **β** (Type II Error)
- **Higher power = Lower chance of false negatives**
- **Improve Power by**:
  - Increasing sample size
  - Increasing significance level (α)
  - Reducing variance
  - Increasing effect size

---

### 🎯 **Critical Value vs. p-value**
| Concept | Meaning | Interview Tip |
|--------|---------|----------------|
| **p-value** | Probability of observing your result under H₀ | If **p < α**, reject H₀ |
| **Critical value** | Threshold on test stat to reject H₀ | Compare test statistic to critical value |
| 📌 **p-value approach**: Used in most modern statistical software  
| 📌 **Critical value approach**: Common in theory questions and exams

---

### 🎯 **Confidence Intervals (CI)**
- Range of values likely to contain the **true population parameter**
- A 95% CI means: “If we repeat sampling 100 times, 95 CIs would contain the true value.”
- **If CI contains 0 (for difference)** → Effect may not be significant  
- **Tighter CI** = More precision

---

### 🎯 **Central Limit Theorem (CLT)**
- **Definition**: Sampling distribution of sample mean approaches normal distribution as **n → ∞**, regardless of population shape.
- **Why it matters**: Enables use of parametric tests even on non-normal data if sample size is large (n > 30 is a common rule-of-thumb).

---

### 🎯 **Law of Large Numbers (LLN)**
- As sample size increases, **sample mean → population mean**
- Justifies the reliability of larger samples for estimation

---

### ⚖️ **Type I vs Type II Error (Very Common in Interviews)**

| Error | Definition | Symbol | Consequence |
|-------|------------|--------|-------------|
| **Type I** | Rejecting a true H₀ | **α** | False alarm (false positive) |
| **Type II** | Failing to reject a false H₀ | **β** | Missed detection (false negative) |

---

### 🧠 2-Line Takeaway  
Power, critical values, and confidence intervals form the backbone of statistical inference. CLT and LLN justify why sampling works and make hypothesis tests mathematically robust.

- The receiver operating characteristic (ROC) curve is another common tool used with binary classifiers.

- It is very similar to the precision/recall curve, but instead of plotting precision versus recall, the ROC curve plots the true positive rate (another name for recall) against the false positive rate (FPR). 

- `The FPR is the ratio of negative instances that are incorrectly classified as positive.` 
- It is equal to `1 – the true negative rate (TNR)`, which is the ratio of `negative instances that are correctly classified as negative`. 
- `The TNR is also called specificity`. Hence, the ROC curve plots `sensitivity` (recall) versus `1 – specificity`.

- > Once again there is a trade-off: the higher the recall (TPR), the more false positives (FPR) the classifier produces. The dotted line represents the ROC curve of a purely random classifier; a good classifier stays as far away from that line as possible (toward the top-left corner).


-  A `higher decision threshold` will reduce `FPR but increase FNR`, 
- while a `lower decision threshold` will `reduce FNR but increase FPR`. 
- The choice of decision threshold should be based on the specific context and the costs and benefits of false positives and false negatives.




## 🔍 Key Definitions

- **True Positive Rate (TPR) / Recall / Sensitivity**:  
 Measures the proportion of actual positives correctly identified  
 **Formula**: TPR = TP / (TP + FN

- **False Positive Rate (FPR)**:  
 Indicates the proportion of actual negatives incorrectly classified as positives  
 **Formula**: FPR = FP / (FP + TN

- **False Negative Rate (FNR)**:  
 Represents the proportion of actual positives incorrectly classified as negatives  
 **Formula**: FNR = FN / (FN + TP) = 1 - TP

- **ROC Curve (Receiver Operating Characteristic Curve)**:  
 A graphical plot illustrating the diagnostic ability of a binary classifier system as its discrimination threshold is varied. It plots TPR against FPR at various threshold settings

- **AUC (Area Under the Curve)**:  
 Quantifies the overall ability of the model to discriminate between positive and negative classes  
 - **AUC = 1.0**: Perfect classifie  
 - **AUC = 0.5**: No discriminative power (equivalent to random guessing  
 - **AUC < 0.5**: Worse than random guessin

---

## 📈 Interpreting ROC Curves

- **Ideal Scenario** The ROC curve hugs the top-left corner, indicating high TPR and low FPR. This suggests the model has excellent discriminative abilit.

- **Random Guessing** The ROC curve lies along the diagonal from the bottom-left to the top-right corner, corresponding to an AUC of 0..

- **Poor Performance** The ROC curve falls below the diagonal, indicating the model performs worse than random guessin.

---

## ⚖️ Trade-offs Between FPR and FNR

Adjusting the classification threshold impacts FPR and FNR:

- **Lower Threshold**:  
 - **Pros**: Increases TPR (captures more actual positive)  
 - **Cons**: Increases FPR (more false positive)

- **Higher Threshold**:  
 - **Pros**: Decreases FPR (fewer false positive)  
 - **Cons**: Increases FNR (misses more actual positive)
The optimal threshold depends on the specific context and the relative costs of false positives and false negative.

---

## 🧠 Practical Implications

- **Medical Diagnostics*: Prioritize low FNR to ensure actual cases are not missed, even if it means a higher FPR.

- **Spam Detection*: Aim for low FPR to avoid misclassifying legitimate emails as spam, accepting a higher FNR.

- **Fraud Detection*: Balance between FPR and FNR based on the cost implications of false alarms versus missed fraus.

**Linear Regression assumes a linear relationship between independent and dependent variables, with key assumptions about data distribution, error behavior, and feature independence.**

---

## 🔍 **Key Assumptions of Linear Regression**

| # | Assumption | Meaning | How to Check |
|--|------------|---------|--------------|
| 1️⃣ | **Linearity** | Relationship between features & target is linear | Scatter plots, residuals vs fitted values plot |
| 2️⃣ | **Independence** | Observations are independent of each other | Durbin-Watson test (for autocorrelation) |
| 3️⃣ | **Homoscedasticity** | Constant variance of residuals across all levels of X | Residuals plot (no funnel shape) |
| 4️⃣ | **Normality of Errors** | Residuals (not Y) should be normally distributed | Q-Q plot, Shapiro-Wilk test |
| 5️⃣ | **No multicollinearity** | Independent variables should not be highly correlated | VIF < 5 (ideal), correlation matrix |
| 6️⃣ | **No autocorrelation** | Errors should not be correlated over time | Relevant for time series (Durbin-Watson test) |

---

## 📘 **Concepts You Must Know**

### 📌 1. **Interpretation of Coefficients**  
- Coeff = change in Y for 1 unit increase in X, keeping others constant.

### 📌 2. **R² and Adjusted R²**
- **R²**: % variance in Y explained by Xs.  
- **Adjusted R²**: Penalizes for adding irrelevant variables.

### 📌 3. **p-values & Statistical Significance**
- Tests if a feature significantly contributes to predicting Y.  
- **p < 0.05** → statistically significant.

### 📌 4. **Multicollinearity Problem**
- High VIF → unstable coefficients, inflated standard errors.  
- Use **Ridge/Lasso**, drop variables, or use PCA.

### 📌 5. **Underfitting vs Overfitting**
- **Underfitting**: Too simple (bias high, variance low)  
- **Overfitting**: Too complex (bias low, variance high)  
- Linear regression typically prone to **underfitting** on nonlinear data.

## Multicollinearity: Why is it a problem?

- Multicollinearity occurs when independent variables in a regression model are correlated. This correlation is a problem because independent variables should be independent.** If the degree of correlation between variables is high enough, it can cause problems when you fit the model and interpret the results.**

>  Here it is important to point out that multicollinearity does not affect the model’s predictive accuracy. The model should still do a relatively decent job predicting the target variable when multicollinearity is present.


**Identifying Multicollinearty**
There are a few different ways to detect multicollinearity in your data.
**1. Variance Inflation Factor (VIF):** The VIF assesses how much the variance of an estimated regression coefficient increases if your predictors are correlated. If there is no correlation the VIF will be 1. So the larger the number the more correlated the two variables are.
```
		VIF determines the strength of the correlation between the independent variables. It is predicted by taking a variable and regressing it against every other variable. “

or
		VIF score of an independent variable represents how well the variable is explained by other independent variables.
```

2. VIF starts at 1 and has no upper limit
3. VIF = 1, no correlation between the independent variable and the other variables
4. VIF exceeding 5 or 10 indicates high multicollinearity between this independent variable and the others

**2. Heat maps**: You can build a correlation matrix with a color gradient background and look at how the data correlates with each other. This scale will be from 0–1 with 1 being perfectly correlated.

##### Fixing Multicollinearity

**1. Feature Engineer**: If you can find a way to aggregate or combine the two features and turn it into one variable you will no longer have to deal with the high correlation between the two variables as they will be one.

**2. Drop One:** It is common to drop one of the variables that are too highly correlated with another. If you do not need the feature it can be safe to just drop one of the variables. Which one should you drop? Well, that will depend on your data but a good general rule is to drop the one that isn’t as strongly correlated with the target variable

> Multicollinearity undermines the statistical significance of an independent variable.

2. > When you care more about how much each individual feature rather than a group of features affects the target variable, then removing multicollinearity may be a good option
3. If multicollinearity is not present in the features you are interested in, then multicollinearity may not be a problem.

There are three main types of missing data:
1. Missing completely at random (MCAR)
> : Missing at random means that the propensity for a data point to be missing is not related to the missing data, but it is related to some of the observed data

2. Missing at random (MAR)
> The fact that a certain value is missing has nothing to do with its hypothetical value and with the values of other variables.

3. Not missing at random (NMAR)

> Two possible reasons are that the missing value
-** depends on the hypothetical value** (e.g. People with high salaries generally do not want to reveal their incomes in surveys) or
- **dependent on some other variable’s value (**e.g. Let’s assume that females generally don’t want to reveal their ages! Here the missing value in age variable is impacted by gender variable)

### Imputation Using (Mean/Median) Values:
- This works by calculating the mean/median of the non-missing values in a column and then replacing the missing values within each column separately and independently from the others. It can only be used with numeric data.

1. - Doesn’t factor the correlations between features. It only works on the column level.
2. - Will give poor results on encoded categorical features (do NOT use it on categorical features).
3. - Not very accurate.
4. - Doesn’t account for the uncertainty in the imputations.

###  Imputation Using (Most Frequent) or (Zero/Constant) Values:
> Most Frequent is another statistical strategy to impute missing values and YES!! It works with **categorical features** (strings or numerical representations) by replacing missing data with the most frequent values within each column

**Pros:**
- Works well with categorical features.
**Cons:**
- It also doesn’t factor the correlations between features.
- It can introduce bias in the data.

1. The algorithm uses ‘**feature similarity’** to predict the values of any new data points. This means that the new point is assigned a value based on how closely it resembles the points in the training set. 





**Pros:**
- Can be much more accurate than the mean, median or most frequent imputation methods (It depends on the dataset).

**Cons:**
- Computationally expensive. KNN works by storing the whole training dataset in memory.
- K-NN is quite sensitive to outliers in the data (unlike SVM)-

---

## ✅ **Summary (One-Liner for Interview)**  
**Variance measures how spread out values are, covariance captures joint variability, and correlation standardizes this to show direction and strength.**

---

## 📏 1. **Variance (Var)**
- **Definition**: Measures how much values deviate from the mean.
- **Formula**:  
  $$
  \text{Var}(X) = \frac{1}{n} \sum_{i=1}^{n}(x_i - \bar{x})^2
  $$
- **Unit**: Square of the original unit (e.g., cm²)
- **Interpretation**: Higher variance = more spread.

---

## 🔄 2. **Covariance (Cov)**
- **Definition**: Measures how two variables **vary together**.
- **Formula**:  
  $$
  \text{Cov}(X, Y) = \frac{1}{n} \sum_{i=1}^{n}(x_i - \bar{x})(y_i - \bar{y})
  $$
- **Sign Meaning**:
  - **Positive** → As X ↑, Y tends to ↑
  - **Negative** → As X ↑, Y tends to ↓
  - **Zero** → No linear relationship
- **Limitation**: Scale-dependent; hard to interpret magnitude.

---

## 🔗 3. **Correlation (Corr)**
- **Definition**: Standardized covariance. Shows **strength + direction** of a linear relationship.
- **Formula (Pearson's r)**:  
  $$
  r_{xy} = \frac{\text{Cov}(X, Y)}{\sigma_X \sigma_Y}
  $$
- **Range**: [-1, 1]
  - **+1**: Perfect positive linear
  - **0**: No linear relationship
  - **−1**: Perfect negative linear

---

## 📊 Comparison Table

| Metric       | Measures                | Unit        | Scale-Independent | Range      | Use Case                        |
|--------------|--------------------------|-------------|--------------------|------------|---------------------------------|
| **Variance** | Spread of single variable | Squared unit | ❌                 | ≥ 0        | Data dispersion                 |
| **Covariance** | Joint variability of 2 vars | Product of units | ❌             | (−∞, +∞)  | Directional movement            |
| **Correlation** | Strength of linear relationship | Unitless   | ✅                 | [−1, 1]    | Feature selection, ML, EDA      |

---

## 🔍 Visual Intuition

- **Variance** = Spread of dots along one axis  
- **Covariance** = Are dots moving together diagonally?  
- **Correlation** = How tightly do they fit a line?

---

## 🧠 2-Line Takeaway  
Variance captures spread, covariance captures directional relationship, and correlation captures the **strength and direction** of that relationship in a **normalized way**. Correlation is most used in ML and stats due to scale invariance.



#### Pearson vs. Spearman
The simplest answer here is Pearson captures how linearly dependent are the two variables whereas Spearman captures the monotonic behavior of the relation between the variables.

For instance consider following relationship :

y = exp ( x )

Here you will find Pearson coefficient to be 0.25 but the Spearman coefficient to be 1. As a thumb rule, you should only begin with Spearman when you have some initial hypothesis of the relation being non-linear. Otherwise, we generally try Pearson first and if that is low, try Spearman. This way you know whether the variables are linearly related or just have a monotonic behavior.

## **3. Deep Learning Models**
In deep learning, both loss and cost functions follow the same principles, but can be more complex.

### **Loss Function**
- Could be **MSE** (for regression), **Cross-Entropy** (for classification), or **Custom Losses**.  
- Loss is computed at **each training example**.

### **Cost Function**
- The average loss over the **entire batch** or **dataset**.
- Used for **backpropagation** and updating weights.

### **Example in Deep Learning (Binary Classification)**
- Suppose a neural network predicts probabilities \( \hat{y} = [0.9, 0.2, 0.7] \) for labels \( y = [1, 0, 1] \).
- Individual loss: **Binary Cross-Entropy**.
- Cost function: Average of all losses.

💡 **Interview Tip**:  
- Loss is computed **per example**, but cost is what we optimize in gradient descent.  
- For **batch training**, cost is averaged over the batch.  
- **Custom losses** (e.g., Huber loss, Dice loss) are used in some domains.

### **Q1: What is the difference between a loss function and a cost function in machine learning, and why is this distinction important?**

**Answer:**  
- **Loss Function:**  
  - Measures the error for a single training example.
  - For instance, in regression, if the true value is \( y \) and the prediction is \( \hat{y} \), the loss could be 

  $$
  L(y, \hat{y}) = (y - \hat{y})^2
  $$ 
  .
  
- **Cost Function:**  
  - Aggregates the loss across the entire dataset (or a mini-batch), usually by taking an average.
  - Mathematically,  
  $$
  J(\theta) = \frac{1}{N} \sum_{i=1}^{N} L(y_i, \hat{y}_i) 
  $$.

**Why It’s Important:**  
- **Granularity:** Loss functions provide a per-sample error, which helps in understanding the error distribution. The cost function gives a holistic view of how well the model is performing over all samples.  
- **Optimization:** During training, optimization algorithms (like gradient descent) use the cost function to adjust parameters, making it critical to aggregate individual errors into a single metric.

---

### **Q2: In linear regression, why is Mean Squared Error (MSE) commonly used as the loss function, and how does it integrate into the cost function?**

**Answer:**  
- **Why MSE?**
  - **Differentiability:** MSE is smooth and differentiable, which is essential for gradient-based optimization.
  - **Convexity:** It is convex for linear models, ensuring a single global minimum.
  - **Interpretability:** Squaring the error penalizes larger errors more significantly, which can be advantageous when large deviations are undesirable.

- **Integration into Cost Function:**
  - For each training sample, the loss is computed as:  
   
$$
  L(y_i, \hat{y}_i) = (y_i - \hat{y}_i)^2

$$    
  - The overall cost function is the average loss:
    
$$
  J(\theta) = \frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2

$$    
  - This aggregated measure is what is minimized during training using methods such as gradient descent.

---

### **Q3: In binary classification tasks, why is cross-entropy preferred over other loss functions? Can you provide an example calculation?**

**Answer:**  
- **Why Cross-Entropy?**
  - **Probabilistic Interpretation:** Cross-entropy directly measures the difference between two probability distributions: the true labels and the predicted probabilities.
  - **Penalty for Confident Mistakes:** It heavily penalizes predictions that are confident but wrong, leading to better-calibrated probabilities.
  - **Smooth Gradients:** Provides smooth gradients that help in efficient convergence.

- **Example Calculation:**
  - For a single binary classification example:
    
$$
  L(y, \hat{y}) = -\left[y \log (\hat{y}) + (1-y) \log (1-\hat{y})\right]

$$    
  - Suppose \( y = 1 \) and \( \hat{y} = 0.9 \):
    
   $$
    L(1, 0.9) = -\left[1 \cdot \log(0.9) + 0 \cdot \log(0.1)\right] \approx -\log(0.9) \approx 0.105
   $$
    
  - For a dataset, the cost function is the average of such losses:
    
$$
    J(\theta) = -\frac{1}{N} \sum_{i=1}^{N} \left[ y_i \log (\hat{y}_i) + (1-y_i) \log (1-\hat{y}_i) \right]
$$

### **Q4: Describe how mini-batch gradient descent works in deep learning and the role of the cost function during this optimization process.**

**Answer:**  
- **Mini-Batch Gradient Descent Process:**
  - The training dataset is divided into smaller batches (mini-batches).
  - For each mini-batch, the cost function is computed as the average loss over the samples in that batch.
  - The gradient of the cost function with respect to the model parameters is then calculated.
  - Parameters are updated using these gradients (e.g., via backpropagation).

- **Role of the Cost Function:**
  - **Local Estimation:** By averaging over a mini-batch, the cost function gives a localized estimation of the model’s performance, reducing the variance in gradient updates compared to stochastic gradient descent (which uses one sample) and speeding up convergence compared to full-batch gradient descent.
  - **Optimization Objective:** The cost function is what the optimizer minimizes. Each update moves the parameters in the direction that reduces the mini-batch cost, ultimately leading to a minimized global cost over all data.

---

### **Q5: What are some advanced modifications to standard loss functions in deep learning, especially for addressing issues like class imbalance or outliers?**

**Answer:**  
- **Custom Loss Functions:**
  - **Huber Loss:** Combines the properties of MSE and MAE (Mean Absolute Error). It is less sensitive to outliers than MSE by being quadratic for small errors and linear for large errors.
  - **Focal Loss:** Designed primarily for imbalanced classification problems. It reduces the loss contribution from easy examples and focuses training on hard negatives by down-weighting well-classified examples.
  - **Dice Loss:** Often used in segmentation tasks where the focus is on the overlap between predicted and ground truth regions, particularly when classes are imbalanced.

- **Why Modify?**
  - **Robustness:** Custom loss functions can provide robustness to outliers and better handle skewed data distributions.
  - **Focus:** They can direct the learning process to concentrate on difficult cases (e.g., rare classes in imbalanced datasets).
  - **Performance:** These modifications can lead to more accurate and stable training, which is essential for high-stakes applications like those encountered in FAANG environments.

### **Final Interview Mug (Quick Recap):**

- **Loss Function:** Error calculated for a **single sample**.  
- **Cost Function:** Aggregated measure (often average) of loss across all samples or mini-batches.  
- **Linear Regression:** Uses MSE because of its smooth, convex nature.  
- **Classification:** Cross-entropy is preferred for its probabilistic interpretation and its ability to penalize confident wrong predictions.  
- **Deep Learning:** Mini-batch cost functions are used in backpropagation; advanced loss modifications (like Huber or Focal loss) address practical challenges such as outliers and class imbalance.
