# What is Recurrent Neural Network (RNN)?

Recurrent Neural Network is a generalization of feedforward neural network that has an internal memory. RNN is recurrent in nature as it performs the same function for every input of data while the output of the current input depends on the past one computation. After producing the output, it is copied and sent back into the recurrent network. For making a decision, it considers the current input and the output that it has learned from the previous input.

Unlike feedforward neural networks,  RNNs can use their internal state (memory) to process sequences of inputs. This makes them applicable to tasks such as unsegmented, connected handwriting recognition or speech recognition. In other neural networks, all the inputs are independent of each other. But in RNN, all the inputs are related to each other.

First, it takes the X(0) from the sequence of input and then it outputs h(0) which together with X(1) is the input for the next step. So, the h(0) and X(1) is the input for the next step. Similarly, h(1) from the next is the input with X(2) for the next step and so on. This way, it keeps remembering the context while training.
The formula for the current state is

### 🔁 RNN (Recurrent Neural Network)
- **Structure:** Simple loop architecture — each unit passes its hidden state to the next timestep.
- **Working:** At each time step, it takes input and previous hidden state to produce a new hidden state.
- **Problem:** Suffers from **vanishing/exploding gradients**, making it hard to learn long-term dependencies.
- **Memory:** Remembers information for short durations.

---

### 🧠 LSTM (Long Short-Term Memory)
- **Structure:** A special kind of RNN that introduces a **memory cell** and three gates: **input gate, forget gate, and output gate**.
- **Working:** Controls the flow of information using gates:
  - **Forget Gate:** Decides what information to discard.
  - **Input Gate:** Decides what new information to store.
  - **Output Gate:** Decides what part of the memory to output.
- **Advantage:** **Solves vanishing gradient problem**, making it better at capturing **long-term dependencies**.
- **Memory:** Maintains both **hidden state** and **cell state** (long-term memory).

# What is a Seq2Seq model?

In a Seq2seq model, a neural machine translation receives an input in the form of a word sequence and generates a word sequence as output.

- The encoder captures the `_context_` of the input sequence in the form of a `_hidden state vector_` and sends it to the decoder, which then produces the output sequence. 
- RNNs by design, take two inputs, the current example they see, and a representation of the previous input. Thus, the output at time step _t_ depends on the current input as well as the input at time _t-1_. 
- This is the reason they perform better when posed with sequence related tasks. The sequential information is preserved in a hidden state of the network and used in the next instance.

- **_Drawback:_** The output sequence relies heavil`**y on the context defined by the hidden state**` in the final output of the encoder, making it challenging for the model to deal with long sentences. 
* `In the case of long sequences, **there is a high probability that the initial context has been lost by the end of the sequence.**`

## Self-Attention

Self-attention is a way for Transformer to convert the “understanding” of other related words into the word we are dealing with.

- First, self-attention calculates three new vectors. In the paper, the dimension of the vector is **512 dimensions**. 
- We call these three vectors Query, Key, and Value respectively. 
These three vectors are produced by multiplying **the word embedding vectors and a ​​randomly matrix (the dimension is (64, 512) in the paper)**whose value is updated during the back-propagation process.
- we calculate the fractional value of the self-attention, which determines how much attention is paid to the rest of the input sentence when we encode a word at a certain location.
- This method of determining the **weight distribution of value by the degree of similarity between query and key is called** scaled dot-product attention.




### Seq2Seq   Model Limitation
In Seq2Seq the input sentence meaning is captured in one vector, so **as the length of the sentence increase, the more difficult it gets for the model to capture the information in this vector.** Consequently, its performance decreases with long sentences as it tends to forget parts of it, **the hidden vector becomes a bottleneck.**



In the Seq2Seq model with attention,` the mechanism determines the relevance of each encoder hidden state to the current decoding step by calculating attention scores`. These scores are then normalized using the softmax function to produce attention weights, which indicate the importance of each encoder hidden state in generating the current output token.

### How Attention Scores and Softmax Work

1.  **Calculating Attention Scores**: At each decoding step, `the decoder's current hidden state is compared with each of the encoder's hidden states.` T`his comparison can be done using various functions, such as the dot product or a feed-forward neural network`, resulting in a scalar value known as the attention score.
    
2.  **Applying Softmax**: `The attention scores are passed through the softmax function to normalize them into a probability distribution`. This ensures that the sum of all attention weights equals 1, making them interpretable as probabilities. 
* The softmax function amplifies the differences between scores, assigning higher probabilities to more relevant encoder states and lower probabilities to less relevant ones. 
    
4.  **Generating the Context Vector**: The attention weights are used to compute a weighted sum of the encoder's hidden states, producing the context vector. This context vector encapsulates the relevant information from the input sequence that the decoder uses to generate the current output token.
    

### Visualizing the Process

Consider the following example:

-   **Encoder Hidden States**: h₀, h₁, h₂, h₃
    
-   **Decoder Hidden State**: s₁
    

The attention scores are computed as:

e₀ = f(s₁, h₀), e₁ = f(s₁, h₁), e₂ = f(s₁, h₂), e₃ = f(s₁, h₃)

Applying softmax:

a₀ = softmax(e₀), a₁ = softmax(e₁), a₂ = softmax(e₂), a₃ = softmax(e₃)

The context vector is then:

c₁ = a₀ * h₀ + a₁ * h₁ + a₂ * h₂ + a₃ * h₃

This context vector is used by the decoder to generate the output token at the current step.


* **The main bottleneck of the Seq2Seq model i**s `the need to compress the entire contents of the source sequence into a fixed-size vector`. If the text is slightly longer,` it is easy to lose some information of the text.`

* In order to solve this problem, **Attention came into being.** The attention mechanism alleviates this problem by allowing the decoder to look back at the `source sequence hidden state`, and then provide its weighted average as an additional input to the decoder.

There are two main differences between **Attention and the traditional Seq2Seq model.** 
- First, the encoder provides more data to the decoder, 
- the encoder will provide the hidden state of `all nodes to the decoder`, not just the hidden state of the last node of the encoder.

####  Why choose Bert over Embedding models?

**1. Contextual Sensitivity:**

-   **Word2Vec:** Generates static, context-independent embeddings. Each word has a single vector representation, regardless of its usage. For example, "bank" will have the same embedding whether referring to a financial institution or the side of a river. 
    
-   **BERT:** Produces dynamic, context-dependent embeddings. The representation of a word varies based on its surrounding words. Thus, "bank" would have different embeddings in "river bank" versus "bank account," accurately reflecting its intended meaning in each context. 
    

**2. Handling Polysemy:**

-   **Word2Vec:** Struggles with polysemous words (words with multiple meanings) since it assigns a single vector to each word, failing to distinguish between different senses. 
    
-   **BERT:** Effectively disambiguates word meanings by considering the full context, allowing it to assign different embeddings to the same word based on its usage. 
    

**3. Incorporation of Word Position:**

-   **Word2Vec:** Does not account for word order or position within a sentence, treating context as a simple bag of words.
    
-   **BERT:** Integrates position embeddings, enabling it to understand the sequential nature of language and the importance of word order in conveying meaning. 
    

**4. Training Approach and Flexibility:**

-   **Word2Vec:** Learns embeddings based on word co-occurrence statistics, capturing general semantic relationships. However, it lacks the ability to adapt to specific tasks without retraining.
    
-   **BERT:** Utilizes a transformer-based architecture with bidirectional context understanding. Pre-trained on vast text corpora, it can be fine-tuned for various downstream tasks, offering flexibility and state-of-the-art performance across numerous NLP applications.

## 🔤 Tokenizers in NLP 

### 🧩 1. **WordPiece Tokenizer (Used in BERT)**
- **Type**: Subword tokenizer
- **Goal**: Breaks unknown or rare words into meaningful subword units.
####  WordPiece

#### ✅ **Working Process**:
1. **Check vocabulary**:
   - If full word exists → use it as is.
   - If not → split word into subwords.
2. **Iterative subword splitting**:
   - Break down until smallest possible valid subwords.
   - If needed, go to character-level.
   
The following steps help us to understand the algorithm better:

1.  Extract the words from the given dataset along with their count.
2.  Define the vocabulary size.
3.  Split the words into a character sequence.
4.  Add all the unique characters in our character sequence to the vocabulary.
5.  Build the language model on the given dataset (training set).
6.  Select and merge the symbol pair that has the maximum likelihood of the language model trained on the training set.
7.  Repeat  _step 6_  until the vocabulary size is reached.

#### 📌 **Special notation**:
- Subwords after the first one are prefixed with `##`.
- Ex: `"pretraining"` → `["pre", "##train", "##ing"]`

#### ✅ **Strengths**:
- Handles out-of-vocabulary (OOV) gracefully.
- Retains meaningful chunks of unknown words.
- Compact vocabulary size.

---

### 🔁 2. **Byte Pair Encoding (BPE)**
- **Used in**: GPT-2, RoBERTa
- **Type**: Subword tokenizer (compression-inspired)

Byte Pair Encoding:
The steps involved in BPE are provided here:

1.  Extract the words from the given dataset along with their count.
2.  Define the vocabulary size.
3.  Split the words into a character sequence.
4.  Add all the unique characters in our character sequence to the vocabulary.
5.  Select and merge the symbol pair that has a high frequency.
6.  Repeat step 5 until the vocabulary size is reached.

  
#### ✅ **Working Process**:
1. Start with characters.
2. Count most frequent pairs (e.g., `'t'` + `'h'` → `'th'`).
3. Merge pairs iteratively to form new subwords.
4. Repeat till desired vocabulary size.

#### 🔎 **Example**:
- `"lowers"` → `["low", "er", "s"]`  
- `"lowering"` → `["low", "er", "ing"]`

#### ✅ **Strengths**:
- Learns common patterns efficiently.
- Faster to train compared to WordPiece.
- What is the purpose of doing byte-level BPE instead of character-level BPE?

  >  Well, byte-level BPE will be very useful in a multilingual setting. It is very effective in handling OOV words and it is great at sharing vocabulary across multiple languages.

---

### 📊 3. **Unigram Language Model Tokenizer (Used in SentencePiece)**
- **Type**: Probabilistic subword tokenizer

#### ✅ **Working Process**:
1. Start with a large list of subwords.
2. Learn the most probable subword segmentation for a sentence.
3. Remove subwords with low probabilities iteratively.

#### 🔎 **Example**:
- `"beautiful"` → `["beaut", "iful"]` or `["be", "aut", "iful"]`  
- Chooses based on probability.

#### ✅ **Strengths**:
- Flexibility in choosing multiple segmentations.
- Can generate more compact token representations.

---

### ⬜ 4. **Whitespace Tokenizer**
- **Simplest tokenizer**
- Splits text at white spaces and punctuations.

#### 🔎 Example:
- `"I love AI."` → `["I", "love", "AI", "."]`

#### 🚫 **Limitations**:
- Doesn’t handle compound or unknown words.
- No understanding of subwords or character patterns.

## 🔍  Tips

> 💡 **Q: Why do we prefer subword tokenizers like WordPiece or BPE over word-level tokenizers?**  
✅ **A**: Subword tokenizers reduce OOV issues, allow efficient vocabulary usage, and retain semantic meaning in word pieces.

> 💡 **Q: What’s the difference between WordPiece and BPE?**  
✅ **A**: WordPiece chooses merges based on likelihood to improve language model accuracy; BPE does so based on frequency counts.

> 💡 **Q: When would Unigram be preferred?**  
✅ **A**: When flexibility in segmentation and probabilistic modeling is needed, e.g., for multilingual text.

### **Next sentence prediction** (**NSP**) is another interesting strategy used for training the BERT model. NSP is a binary classification task.

### 🔍 What Is Learning Rate Warm-Up?

**Learning rate warm-up** is a technique where the learning rate starts from a small value and increases linearly over a set number of steps (the warm-up period) until it reaches a predefined maximum.After this period, the learning rate typically follows a decay schedule

### 🧠 Why Start with a Low Learning Rate?

Starting with a high learning rate can cause instability in the early stages of training because:

-   **Random Initialization** Model parameters are randomly initialized, and large updates can lead to divergence too early.
    
-   **Unstable Gradients** Early in training, gradients can be large and erratic. A high learning rate amplifies this issue.
    
-   **Loss Landscape** The optimization landscape can be sharp or uneven initially, and large steps might overshoot minim.By starting with a low learning rate, the model makes smaller, more controlled updates, allowing it to stabilize before larger updates are applie.
    

### 📈 How Does Warm-Up Work

If the target learning rate is ηmax\eta_{\text{max}} and the warm-up period is NN steps, the learning rate at step ii s:

$$ηi=iN⋅ηmax \eta_i = \frac{i}{N} \cdot \eta_{\text{max}}$$

This linear increase continues until step NN, after which the learning rate may decay according to a chosen schedule (e.g., linear decay, cosine deca)



### 🚀 Why Use Warm-Up?

-   *_Start Low_: Begin training with a low learning rate to ensure stabilty.
    
-   *_Gradual Increase_: Use warm-up to linearly increase the learning rate to its optimal vaue.
    
-   *_Enhanced Stability_: This approach helps prevent divergence and leads to better convergece.


-   **Stabilizes Training** Prevents large weight updates that could destabilize the model in early training stage.
    
-   **Improves Convergence** Facilitates smoother and more reliable convergence to optimal solution.
    
-   **Enhances Performance** Often leads to better final model performance by avoiding early training pitfall.

### 🧠 How BERT Predicts Answer Spans

1. **Input Preparation**:
   -The input to BERT is a concatenation of the question and the context (paragraph), separated by special tokens
     ```
     [CLS] Question [SEP] Context [SEP]
     ```
   -Each token in this sequence is embedded with positional and segment embeddings to help BERT distinguish between the question and the context

2. **Token Representation**:
   -BERT processes the input sequence and generates contextualized embeddings for each token

3. **Start and End Token Prediction**:
   -Two separate linear layers (often referred to as "heads") are added on top of BERT's output
     - **Start Token Classifier**: Computes a score for each token being the start of the answer span.
     - **End Token Classifier**: Computes a score for each token being the end of the answer span.
   -These scores are obtained by taking the dot product between each token's embedding and a learned weight vector for start or end positions
4. **Probability Distribution**:
   -The scores from the start and end classifiers are passed through a softmax function to obtain probability distributions over all tokens
     - `P_start = softmax(scores_start)`
     - `P_end = softmax(scores_end)`

5. **Answer Span Selection**:
   -The model selects the span with the highest joint probability
     - `score(i, j) = P_start[i] * P_end[j]`
     - Subject to the constraint that `j >= i` and `(j - i + 1) <= max_answer_length`
   -The tokens from position `i` to `j` in the context are then extracted as the predicted answer

### ## Factorized embedding parameterization

 > ALBERT decouples these sizes by factorizing the embedding matrix into two smaller matrices: one for vocabulary embeddings and another for projecting these embeddings into the hidden space.
With factorization, instead of directly projecting the one-hot-encoded vectors of vocabulary to the hidden space (_V_ x _H_).

	-  first, we project one-hot-encoded vectors to low-dimensional embedding space (_V_ x _E_), and then we project this

	- low-dimensional embedding to hidden space (_E_ x _H_). 
		- _E_ x _H._ The dimension becomes _E_ x _H =_ 128 x 768.
	- That is, instead of projecting _V_ x _H_ directly, we factorize this step into _V_ x _E_ and  _E_ x _H_.

### **ELECTRA: Efficiently Learning an Encoder that Classifies Token Replacements Accurately**

**1. Introduction: What is ELECTRA?**

-   **Definition:** ELECTRA is a pre-training approach for language models (like BERT) that aims to be significantly more computationally efficient and achieve strong performance on downstream NLP tasks.
    
-   **Core Idea:** Instead of the traditional Masked Language Modeling (MLM) used by BERT, ELECTRA introduces a new pre-training task called **Replaced Token Detection (RTD)**.
    
    

**2. The Problem with Masked Language Modeling (MLM)**

-   **How MLM Works (BERT):** Randomly select ~15% of input tokens, replace them with a special [MASK] token (or sometimes random words/original words), and train the model to predict the original masked tokens.
    
-   **The Issue:**
    
    -   **Computational Inefficiency:** The model only learns from the masked tokens (~15% of the input), making the learning process sample-inefficient.
        
    -   **Pre-train/Fine-tune Mismatch:** The artificial [MASK] token is present during pre-training but never seen during fine-tuning on downstream tasks (like sentiment analysis, question answering). This discrepancy can hinder performance.
        

**3. ELECTRA's Solution: Replaced Token Detection (RTD)**

-   **The Core Task:** Instead of masking tokens, ELECTRA replaces some input tokens with plausible alternatives generated by another small neural network. The main ELECTRA model is then trained to **distinguish between the original tokens and the replaced (fake) tokens** for every token in the sequence.
    
-   **How it Addresses MLM's Problems:**
    
    -   **Efficiency:** The model learns from all input tokens (classifying each as 'original' or 'replaced'), not just the 15% that were masked. This makes the learning signal much denser and the pre-training significantly more computationally efficient.
        
    -   **Pre-train/Fine-tune Mismatch:** Because the input to the main ELECTRA model consists of actual words (some original, some plausible but incorrect replacements), it doesn't rely on the artificial [MASK] token. `This makes the pre-training task distribution closer to the distributions encountered during fine-tuning.`
       


**4. The Generator-Discriminator Architecture**

ELECTRA uses a two-model setup for pre-training:

-   **Generator (G):**
    
    -   Typically a small Masked Language Model (e.g., a small BERT).
        
    -   **Function:** Takes a masked version of the input sequence and predicts the original tokens for the masked positions.
        
    -   **Example:**
        
        -   Original Input: the quick brown fox jumps
            
        -   Masked Input for Generator: the [MASK] brown [MASK] jumps
            
        -   Generator Output (Potential Prediction): the **fast** brown **dog** jumps
            
-   **Discriminator (D):**
    
    -   This is the main **ELECTRA model** that we actually want to train and use for downstream tasks.
        
    -   **Function:** Takes the output sequence from the Generator (where some tokens have been replaced) and predicts, for each token, whether it was part of the original input or if it was replaced by the Generator.
        
    -   **Example:**
        
        -   Input to Discriminator: the fast brown dog jumps
            
        -   Discriminator Output (Prediction): the (original) fast (replaced) brown (original) dog (replaced) jumps (original)
            

**Important Note:** During pre-training, the loss is calculated based on the Discriminator's performance. The Generator is only trained to produce plausible replacements (via its MLM objective), but its weights are often not used after pre-training. The Discriminator is the final ELECTRA model.

**5. The Pre-training Process (Simplified)**

1.  **Input:** Take a sentence: The chef cooked the meal.
    
2.  **Masking:** Select tokens to replace (e.g., chef, meal): The [MASK] cooked the [MASK].
    
3.  **Generation:** The Generator (small MLM) predicts replacements: The **baker** cooked the **food**.
    
4.  **Corruption:** Create the input for the Discriminator by replacing the original tokens at the masked positions with the Generator's predictions: The baker cooked the food.
    
5.  **Discrimination:** The Discriminator (ELECTRA model) processes the corrupted sentence and outputs a probability for each token indicating if it's 'original' or 'replaced'.
    
    -   The -> original
        
    -   baker -> replaced
        
    -   cooked -> original
        
    -   the -> original
        
    -   food -> replaced
        
    -   . -> original
        
6.  **Loss Calculation:** Calculate the Binary Cross-Entropy loss based on the Discriminator's predictions vs. the actual labels (original/replaced). The model learns to improve its discrimination ability.
    

    
-   **Loss (Discriminator - RTD):** Sum over all tokens t:  
    L_RTD = Σ - [ I(x_corrupt_t = x_original_t) * log(D(X_corrupt, t)) + I(x_corrupt_t != x_original_t) * log(1 - D(X_corrupt, t)) ]  
    (This is essentially a binary classification loss for each token)
    
    (Optionally, there's also a loss for the Generator, L_MLM, based on its predictions)
    

**7. Advantages Summarized**

-   **Sample Efficiency:** Learns from all input tokens, leading to faster training and requiring less data/compute for comparable performance to MLM-based models.
    
-   **Performance:** Often outperforms models like BERT trained for the same amount of compute, especially on smaller model sizes.
    
-   **No [MASK] Mismatch:** Avoids the pre-train/fine-tune discrepancy caused by the [MASK] token.
    


### **DistilBERT: A Distilled Version of BERT**

**1. Introduction: What is DistilBERT and Why Was It Created?**

-   **Definition:** DistilBERT is a smaller, faster, cheaper, and lighter version of the BERT (Bidirectional Encoder Representations from Transformers) model.
    
-   **Problem Solved:** Standard BERT models are very large (millions of parameters) and computationally expensive. This makes them difficult and costly to train and deploy, especially in resource-constrained environments like mobile devices or applications requiring low latency.
    
-   **Goal:** To create a model that retains a significant portion of BERT's language understanding capabilities while being substantially smaller and faster. DistilBERT aims to reduce computation time and model size, making powerful transformer models more practical.
    


**Knowledge distillation** is a model compression technique in which a small model is trained to reproduce the behavior of a large pre-trained model.

> It is also referred to as teacher-student learning, where the large pre-trained model is the teacher and the small model is the student.

This indicates that apart from the word **H****omework**_,_ the words **Book** and **Assignment** are more relevant to the given sentence compared to words like **Cake** and **Car**. This is known as **dark knowledge.**

To Extract Dark Knoweledge we use, we use the softmax function with a temperature.

`It is commonly referred to as softmax temperature. We use the softmax temperature in the output layer. It is used to smooth the probability distribution.`



When T = 1, we have the same probability distribution returned by our network with the standard softmax function. 

When T = 2,the probability distribution is smoothed

- we get a smoothed probability distribution, which gives more information about other classes.

- First, we pre-train the teacher network with softmax temperature to obtain dark knowledge. 
- Then, during knowledge distillation (transferring knowledge from the teacher to the student), we transfer this dark knowledge from the teacher to the student.

`The output of the teacher network is called a **soft target**, and the prediction made by the student network is called a **soft prediction**.`




**2. Core Concept: Knowledge Distillation (KD)**

-   **Analogy:** DistilBERT uses a "teacher-student" framework based on the concept of Knowledge Distillation.
    
-   **Teacher Model:** A large, pre-trained, high-performing BERT model (e.g., BERT-base). This model has already learned rich language representations.
    
-   **Student Model:** A smaller transformer model (DistilBERT) with fewer layers and parameters.
    
-   **Process:** The goal is to "distill" or transfer the knowledge embedded within the large Teacher model to the much smaller Student model. The Teacher model guides the training of the Student model.
    

# DistilBERT – the distilled version of BERT

As the name suggests, DistilBERT uses knowledge distillation. The ultimate idea of DistilBERT is that we take a 

	- large pre-trained BERT model and transfer its knowledge to a small BERT through knowledge distillation. 
	- The large pre-trained BERT is called a **teacher BERT** and the small BERT is called a **student BERT**.



Thus, our loss function is the sum of the following three losses:

-   Distillation loss
-   Masked language modeling loss (student loss)
-   Cosine embedding loss

> Know the **three components of the loss function:** Distillation Loss (soft labels), MLM Loss (hard labels), and Cosine Embedding Loss (hidden states).


**3. How DistilBERT is Trained (The Distillation Process)**

-   **Initialization:** The Student model (DistilBERT) is typically initialized with a subset of the Teacher's weights (e.g., taking every other layer).
    
-   **Training Data:** The Student is trained on the same large corpus of text that the Teacher was originally trained on.
    
-   **Mimicking the Teacher:** The key idea is to train the Student model not just on the "hard" ground-truth labels (like in standard MLM), but also to mimic the output behavior and internal representations of the Teacher model.
    
-   **Teacher's Role:** The pre-trained Teacher model processes the training data, and its outputs (probabilities and hidden states) serve as "soft targets" or guidance for the Student.
    



### 🔍 Components of DistilBERT's Triple Loss Function

1. **Distillation Loss (L<sub>ce</sub>)**:
   - **Purpose**:Encourages the student model to mimic the teacher model's output probabilities
   - **Mechanism**:Applies a softmax function with a temperature parameter (T) to both the teacher's and student's logits, producing softened probability distributions. The Kullback-Leibler (KL) divergence is then computed between these distributions
   - **Formula**:
     $$
     L_{ce} = \text{KL}(\text{softmax}(z_t / T) \parallel \text{softmax}(z_s / T))
     $$
     Where \( z_t \) and \( z_s \) are the logits from the teacher and student models, respectively.
   - **Insight**:Using a higher temperature (e.g., T=2.0) produces softer probability distributions, providing more informative gradients for the student model during training 

2. **Masked Language Modeling Loss (L<sub>mlm</sub>)**:
   - **Purpose**:Ensures the student model retains the ability to perform the masked language modeling task, similar to BERT
   - **Mechanism**:Calculates the cross-entropy loss between the student's predictions and the actual masked tokens in the input
   - **Formula**:
     $$
     L_{mlm} = \text{CrossEntropy}(y, \text{softmax}(z_s))
     $$
     Where \( y \) represents the true labels for the masked tokens.
   - **Insight**:This component ensures that DistilBERT maintains strong language understanding capabilities citeturn0search4

3. **Cosine Embedding Loss (L<sub>cos</sub>)**:
   - **Purpose**:Aligns the internal representations (hidden states) of the student model with those of the teacher model
   - **Mechanism**:Computes the cosine similarity between the hidden states of the teacher and student models for the same inputs. The loss encourages these representations to be directionally similar
   - **Formula**:
     $$
     L_{cos} = 1 - \cos(h_t, h_s)
     $$
     Where \( h_t \) and \( h_s \) are the hidden state vectors from the teacher and student models, respectively.
   - **Insight**:By aligning hidden states, the student model captures nuanced information learned by the teacher model citeturn0search4

### 🧮 Total Loss Function
The overall loss function used to train DistilBERT is a weighted sum of the three component:
$$
L_{total} = \lambda_{ce} \cdot L_{ce} + \lambda_{mlm} \cdot L_{mlm} + \lambda_{cos} \cdot L_{cos}
$$
Where \( \lambda_{ce} \), \( \lambda_{mlm} \), and \( \lambda_{cos} \) are hyperparameters that balance the contribution of each loss componen. In the original implementation, typical values are \( \lambda_{ce} = 5.0 \), \( \lambda_{mlm} = 2.0 \), and \( \lambda_{cos} = 1.0 \.

### 📌 Summary Tabl

| Loss Component             | Purpose                                      | Calculation Method                        |
|----------------------------|----------------------------------------------|-------------------------------------------|
| **Distillation Loss (L<sub>ce</sub>)** | Mimic teacher's output probabilities       | KL divergence between softened logits     |
| **MLM Loss (L<sub>mlm</sub>)**         | Retain language modeling capabilities      | Cross-entropy with true masked tokens     |
| **Cosine Loss (L<sub>cos</sub>)**      | Align internal representations             | 1 - Cosine similarity between hidden state |

This triple loss strategy enables DistilBERT to achieve approximately 97% of BERT's performance while being 40% smaller and 60% fastr.





## The Process

So we understand what MLM is doing, but how does this actually work? What are the logical steps that we’ll need to follow in code?

1.We  **tokenize**  our text. Just like we usually would with transformers, we begin with text tokenization.

From tokenization we will receive three different tensors:

-   _input_ids_
-   _token_type_ids_
-   _attention_mask_

We don’t need  _token_type_ids_  for MLM — and in this example  _attention_mask_  is not so important.

For us, the  _input_ids_  tensor is most important. Here, we will have a tokenized representation of our text — which is what we will be modifying moving forwards.

2.Create a  **_labels_**  tensor. We’re training our model here, so we need a  _labels_  tensor to calculate loss against — and optimize towards.

The  _labels_  tensor is simply  _input_ids_  — so all we need to do is make a copy.

3.**Mask** tokens in  _input_ids_. Now that we’ve created a copy of  _input_ids_  for  _labels_, we can go ahead and mask a random selection of tokens.

The BERT paper uses a 15% probability of masking each token during model pre-training, with a few additional rules — we’ll use a simplified version of this and assign a 15% probability of each word being masked.

4.Calculate  **loss**. We process the  _input_ids_  and  _labels_  tensors through our BERT model and calculate the loss between them both.

Using this loss, we calculate the required gradient changes with BERT — and optimize our model weights.

> The loss is calculated as the difference between the output probability distributions for each output ‘token’, and the true one-hot encoded  _labels_.

## Residual Connection and Layer Normalization
- Skip connections or residual connections are used to allow gradients to flow through a network directly, ***without passing through non-linear activation functions***.
-  **Non-linear activation functions, by nature of being non-linear, cause the gradients to explode or vanish (depending on the weights).**

###  internal covariate shift
**Normalization** *helps with the problem called internal covariate shift*. 

> Internal covariate shift refers to covariate shift occurring within a
> neural network, i.e. going from (say) layer 2 to layer 3. the network
> learns and the weights are updated, the distribution of outputs of a
> specific layer in the network changes. This forces the higher layers
> to adapt to that drift, which slows down learning.

###  Batch normalization
- Batch normalization normalizes the input features across the batch dimension.
- -In batch normalization, the statistics are computed across the batch and are the same for each example in the batch.

### lAYER nOMALUZATION
- the key feature of layer normalization is that it normalizes the inputs across the features.
-  in layer normalization, the statistics are computed across each feature and are independent of other examples.
