# Evaluating LLM and RAG Systems
1.  **Retriever:** Finds relevant information from a knowledge source (database, documents).
2.  **Generator (LLM):** Uses the retrieved information and the original query to generate a final answer.

**Retrieval-Augmented Generation (RAG)** combines an information retrieval module with a text generation model. Hence, evaluation must measure:
- 🔍 **Retrieval Quality** (Was the right context retrieved?)
- ✍️ **Generation Quality** (Was the answer fluent, relevant, and correct?)

---


## 📚 Retrieval Evaluation Metrics

### 1. **Recall@K**

**Formula:**
$$
\text{Recall@K} = \frac{\text{Number of relevant documents in top K}}{\text{Total number of relevant documents}}
$$ 

**Example:**
- Total relevant documents for a query: 5
- Top 3 retrieved documents: 2 are relevant 

$$
\text{Recall@3} = \frac{2}{5} = 0.4
$$ 

### 2. **Precision@K**

**Formula:**
$$
\text{Precision@K} = \frac{\text{Number of relevant documents in top K}}{K}
$$ ([Metrics for Information Retrieval - Fabian Hertwig's Blog](https://fabianhertwig.com/blog/information-retrieval-metrics/?utm_source=chatgpt.com))

**Example:**
- Top 3 retrieved documents: 2 are relevant

$$
\text{Precision@3} = \frac{2}{3} \approx 0.6667
$$

### 3. **Mean Reciprocal Rank (MRR)**

**Formula:**
$$
\text{MRR} = \frac{1}{N} \sum_{i=1}^{N} \frac{1}{\text{rank}_i}
$$

Where:
- \( N $$is the total number of queries
- \( \text{rank}_i $$is the rank position of the first relevant document for the \( i^{th} $$query

**Example:**
- Query 1: First relevant document at rank 2 → \( \frac{1}{2} \)
- Query 2: First relevant document at rank 1 → \( 1 \)
- Query 3: First relevant document at rank 4 → \( \frac{1}{4} \)

$$
\text{MRR} = \frac{1}{3} \left( \frac{1}{2} + 1 + \frac{1}{4} \right) = \frac{1}{3} \times \frac{7}{4} = \frac{7}{12} \approx 0.5833
$$

### 4. **Normalized Discounted Cumulative Gain (nDCG@K)**

**Formula:**
$$
\text{nDCG@K} = \frac{\text{DCG@K}}{\text{IDCG@K}}
$$ 
Where:
- DCG@K (Discounted Cumulative Gain) is:

$$
\text{DCG@K} = \sum_{i=1}^{K} \frac{2^{\text{rel}_i} - 1}{\log_2(i + 1)}
$$ 

- IDCG@K is the ideal DCG (maximum possible DCG) for the top K documents

**Example:**
- Top 3 retrieved documents with relevance scores: [3, 2, 1]

$$
\text{DCG@3} = \frac{2^3 - 1}{\log_2(1 + 1)} + \frac{2^2 - 1}{\log_2(2 + 1)} + \frac{2^1 - 1}{\log_2(3 + 1)} = \frac{7}{1} + \frac{3}{1.58496} + \frac{1}{2} \approx 7 + 1.8928 + 0.5 = 9.3928
$$

- Assuming this is the ideal ranking, IDCG@3 = DCG@3 

$$
\text{nDCG@3} = \frac{9.3928}{9.3928} = 1.0
$$





## 🔍 2. Retrieval Evaluation Metrics

| Metric          | Description                                                                 | Good For                       | Limitation                         |
|------------------|-----------------------------------------------------------------------------|----------------------------------|-------------------------------------|
| **Recall@k**     | % of times ground truth document is in top-k retrieved docs                 | Recall-focused tasks             | Doesn’t consider ranking order      |
| **Precision@k**  | % of top-k retrieved docs that are relevant                                 | High-precision requirements      | May penalize models for redundancy |
| **Mean Reciprocal Rank (MRR)** | 1/rank of first relevant document                                          | Prioritizing early correct hits  | Sensitive to only first correct hit |
| **Normalized Discounted Cumulative Gain (nDCG)** | Takes into account ranking + relevance scores                              | Graded relevance (e.g., 3 vs 5)  | Needs relevance annotations         |
| **Hit@k**        | Binary: 1 if relevant doc in top-k, else 0                                   | Binary relevance tasks           | Doesn't measure ranking quality     |

📌 **Use Case Tip**:  
- For **open-domain QA**, use `Recall@k` and `nDCG`.  
- For **fact-checking**, `Precision@k` and `MRR` are more critical.

# Preparation Guide: The ROUGE Metric

## 1. What is ROUGE?

*   **ROUGE** stands for **R**ecall-**O**riented **U**nderstudy for **G**isting **E**valuation.
*   It's a set of metrics used primarily for evaluating **automatic summarization** and sometimes machine translation.
*   **Core Idea:** ROUGE measures the quality of a system-generated summary by comparing it to one or more ideal "reference" summaries (usually human-written). It counts the number of overlapping units (like n-grams, word sequences) between the system summary and the reference summaries.

## 2. How Does it Work? (Recall Focus)

*   ROUGE primarily focuses on **recall**: how much of the information present in the reference summaries is captured by the system summary?
*   The general formula emphasizes recall:
    `ROUGE Score = (Number of Overlapping Units) / (Total Number of Units in Reference Summary)`

## 3. Key ROUGE Variants

*   **ROUGE-N:** Measures the overlap of **n-grams** (sequences of N words).
    *   `ROUGE-1`: Overlap of **unigrams** (single words). Measures word frequency overlap.
    *   `ROUGE-2`: Overlap of **bigrams** (pairs of adjacent words). Better reflects phrase overlap and local coherence.
*   **ROUGE-L:** Measures the **Longest Common Subsequence (LCS)**. It finds the longest sequence of words that appear in both the system and reference summaries, *in the same order*, but not necessarily consecutively. This better captures sentence-level structure.
*   **ROUGE-S/SU:** Measures overlap of **skip-bigrams** (word pairs allowing gaps) or skip-bigrams plus unigrams (SU). Less commonly used than ROUGE-N/L.

***Highlight:*** `ROUGE-1 (unigram), ROUGE-2 (bigram), and ROUGE-L (LCS) are the most commonly reported variants.`

## 4. Simple Example (ROUGE-1 Recall)

Let's calculate ROUGE-1 Recall:

*   **Reference Summary:** `the quick brown fox jumps over the lazy dog`
    *   *Unigrams:* {the: 2, quick: 1, brown: 1, fox: 1, jumps: 1, over: 1, lazy: 1, dog: 1} -> Total: 9 unigrams
*   **System Summary:** `the quick fox jumps over the dog`
    *   *Unigrams:* {the: 2, quick: 1, fox: 1, jumps: 1, over: 1, dog: 1}
*   **Overlapping Unigrams:** {the: 2, quick: 1, fox: 1, jumps: 1, over: 1, dog: 1} -> Total: 7 overlapping unigrams

*   **ROUGE-1 Recall** = (Number of overlapping unigrams) / (Total unigrams in *Reference*)
    *   = 7 / 9
    *   ≈ 0.78

***Interpretation:*** `The system summary captured approximately 78% of the words present in the reference summary.`

*(Note: ROUGE calculations often also include Precision and F1-Score, but the core ROUGE concept emphasizes Recall).*

## 5. Usage and Limitations

*   **Primary Usage:** Evaluating text summarization tasks where capturing key content from the reference is important. Also used in machine translation.
*   **Advantages:**
    *   Automatic and computationally inexpensive.
    *   Correlates reasonably well (especially ROUGE-2, ROUGE-L) with human judgments of summary *content/informativeness*.
    *   Standardized and widely used, making comparisons easier.
*   **Limitations:**
    *   **Doesn't measure:** Fluency, coherence, readability, or factual accuracy. A summary can have a high ROUGE score but be grammatically poor or nonsensical.
    *   **Requires Reference Summaries:** Depends heavily on having high-quality human references, which can be subjective and expensive to create.
    *   **Poor at Semantics:** Penalizes valid paraphrasing. Two summaries meaning the same thing but using different words will get a low ROUGE score.
    *   **Sensitivity:** Can be influenced by factors like summary length.

***Highlight:*** `ROUGE is a useful tool for assessing content overlap in summaries but should NOT be the sole metric for evaluating overall summary quality. Combine it with other metrics or human evaluation.`

# BLEU Metric

## 1. What is BLEU?

*   **BLEU** stands for **B**ilingual **E**valuation **U**nderstudy.
*   It's a metric primarily used for evaluating the quality of **Machine Translation (MT)** output, but also sometimes used for other text generation tasks (like summarization, dialogue).


## 2. How Does it Work? (Precision Focus)

*   BLEU calculates` the precision` of n-grams (sequences of 1, 2, 3, and typically 4 words) found in the candidate text compared to their presence in the reference text(s).
*   It combines these n-gram precisions `(usually via a geometric mean).`
*   It includes a **Brevity Penalty (BP)** to penalize candidate translations that are much shorter than their references, as precision alone would unfairly favor short, incomplete outputs.

## 3. Key Components

*   **N-gram Precision (Modified):** Calculates the proportion of candidate n-grams (typically for n=1 to 4) that appear in *any* of the reference translations. The counts are "clipped" – an n-gram in the candidate can only be counted as many times as its maximum frequency in *any single* reference. This prevents rewarding over-generation of common words/phrases.
*   **Brevity Penalty (BP):**
    *   If candidate length > reference length, BP = 1 (no penalty).
    *   If candidate length <= reference length, `BP = exp(1 - reference_length / candidate_length)`. This penalty increases exponentially as the candidate gets shorter than the reference.
*   **Final Score:** BLEU = BP * geometric_mean(precision_1, precision_2, ..., precision_N)

## 4. Simple Example (Conceptual)

*   **Reference 1:** `the cat sat on the mat`
*   **Reference 2:** `there is a cat on the mat`
*   **Candidate:** `the the the cat`

*   **Unigram Precision (Clipped):** 
- "the" appears 3 times in candidate, max 2 times in Ref 1, max 1 time in Ref 2 -> max count is 2.
-  "cat" appears 1 time in candidate, max 1 time in refs. Clipped count = 2 ("the") + 1 ("cat") = 3. 
- Candidate length = 4. Precision-1 = 3/4.
*   **Bigram Precision (Clipped):** "the the" appears 2 times in candidate, max 0 times in refs. "the cat" appears 1 time in candidate, max 1 time in refs. Clipped count = 0 + 1 = 1. Candidate bigrams = 3. Precision-2 = 1/3.
*   **Brevity Penalty:** Candidate length (4) is shorter than reference lengths (6). BP will be < 1.
*   **BLEU Score:** Will be low due to poor higher-order n-gram matches and the brevity penalty.

## 5. Usage and Limitations

*   **Primary Usage:** Standard benchmark for Machine Translation evaluation at the corpus level.
*   **Advantages:**
    *   Fast, cheap, language-independent, and widely used.
    *   Correlates moderately well with human judgments of fluency/adequacy at the *corpus* level.
*   **Limitations:**
    *   Poor correlation with human judgments at the *sentence* level.
    *   Penalizes legitimate paraphrasing and synonym use heavily.
    *   `Doesn't directly measure recall (how much of the reference is covered).`
    *   Can be "gamed" by systems producing repetitive but precise n-grams.

***Highlight:*** `BLEU is a precision-focused metric good for comparing MT systems across large datasets, but poor at judging semantic meaning or sentence-level quality.`

# METEOR Metric

## 1. What is METEOR?

*   **METEOR** stands for **M**etric for **E**valuation of **T**ranslation with **E**xplicit **OR**dering.
*   `It's designed to evaluate machine translation (and other generation tasks) by addressing some of BLEU's shortcomings.`
*   **Core Idea:** METEOR creates an alignment between the candidate text and reference text(s) based on **exact word matches, stems, and synonyms**. It then calculates scores based on this alignment, explicitly considering both precision and recall.

## 2. How Does it Work? (Alignment & F-Score Focus)

*   METEOR aligns unigrams between the candidate and reference. An alignment prioritizes exact matches, then stemmed matches (using a stemmer like Porter), then synonym matches (using WordNet or similar resources).
*   Based on these alignments, it calculates unigram 
	* **Precision** (aligned words / total words in candidate) and 
	* **Recall** (aligned words / total words in reference).
*   It combines Precision (P) and Recall (R) using a **harmonic mean (F-mean)**, typically weighting recall higher than precision (e.g., `F_mean = (10*P*R) / (R + 9*P)`).
*   It applies a **Fragmentation Penalty** to penalize alignments where the matched words are not in contiguous chunks or in the same order in both sentences.

## 3. Key Components

*   **Alignment Module:** Matches words based on exact form, stemmed form, and synonymy (requires external resources like WordNet).
*   **Precision & Recall Calculation:** Based on the number of aligned unigrams relative to sentence lengths.
*   **Harmonic Mean:** Combines P and R, often favouring recall.
*   **Fragmentation Penalty:** Penalizes gaps and word order differences in the alignment. `Penalty = (num_chunks / num_aligned_unigrams)^alpha`. Fewer chunks = lower penalty.

## 4. Simple Example (Conceptual)

*   **Reference:** `the quick brown dog jumps`
*   **Candidate:** `the speedy brown canine leaps`

*   **Alignment:**
    *   `the` <-> `the` (exact)
    *   `brown` <-> `brown` (exact)
    *   `quick` <-> `speedy` (synonym via WordNet)
    *   `dog` <-> `canine` (synonym via WordNet)
    *   `jumps` <-> `leaps` (synonym via WordNet)
*   **P & R:** All words align, so Precision = 5/5 = 1, Recall = 5/5 = 1. F-mean = 1.
*   **Fragmentation:** The aligned words are in the same order, likely forming one contiguous chunk. Penalty is minimal.
*   **METEOR Score:** Will be high, close to 1.0, because it recognizes the synonyms. BLEU would score this much lower.

## 5. Usage and Limitations

*   **Primary Usage:** Machine Translation, summarization, paraphrase generation, image captioning.
*   **Advantages:**
    *   Correlates better with human judgments than BLEU, especially at the sentence level.
    *   Explicitly considers recall alongside precision.
    *   Handles synonyms and stemming.
    *   Includes penalty for incorrect word order (fragmentation).
*   **Limitations:**
    *   More complex computation than BLEU.
    *   Relies on language-specific resources (stemmers, WordNet) which may not exist for all languages.
    *   Still primarily focused on word-level matching, though more flexible than strict n-grams.

***Highlight:*** `METEOR improves upon BLEU by incorporating synonyms, stemming, recall, and word order, generally providing better correlation with human quality assessments.`

#  BERTScore

## 1. What is BERTScore?

*   **BERTScore** is an automatic evaluation metric for text generation tasks.
*   It leverages pre-trained contextual embeddings from models like **BERT** (Bidirectional Encoder Representations from Transformers).


## 2. How Does it Work? (Semantic Similarity Focus)

*   It represents each token (word or subword) in both the candidate and reference sentences as a high-dimensional vector (contextual embedding) obtained from a pre-trained transformer model (e.g., BERT, RoBERTa).
*   It computes the **cosine similarity** between each token in the candidate and each token in the reference.
*   It **greedily matches** each candidate token to the most similar reference token (for recall) and each reference token to the most similar candidate token (for precision).
*   It calculates overall **Precision, Recall, and F1-score** based on these similarity scores. Optionally, scores can be weighted by Inverse Document Frequency (IDF) to emphasize important words.

## 3. Key Components

*   **Contextual Embeddings:** Uses pre-trained models (like BERT) to get vector representations that capture word meaning *in context*.
*   **Cosine Similarity:** Measures the similarity between embedding vectors.
*   **Greedy Matching:** Finds the best semantic match for each token across the candidate and reference sentences.
*   **Precision, Recall, F1 Calculation:** Aggregates the similarity scores of the matched tokens. (BERTScore-P, BERTScore-R, BERTScore-F1).

## 4. Simple Example (Conceptual)

*   **Reference:** `the puppy is adorable and playful`
*   **Candidate:** `the little dog is cute and active`

*   **Embeddings:** BERT generates vectors for each word. The vector for "puppy" will be contextually similar to "dog"; "adorable" similar to "cute"; "playful" similar to "active".
*   **Similarity Matching:** Cosine similarity between ("puppy", "dog"), ("adorable", "cute"), ("playful", "active") will be high. Matches for "the", "is", "and" will also be high (exact matches). "little" doesn't have a strong match in the reference.
*   **Scores:** BERTScore-R will be high because most reference words have good semantic matches in the candidate. BERTScore-P will be slightly lower because "little" in the candidate doesn't strongly match anything in the reference. BERTScore-F1 combines these.

## 5. Usage and Limitations

*   **Primary Usage:** Evaluating tasks where semantic equivalence is important: translation, summarization, dialogue, paraphrase generation, image captioning.
*   **Advantages:**
    *   Captures `semantic similarity much better than n-gram metrics (ROUGE, BLEU).`
    *   `Robust to paraphrasing and use of synonyms.`
    *   Often shows `higher correlation with human judgments regarding` semantic adequacy than BLEU/ROUGE.
    *   No reliance on external knowledge bases like WordNet (uses pre-trained model knowledge).
*   **Limitations:**
    *   Computationally more expensive than n-gram metrics (requires running inference with a large transformer model).
    *   Results can depend on the specific pre-trained model used.
    *   Doesn't inherently measure factual correctness or fluency directly (though semantic errors or disfluencies might lower the score).
    *   May give high scores to semantically plausible but factually incorrect statements if the reference is also flawed or limited.

***Highlight:*** `BERTScore excels at measuring semantic similarity using contextual embeddings, overcoming lexical limitations of older metrics and correlating well with human judgment on meaning.`

# Normalized Discounted Cumulative Gain (NDCG)

## 1. What is NDCG?

*   **NDCG** stands for **N**ormalized **D**iscounted **C**umulative **G**ain.
*   It's a metric used to evaluate the quality of **ranked lists**, most commonly in Information Retrieval (search engine results) and Recommender Systems.
*   **Core Idea:** NDCG assesses a ranked list based on two principles:
    1.  Highly relevant items are more valuable than marginally relevant items.
    2.  Relevant items appearing higher in the ranking (earlier positions) are more valuable than those appearing lower down.
    *   It provides a score between 0.0 and 1.0, where 1.0 represents the ideal ranking.

## 2. How Does it Work? (Building Blocks)

NDCG is built upon several simpler concepts:

*   **Cumulative Gain (CG):** The sum of the relevance scores of the items in the list up to a certain rank `K`.
    *   Formula: `CG@K = Σ (from i=1 to K) rel_i`
    *   *Limitation:* Treats all positions equally; doesn't care about the order.
*   **Discounted Cumulative Gain (DCG):** Modifies CG by applying a logarithmic discount based on rank position `i`. Items ranked lower have their relevance scores "discounted" more heavily.
    *   Formula: `DCG@K = Σ (from i=1 to K) (rel_i / log2(i+1))`
    *   *Alternative Formula (emphasizes highly relevant items more):* `DCG@K = Σ (from i=1 to K) ((2^rel_i - 1) / log2(i+1))`
    *   *Limitation:* DCG scores aren't directly comparable across queries with different numbers of relevant items.
*   **Ideal Discounted Cumulative Gain (IDCG):** The maximum possible DCG score for a given query up to rank `K`. This is calculated by imagining the *perfect* ranking where all relevant items are sorted by relevance score (highest first) at the top of the list.
*   **Normalized Discounted Cumulative Gain (NDCG):** The final metric, obtained by dividing the actual DCG score by the ideal DCG score. This normalizes the score to the range [0, 1].
    *   Formula: `NDCG@K = DCG@K / IDCG@K`

## 3. Key Components Recap

*   **Relevance Scores (`rel_i`):** Numerical scores assigned to each item indicating its relevance to the query (e.g., 0=irrelevant, 1=fair, 2=good, 3=perfect).
*   **Rank Position (`i`):** The position of an item in the ranked list (1, 2, 3,...).
*   **Discount (`log2(i+1)`):** Reduces the contribution of items at lower ranks.
*   **Normalization (dividing by IDCG):** Allows for fair comparison across different queries or lists of different lengths.

## 4. Simple Example (NDCG@5)

Suppose for a query, we get 5 results with the following relevance scores (0=min, 3=max):
`Ranked List: [DocA(3), DocB(1), DocC(2), DocD(0), DocE(2)]`

Let's calculate NDCG@5 using `DCG = Σ (rel_i / log2(i+1))`:

1.  **Calculate DCG@5 for the Ranked List:**
    *   Pos 1 (DocA): `3 / log2(1+1) = 3 / 1 = 3.0`
    *   Pos 2 (DocB): `1 / log2(2+1) = 1 / 1.585 = 0.631`
    *   Pos 3 (DocC): `2 / log2(3+1) = 2 / 2 = 1.0`
    *   Pos 4 (DocD): `0 / log2(4+1) = 0 / 2.322 = 0.0`
    *   Pos 5 (DocE): `2 / log2(5+1) = 2 / 2.585 = 0.774`
    *   **DCG@5 = 3.0 + 0.631 + 1.0 + 0.0 + 0.774 = 5.405**

2.  **Determine the Ideal Ranking @5:** Sort the available documents by relevance: `[DocA(3), DocC(2), DocE(2), DocB(1), DocD(0)]`

3.  **Calculate IDCG@5 (DCG of the Ideal Ranking):**
    *   Pos 1 (DocA): `3 / log2(1+1) = 3 / 1 = 3.0`
    *   Pos 2 (DocC): `2 / log2(2+1) = 2 / 1.585 = 1.262`
    *   Pos 3 (DocE): `2 / log2(3+1) = 2 / 2 = 1.0`
    *   Pos 4 (DocB): `1 / log2(4+1) = 1 / 2.322 = 0.431`
    *   Pos 5 (DocD): `0 / log2(5+1) = 0 / 2.585 = 0.0`
    *   **IDCG@5 = 3.0 + 1.262 + 1.0 + 0.431 + 0.0 = 5.693**

4.  **Calculate NDCG@5:**
    *   `NDCG@5 = DCG@5 / IDCG@5 = 5.405 / 5.693 ≈ 0.95`

***Interpretation:*** `The generated ranking achieved about 95% of the potential score attainable by the ideal ranking for the top 5 positions.`

## 5. Usage and Limitations

*   **Primary Usage:** Evaluating search engine results, product recommendations, and other systems that produce ranked lists. Standard metric in information retrieval benchmarks.
*   **Advantages:**
    *   Accounts for graded relevance (not just relevant/irrelevant).
    *   Position-aware (rewards relevant items ranked higher).
    *   Normalized score allows comparison across queries.
*   **Limitations:**
    *   Requires reliable, often graded, relevance judgments, which can be expensive and time-consuming to obtain.
    *   Doesn't explicitly measure or reward result *diversity* or *novelty*.
    *   Can be less intuitive to interpret than simpler metrics like Precision@K or Recall@K.

***Highlight:*** `NDCG is considered the most informative standard metric for evaluating ranked lists when graded relevance judgments are available, as it balances relevance level and rank position.`

## 📚 Retrieval Evaluation Metrics

### 1. **Recall@K**

**Formula:**
$$
\text{Recall@K} = \frac{\text{Number of relevant documents in top K}}{\text{Total number of relevant documents}}
$$ 

**Example:**
- Total relevant documents for a query: 5
- Top 3 retrieved documents: 2 are relevant 

$$
\text{Recall@3} = \frac{2}{5} = 0.4
$$ 

### 2. **Precision@K**

**Formula:**
$$
\text{Precision@K} = \frac{\text{Number of relevant documents in top K}}{K}
$$ ([Metrics for Information Retrieval - Fabian Hertwig's Blog](https://fabianhertwig.com/blog/information-retrieval-metrics/?utm_source=chatgpt.com))

**Example:**
- Top 3 retrieved documents: 2 are relevant

$$
\text{Precision@3} = \frac{2}{3} \approx 0.6667
$$

### 3. **Mean Reciprocal Rank (MRR)**

**Formula:**
$$
\text{MRR} = \frac{1}{N} \sum_{i=1}^{N} \frac{1}{\text{rank}_i}
$$

Where:
- \( N $$is the total number of queries
- \( \text{rank}_i $$is the rank position of the first relevant document for the \( i^{th} $$query

**Example:**
- Query 1: First relevant document at rank 2 → \( \frac{1}{2} \)
- Query 2: First relevant document at rank 1 → \( 1 \)
- Query 3: First relevant document at rank 4 → \( \frac{1}{4} \)

$$
\text{MRR} = \frac{1}{3} \left( \frac{1}{2} + 1 + \frac{1}{4} \right) = \frac{1}{3} \times \frac{7}{4} = \frac{7}{12} \approx 0.5833
$$

### 4. **Normalized Discounted Cumulative Gain (nDCG@K)**

**Formula:**
$$
\text{nDCG@K} = \frac{\text{DCG@K}}{\text{IDCG@K}}
$$ 
Where:
- DCG@K (Discounted Cumulative Gain) is:

$$
\text{DCG@K} = \sum_{i=1}^{K} \frac{2^{\text{rel}_i} - 1}{\log_2(i + 1)}
$$ 

- IDCG@K is the ideal DCG (maximum possible DCG) for the top K documents

**Example:**
- Top 3 retrieved documents with relevance scores: [3, 2, 1]

$$
\text{DCG@3} = \frac{2^3 - 1}{\log_2(1 + 1)} + \frac{2^2 - 1}{\log_2(2 + 1)} + \frac{2^1 - 1}{\log_2(3 + 1)} = \frac{7}{1} + \frac{3}{1.58496} + \frac{1}{2} \approx 7 + 1.8928 + 0.5 = 9.3928
$$

- Assuming this is the ideal ranking, IDCG@3 = DCG@3 

$$
\text{nDCG@3} = \frac{9.3928}{9.3928} = 1.0
$$

## ✍️ Generation Evaluation Metrics

### 1. **BLEU (Bilingual Evaluation Understudy)**

**Formula:**
$$
\text{BLEU} = \text{BP} \times \exp\left( \sum_{n=1}^{N} w_n \log p_n \right)
$$ 

Where:
- \( p_n $$is the precision for n-grams
- \( w_n $$is the weight for each n-gram (commonly uniform)
- BP is the brevity penalty to penalize short translations ([Demystifying the BLEU Metric: A Comprehensive Guide to Machine ...](https://www.traceloop.com/blog/demystifying-the-bleu-metric?utm_source=chatgpt.com))

**Example:**
- Candidate: "the cat is on the mat"
- Reference: "there is a cat on the mat" ([Two minutes NLP — Learn the ROUGE metric by examples - Medium](https://medium.com/nlplanet/two-minutes-nlp-learn-the-rouge-metric-by-examples-f179cc285499?utm_source=chatgpt.com))

Calculate unigram precision, bigram precision, etc., apply brevity penalty, and compute BLEU score accordingly.

### 2. **ROUGE-L (Recall-Oriented Understudy for Gisting Evaluation - Longest Common Subsequence)**

**Formula:**
$$
\text{ROUGE-L} = \frac{(1 + \beta^2) \cdot \text{Precision} \cdot \text{Recall}}{\text{Recall} + \beta^2 \cdot \text{Precision}}
$$ ([Evaluation measures (information retrieval)](https://en.wikipedia.org/wiki/Evaluation_measures_%28information_retrieval%29?utm_source=chatgpt.com))

Where:
- Precision = LCS / length of candidate
- Recall = LCS / length of reference
- \( \beta $$is a parameter to balance precision and recall (commonly set to 1) ([Two minutes NLP — Learn the ROUGE metric by examples - Medium](https://medium.com/nlplanet/two-minutes-nlp-learn-the-rouge-metric-by-examples-f179cc285499?utm_source=chatgpt.com), [BERTScore in AI: Transforming Semantic Text Evaluation and Quality](https://www.galileo.ai/blog/bert-score-explained-guide?utm_source=chatgpt.com))

**Example:**
- Candidate: "the cat sat on the mat"
- Reference: "the cat is on the mat"
- LCS: "the cat on the mat" (length 5)
- Candidate length: 6, Reference length: 6

$$
\text{Precision} = \frac{5}{6}, \quad \text{Recall} = \frac{5}{6}
$$

## ✍️ Generation Evaluation Metrics (Continued)

### 3. **METEOR (Metric for Evaluation of Translation with Explicit ORdering)**

**Formula:**

1. **Precision (P):** $$
P = \frac{m}{w_t}
$$ ([METEOR](https://en.wikipedia.org/wiki/METEOR?utm_source=chatgpt.com))

2. **Recall (R):** $$
R = \frac{m}{w_r}
$$

3. **F-mean:** $$
F_{mean} = \frac{10 \cdot P \cdot R}{R + 9P}
$$ ([METEOR](https://en.wikipedia.org/wiki/METEOR?utm_source=chatgpt.com))

4. **Penalty (p):** $$
p = 0.5 \left( \frac{c}{m} \right)^3
$$ ([METEOR](https://en.wikipedia.org/wiki/METEOR?utm_source=chatgpt.com))

5. **METEOR Score (M):** $$
M = F_{mean} \cdot (1 - p)
$$ ([METEOR](https://en.wikipedia.org/wiki/METEOR?utm_source=chatgpt.com))

Where:
- \( m $$is the number of matched unigrams
- \( w_t $$is the number of unigrams in the candidate
- \( w_r $$is the number of unigrams in the reference
- \( c $$is the number of chunks (contiguous matched sequences) ([Understanding BLEU and ROUGE score for NLP evaluation](https://www.geeksforgeeks.org/understanding-bleu-and-rouge-score-for-nlp-evaluation/?utm_source=chatgpt.com), [METEOR](https://en.wikipedia.org/wiki/METEOR?utm_source=chatgpt.com))

**Example:**
- Candidate: "the cat sat on the mat"
- Reference: "the cat is on the mat"
- Matched unigrams (m): 5
- Candidate length (w_t): 6
- Reference length (w_r): 6
- Chunks (c): 2 ([Evaluating Model Performance with the ROUGE Metric - Traceloop](https://www.traceloop.com/blog/evaluating-model-performance-with-the-rouge-metric-a-comprehensive-guide?utm_source=chatgpt.com))

Calculations:
- \( P = \frac{5}{6} \approx 0.833 \)
- \( R = \frac{5}{6} \approx 0.833 \)
- \( F_{mean} = \frac{10 \cdot 0.833 \cdot 0.833}{0.833 + 9 \cdot 0.833} \approx 0.833 \)
- \( p = 0.5 \left( \frac{2}{5} \right)^3 = 0.5 \cdot 0.064 = 0.032 \)
- \( M = 0.833 \cdot (1 - 0.032) \approx 0.806 $$([METEOR](https://en.wikipedia.org/wiki/METEOR?utm_source=chatgpt.com))

### 4. **BERTScore**

**Formula:**

BERTScore computes similarity between candidate and reference sentences using BERT embeddings.

1. **Precision (P):** $$
P = \frac{1}{|x|} \sum_{x_i \in x} \max_{y_j \in y} \text{cosine}(x_i, y_j)
$$

2. **Recall (R):** $$
R = \frac{1}{|y|} \sum_{y_j \in y} \max_{x_i \in x} \text{cosine}(y_j, x_i)
$$

3. **F1 Score:** $$
F1 = \frac{2PR}{P + R}
$$ ([Evaluating AI Text Summarization: Understanding the ROUGE Metric](https://www.galileo.ai/blog/rouge-metric?utm_source=chatgpt.com))

Where:
- \( x $$and \( y $$are the sets of tokens in the candidate and reference sentences, respectively
- \( x_i $$and \( y_j $$are the embeddings of the tokens ([BERTScore – A Powerful NLP Metric Explained & How To Tutorial](https://spotintelligence.com/2024/08/20/bertscore/?utm_source=chatgpt.com))

**Example:**
- Candidate: "the cat sat on the mat"
- Reference: "the cat is on the mat"

Compute BERT embeddings for each token, calculate the maximum cosine similarity for each token in the candidate against all tokens in the reference (for precision), and vice versa for recall. Then compute the F1 score.

---

### 5. **BLEURT**

**Description:**

BLEURT is a learned evaluation metric that uses a pre-trained BERT model fine-tuned on human judgments to predict the quality of a candidate sentence given a reference.

**Usage:**

Given a candidate and reference sentence, BLEURT outputs a score between 0 and 1, where higher scores indicate better quality.

**Example:**
- Candidate: "the cat sat on the mat"
- Reference: "the cat is on the mat"

Input both sentences into the BLEURT model to obtain a similarity score.

---

### 6. **COMET**

**Description:**

COMET is a learned metric that predicts the quality of a translation by considering the source sentence, the machine translation (candidate), and the reference translation.

**Usage:**

Input the source sentence, candidate translation, and reference translation into the COMET model to obtain a quality score.

**Example:**
- Source: "le chat est sur le tapis"
- Candidate: "the cat sat on the mat"
- Reference: "the cat is on the mat"

Input these into the COMET model to get a quality score.

---

### 7. **chrF++**

**Formula:**

chrF++ combines character n-gram F-scores with word n-gram F-scores.

$$
\text{chrF}_{\beta} = \frac{(1 + \beta^2) \cdot \text{Precision} \cdot \text{Recall}}{\beta^2 \cdot \text{Precision} + \text{Recall}}
$$

Where:
- Precision and Recall are computed over character n-grams and word n-grams
- \( \beta $$is a parameter to weight recall higher than precision (commonly set to 2) ([[PDF] chrF++: words helping character n-grams - Statmt.org](https://statmt.org/wmt17/pdf/WMT70.pdf?utm_source=chatgpt.com), [chrf.py - Hugging Face](https://huggingface.co/spaces/evaluate-metric/chrf/blob/0b3fdd5f0929b50864fed72b059bd6f72625036c/chrf.py?utm_source=chatgpt.com))

**Example:**
- Candidate: "the cat sat on the mat"
- Reference: "the cat is on the mat"

Compute character and word n-gram overlaps to calculate precision and recall, then apply the formula to get the chrF++ score.

| **ROUGE-L**     | Measures the longest common subsequence (LCS) between candidate and reference texts, focusing on recall. | $$\(\text{ROUGE-L} = \frac{(1 + \beta^2) \cdot \text{Precision} \cdot \text{Recall}}{\text{Recall} + \beta^2 \cdot \text{Precision}}\)$$ | *Candidate*: "the cat sat on the mat" <br> *Reference*: "the cat is on the mat" <br> LCS: "the cat on the mat" (length 5) <br> Precision = 5/6, Recall = 5/6, ROUGE-L ≈ 0.833 | Evaluating summarization and text generation tasks. |
| **METEOR**      | Considers exact, stem, synonym, and paraphrase matches between candidate and reference, combining precision and recall with a penalty for fragmentation. | $$\(\text{METEOR} = F_{mean} \cdot (1 - p)\)$$ <br> $$\(F_{mean} = \frac{10 \cdot P \cdot R}{R + 9P}\)$$ <br> $$\(p = 0.5 \left( \frac{c}{m} \right)^3\)$$ | *Candidate*: "the cat sat on the mat" <br> *Reference*: "the cat is on the mat" <br> Matched unigrams: 5, Chunks: 2 <br> Compute precision, recall, F_mean, penalty, and METEOR score. | Evaluating machine translation and paraphrase generation. |
| **BERTScore**   | Computes similarity between candidate and reference texts using contextual embeddings from BERT, capturing semantic similarity. | $$\(\text{Precision} = \frac{1}{|x|} \sum_{x_i \in x} \max_{y_j \in y} \cos(x_i, y_j)\)$$ <br> $$\(\text{Recall} = \frac{1}{|y|} \sum_{y_j \in y} \max_{x_i \in x} \cos(y_j, x_i)\)$$ <br> $$\(\text{F1} = \frac{2 \cdot \text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}\)$$ | *Candidate*: "the cat sat on the mat" <br> *Reference*: "the cat is on the mat" <br> Compute BERT embeddings, calculate precision, recall, and F1 score. | Evaluating text generation with a focus on semantic similarity. |
| **BLEURT**      | A learned evaluation metric based on BERT, fine-tuned on human judgments to predict the quality of text generation. | Outputs a score between 0 and 1 indicating the quality of the candidate sentence compared to the reference. | *Candidate*: "the cat sat on the mat" <br> *Reference*: "the cat is on the mat" <br> Input both sentences into BLEURT model to obtain a similarity score. | Evaluating text generation with a focus on human judgment correlation. |
| **COMET**       | A neural framework for evaluating machine translation, considering source, reference, and hypothesis to predict quality. | Outputs a score between 0 and 1 indicating the quality of the translation. | *Source*: "le chat est sur le tapis" <br> *Candidate*: "the cat sat on the mat" <br> *Reference*: "the cat is on the mat" <br> Input into COMET model to obtain a quality score. | Evaluating machine translation quality with contextual understanding. |
| **chrF++**      | Combines character n-gram F-score with word n-gram F-score to evaluate the similarity between candidate and reference texts. | $$\(\text{chrF}_{\beta} = \frac{(1 + \beta^2) \cdot \text{Precision} \cdot \text{Recall}}{\beta^2 \cdot \text{Precision} + \text{Recall}}\)$$ | *Candidate*: "the cat sat on the mat" <br> *Reference*: "the cat is on the mat" <br> Compute character and word n-gram overlaps, calculate precision and recall, then apply formula. | Evaluating machine translation, especially for morphologically rich languages. | ([BERTScore: A Contextual Metric for LLM Evaluation - Analytics Vidhya](https://www.analyticsvidhya.com/blog/2025/04/bertscore-a-contextual-metric-for-llm-evaluation/?utm_source=chatgpt.com), [The METEOR Metric — An NLP Classic | by Jorgecardete - Medium](https://medium.com/thedeephub/the-meteor-metric-an-nlp-classic-42552cb6ce69?utm_source=chatgpt.com), [ROUGE Metric: Definition & Examples | Vaia](https://www.vaia.com/en-us/explanations/engineering/artificial-intelligence-engineering/rouge-metric/?utm_source=chatgpt.com), [README.md · evaluate-metric/meteor at ... - Hugging Face](https://huggingface.co/spaces/evaluate-metric/meteor/blob/c6d7489d3f077a3f905e7897ad8d3110838e471e/README.md?utm_source=chatgpt.com), [BERTScore Explained in 5 minutes - Medium](https://medium.com/%40abonia/bertscore-explained-in-5-minutes-0b98553bfb71?utm_source=chatgpt.com), [[PDF] BLEURT: Learning Robust Metrics for Text Generation](https://aclanthology.org/2020.acl-main.704.pdf?utm_source=chatgpt.com), [COMET: a Neural Framework for State-of-the-art MT Evaluation](https://www.youtube.com/watch?v=FrOmIZSStvk&utm_source=chatgpt.com))

### **Level 1: Generator Agent Overview**

#### ✅ **Core Capabilities**
- **Text Generation:** Responds to prompts with coherent, fluent language.
- **Retrieval-Augmented Generation (RAG):** Can pull in external documents or knowledge from a vector store or database to improve answer accuracy.
- **Context Awareness (Limited):** Can handle short-term context within a session but lacks persistence across sessions.

#### ❌ **Key Limitations**
- **No Memory:** Cannot retain or evolve based on past interactions. Every prompt is treated independently unless explicitly provided with history.
- **No Planning:** Operates in a purely reactive mode, with no multi-step reasoning or goal-oriented behavior.
- **No Autonomy:** It doesn't initiate actions, set goals, or self-correct. All behavior is user-triggered.

#### 📐 **Typical Architecture Components**
1. **Input Interface (e.g., Chat UI / API call)**
2. **Preprocessing Layer:** Formatting, prompt templating, or embeddings if using RAG.
3. **LLM Core (e.g., GPT, Claude, PaLM)**
4. **Retrieval System (Optional):** Vector store like FAISS, Weaviate, or Pinecone.
5. **Postprocessing:** May include ranking, filtering, or formatting outputs.

---

### 🔧 Example Use Cases
- Customer support bots that handle FAQs.
- Assistants that answer questions using product documentation (via RAG).
- Simple task handlers like “summarize this email” or “generate a blog title.”









### **Level 2: Tool-Calling Agent Overview**

#### ✅ **Core Capabilities**
- **Tool Invocation:** Can decide *when* and *how* to call external tools, such as:
  - APIs (e.g., weather, booking systems)
  - Code interpreters or calculators
  - Web search/browsing
  - Databases or internal services
- **Dynamic Execution:** Uses responses from tool calls in real-time to guide answers or continue workflows.
- **Multi-step Tasks (Limited):** Can complete multi-part commands using tool chains (e.g., "search flights → book cheapest one").

#### ❌ **Key Limitations**
- **Still Reactive:** Acts only in response to user prompts—doesn’t self-initiate.
- **No Long-Term Memory:** Doesn’t retain context beyond the session unless a memory system is manually integrated.
- **Minimal Self-Correction:** Won’t reflect or retry unless explicitly told to (e.g., "try again").

---

### 🧠 **Architectural Components**
1. **User Input Handler**
2. **Orchestrator Layer:**
   - Decides if tool invocation is needed
   - May include planner logic (simple chains or decision trees)
3. **Tool Interfaces:**
   - API wrappers, database connectors, custom plugins
4. **LLM Core (GPT, Claude, etc.):**
   - Handles natural language reasoning and generation
5. **Response Synthesizer:**
   - Merges tool outputs into natural language

---

### 📌 **Example Agents**
- **Web-Browsing Agents:** Use tools to search and cite real-time web info.
- **Coding Assistants (e.g., Copilot):** Call compilers, linters, and file systems.
- **AI Research Assistants:** Integrate PDFs, web sources, and citation tools.
- **AI Workflow Orchestrators:** Trigger Slack messages, calendar events, etc.

---

### 🔄 **Still Missing in Action**

|--------------------------|----------|

| Persistent Memory        | ❌       |
| Adaptive Planning        | ⚠️ (rudimentary chaining) |

### **Level 3: Planning Agent Overview**

#### ✅ **Core Capabilities**
- **Task Decomposition:** Breaks high-level goals into subtasks (e.g., via recursive reasoning or tree search).
- **Outcome Evaluation:** Assesses the success of tool use and adapts accordingly (e.g., retrying failed API calls).
- **Dynamic Sequencing:** Adjusts task order based on context, success, or newly discovered information.
- **Looping & Branching:** Can retry, skip, or replan steps based on intermediate results (basic control flow logic).

---

### ❌ **Limitations**
- **Still Reactive:** Operates only when prompted—no background or proactive execution.
- **No Persistence Across Sessions:** Once a plan completes, the system resets unless explicitly re-invoked.
- **No Long-Term Learning or Memory:** Cannot improve strategies over time without external feedback loops.

---

### 🧠 **Planner Agent Architecture**
1. **Input Interpreter / Goal Parser:** Understands the user’s overall intent.
2. **Planner Module:**
   - Generates task graph or sequence
   - May use few-shot or chain-of-thought prompting to plan steps
3. **Executor:**
   - Calls tools/APIs for each step
   - Monitors outcomes and state transitions
4. **Evaluator:**
   - Checks if each action succeeded
   - Triggers replanning or fallback strategies if needed
5. **LLM Core:** Drives reasoning, planning, and synthesis.

---

### 📌 **Example Use Cases**
- **AI Debugger:** Reads logs → identifies bug → edits code → reruns tests → confirms fix.
- **Workflow Orchestrator:** Checks inventory → places orders → confirms delivery → alerts stakeholders.
- **Data Pipeline Agent:** Downloads data → validates schema → transforms → loads into DB.

## 🔍 **LLM Agent Evaluation Dimensions**

### 1️⃣ **Tool Correctness**
Assesses **whether the right tools were called**, and **how correctly they were used**.

**Key Dimensions:**
- **Tool Selection**: Did the agent select the correct tools?
- **Input Parameters**: Were parameters passed into the tools accurate?
- **Output Accuracy**: Was the tool’s output correct relative to expected result?

📌 *Partial correctness allowed:*  
- Order may not matter → use set comparison.  
- Some params may be incorrect → % of correct params.

🧪 **Code Example**
```python
from deepeval.metrics import ToolCorrectnessMetric
from deepeval.test_case import LLMTestCase, ToolCallParams, ToolCall

test_case = LLMTestCase(
    input="What if these shoes don't fit?",
    actual_output="We offer a 30-day full refund.",
    tools_called=[ToolCall(name="WebSearchTool"), ToolCall(name="QueryTool")],
    expected_tools=[ToolCall(name="WebSearchTool")]
)

metric = ToolCorrectnessMetric(
    evaluation_params=[ToolCallParams.TOOL, ToolCallParams.INPUT_PARAMETERS],
    should_consider_ordering=False
)
metric.measure(test_case)
print(metric.score, metric.reason)
```

---

### 2️⃣ **Tool Efficiency**
Evaluates **how optimal the tool usage path was**, given available tools.

**Key Metrics:**
- ✅ **Redundant Tool Usage** – Unnecessary tools used?
- ✅ **Tool Frequency** – Tools called more times than needed?
- ✅ **Goal-oriented Efficiency** – Was the tool trajectory optimal for the user goal?

🧪 **Code Example**
```python
from deepeval.metrics import ToolEffiencyMetric
from deepeval.test_case import LLMTestCase, ToolCall

test_case = LLMTestCase(
    input="What if these shoes don't fit?",
    actual_output="We offer a 30-day full refund.",
    tools_called=[ToolCall(name="WebSearchTool")],
)

metric = ToolEffiencyMetric(
    available_tools=[ToolCall(name="WebSearchTool"), ToolCall(name="QueryTool")]
)
metric.measure(test_case)
print(metric.score, metric.reason)
```

> 🧠 *LLM-as-Judge* approach can be used for more complex multi-step tasks to evaluate trajectory relevance and efficiency.

## 🧩 Why Evaluate Agentic Workflows?

While **Tool Correctness** and **Tool Efficiency** tell us **how** the agent used tools, **Agentic Workflow Evaluation** answers a higher-order question:

> ✅ **Did the agent actually complete the task successfully?**

Agentic workflows include:
1. Understanding the task from the user’s prompt.
2. Planning a reasoning chain (if applicable).
3. Calling tools or taking steps.
4. Producing a final result that solves the user's need.

## 🎯 **Metric: Task Completion (a.k.a. Goal Accuracy)**

This metric evaluates **end-to-end success**—from user input to final output—**with reasoning context and tool usage included**.

### ✅ Applicable Scenarios
| Task Type                    | Task Completion Definition                                   |
|-----------------------------|--------------------------------------------------------------|
| Web Shopping Agent          | Did the agent buy the correct product (based on attributes)?|
| Travel Planner Agent        | Did it create a valid itinerary matching input criteria?     |
| Debugging Agent             | Did it fix the bug and verify the fix?                      |
| General Chat-based Agents   | Was the response appropriate and actionable?                |

---

## ⚙️ How Task Completion Works in **DeepEval**

### 1. **LLM-Derived Task Understanding**  
   Infers what the task is from the input string.

### 2. **Tool-Use + Reasoning Assessment**  
   Evaluates tools used, steps taken, and checks final outcome.

### 3. **No Ground Truth Needed**  
   Uses a language model (e.g., GPT-4) to **judge correctness based on reasoning**, even when predefined outputs are unavailable.

---

## 🧪 **Code Example**

```python
from deepeval import evaluate

    threshold=0.7,   # Minimum acceptable success score
    model="gpt-4",   # Model used to reason and judge completion
    include_reason=True
)

print(metric.score)
print(metric.reason)
```
