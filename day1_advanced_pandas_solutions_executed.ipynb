```python
import pandas as pd
import numpy as np
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)
pd.set_option("display.max_colwidth", 80)
```

## Create the dataframe

```python
df = pd.DataFrame({
    "emp_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Helen", "Ian", "Jane"],
    "dept": ["Data", "Data", "Web", "Web", "Data", "ML", "ML", "Web", "Data", "ML"],
    "city": ["Bangalore", "Bangalore", "Delhi", "Delhi", "Mumbai", "Bangalore", "Mumbai", "Delhi", "Mumbai", "Bangalore"],
    "salary": [120000, 95000, 110000, 105000, 130000, 150000, 140000, 99000, 125000, 155000],
    "bonus": [10000, 8000, 12000, 7000, 15000, 20000, 18000, 6000, 14000, 22000],
    "rating": [5, 3, 4, 2, 5, 4, 5, 3, 4, 5],
    "experience": [6, 3, 5, 2, 7, 8, 6, 3, 5, 9],
    "joined": ["2020-01-15", "2021-06-10", "2019-09-23", "2022-03-01", "2018-11-12",
               "2017-07-19", "2020-08-25", "2021-01-11", "2019-05-30", "2016-12-05"],
    "skills": [
        ["python", "sql", "pandas"],
        ["python", "excel"],
        ["javascript", "react"],
        ["javascript", "css"],
        ["python", "sql", "spark"],
        ["python", "ml", "deep-learning"],
        ["python", "ml", "sql"],
        ["html", "css", "javascript"],
        ["python", "sql", "tableau"],
        ["python", "ml", "nlp"]
    ],
    "projects": [
        [{"name": "fraud", "hours": 120}, {"name": "churn", "hours": 80}],
        [{"name": "sales_dashboard", "hours": 60}],
        [{"name": "website_revamp", "hours": 100}],
        [{"name": "landing_page", "hours": 40}, {"name": "seo_fix", "hours": 30}],
        [{"name": "recommendation", "hours": 150}],
        [{"name": "forecasting", "hours": 200}, {"name": "vision", "hours": 90}],
        [{"name": "classification", "hours": 170}],
        [{"name": "frontend_migration", "hours": 110}],
        [{"name": "reporting", "hours": 95}, {"name": "automation", "hours": 50}],
        [{"name": "llm_search", "hours": 220}, {"name": "chatbot", "hours": 140}]
    ]
})

df["joined"] = pd.to_datetime(df["joined"])
df
```

## 8. Top 2 salaries in each department

```python
q8 = df.groupby("dept", group_keys=False).apply(lambda g: g.nlargest(2, "salary"))
q8[["dept", "emp_id", "name", "salary"]]
```

## 9. Departments with average rating above 4

```python
q9 = df.groupby("dept").filter(lambda g: g["rating"].mean() > 4)
q9[["dept", "emp_id", "name", "rating"]]
```

## 10. Salary range by department

```python
q10 = df.groupby("dept", as_index=False).agg(
    salary_range=("salary", lambda x: x.max() - x.min())
)
q10
```

## 11. Weighted compensation

```python
df["total_comp"] = df["salary"] + df["bonus"]
q11 = df.groupby("dept", as_index=False).agg(
    avg_total_comp=("total_comp", "mean"),
    max_total_comp=("total_comp", "max")
)
q11
```

## 12. Multi-condition performance flag

```python
conditions = [
    (df["rating"] >= 5) & (df["salary"] >= 130000),
    (df["rating"] >= 4)
]
choices = ["star", "good"]
df["performance_flag"] = np.select(conditions, choices, default="average")
df[["emp_id", "name", "rating", "salary", "performance_flag"]]
```

## 13. Employees whose salary is above their department median

```python
df["dept_median_salary"] = df.groupby("dept")["salary"].transform("median")
df["above_dept_median"] = df["salary"] > df["dept_median_salary"]
df[["emp_id", "name", "dept", "salary", "dept_median_salary", "above_dept_median"]]
```

## 14. Rank salary within each department

```python
df["salary_rank_in_dept"] = df.groupby("dept")["salary"].rank(method="min", ascending=False)
df[["dept", "name", "salary", "salary_rank_in_dept"]].sort_values(["dept", "salary_rank_in_dept", "salary"], ascending=[True, True, False])
```

## 15. Dense rank by rating within each department

```python
df["rating_rank_dense"] = df.groupby("dept")["rating"].rank(method="dense", ascending=False)
df[["dept", "name", "rating", "rating_rank_dense"]].sort_values(["dept", "rating_rank_dense", "name"])
```

## 16. Find employees with more than 2 skills

```python
q16 = df[df["skills"].apply(len) > 2]
q16[["emp_id", "name", "dept", "skills"]]
```

## 17. Explode skills into rows

```python
skills_long = df[["emp_id", "name", "dept", "skills"]].explode("skills").rename(columns={"skills": "skill"}).reset_index(drop=True)
skills_long.head(15)
```

## 18. Count how many employees have each skill

```python
q18 = skills_long["skill"].value_counts().rename_axis("skill").reset_index(name="employee_count")
q18
```

## 19. Department-wise unique skill count

```python
q19 = skills_long.groupby("dept", as_index=False).agg(unique_skill_count=("skill", "nunique"))
q19
```

## 20. Find employees who have both python and sql

```python
q20 = df[df["skills"].apply(lambda x: {"python", "sql"}.issubset(set(x)))]
q20[["emp_id", "name", "dept", "skills"]]
```

## 21. Explode projects into rows

```python
projects_long = df[["emp_id", "name", "dept", "projects"]].explode("projects").reset_index(drop=True)
projects_long.head(12)
```

## 22. Flatten project dictionaries

```python
project_details = pd.json_normalize(projects_long["projects"]).rename(columns={"name": "project_name"})
projects_flat = pd.concat([projects_long.drop(columns="projects"), project_details], axis=1)
projects_flat.head(12)
```

## 23. Total project hours per employee

```python
q23 = projects_flat.groupby(["emp_id", "name"], as_index=False).agg(total_project_hours=("hours", "sum"))
q23.sort_values("total_project_hours", ascending=False)
```

## 24. Department-wise total project hours

```python
q24 = projects_flat.groupby("dept", as_index=False).agg(total_project_hours=("hours", "sum"))
q24
```

## 25. Average project hours per project per department

```python
q25 = projects_flat.groupby(["dept", "project_name"], as_index=False).agg(avg_project_hours=("hours", "mean"))
q25
```

## 26. Most time-consuming project for each department

```python
q26 = projects_flat.sort_values(["dept", "hours"], ascending=[True, False]).drop_duplicates("dept")
q26[["dept", "emp_id", "project_name", "hours"]]
```

## 27. Pivot table: average salary by dept and city

```python
q27 = df.pivot_table(index="dept", columns="city", values="salary", aggfunc="mean")
q27
```

## 28. Pivot table: count of employees by dept and rating

```python
q28 = pd.crosstab(df["dept"], df["rating"])
q28
```

## 29. Latest joined employee in each department

```python
q29 = df.sort_values(["dept", "joined"], ascending=[True, False]).drop_duplicates("dept")
q29[["dept", "emp_id", "name", "joined"]]
```

## 30. Year-wise joining count

```python
q30 = df.assign(join_year=df["joined"].dt.year).groupby("join_year", as_index=False).agg(employee_count=("emp_id", "count"))
q30
```

## 31. Employees above city average salary

```python
df["city_avg_salary"] = df.groupby("city")["salary"].transform("mean")
df["above_city_avg_salary"] = df["salary"] > df["city_avg_salary"]
df[["emp_id", "name", "city", "salary", "city_avg_salary", "above_city_avg_salary"]]
```

## 32. Employees who are top earners in both city and department

```python
df["city_max_salary"] = df.groupby("city")["salary"].transform("max")
df["dept_max_salary"] = df.groupby("dept")["salary"].transform("max")
q32 = df[(df["salary"] == df["city_max_salary"]) & (df["salary"] == df["dept_max_salary"])]
q32[["emp_id", "name", "dept", "city", "salary"]]
```

## 33. Conditional department summary

```python
q33 = df.groupby("dept", as_index=False).agg(
    employees_salary_gt_120k=("salary", lambda x: (x > 120000).sum()),
    employees_bonus_ge_15k=("bonus", lambda x: (x >= 15000).sum()),
    avg_rating_exp_ge_5=("rating", lambda x: x[df.loc[x.index, "experience"] >= 5].mean())
)
q33
```

## 34. Department skill heat problem: top 2 most common skills

```python
skill_counts = skills_long.groupby(["dept", "skill"], as_index=False).size().rename(columns={"size": "count"})
q34 = skill_counts.sort_values(["dept", "count", "skill"], ascending=[True, False, True]).groupby("dept", group_keys=False).head(2)
q34
```

## 35. Complex final department summary

```python
# normal department summary
base_summary = df.groupby("dept", as_index=False).agg(
    employee_count=("emp_id", "count"),
    avg_salary=("salary", "mean"),
    avg_bonus=("bonus", "mean"),
    avg_rating=("rating", "mean"),
    high_performer_count=("rating", lambda x: (x >= 4).sum())
)

# employees with python skill
python_skill_count = (
    skills_long.assign(has_python=lambda x: x["skill"].eq("python"))
    .groupby(["dept", "emp_id"], as_index=False)["has_python"].max()
    .groupby("dept", as_index=False)
    .agg(python_skill_employee_count=("has_python", "sum"))
)

# project hour summaries
project_summary = projects_flat.groupby("dept", as_index=False).agg(
    total_project_hours=("hours", "sum"),
    max_single_project_hours=("hours", "max")
)

q35 = (
    base_summary
    .merge(python_skill_count, on="dept", how="left")
    .merge(project_summary, on="dept", how="left")
)
q35
```

## Notes

- `agg` returns one row per group.
- `transform` returns the same number of rows as the original dataframe.
- `filter` keeps or removes whole groups.
- `explode` turns list values into separate rows.
- `json_normalize` flattens dictionaries into columns.
