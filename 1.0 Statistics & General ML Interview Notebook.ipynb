```python
from __future__ import annotations

import math
import random
from statistics import fmean, pstdev

RNG = random.Random(21)

def percentile(values: list[float], q: float) -> float:
    ordered = sorted(values)
    if not ordered:
        raise ValueError("percentile() requires at least one value")
    pos = (len(ordered) - 1) * (q / 100)
    lo = math.floor(pos)
    hi = math.ceil(pos)
    if lo == hi:
        return float(ordered[lo])
    lower = ordered[lo]
    upper = ordered[hi]
    return float(lower + (upper - lower) * (pos - lo))

try:
    import matplotlib
    matplotlib.use("Agg")
    import matplotlib.pyplot as plt
    try:
        plt.style.use("seaborn-v0_8-whitegrid")
    except OSError:
        pass
    HAS_MATPLOTLIB = True
except ModuleNotFoundError:
    plt = None
    HAS_MATPLOTLIB = False
    print("matplotlib not installed; visual cells will print numeric summaries instead of plots.")
```

<li style="margin:6px 0;">Use <strong>Poisson</strong> when the count itself is random over a fixed interval.</li>

```python
x = [-4 + 8 * i / 399 for i in range(400)]
normal_pdf = [(1 / math.sqrt(2 * math.pi)) * math.exp(-(val**2) / 2) for val in x]
exp_x = [5 * i / 399 for i in range(400)]
exp_pdf = [math.exp(-val) for val in exp_x]

k_pois = list(range(13))
lam = 4
poisson_pmf = [math.exp(-lam) * (lam**k) / math.factorial(k) for k in k_pois]

k_bin = list(range(11))
n, p = 10, 0.5
binom_pmf = [math.comb(n, k) * (p**k) * ((1 - p) ** (n - k)) for k in k_bin]

if HAS_MATPLOTLIB:
    fig, axes = plt.subplots(2, 2, figsize=(12, 8))

    axes[0, 0].plot(x, normal_pdf, color="#2563eb", lw=2.5)
    axes[0, 0].set_title("Normal")

    axes[0, 1].bar(k_pois, poisson_pmf, color="#ea580c", alpha=0.85)
    axes[0, 1].set_title("Poisson (lambda=4)")

    axes[1, 0].bar(k_bin, binom_pmf, color="#16a34a", alpha=0.85)
    axes[1, 0].set_title("Binomial (n=10, p=0.5)")

    axes[1, 1].plot(exp_x, exp_pdf, color="#9333ea", lw=2.5)
    axes[1, 1].set_title("Exponential (lambda=1)")

    for ax in axes.flat:
        ax.set_xlabel("Value")
        ax.set_ylabel("Density / probability")

    fig.suptitle("Distributions worth recognizing instantly in interviews", fontsize=14, y=1.02)
    fig.tight_layout()
    plt.show()
else:
    print({
        "normal_peak": round(max(normal_pdf), 4),
        "poisson_mean_var": (lam, lam),
        "binomial_mean": round(n * p, 2),
        "binomial_var": round(n * p * (1 - p), 2),
        "exponential_mean": 1.0,
    })
```

<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Why divide by <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">n-1</code> in sample variance?</strong></p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">Because the sample mean is estimated from the same data, which uses up one degree of freedom.</li>
<li style="margin:6px 0;">This Bessel correction makes the sample variance a better estimator of population variance.</li>
</ul>
<h3 style="display:inline-block;color:#d8e4ea;background:#22303a;border-left:5px solid #7fa89e;padding:8px 14px;border-radius:12px;border:1px solid #3a4a52;margin:18px 0 10px 0;">Variance vs covariance vs correlation</h3>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;"><strong>Variance</strong>: spread of one variable</li>
<li style="margin:6px 0;"><strong>Covariance</strong>: whether two variables move together, but still in scale-dependent units</li>
<li style="margin:6px 0;"><strong>Correlation</strong>: normalized covariance, so it is easier to compare across features</li>
</ul>
<h3 style="display:inline-block;color:#d8e4ea;background:#22303a;border-left:5px solid #7fa89e;padding:8px 14px;border-radius:12px;border:1px solid #3a4a52;margin:18px 0 10px 0;">When correlation can mislead</h3>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">Nonlinear relationship: Pearson can be near zero while a real signal exists</li>
<li style="margin:6px 0;">Confounding: both variables move with a hidden third factor</li>
<li style="margin:6px 0;">Aggregation effects: Simpson's paradox can reverse the story</li>
</ul>
<h3 style="display:inline-block;color:#d8e4ea;background:#22303a;border-left:5px solid #7fa89e;padding:8px 14px;border-radius:12px;border:1px solid #3a4a52;margin:18px 0 10px 0;">Interview line worth memorizing</h3>
<blockquote style="margin:14px 0;padding:12px 16px;background:#24313d;border-left:5px solid #90b7a2;border-radius:14px;border:1px solid #3c4d5a;color:#d9e4ec;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">&quot;Correlation is a fast linear signal check, not proof of causality and not a complete measure of dependence.&quot;</p>
</blockquote>
</div>


<div data-notebook-cell-style-version="2026-04-v2" style="color:#d7e0e7;line-height:1.8;font-family:ui-sans-serif,system-ui,-apple-system,BlinkMacSystemFont,'Segoe UI',sans-serif;">
<h2 style="background:linear-gradient(90deg,#22313d 0%,#273743 100%);border-left:8px solid #72aa9b;color:#e7eff3;padding:11px 16px;border-radius:14px;border:1px solid #3a4b58;box-shadow:0 6px 18px rgba(0,0,0,0.18);margin:22px 0 14px 0;">Sampling and the Central Limit Theorem</h2>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>CLT in one sentence</strong></p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">As the sample size grows, the sampling distribution of the sample mean tends toward a normal distribution, even if the raw data are not normal.</li>
</ul>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>What this does <em>not</em> mean</strong></p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">It does <strong>not</strong> mean the original dataset becomes normal.</li>
<li style="margin:6px 0;">It does <strong>not</strong> rescue clearly broken assumptions in every setting.</li>
<li style="margin:6px 0;">It does <strong>not</strong> imply <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">n = 30</code> is always enough for extremely skewed or heavy-tailed data.</li>
</ul>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Why interviewers care</strong></p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">It justifies why averages often behave predictably</li>
<li style="margin:6px 0;">It explains why confidence intervals and many mean-based tests become practical</li>
<li style="margin:6px 0;">It gives you a way to talk about sample size, uncertainty, and precision in the same answer</li>
</ul>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Good applied answer</strong></p>
<blockquote style="margin:14px 0;padding:12px 16px;background:#24313d;border-left:5px solid #90b7a2;border-radius:14px;border:1px solid #3c4d5a;color:#d9e4ec;">
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;">&quot;CLT is why I can reason about the sampling distribution of a mean even when the feature itself is skewed. But I would still check sample size, tail behavior, and whether the statistic I care about is actually the mean.&quot;</p>
</blockquote>
</div>


```python
original = [RNG.expovariate(1.0) for _ in range(5000)]
sample_sizes = [5, 20, 50]

summaries = {}
mean_samples = {}

for n in sample_sizes:
    means = [fmean(RNG.choices(original, k=n)) for _ in range(3000)]
    mean_samples[n] = means
    summaries[n] = {
        "mean_of_means": round(fmean(means), 4),
        "std_of_means": round(pstdev(means), 4),
    }

if HAS_MATPLOTLIB:
    fig, axes = plt.subplots(1, 4, figsize=(16, 3.8))
    axes[0].hist(original, bins=40, color="#0f766e", alpha=0.85)
    axes[0].set_title("Original data\n(exponential, skewed)")

    for ax, n in zip(axes[1:], sample_sizes):
        ax.hist(mean_samples[n], bins=35, color="#2563eb", alpha=0.85)
        ax.set_title(f"Sample means\nn={n}")
        ax.set_xlabel("mean")

    for ax in axes:
        ax.set_ylabel("frequency")

    fig.suptitle("CLT intuition: the mean stabilizes faster than the raw data", fontsize=14, y=1.05)
    fig.tight_layout()
    plt.show()
else:
    print({
        "raw_mean": round(fmean(original), 4),
        "raw_std": round(pstdev(original), 4),
        "sample_mean_summaries": summaries,
    })
```

<div data-notebook-cell-style-version="2026-04-v2" style="color:#d7e0e7;line-height:1.8;font-family:ui-sans-serif,system-ui,-apple-system,BlinkMacSystemFont,'Segoe UI',sans-serif;">
<h2 style="background:linear-gradient(90deg,#22313d 0%,#273743 100%);border-left:8px solid #72aa9b;color:#e7eff3;padding:11px 16px;border-radius:14px;border:1px solid #3a4b58;box-shadow:0 6px 18px rgba(0,0,0,0.18);margin:22px 0 14px 0;">Hypothesis Testing Without Common Mistakes</h2>
<table style="width:100%;border-collapse:collapse;background:#ffffff;font-size:0.96rem;margin:16px 0;border:1px solid #dde6ec;border-radius:16px;overflow:hidden;">
<thead>
<tr>
<th style="background:#f3f6f8;color:#24415c;padding:10px 12px;text-align:left;border:1px solid #dde6ec;font-weight:700;">Term</th>
<th style="background:#f3f6f8;color:#24415c;padding:10px 12px;text-align:left;border:1px solid #dde6ec;font-weight:700;">What to say</th>
<th style="background:#f3f6f8;color:#24415c;padding:10px 12px;text-align:left;border:1px solid #dde6ec;font-weight:700;">What not to say</th>
</tr>
</thead>
<tbody>
<tr style="background:#f8fbff;">
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>Null hypothesis (<code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">H0</code>)</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">A baseline claim, often &quot;no effect&quot; or &quot;no difference.&quot;</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">&quot;The null is probably false because my model is better.&quot;</td>
</tr>
<tr>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>Alternative (<code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">H1</code>)</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">The competing claim: some effect, difference, or association exists.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">&quot;This proves causality.&quot;</td>
</tr>
<tr style="background:#f8fbff;">
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>p-value</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Probability of data this extreme assuming <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">H0</code> is true.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">&quot;Probability the null is true.&quot;</td>
</tr>
<tr>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>alpha</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Pre-chosen false-positive tolerance.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">&quot;Magic significance threshold that decides importance.&quot;</td>
</tr>
<tr style="background:#f8fbff;">
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>effect size</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Magnitude of the difference or association.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">&quot;Optional extra after significance.&quot;</td>
</tr>
<tr>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>confidence interval</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Plausible range for the parameter under repeated-sampling logic.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">&quot;95% chance the true value is in this exact interval.&quot;</td>
</tr>
<tr style="background:#f8fbff;">
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>power</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Probability of detecting a real effect.</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">&quot;Same thing as confidence level.&quot;</td>
</tr>
</tbody>
</table>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Rules that make answers stronger</strong></p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">Pair <strong>p-value + effect size + confidence interval</strong></li>
<li style="margin:6px 0;">Mention whether the result is <strong>practically meaningful</strong>, not just statistically significant</li>
<li style="margin:6px 0;">Tie the test choice back to <strong>data type, independence, sample size, and shape</strong></li>
</ul>
<p style="margin:10px 0 15px 0;line-height:1.82;color:#d7e0e7;font-size:1rem;"><strong>Common interview trap</strong></p>
<ul style="margin:10px 0 16px 20px;padding-left:18px;line-height:1.9;color:#d2dce4;">
<li style="margin:6px 0;">&quot;We got <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">p &lt; 0.001</code>, so the change is important.&quot;</li>
<li style="margin:6px 0;">Better answer: significance can happen for trivial lifts when sample size is huge; I still need effect size, uncertainty bounds, and business impact.</li>
</ul>
</div>


```python
baseline = [1 if RNG.random() < 0.72 else 0 for _ in range(2500)]
challenger = [1 if RNG.random() < 0.75 else 0 for _ in range(2500)]

observed_lift = fmean(challenger) - fmean(baseline)
bootstrap = []

for _ in range(3000):
    a = fmean(RNG.choices(baseline, k=len(baseline)))
    b = fmean(RNG.choices(challenger, k=len(challenger)))
    bootstrap.append(b - a)

ci_low, ci_high = percentile(bootstrap, 2.5), percentile(bootstrap, 97.5)

if HAS_MATPLOTLIB:
    plt.figure(figsize=(9, 4))
    plt.hist(bootstrap, bins=40, color="#475569", alpha=0.85)
    plt.axvline(observed_lift, color="#16a34a", lw=2, label=f"observed lift = {observed_lift:.3f}")
    plt.axvline(ci_low, color="#dc2626", ls="--", lw=2, label=f"95% CI low = {ci_low:.3f}")
    plt.axvline(ci_high, color="#dc2626", ls="--", lw=2, label=f"95% CI high = {ci_high:.3f}")
    plt.title("Bootstrap view of uncertainty for an A/B-style model comparison")
    plt.xlabel("challenger - baseline accuracy")
    plt.ylabel("frequency")
    plt.legend()
    plt.tight_layout()
    plt.show()

print({"baseline_acc": round(fmean(baseline), 4),
       "challenger_acc": round(fmean(challenger), 4),
       "observed_lift": round(observed_lift, 4),
       "ci_95": (round(ci_low, 4), round(ci_high, 4))})
```

<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Check association between categorical variables</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>Chi-square test of independence</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Counts in a contingency table</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Expected counts should not be too small</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Use Fisher's exact test for tiny 2x2 tables.</td>
</tr>
<tr style="background:#f8fbff;">
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Check linear relationship</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>Pearson correlation</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Numeric variables, approximately linear relationship</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Sensitive to outliers and nonlinearity</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Use scatter plots, not coefficient alone.</td>
</tr>
<tr>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Check monotonic relationship</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>Spearman correlation</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Ranks matter more than exact distance</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Monotonic trend</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Better than Pearson for monotonic but nonlinear relationships.</td>
</tr>
<tr style="background:#f8fbff;">
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Check normality</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>Shapiro-Wilk + QQ / histogram</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Small to medium samples where formal check helps</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Very large <code style="background:#27313a;color:#e8eef4;padding:2px 6px;border-radius:6px;border:1px solid #3c4a55;">n</code> can reject tiny departures</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Do not use a p-value-only normality decision.</td>
</tr>
<tr>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Check equal variances</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;"><strong>Levene's test</strong></td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Before mean-comparison workflows</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">Independence</td>
<td style="padding:10px 12px;border:1px solid #e4edf2;vertical-align:top;color:#31475b;">More robust than Bartlett when normality is questionable.</td>
</tr>
</tbody>
</table>

## Interview Assessment: Sharp Questions, Not Vague Ones

These are the kinds of questions that expose whether the understanding is real.

<details>
<summary><strong>1. Your p-value is 0.03. What exactly can you conclude?</strong></summary>

You can say the observed data would be fairly unlikely under the null hypothesis. You cannot say there is a 3% probability the null is true. A strong follow-up answer also mentions effect size, confidence interval, and whether the result matters in practice.
</details>

<details>
<summary><strong>2. When does CLT help, and when is "n = 30" too simplistic?</strong></summary>

CLT helps when the statistic of interest is the sample mean and the sample size is large enough for the sampling distribution to stabilize. `n = 30` is only a rule of thumb; highly skewed or heavy-tailed data may need more.
</details>

<details>
<summary><strong>3. Why might a statistically significant model improvement still be a bad launch?</strong></summary>

Because significance can appear for tiny effects with enough data. If latency, fairness, cost, or operational complexity worsen more than the lift helps, it is still a bad decision.
</details>

<details>
<summary><strong>4. How would you choose between Welch's t-test and Mann-Whitney U for two model variants?</strong></summary>

If the outcome is numeric and mean comparison is the right target, Welch's t-test is the default first choice when groups are independent and reasonably well-behaved. If the data are strongly skewed, ordinal, or dominated by outliers, Mann-Whitney U is safer as a rank-based alternative.
</details>

<details>
<summary><strong>5. Why report confidence intervals with A/B tests instead of only p-values?</strong></summary>

Because confidence intervals communicate uncertainty and the plausible effect range. They help decide whether the lift is practically large enough to matter.
</details>

<details>
<summary><strong>6. When would Pearson correlation miss a useful feature that mutual information could catch?</strong></summary>

When the feature has a nonlinear but real dependency with the target. Pearson only sees linear association, while mutual information captures broader dependence patterns.
</details>

<details>
<summary><strong>7. Why is correlation not enough for feature selection?</strong></summary>

Correlation is pairwise and linear. It misses nonlinear structure, interactions, leakage, redundancy across multiple features, and deployment constraints.
</details>

<details>
<summary><strong>8. What is a practical explanation of Type I vs Type II error in ML?</strong></summary>

Type I error means you ship a model change that is not actually better. Type II error means you miss a truly better model because your test lacked sensitivity or power.
</details>

<details>
<summary><strong>9. How would you explain why `n-1` appears in sample variance?</strong></summary>

Because one degree of freedom is spent estimating the sample mean first. Dividing by `n-1` corrects the tendency to underestimate population variance.
</details>

<details>
<summary><strong>10. ANOVA vs chi-square: what is the real difference?</strong></summary>

ANOVA compares means of a numeric outcome across groups. Chi-square tests association between categorical variables using counts in a contingency table. Same interview family, different data types and null hypotheses.
</details>
