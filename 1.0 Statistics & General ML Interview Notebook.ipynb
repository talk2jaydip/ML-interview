## Interview Answer Frame

A strong spoken answer is usually:

**1. Definition**
- What is it in one clean sentence?

**2. When to use it**
- What decision or scenario makes it relevant?

**3. Assumptions / limits**
- What must roughly be true for it to make sense?

**4. Practical caveat**
- What would you still check before acting on it?

Example:

> "A p-value is the probability of observing data this extreme, or more extreme, assuming the null hypothesis is true. I use it as evidence strength against the null, not as proof the null is false. I still pair it with effect size, confidence interval, and sample-size context because significance alone can be misleading."


## Fast Recall Cheat Sheet

| Topic | Interview-safe answer | ML / product relevance |
|---|---|---|
| **Normal distribution** | Continuous bell-shaped distribution defined by mean and variance. | Noise models, residual assumptions, z-scores, initialization intuition. |
| **Binomial distribution** | Number of successes in `n` independent Bernoulli trials with success probability `p`. | Click/no-click, fraud/not-fraud, correct/incorrect predictions. |
| **Poisson distribution** | Count of events in a fixed interval when events occur at an average rate `lambda`. | Requests per minute, incidents per hour, arrivals to a queue. |
| **Exponential distribution** | Time between Poisson events; memoryless. | Waiting times, uptime / inter-arrival gaps. |
| **Variance** | Average squared spread around the mean; hard to interpret because units are squared. | Outlier sensitivity, feature scaling, PCA intuition. |
| **Standard deviation** | Square root of variance; spread in the original units. | More interpretable uncertainty summary than variance alone. |
| **Covariance** | Direction of joint movement between two variables. | Feature relationships, but scale-dependent. |
| **Correlation** | Normalized linear association in `[-1, 1]`. | Quick signal check, multicollinearity screening. |
| **CLT** | The sampling distribution of the mean tends toward normal as sample size grows. | Why averages and many parametric procedures become usable at scale. |
| **Confidence interval** | A range produced by a procedure that captures the true parameter at the stated long-run rate. | Report uncertainty for lift, accuracy, CTR, latency. |
| **p-value** | Probability of data this extreme assuming the null is true. | Evidence against the null, not evidence that the null is false. |
| **Type I error** | False positive: rejecting a true null. | Shipping a model change that is not actually better. |
| **Type II error** | False negative: missing a real effect. | Failing to ship a truly better model. |
| **Power** | Probability of detecting a real effect. | Vital in experiment design and offline benchmark planning. |
| **Mutual information** | Nonlinear dependence score; zero means independence. | Useful when Pearson misses nonlinear signal. |


```python
from __future__ import annotations

import math
import random
from statistics import fmean, pstdev

RNG = random.Random(21)

def percentile(values: list[float], q: float) -> float:
    ordered = sorted(values)
    if not ordered:
        raise ValueError("percentile() requires at least one value")
    pos = (len(ordered) - 1) * (q / 100)
    lo = math.floor(pos)
    hi = math.ceil(pos)
    if lo == hi:
        return float(ordered[lo])
    lower = ordered[lo]
    upper = ordered[hi]
    return float(lower + (upper - lower) * (pos - lo))

try:
    import matplotlib
    matplotlib.use("Agg")
    import matplotlib.pyplot as plt
    try:
        plt.style.use("seaborn-v0_8-whitegrid")
    except OSError:
        pass
    HAS_MATPLOTLIB = True
except ModuleNotFoundError:
    plt = None
    HAS_MATPLOTLIB = False
    print("matplotlib not installed; visual cells will print numeric summaries instead of plots.")
```

## Core Distributions You Should Explain Cleanly

| Distribution | Type | What it models | Parameters | Interview notes |
|---|---|---|---|---|
| **Normal** | Continuous | Symmetric real-valued measurements | `mu`, `sigma^2` | Mention symmetry, bell shape, and why it matters for sampling distributions. |
| **Binomial** | Discrete | Number of successes in fixed trials | `n`, `p` | Say "fixed number of independent trials" explicitly. |
| **Poisson** | Discrete | Count of events in a fixed interval | `lambda` | Mean and variance are both `lambda`. |
| **Exponential** | Continuous | Time between Poisson events | `lambda` | Memoryless property is the memorable interview hook. |

**What interviewers often test**
- Whether you confuse **what is being counted** (events vs successes)
- Whether you know **discrete vs continuous**
- Whether you can connect a distribution to a believable business / ML example

**A clean contrast**
- Use **binomial** when the number of trials is fixed.
- Use **Poisson** when the count itself is random over a fixed interval.
- Use **exponential** when the question is about time between Poisson events.


```python
x = [-4 + 8 * i / 399 for i in range(400)]
normal_pdf = [(1 / math.sqrt(2 * math.pi)) * math.exp(-(val**2) / 2) for val in x]
exp_x = [5 * i / 399 for i in range(400)]
exp_pdf = [math.exp(-val) for val in exp_x]

k_pois = list(range(13))
lam = 4
poisson_pmf = [math.exp(-lam) * (lam**k) / math.factorial(k) for k in k_pois]

k_bin = list(range(11))
n, p = 10, 0.5
binom_pmf = [math.comb(n, k) * (p**k) * ((1 - p) ** (n - k)) for k in k_bin]

if HAS_MATPLOTLIB:
    fig, axes = plt.subplots(2, 2, figsize=(12, 8))

    axes[0, 0].plot(x, normal_pdf, color="#2563eb", lw=2.5)
    axes[0, 0].set_title("Normal")

    axes[0, 1].bar(k_pois, poisson_pmf, color="#ea580c", alpha=0.85)
    axes[0, 1].set_title("Poisson (lambda=4)")

    axes[1, 0].bar(k_bin, binom_pmf, color="#16a34a", alpha=0.85)
    axes[1, 0].set_title("Binomial (n=10, p=0.5)")

    axes[1, 1].plot(exp_x, exp_pdf, color="#9333ea", lw=2.5)
    axes[1, 1].set_title("Exponential (lambda=1)")

    for ax in axes.flat:
        ax.set_xlabel("Value")
        ax.set_ylabel("Density / probability")

    fig.suptitle("Distributions worth recognizing instantly in interviews", fontsize=14, y=1.02)
    fig.tight_layout()
    plt.show()
else:
    print({
        "normal_peak": round(max(normal_pdf), 4),
        "poisson_mean_var": (lam, lam),
        "binomial_mean": round(n * p, 2),
        "binomial_var": round(n * p * (1 - p), 2),
        "exponential_mean": 1.0,
    })
```

## Descriptive Statistics: What to Say Precisely

### Sample vs population

| Concept | Population | Sample |
|---|---|---|
| Mean | `mu` | `x_bar` |
| Variance | `sigma^2` (divide by `N`) | `s^2` (divide by `n-1`) |
| Standard deviation | `sigma` | `s` |

**Why divide by `n-1` in sample variance?**
- Because the sample mean is estimated from the same data, which uses up one degree of freedom.
- This Bessel correction makes the sample variance a better estimator of population variance.

### Variance vs covariance vs correlation
- **Variance**: spread of one variable
- **Covariance**: whether two variables move together, but still in scale-dependent units
- **Correlation**: normalized covariance, so it is easier to compare across features

### When correlation can mislead
- Nonlinear relationship: Pearson can be near zero while a real signal exists
- Confounding: both variables move with a hidden third factor
- Aggregation effects: Simpson's paradox can reverse the story

### Interview line worth memorizing
> "Correlation is a fast linear signal check, not proof of causality and not a complete measure of dependence."


## Sampling and the Central Limit Theorem

**CLT in one sentence**
- As the sample size grows, the sampling distribution of the sample mean tends toward a normal distribution, even if the raw data are not normal.

**What this does _not_ mean**
- It does **not** mean the original dataset becomes normal.
- It does **not** rescue clearly broken assumptions in every setting.
- It does **not** imply `n = 30` is always enough for extremely skewed or heavy-tailed data.

**Why interviewers care**
- It justifies why averages often behave predictably
- It explains why confidence intervals and many mean-based tests become practical
- It gives you a way to talk about sample size, uncertainty, and precision in the same answer

**Good applied answer**
> "CLT is why I can reason about the sampling distribution of a mean even when the feature itself is skewed. But I would still check sample size, tail behavior, and whether the statistic I care about is actually the mean."


```python
original = [RNG.expovariate(1.0) for _ in range(5000)]
sample_sizes = [5, 20, 50]

summaries = {}
mean_samples = {}

for n in sample_sizes:
    means = [fmean(RNG.choices(original, k=n)) for _ in range(3000)]
    mean_samples[n] = means
    summaries[n] = {
        "mean_of_means": round(fmean(means), 4),
        "std_of_means": round(pstdev(means), 4),
    }

if HAS_MATPLOTLIB:
    fig, axes = plt.subplots(1, 4, figsize=(16, 3.8))
    axes[0].hist(original, bins=40, color="#0f766e", alpha=0.85)
    axes[0].set_title("Original data\n(exponential, skewed)")

    for ax, n in zip(axes[1:], sample_sizes):
        ax.hist(mean_samples[n], bins=35, color="#2563eb", alpha=0.85)
        ax.set_title(f"Sample means\nn={n}")
        ax.set_xlabel("mean")

    for ax in axes:
        ax.set_ylabel("frequency")

    fig.suptitle("CLT intuition: the mean stabilizes faster than the raw data", fontsize=14, y=1.05)
    fig.tight_layout()
    plt.show()
else:
    print({
        "raw_mean": round(fmean(original), 4),
        "raw_std": round(pstdev(original), 4),
        "sample_mean_summaries": summaries,
    })
```

## Hypothesis Testing Without Common Mistakes

| Term | What to say | What not to say |
|---|---|---|
| **Null hypothesis (`H0`)** | A baseline claim, often "no effect" or "no difference." | "The null is probably false because my model is better." |
| **Alternative (`H1`)** | The competing claim: some effect, difference, or association exists. | "This proves causality." |
| **p-value** | Probability of data this extreme assuming `H0` is true. | "Probability the null is true." |
| **alpha** | Pre-chosen false-positive tolerance. | "Magic significance threshold that decides importance." |
| **effect size** | Magnitude of the difference or association. | "Optional extra after significance." |
| **confidence interval** | Plausible range for the parameter under repeated-sampling logic. | "95% chance the true value is in this exact interval." |
| **power** | Probability of detecting a real effect. | "Same thing as confidence level." |

**Rules that make answers stronger**
- Pair **p-value + effect size + confidence interval**
- Mention whether the result is **practically meaningful**, not just statistically significant
- Tie the test choice back to **data type, independence, sample size, and shape**

**Common interview trap**
- "We got `p < 0.001`, so the change is important."
- Better answer: significance can happen for trivial lifts when sample size is huge; I still need effect size, uncertainty bounds, and business impact.


```python
baseline = [1 if RNG.random() < 0.72 else 0 for _ in range(2500)]
challenger = [1 if RNG.random() < 0.75 else 0 for _ in range(2500)]

observed_lift = fmean(challenger) - fmean(baseline)
bootstrap = []

for _ in range(3000):
    a = fmean(RNG.choices(baseline, k=len(baseline)))
    b = fmean(RNG.choices(challenger, k=len(challenger)))
    bootstrap.append(b - a)

ci_low, ci_high = percentile(bootstrap, 2.5), percentile(bootstrap, 97.5)

if HAS_MATPLOTLIB:
    plt.figure(figsize=(9, 4))
    plt.hist(bootstrap, bins=40, color="#475569", alpha=0.85)
    plt.axvline(observed_lift, color="#16a34a", lw=2, label=f"observed lift = {observed_lift:.3f}")
    plt.axvline(ci_low, color="#dc2626", ls="--", lw=2, label=f"95% CI low = {ci_low:.3f}")
    plt.axvline(ci_high, color="#dc2626", ls="--", lw=2, label=f"95% CI high = {ci_high:.3f}")
    plt.title("Bootstrap view of uncertainty for an A/B-style model comparison")
    plt.xlabel("challenger - baseline accuracy")
    plt.ylabel("frequency")
    plt.legend()
    plt.tight_layout()
    plt.show()

print({"baseline_acc": round(fmean(baseline), 4),
       "challenger_acc": round(fmean(challenger), 4),
       "observed_lift": round(observed_lift, 4),
       "ci_95": (round(ci_low, 4), round(ci_high, 4))})
```

## Test Selection Matrix

| Goal | Recommended first choice | Use when | Key assumptions / checks | Practical note |
|---|---|---|---|---|
| Compare 2 independent means | **Welch's t-test** | Numeric outcome, roughly symmetric groups | Independence; t-test is fairly robust, especially with reasonable `n` | Prefer Welch over pooled t-test when variances may differ. |
| Compare paired measurements | **Paired t-test** | Before/after or same-unit repeated measure | Pairing is real and meaningful | Great for offline-vs-online or pre/post comparisons on matched units. |
| Compare 2 independent non-normal / ordinal groups | **Mann-Whitney U** | Skewed data, ordinal outcomes, or outlier-heavy data | Independent samples; interpret carefully as rank-based comparison | Good fallback when mean-based assumptions are weak. |
| Compare 3+ group means | **ANOVA** | Numeric outcome across 3 or more groups | Independence, variance assumptions, shape not wildly broken | If significant, follow with post-hoc comparisons. |
| Compare 3+ non-normal groups | **Kruskal-Wallis** | Rank-based comparison across several groups | Independent samples | Safer when ANOVA assumptions are poor. |
| Check association between categorical variables | **Chi-square test of independence** | Counts in a contingency table | Expected counts should not be too small | Use Fisher's exact test for tiny 2x2 tables. |
| Check linear relationship | **Pearson correlation** | Numeric variables, approximately linear relationship | Sensitive to outliers and nonlinearity | Use scatter plots, not coefficient alone. |
| Check monotonic relationship | **Spearman correlation** | Ranks matter more than exact distance | Monotonic trend | Better than Pearson for monotonic but nonlinear relationships. |
| Check normality | **Shapiro-Wilk + QQ / histogram** | Small to medium samples where formal check helps | Very large `n` can reject tiny departures | Do not use a p-value-only normality decision. |
| Check equal variances | **Levene's test** | Before mean-comparison workflows | Independence | More robust than Bartlett when normality is questionable. |


## ML-Linked Decisions Interviewers Expect You to Justify

### Feature scaling
- Usually important for distance-based or gradient-based models: k-NN, k-means, PCA, SVM, logistic regression, linear regression, neural nets.
- Usually less important for tree-based models: decision trees, random forests, gradient boosting.
- Strong answer: explain scaling as an optimization / geometry issue, not a ritual.

### Missing data
- **Low missingness**: simple imputation is often fine.
- **Structured missingness**: add a missing-indicator when "missing" itself may carry signal.
- **Very high missingness**: dropping the feature can be cleaner than inventing too much synthetic structure.

### Feature relevance metrics

| Metric | Best for | What it tells you | Limitation |
|---|---|---|---|
| **Pearson** | Numeric feature vs numeric target / feature | Linear signal | Misses nonlinear effects. |
| **Spearman** | Ranked or monotonic relationships | Monotonic signal | Still not a full dependence measure. |
| **Chi-square** | Categorical feature vs categorical target | Dependence in contingency tables | Count-based; expected cell counts matter. |
| **ANOVA F / `f_classif`** | Numeric feature vs categorical target | Whether group means differ | Captures mean separation, not arbitrary nonlinear dependence. |
| **Mutual information** | Mixed / nonlinear dependence | Any dependency pattern, not just linear | Harder to interpret as a direct effect size. |
| **VIF** | Sets of numeric predictors | Multicollinearity severity | Diagnostic, not a direct relevance score. |
| **WoE / IV** | Binary target scorecard-style problems | Ordered category risk signal | Domain-specific; not a universal default. |

**Strong interview line**
> "I choose the relevance metric based on feature type, target type, and whether I care about linear signal, monotonic signal, or any dependency at all."


## Interview Assessment: Sharp Questions, Not Vague Ones

These are the kinds of questions that expose whether the understanding is real.

<details>
<summary><strong>1. Your p-value is 0.03. What exactly can you conclude?</strong></summary>

You can say the observed data would be fairly unlikely under the null hypothesis. You cannot say there is a 3% probability the null is true. A strong follow-up answer also mentions effect size, confidence interval, and whether the result matters in practice.
</details>

<details>
<summary><strong>2. When does CLT help, and when is "n = 30" too simplistic?</strong></summary>

CLT helps when the statistic of interest is the sample mean and the sample size is large enough for the sampling distribution to stabilize. `n = 30` is only a rule of thumb; highly skewed or heavy-tailed data may need more.
</details>

<details>
<summary><strong>3. Why might a statistically significant model improvement still be a bad launch?</strong></summary>

Because significance can appear for tiny effects with enough data. If latency, fairness, cost, or operational complexity worsen more than the lift helps, it is still a bad decision.
</details>

<details>
<summary><strong>4. How would you choose between Welch's t-test and Mann-Whitney U for two model variants?</strong></summary>

If the outcome is numeric and mean comparison is the right target, Welch's t-test is the default first choice when groups are independent and reasonably well-behaved. If the data are strongly skewed, ordinal, or dominated by outliers, Mann-Whitney U is safer as a rank-based alternative.
</details>

<details>
<summary><strong>5. Why report confidence intervals with A/B tests instead of only p-values?</strong></summary>

Because confidence intervals communicate uncertainty and the plausible effect range. They help decide whether the lift is practically large enough to matter.
</details>

<details>
<summary><strong>6. When would Pearson correlation miss a useful feature that mutual information could catch?</strong></summary>

When the feature has a nonlinear but real dependency with the target. Pearson only sees linear association, while mutual information captures broader dependence patterns.
</details>

<details>
<summary><strong>7. Why is correlation not enough for feature selection?</strong></summary>

Correlation is pairwise and linear. It misses nonlinear structure, interactions, leakage, redundancy across multiple features, and deployment constraints.
</details>

<details>
<summary><strong>8. What is a practical explanation of Type I vs Type II error in ML?</strong></summary>

Type I error means you ship a model change that is not actually better. Type II error means you miss a truly better model because your test lacked sensitivity or power.
</details>

<details>
<summary><strong>9. How would you explain why `n-1` appears in sample variance?</strong></summary>

Because one degree of freedom is spent estimating the sample mean first. Dividing by `n-1` corrects the tendency to underestimate population variance.
</details>

<details>
<summary><strong>10. ANOVA vs chi-square: what is the real difference?</strong></summary>

ANOVA compares means of a numeric outcome across groups. Chi-square tests association between categorical variables using counts in a contingency table. Same interview family, different data types and null hypotheses.
</details>


## Applied Interview Drills

| Prompt | What a strong answer should include |
|---|---|
| **"Offline model A beats model B by 0.4 accuracy points. Are you done?"** | Variance across folds, confidence interval, practical lift, slice-level behavior, leakage checks, and business cost. |
| **"A feature has weak Pearson correlation but improves tree-based validation. Why?"** | Nonlinearity, threshold effects, interactions, and model class differences. |
| **"Your experiment has p > 0.05 but the lift direction is consistently positive."** | Possible underpower, effect size still worth examining, maybe extend test or perform power analysis. |
| **"How do you justify median imputation over mean?"** | Robustness to skew and outliers, especially for long-tailed numeric features. |
| **"When would you not standardize a feature?"** | Tree-based models, or cases where original units matter for interpretation and model family does not require scaling. |
| **"Why can high sample size be dangerous in interpretation?"** | It can make trivial effects look highly significant; importance still depends on effect size and impact. |


## One-Page Recap

**Always say**
- what the method is
- when to use it
- which assumptions matter
- what the result does and does not imply

**Fast lookup**
- 2 independent means -> Welch's t-test
- 2 paired measurements -> paired t-test
- 2 skewed / ordinal groups -> Mann-Whitney U
- 3+ means -> ANOVA
- 3+ skewed groups -> Kruskal-Wallis
- categorical association -> Chi-square / Fisher exact
- linear association -> Pearson
- monotonic association -> Spearman
- nonlinear feature relevance -> mutual information

**Lines that usually score well**
- "Statistical significance is not the same as business significance."
- "CLT is about the sampling distribution of the mean, not the raw data becoming normal."
- "A p-value is computed under the null; it is not the probability the null is true."
- "Confidence intervals are usually more decision-useful than p-values alone."
- "I choose the test based on target type, feature type, dependence structure, and robustness needs."
