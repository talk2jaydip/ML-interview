### **1. Word2Vec**

#### **Working:**
Word2Vec is based on **predictive models**. It learns word representations by predicting the context words (surrounding words) given a target word (or vice versa). There are two main architectures within Word2Vec:

- **Continuous Bag of Words (CBOW)**: Predicts the target word based on the context words surrounding it. For instance, in the sentence “The cat sat on the mat,” the context might be the words surrounding "sat," and the model learns to predict "sat" from those context words.
  
- **Skip-gram**: In this approach, the model does the inverse by predicting the surrounding context words given a target word. So, given the word “sat,” it tries to predict the surrounding words like “The,” “cat,” “on,” and “mat.”

The objective is to minimize the **loss function** (usually negative log-likelihood) based on the probability of predicting context words from the target word. The training uses a large corpus and updates the word embeddings iteratively.

#### **Key Points of Word2Vec**:
- **Training Objective**: Predict context words from a target word (Skip-gram) or predict a target word from context words (CBOW).
- **Optimization**: Typically uses negative sampling or hierarchical softmax to speed up training.
- **Local Context**: Focuses on the local context of words, which means that the model learns word embeddings by looking at a small window of surrounding words.

---

### **2. GloVe**

#### **Working:**
GloVe (Global Vectors for Word Representation) takes a different approach by modeling the **global co-occurrence statistics** of words in a corpus. Rather than predicting words given their context like in Word2Vec, GloVe creates word embeddings by capturing the relationships between words based on how often they co-occur across the entire corpus.

GloVe starts with a co-occurrence matrix $$ X $$ where each entry $$ X_{ij} $$ represents how often word $$ j $$ appears in the context of word $$ i $$ across the entire corpus. The basic idea behind GloVe is that the ratio of co-occurrence probabilities between words contains semantic relationships, so it factorizes the co-occurrence matrix into lower-dimensional matrices that represent words as vectors.

The key equation for GloVe is:

$$
\left( \mathbf{w}_i^\top \mathbf{w}_j \right) \approx \log X_{ij}
$$

Where:
- $$ \mathbf{w}_i $$ and $$ \mathbf{w}_j $$ are the word vectors for words $$ i $$ and $$ j $$,
- $$ X_{ij} $$ is the number of times word $$ j $$ appears in the context of word $$ i $$,
- $$ \log X_{ij} $$ is the logarithm of the co-occurrence probability.

The goal is to minimize the difference between the dot product of the word vectors and the logarithm of the co-occurrence probability, which captures the global statistical relationship between words.

#### **Key Points of GloVe**:
- **Training Objective**: Factorize the word co-occurrence matrix such that the dot product of word vectors approximates the log of the co-occurrence probabilities.
- **Optimization**: Minimizes a weighted least squares error.
- **Global Context**: Focuses on the entire corpus's word co-occurrence statistics rather than a local context window.

---

### **3. Triangular Differences (Relationships Between Word2Vec and GloVe)**

Both Word2Vec and GloVe aim to achieve a similar result: representing words as vectors in a continuous vector space such that semantically similar words are close to each other. However, they have different focuses and methods.

- **Word2Vec** learns word embeddings by **local context** (nearby words) and is **predictive**, while GloVe uses a **global co-occurrence matrix** and is more **statistical** in nature.
  
- In both models, **vector arithmetic** is often used to explore relationships between words. A popular example is the **"king - man + woman = queen"** analogy, where vector operations reveal semantic relationships between words.

- **Triangular Differences in Geometric Terms**:
  If we think of the word embeddings as points in a vector space, the geometric relationship between them can be described by **triangles** in that space. For example:
  
  - In **Word2Vec**, words that appear in similar contexts are closer together, forming a geometric structure where similar words (in terms of usage) will be clustered, and arithmetic between vectors captures the relationships (e.g., "king" - "man" + "woman" approximates "queen").
  
  - In **GloVe**, the relationship between word vectors also reflects the statistical structure of co-occurrence patterns. In a sense, GloVe models the **global statistical "triangles"** of word relationships through the co-occurrence matrix factorization.

Despite the differences in training methods, both Word2Vec and GloVe rely on the idea that word vectors capture meaningful geometric relationships (triangular differences) in a continuous space. They might learn slightly different relationships due to their methods, but both models are effective in capturing word semantics and can be used to explore analogies, similarities, and other relationships in the language.

---

### **Summary of Key Differences**:

| Feature              | **Word2Vec**                                      | **GloVe**                                        |
|----------------------|--------------------------------------------------|------------------------------------------------|
| **Approach**          | Predictive (local context)                       | Statistical (global co-occurrence)              |
| **Model Type**        | Neural network (Skip-gram or CBOW)               | Matrix factorization                             |
| **Focus**             | Local context relationships (context window)     | Global statistical relationships (co-occurrence)|
| **Objective**         | Maximize probability of context words             | Factorize co-occurrence matrix to minimize error|
| **Training Speed**    | Can be faster with negative sampling             | Slower due to matrix factorization              |

In terms of **triangular differences**, both models represent relationships between words as geometric differences in a continuous vector space, and vector arithmetic can be used to explore these relationships. Word2Vec focuses more on **local context**, whereas GloVe emphasizes **global co-occurrence** statistics, leading to different but complementary approaches to learning word embeddings.

##### LDA Topic Modeling: An Explanation
> Topic modeling is the process of identifying topics in a set of documents. This can be useful for search engines, customer service automation, and any other instance where knowing the topics of documents is important.

The Algorithm

<b> LDA is a form of unsupervised learning that views documents as bags of words (ie order does not matter). LDA works by first making a key assumption: the way a document was generated was by picking a set of topics and then for each topic picking a set of words</b>
<div class="alert alert-info">Document->Topic->Word</div>

1. Assume there are** k topics** across all of the documents
2. Distribute these k topics across **document m** (t**his distribution is known as α **and can be symmetric or asymmetric, more on this later) by assigning each word a topic.
3. For each **word w in document m**, assume its topic is wrong but every other word is assigned the correct topic.
4. Probabilistically assign word w a topic based on two things:
	. - what topics are in document m
 	- **how many times word w has been assigned a particular topic across all of the documents (this distribution is called β, more on this late**r)
	
7. Repeat this process a number of times for each document and you’re done!





- α is the per-document topic distributions,
- β is the per-topic word distribution,
- θ is the topic distribution for document m,
- φ is the word distribution for topic k,
- z is the topic for the n-th word in document m, and
- w is the specific word


- α is a matrix where each row is a document and each column represents a topic.
- **DocumentXTopic** 
> A value in row i and column j represents how likely document i contains topic j.

- β is a matrix where each row represents a topic and each column represents a word
- **TopicXWord**
> A value in row i and column j represents how likely that topic i contains word j

##### What is the big idea behind LDA?


> Each document can be described by a distribution of topics and each topic can be described by a distribution of words

1. α — Distribution related parameter that governs what the distribution of topics is for all the documents in the corpus looks like
2. θ — Random matrix where θ(i,j) represents the probability of the i th document to containing the j th topic
3. η — Distribution related parameter that governs what the distribution of words in each topic looks like
4. β — A random matrix where β(i,j) represents the probability of i th topic containing the j th word.
